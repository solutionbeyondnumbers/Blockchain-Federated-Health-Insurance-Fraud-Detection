```python
# ============================================================
# CELL 01/15
# PROJECT UPLOAD, DRIVE SETUP AND ENVIRONMENT VERIFICATION
# Reviewer: Major Comment 7
# ============================================================

import os
import sys
import json
import shutil
import zipfile
import hashlib
import platform
import subprocess
import psutil
from pathlib import Path
from datetime import datetime
from google.colab import drive, files

# ------------------------------------------------------------
# 1. Mount Google Drive
# ------------------------------------------------------------

drive.mount("/content/drive")

PROJECT_ROOT = Path(
    "/content/drive/MyDrive/Narayan_Patil_Reviewer_Revision"
)

ORIGINAL_DIR = PROJECT_ROOT / "01_Original_Project"
REVISED_DIR = PROJECT_ROOT / "02_Revised_Project"
CODE_DIR = REVISED_DIR / "code"
DATA_DIR = REVISED_DIR / "data"
MODEL_DIR = REVISED_DIR / "models"
RESULT_DIR = REVISED_DIR / "results"
TABLE_DIR = RESULT_DIR / "tables"
FIGURE_DIR = RESULT_DIR / "figures"
REPORT_DIR = REVISED_DIR / "reports"
CHECKPOINT_DIR = REVISED_DIR / "checkpoints"
ARCHIVE_DIR = PROJECT_ROOT / "00_Original_Archive"

directories = [
    ORIGINAL_DIR,
    REVISED_DIR,
    CODE_DIR,
    DATA_DIR,
    MODEL_DIR,
    RESULT_DIR,
    TABLE_DIR,
    FIGURE_DIR,
    REPORT_DIR,
    CHECKPOINT_DIR,
    ARCHIVE_DIR
]

for directory in directories:
    directory.mkdir(parents=True, exist_ok=True)

print("=" * 70)
print("GOOGLE DRIVE WORKSPACE CREATED")
print("=" * 70)
print(PROJECT_ROOT)

# ------------------------------------------------------------
# 2. Verify hardware configuration
# ------------------------------------------------------------

gpu_check = subprocess.run(
    [
        "nvidia-smi",
        "--query-gpu=name,memory.total,driver_version",
        "--format=csv,noheader"
    ],
    capture_output=True,
    text=True
) if shutil.which("nvidia-smi") else None

gpu_info = (
    gpu_check.stdout.strip()
    if gpu_check and gpu_check.returncode == 0
    else "GPU not available"
)

cpu_model = platform.processor()

if Path("/proc/cpuinfo").exists():
    for line in Path("/proc/cpuinfo").read_text().splitlines():
        if line.startswith("model name"):
            cpu_model = line.split(":", 1)[1].strip()
            break

environment = {
    "timestamp": datetime.now().isoformat(),
    "python_version": sys.version,
    "operating_system": platform.platform(),
    "cpu_model": cpu_model,
    "physical_cpu_cores": psutil.cpu_count(logical=False),
    "logical_cpu_cores": psutil.cpu_count(logical=True),
    "ram_gb": round(
        psutil.virtual_memory().total / 1024**3, 2
    ),
    "gpu": gpu_info,
    "planned_random_seeds": [42, 52, 62, 72, 82]
}

print("\nHARDWARE CONFIGURATION")
print("-" * 70)

for key, value in environment.items():
    print(f"{key}: {value}")

if gpu_info == "GPU not available":
    raise RuntimeError(
        "Enable T4 GPU in Colab before continuing."
    )

# ------------------------------------------------------------
# 3. Upload original project ZIP
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("UPLOAD THE COMPLETE ORIGINAL PROJECT ZIP")
print("=" * 70)

uploaded = files.upload()

zip_files = [
    name for name in uploaded
    if name.lower().endswith(".zip")
]

if len(zip_files) != 1:
    raise ValueError(
        "Please upload exactly one ZIP archive."
    )

archive_name = zip_files[0]
local_archive = Path("/content") / archive_name
drive_archive = ARCHIVE_DIR / archive_name

# Save original archive without overwriting another version
if drive_archive.exists():
    if hashlib.sha256(
        drive_archive.read_bytes()
    ).hexdigest() != hashlib.sha256(
        local_archive.read_bytes()
    ).hexdigest():
        raise FileExistsError(
            "A different archive with the same name exists."
        )
else:
    shutil.copy2(local_archive, drive_archive)

print(f"\nOriginal archive saved: {drive_archive}")

# ------------------------------------------------------------
# 4. Verify ZIP integrity and extract
# ------------------------------------------------------------

archive_hash = hashlib.sha256(
    drive_archive.read_bytes()
).hexdigest()

with zipfile.ZipFile(local_archive, "r") as z:

    corrupted_file = z.testzip()

    if corrupted_file is not None:
        raise RuntimeError(
            f"Corrupted archive member: {corrupted_file}"
        )

    # Prevent unsafe archive paths
    for member in z.infolist():
        target = (ORIGINAL_DIR / member.filename).resolve()

        if not target.is_relative_to(ORIGINAL_DIR.resolve()):
            raise RuntimeError(
                f"Unsafe archive path: {member.filename}"
            )

    if not any(ORIGINAL_DIR.iterdir()):
        z.extractall(ORIGINAL_DIR)
        print("Archive successfully extracted.")
    else:
        print("Existing original extraction preserved.")

print("ZIP integrity: PASS")
print("SHA-256:", archive_hash)

# ------------------------------------------------------------
# 5. Inspect project contents
# ------------------------------------------------------------

all_files = [
    p for p in ORIGINAL_DIR.rglob("*")
    if p.is_file()
]

python_files = [
    p for p in all_files if p.suffix.lower() == ".py"
]

csv_files = [
    p for p in all_files if p.suffix.lower() == ".csv"
]

notebooks = [
    p for p in all_files if p.suffix.lower() == ".ipynb"
]

model_files = [
    p for p in all_files
    if p.suffix.lower() in [
        ".pkl", ".joblib", ".cbm", ".ubj",
        ".model", ".pt", ".pth"
    ]
]

requirements_files = [
    p for p in all_files
    if p.name.lower() == "requirements.txt"
]

print("\n" + "=" * 70)
print("ORIGINAL PROJECT INVENTORY")
print("=" * 70)

print("Total files        :", len(all_files))
print("Python scripts     :", len(python_files))
print("CSV datasets       :", len(csv_files))
print("Jupyter notebooks  :", len(notebooks))
print("Saved models       :", len(model_files))
print("Requirements files :", len(requirements_files))

print("\nPYTHON SCRIPTS:")
print("-" * 70)

for p in sorted(python_files):
    print(p.relative_to(ORIGINAL_DIR))

print("\nDATASET FILES:")
print("-" * 70)

for p in sorted(csv_files):
    size_mb = p.stat().st_size / 1024**2

    print(
        f"{p.relative_to(ORIGINAL_DIR)}"
        f" | {size_mb:.2f} MB"
    )

# ------------------------------------------------------------
# 6. Save environment and project manifest
# ------------------------------------------------------------

environment["archive_name"] = archive_name
environment["archive_sha256"] = archive_hash

with open(
    REPORT_DIR / "cell01_environment.json", "w"
) as f:
    json.dump(environment, f, indent=4)

manifest = [
    {
        "path": str(p.relative_to(ORIGINAL_DIR)),
        "size_bytes": p.stat().st_size
    }
    for p in sorted(all_files)
]

with open(
    REPORT_DIR / "cell01_original_file_manifest.json", "w"
) as f:
    json.dump(manifest, f, indent=4)

# ------------------------------------------------------------
# 7. Save execution checkpoint
# ------------------------------------------------------------

checkpoint = {
    "cell": 1,
    "status": "COMPLETED",
    "reviewer_comment": "Major Comment 7",
    "original_project": str(ORIGINAL_DIR),
    "revised_project": str(REVISED_DIR),
    "archive_sha256": archive_hash,
    "python_scripts": len(python_files),
    "csv_files": len(csv_files),
    "gpu": gpu_info,
    "original_files_modified": False
}

with open(
    CHECKPOINT_DIR / "cell01_completed.json", "w"
) as f:
    json.dump(checkpoint, f, indent=4)

print("\n" + "=" * 70)
print("CELL 01 COMPLETED SUCCESSFULLY")
print("=" * 70)
print("Original project preserved.")
print("Archive and reports saved in Google Drive.")
print("No original code modified.")
print("No models retrained.")
print("No additional experiments performed.")
print("\nREADY FOR CELL 02: ORIGINAL CODE AND DATA AUDIT")
```

```python
# ============================================================
# CELL 01 - CORRECTED CONTINUATION
# SUPPORT RAR AND ZIP ARCHIVES
# Reviewer: Major Comment 7
# ============================================================

import os
import sys
import json
import shutil
import zipfile
import hashlib
import subprocess
import ast

from pathlib import Path

# ------------------------------------------------------------
# 1. Locate already uploaded archive
# ------------------------------------------------------------

archive_candidates = [
    Path("/content/Narayan_Patil.rar"),
    Path("/content/Narayan_Patil.zip")
]

local_archive = next(
    (p for p in archive_candidates if p.is_file()),
    None
)

if local_archive is None:
    raise FileNotFoundError(
        "Previously uploaded archive not found."
    )

archive_name = local_archive.name

print("Archive located:", local_archive)
print(
    "Archive size:",
    round(local_archive.stat().st_size / 1024**2, 2),
    "MB"
)

# ------------------------------------------------------------
# 2. Calculate archive hash
# ------------------------------------------------------------

def calculate_sha256(path):
    sha256 = hashlib.sha256()

    with open(path, "rb") as f:
        for chunk in iter(
            lambda: f.read(1024 * 1024), b""
        ):
            sha256.update(chunk)

    return sha256.hexdigest()


archive_hash = calculate_sha256(local_archive)

# ------------------------------------------------------------
# 3. Preserve original archive in Google Drive
# ------------------------------------------------------------

drive_archive = ARCHIVE_DIR / archive_name

if drive_archive.exists():

    if calculate_sha256(drive_archive) != archive_hash:
        raise RuntimeError(
            "A different archive with the same name "
            "already exists in Google Drive."
        )

    print("Original archive already preserved.")

else:
    shutil.copy2(local_archive, drive_archive)

assert calculate_sha256(drive_archive) == archive_hash

print("Archive backup verified.")
print("SHA-256:", archive_hash)

# ------------------------------------------------------------
# 4. Install RAR extraction support if required
# ------------------------------------------------------------

if local_archive.suffix.lower() == ".rar":

    if shutil.which("unrar") is None:

        print("\nInstalling RAR extraction utility...")

        subprocess.run(
            ["apt-get", "update", "-qq"],
            check=True
        )

        subprocess.run(
            ["apt-get", "install", "-y", "-qq", "unrar"],
            check=True
        )

    if shutil.which("unrar") is None:
        raise RuntimeError("RAR extraction utility unavailable.")

    print("RAR extraction utility: READY")

# ------------------------------------------------------------
# 5. Verify archive integrity
# ------------------------------------------------------------

print("\nVERIFYING ARCHIVE INTEGRITY")

if local_archive.suffix.lower() == ".rar":

    test_result = subprocess.run(
        ["unrar", "t", "-p-", str(local_archive)],
        capture_output=True,
        text=True
    )

    if test_result.returncode != 0:
        raise RuntimeError(
            "RAR integrity verification failed:\n"
            + test_result.stdout
            + test_result.stderr
        )

else:

    with zipfile.ZipFile(local_archive, "r") as z:

        corrupted_file = z.testzip()

        if corrupted_file is not None:
            raise RuntimeError(
                f"Corrupted file: {corrupted_file}"
            )

print("Archive integrity: PASS")

# ------------------------------------------------------------
# 6. Extract into temporary workspace
# ------------------------------------------------------------

TEMP_EXTRACT = Path(
    "/content/Narayan_Patil_Original_Extraction"
)

if TEMP_EXTRACT.exists():
    shutil.rmtree(TEMP_EXTRACT)

TEMP_EXTRACT.mkdir(parents=True, exist_ok=True)

print("\nEXTRACTING ORIGINAL PROJECT")

if local_archive.suffix.lower() == ".rar":

    subprocess.run(
        [
            "unrar",
            "x",
            "-o-",
            "-p-",
            str(local_archive),
            str(TEMP_EXTRACT) + "/"
        ],
        check=True,
        stdout=subprocess.DEVNULL
    )

else:

    with zipfile.ZipFile(local_archive, "r") as z:

        for member in z.infolist():

            target = (
                TEMP_EXTRACT / member.filename
            ).resolve()

            if not target.is_relative_to(
                TEMP_EXTRACT.resolve()
            ):
                raise RuntimeError(
                    f"Unsafe archive path: {member.filename}"
                )

        z.extractall(TEMP_EXTRACT)

print("Extraction: SUCCESS")

# ------------------------------------------------------------
# 7. Preserve extracted original project in Drive
# ------------------------------------------------------------

if not any(ORIGINAL_DIR.iterdir()):

    shutil.copytree(
        TEMP_EXTRACT,
        ORIGINAL_DIR,
        dirs_exist_ok=True
    )

    print("Original project saved in Google Drive.")

else:

    print("Verifying existing original project...")

    extracted_files = [
        p for p in TEMP_EXTRACT.rglob("*")
        if p.is_file()
    ]

    for source in extracted_files:

        destination = (
            ORIGINAL_DIR
            / source.relative_to(TEMP_EXTRACT)
        )

        if not destination.is_file():
            raise RuntimeError(
                f"Missing original file: {destination}"
            )

        if calculate_sha256(source) != calculate_sha256(
            destination
        ):
            raise RuntimeError(
                f"Original file mismatch: {destination}"
            )

    print("Existing original project verified.")

# ------------------------------------------------------------
# 8. Project inventory
# ------------------------------------------------------------

all_files = [
    p for p in ORIGINAL_DIR.rglob("*")
    if p.is_file()
]

python_files = [
    p for p in all_files
    if p.suffix.lower() == ".py"
]

csv_files = [
    p for p in all_files
    if p.suffix.lower() == ".csv"
]

notebooks = [
    p for p in all_files
    if p.suffix.lower() == ".ipynb"
]

model_files = [
    p for p in all_files
    if p.suffix.lower() in [
        ".pkl", ".joblib", ".cbm",
        ".ubj", ".model", ".pt", ".pth"
    ]
]

requirements_files = [
    p for p in all_files
    if p.name.lower() == "requirements.txt"
]

print("\n" + "=" * 70)
print("ORIGINAL PROJECT INVENTORY")
print("=" * 70)

print("Total files        :", len(all_files))
print("Python scripts     :", len(python_files))
print("CSV files          :", len(csv_files))
print("Notebooks          :", len(notebooks))
print("Saved models       :", len(model_files))
print("Requirements files :", len(requirements_files))

# ------------------------------------------------------------
# 9. Python syntax verification
# ------------------------------------------------------------

syntax_results = []

print("\nPYTHON SCRIPT VERIFICATION")
print("-" * 70)

for p in sorted(python_files):

    try:

        ast.parse(
            p.read_text(
                encoding="utf-8-sig"
            )
        )

        status = "PASS"
        error = None

    except Exception as e:

        status = "FAIL"
        error = str(e)

    relative_path = str(
        p.relative_to(ORIGINAL_DIR)
    )

    syntax_results.append({
        "file": relative_path,
        "status": status,
        "error": error
    })

    print(f"{status} | {relative_path}")

syntax_passed = sum(
    item["status"] == "PASS"
    for item in syntax_results
)

print(
    "\nSyntax verification:",
    syntax_passed,
    "/",
    len(python_files)
)

# ------------------------------------------------------------
# 10. Dataset inventory
# ------------------------------------------------------------

print("\nDATASET FILES")
print("-" * 70)

for p in sorted(csv_files):

    print(
        f"{p.relative_to(ORIGINAL_DIR)}"
        f" | {p.stat().st_size / 1024**2:.2f} MB"
    )

print("\nSAVED MODEL FILES")
print("-" * 70)

for p in sorted(model_files):
    print(p.relative_to(ORIGINAL_DIR))

# ------------------------------------------------------------
# 11. Save environment and audit reports
# ------------------------------------------------------------

environment["archive_name"] = archive_name
environment["archive_sha256"] = archive_hash
environment["archive_format"] = local_archive.suffix

manifest = [
    {
        "path": str(p.relative_to(ORIGINAL_DIR)),
        "size_bytes": p.stat().st_size
    }
    for p in sorted(all_files)
]

reports = {
    "cell01_environment.json": environment,
    "cell01_original_file_manifest.json": manifest,
    "cell01_python_syntax_audit.json": syntax_results
}

for filename, content in reports.items():

    with open(
        REPORT_DIR / filename,
        "w",
        encoding="utf-8"
    ) as f:

        json.dump(content, f, indent=4)

# ------------------------------------------------------------
# 12. Save checkpoint
# ------------------------------------------------------------

checkpoint = {
    "cell": 1,
    "status": "COMPLETED",
    "reviewer_comment": "Major Comment 7",
    "original_project": str(ORIGINAL_DIR),
    "revised_project": str(REVISED_DIR),
    "archive_name": archive_name,
    "archive_sha256": archive_hash,
    "archive_integrity": "PASS",
    "python_scripts": len(python_files),
    "syntax_passed": syntax_passed,
    "csv_files": len(csv_files),
    "saved_models": len(model_files),
    "gpu": gpu_info,
    "original_files_modified": False
}

with open(
    CHECKPOINT_DIR / "cell01_completed.json",
    "w",
    encoding="utf-8"
) as f:

    json.dump(checkpoint, f, indent=4)

print("\n" + "=" * 70)
print("CELL 01 COMPLETED SUCCESSFULLY")
print("=" * 70)

print("Original archive preserved.")
print("Original project extracted.")
print("Python scripts inspected.")
print("Audit reports saved in Google Drive.")
print("Checkpoint saved.")
print("No original code modified.")
print("No models retrained.")

print("\nREADY FOR CELL 02/15")
```

```python
# ============================================================
# CELL 02/15
# ORIGINAL CODE, DATASET AND METHODOLOGY AUDIT
#
# Reviewer Comments:
# Major 5, 7, 10
# Minor 26, 27
# ============================================================

import os
import sys
import ast
import json
import hashlib
import shutil
import platform
import importlib.metadata as metadata
from pathlib import Path

import numpy as np
import pandas as pd

# ------------------------------------------------------------
# 1. Restore project paths
# ------------------------------------------------------------

PROJECT_ROOT = Path(
    "/content/drive/MyDrive/Narayan_Patil_Reviewer_Revision"
)

ORIGINAL_DIR = PROJECT_ROOT / "01_Original_Project"
REVISED_DIR = PROJECT_ROOT / "02_Revised_Project"

SOURCE_ROOT = ORIGINAL_DIR / "SBN"

RAW_DIR = SOURCE_ROOT / "data" / "raw"
PROCESSED_DIR = SOURCE_ROOT / "data" / "processed"

CODE_DIR = REVISED_DIR / "code"
DATA_DIR = REVISED_DIR / "data"
MODEL_DIR = REVISED_DIR / "models"
RESULT_DIR = REVISED_DIR / "results"
TABLE_DIR = RESULT_DIR / "tables"
FIGURE_DIR = RESULT_DIR / "figures"
REPORT_DIR = REVISED_DIR / "reports"
CHECKPOINT_DIR = REVISED_DIR / "checkpoints"

assert SOURCE_ROOT.exists(), "Original project not found."

assert (
    CHECKPOINT_DIR / "cell01_completed.json"
).exists(), "Cell 01 checkpoint missing."

print("=" * 75)
print("CELL 02: ORIGINAL PROJECT AUDIT")
print("=" * 75)

print("Original project:", SOURCE_ROOT)
print("Revised project :", REVISED_DIR)

# ------------------------------------------------------------
# 2. Copy original source code into revision workspace
# ------------------------------------------------------------

SOURCE_SNAPSHOT = CODE_DIR / "original_source_snapshot"

SOURCE_SNAPSHOT.mkdir(
    parents=True,
    exist_ok=True
)

def sha256_file(path):

    h = hashlib.sha256()

    with open(path, "rb") as f:

        for chunk in iter(
            lambda: f.read(1024 * 1024), b""
        ):
            h.update(chunk)

    return h.hexdigest()

source_manifest = []

for source in sorted(SOURCE_ROOT.rglob("*.py")):

    relative_path = source.relative_to(SOURCE_ROOT)

    destination = SOURCE_SNAPSHOT / relative_path

    destination.parent.mkdir(
        parents=True,
        exist_ok=True
    )

    if not destination.exists():
        shutil.copy2(source, destination)

    assert sha256_file(source) == sha256_file(destination), (
        f"Source snapshot mismatch: {relative_path}"
    )

    source_manifest.append({
        "file": str(relative_path),
        "sha256": sha256_file(source),
        "size_bytes": source.stat().st_size
    })

print("\nOriginal source scripts preserved:",
      len(source_manifest))

# ------------------------------------------------------------
# 3. Inspect original preprocessing and training scripts
# ------------------------------------------------------------

audit_scripts = [
    "preprocessing/load_data.py",
    "preprocessing/merge_data.py",
    "preprocessing/clean_data.py",
    "preprocessing/feature_engineering.py",
    "preprocessing/encode_and_balance.py",
    "preprocessing/train_test_split.py",
    "models/train_lightgbm.py",
    "models/train_catboost.py",
    "models/train_xgboost.py",
    "models/train_model.py",
    "evaluation/evaluate_model.py",
    "blockchain/block.py",
    "blockchain/blockchain.py",
    "blockchain/fraud_chain.py",
    "explainability/shap_analysis.py"
]

script_contents = {}

print("\n" + "=" * 75)
print("ORIGINAL IMPLEMENTATION")
print("=" * 75)

for relative_path in audit_scripts:

    path = SOURCE_ROOT / relative_path

    if not path.exists():

        print("\nMISSING:", relative_path)
        continue

    content = path.read_text(
        encoding="utf-8-sig"
    )

    script_contents[relative_path] = content

    print("\n" + "-" * 75)
    print("FILE:", relative_path)
    print("-" * 75)

    for number, line in enumerate(
        content.splitlines(), start=1
    ):
        print(f"{number:03d} | {line}")

# ------------------------------------------------------------
# 4. Inspect raw dataset dimensions and missing values
# ------------------------------------------------------------

raw_inventory = []
missing_reports = []

print("\n" + "=" * 75)
print("RAW DATASET AUDIT")
print("=" * 75)

for path in sorted(RAW_DIR.glob("*.csv")):

    df = pd.read_csv(
        path,
        low_memory=False
    )

    missing_count = df.isna().sum()

    total_cells = df.shape[0] * df.shape[1]

    missing_percentage = (
        missing_count.sum() / max(total_cells, 1)
    ) * 100

    record = {
        "file": path.name,
        "rows": len(df),
        "columns": df.shape[1],
        "missing_values": int(missing_count.sum()),
        "missing_percentage": float(missing_percentage),
        "duplicate_rows": int(df.duplicated().sum())
    }

    raw_inventory.append(record)

    for column in df.columns:

        missing_reports.append({
            "file": path.name,
            "column": column,
            "missing_count": int(missing_count[column]),
            "missing_percentage": float(
                missing_count[column] /
                max(len(df), 1) * 100
            )
        })

    print(
        f"\n{path.name}"
        f"\nRows: {len(df):,}"
        f" | Columns: {df.shape[1]}"
        f" | Missing: {missing_percentage:.4f}%"
        f" | Duplicates: {record['duplicate_rows']:,}"
    )

    if "PotentialFraud" in df.columns:

        print("Fraud distribution:")

        print(
            df["PotentialFraud"]
            .value_counts(dropna=False)
            .to_string()
        )

    del df

raw_inventory_df = pd.DataFrame(raw_inventory)

missing_report_df = pd.DataFrame(missing_reports)

raw_inventory_df.to_csv(
    TABLE_DIR / "cell02_raw_dataset_inventory.csv",
    index=False
)

missing_report_df.to_csv(
    TABLE_DIR / "cell02_raw_missing_values.csv",
    index=False
)

# ------------------------------------------------------------
# 5. Identify fraud target column
# ------------------------------------------------------------

TARGET_CANDIDATES = [
    "PotentialFraud",
    "FraudLabel",
    "Fraud",
    "fraud",
    "fraud_label",
    "target",
    "Target",
    "label"
]

def detect_target(columns):

    lookup = {
        str(c).strip().lower(): c
        for c in columns
    }

    for candidate in TARGET_CANDIDATES:

        if candidate.lower() in lookup:
            return lookup[candidate.lower()]

    return None

# ------------------------------------------------------------
# 6. Audit original processed datasets using chunks
# ------------------------------------------------------------

processed_inventory = []

processed_columns = {}

processed_distributions = {}

print("\n" + "=" * 75)
print("ORIGINAL PROCESSED DATASET AUDIT")
print("=" * 75)

for path in sorted(PROCESSED_DIR.glob("*.csv")):

    header = pd.read_csv(
        path,
        nrows=0
    )

    columns = header.columns.tolist()

    processed_columns[path.name] = columns

    target = detect_target(columns)

    if path.stem in ["y_train", "y_test"]:

        if len(columns) == 1:
            target = columns[0]

    row_count = 0
    total_missing = 0
    distribution = {}

    for chunk in pd.read_csv(
        path,
        chunksize=50000,
        low_memory=False
    ):

        row_count += len(chunk)

        total_missing += int(
            chunk.isna().sum().sum()
        )

        if target is not None:

            counts = (
                chunk[target]
                .astype(str)
                .value_counts(dropna=False)
            )

            for label, count in counts.items():

                distribution[label] = (
                    distribution.get(label, 0)
                    + int(count)
                )

    record = {
        "file": path.name,
        "rows": row_count,
        "columns": len(columns),
        "missing_values": total_missing,
        "target_column": target,
        "file_size_mb": round(
            path.stat().st_size / 1024**2,
            2
        )
    }

    processed_inventory.append(record)

    processed_distributions[path.name] = distribution

    print("\nFile:", path.name)
    print("Rows:", f"{row_count:,}")
    print("Columns:", len(columns))
    print("Missing values:", f"{total_missing:,}")

    if distribution:

        print("Target:", target)
        print("Class distribution:", distribution)

processed_inventory_df = pd.DataFrame(
    processed_inventory
)

processed_inventory_df.to_csv(
    TABLE_DIR / "cell02_processed_dataset_inventory.csv",
    index=False
)

# ------------------------------------------------------------
# 7. Verify manuscript dataset dimensions
# ------------------------------------------------------------

expected_dimensions = {
    "master_dataset.csv": (558211, 55),
    "cleaned_dataset.csv": (558211, 38),
    "featured_dataset.csv": (558211, 43),
    "balanced_dataset.csv": (690830, None),
    "X_train.csv": (552664, None),
    "X_test.csv": (138166, None)
}

dimension_checks = []

print("\n" + "=" * 75)
print("MANUSCRIPT DATASET DIMENSION VERIFICATION")
print("=" * 75)

for filename, expected in expected_dimensions.items():

    matching = processed_inventory_df[
        processed_inventory_df["file"] == filename
    ]

    if matching.empty:

        dimension_checks.append({
            "file": filename,
            "status": "MISSING"
        })

        print(filename, ": MISSING")
        continue

    actual_rows = int(
        matching.iloc[0]["rows"]
    )

    actual_cols = int(
        matching.iloc[0]["columns"]
    )

    expected_rows, expected_cols = expected

    match = (
        actual_rows == expected_rows
        and (
            expected_cols is None
            or actual_cols == expected_cols
        )
    )

    status = "MATCH" if match else "MISMATCH"

    dimension_checks.append({
        "file": filename,
        "expected_rows": expected_rows,
        "actual_rows": actual_rows,
        "expected_columns": expected_cols,
        "actual_columns": actual_cols,
        "status": status
    })

    print(
        f"{status} | {filename}"
        f" | Rows: {actual_rows:,}"
        f" | Columns: {actual_cols}"
    )

pd.DataFrame(dimension_checks).to_csv(
    TABLE_DIR / "cell02_manuscript_dimension_checks.csv",
    index=False
)

# ------------------------------------------------------------
# 8. Identify removed and engineered features
# ------------------------------------------------------------

master_cols = processed_columns.get(
    "master_dataset.csv", []
)

cleaned_cols = processed_columns.get(
    "cleaned_dataset.csv", []
)

featured_cols = processed_columns.get(
    "featured_dataset.csv", []
)

removed_columns = [
    c for c in master_cols
    if c not in cleaned_cols
]

engineered_columns = [
    c for c in featured_cols
    if c not in cleaned_cols
]

print("\n" + "=" * 75)
print("FEATURE SELECTION AND ENGINEERING AUDIT")
print("=" * 75)

print("\nRemoved columns:")

for column in removed_columns:
    print(" -", column)

print("\nEngineered columns:")

for column in engineered_columns:
    print(" +", column)

feature_audit = []

for column in removed_columns:

    feature_audit.append({
        "column": column,
        "operation": "removed"
    })

for column in engineered_columns:

    feature_audit.append({
        "column": column,
        "operation": "engineered"
    })

pd.DataFrame(
    feature_audit,
    columns=["column", "operation"]
).to_csv(
    TABLE_DIR / "cell02_feature_transformation_audit.csv",
    index=False
)

# ------------------------------------------------------------
# 9. Identify original identifier features
# ------------------------------------------------------------

original_predictors = processed_columns.get(
    "X_train.csv", []
)

identifier_keywords = [
    "provider",
    "beneid",
    "claimid",
    "physician",
    "patientid"
]

identifier_features = [
    column
    for column in original_predictors
    if any(
        keyword in column.lower()
        for keyword in identifier_keywords
    )
]

print("\nPOTENTIAL IDENTIFIER-RELATED FEATURES:")

for column in identifier_features:
    print(" -", column)

print(
    "\nNote: Identifier-related features require "
    "individual leakage assessment."
)

# ------------------------------------------------------------
# 10. Audit original model hyperparameters using AST
# ------------------------------------------------------------

model_constructors = {
    "XGBClassifier",
    "LGBMClassifier",
    "CatBoostClassifier"
}

model_hyperparameters = []

print("\n" + "=" * 75)
print("ORIGINAL MODEL HYPERPARAMETERS")
print("=" * 75)

for path in sorted(
    (SOURCE_ROOT / "models").glob("*.py")
):

    content = path.read_text(
        encoding="utf-8-sig"
    )

    tree = ast.parse(content)

    for node in ast.walk(tree):

        if not isinstance(node, ast.Call):
            continue

        if isinstance(node.func, ast.Name):

            model_name = node.func.id

        elif isinstance(node.func, ast.Attribute):

            model_name = node.func.attr

        else:
            continue

        if model_name not in model_constructors:
            continue

        parameters = {}

        for keyword in node.keywords:

            key = keyword.arg or "**kwargs"

            try:
                value = ast.literal_eval(
                    keyword.value
                )
            except Exception:
                value = ast.unparse(
                    keyword.value
                )

            parameters[key] = value

        model_hyperparameters.append({
            "script": path.name,
            "model": model_name,
            "line": node.lineno,
            "parameters": parameters
        })

        print("\nModel:", model_name)
        print("Script:", path.name)

        for key, value in parameters.items():
            print(f"  {key}: {value}")

# ------------------------------------------------------------
# 11. Identify reviewer-related implementation evidence
# ------------------------------------------------------------

audit_keywords = {
    "SMOTE": ["SMOTE", "fit_resample"],
    "Data splitting": [
        "train_test_split",
        "stratify",
        "GroupShuffleSplit",
        "StratifiedGroupKFold"
    ],
    "Scaling": [
        "StandardScaler",
        "fit_transform",
        "MinMaxScaler"
    ],
    "Feature selection": [
        "SelectKBest",
        "mutual_info_classif",
        "RFE(",
        "SelectFromModel",
        "dropna",
        ".drop("
    ],
    "Federated learning": [
        "FedAvg",
        "federated",
        "client_models",
        "communication_round"
    ],
    "Differential privacy": [
        "PrivacyEngine",
        "privacy_accountant",
        "noise_multiplier",
        "differential_privacy"
    ],
    "Blockchain consensus": [
        "PBFT",
        "ProofOfAuthority",
        "consensus",
        "merkle"
    ],
    "Statistical testing": [
        "wilcoxon",
        "ttest_rel",
        "bootstrap"
    ]
}

keyword_evidence = []

print("\n" + "=" * 75)
print("REVIEWER-RELATED SOURCE CODE EVIDENCE")
print("=" * 75)

for path in sorted(SOURCE_ROOT.rglob("*.py")):

    lines = path.read_text(
        encoding="utf-8-sig"
    ).splitlines()

    for category, keywords in audit_keywords.items():

        for line_no, line in enumerate(
            lines, start=1
        ):

            if any(
                keyword.lower() in line.lower()
                for keyword in keywords
            ):

                keyword_evidence.append({
                    "category": category,
                    "file": str(
                        path.relative_to(SOURCE_ROOT)
                    ),
                    "line": line_no,
                    "code": line.strip()
                })

evidence_df = pd.DataFrame(
    keyword_evidence,
    columns=[
        "category", "file", "line", "code"
    ]
)

evidence_df.to_csv(
    TABLE_DIR / "cell02_source_code_evidence.csv",
    index=False
)

for category in audit_keywords:

    matches = evidence_df[
        evidence_df["category"] == category
    ]

    print(
        f"{category:25s}: "
        f"{len(matches)} source references"
    )

# ------------------------------------------------------------
# 12. Original SMOTE and test distribution audit
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("ORIGINAL CLASS BALANCING AUDIT")
print("=" * 75)

for filename in [
    "master_dataset.csv",
    "balanced_dataset.csv",
    "y_train.csv",
    "y_test.csv"
]:

    print(
        filename,
        ":",
        processed_distributions.get(
            filename, {}
        )
    )

print(
    "\nOriginal preprocessing order must be determined "
    "from encode_and_balance.py and train_test_split.py."
)

print(
    "A balanced test set alone is not proof of leakage; "
    "the preprocessing source establishes operation order."
)

# ------------------------------------------------------------
# 13. Record currently installed library versions
# ------------------------------------------------------------

packages = [
    "numpy",
    "pandas",
    "scipy",
    "scikit-learn",
    "imbalanced-learn",
    "xgboost",
    "lightgbm",
    "catboost",
    "torch",
    "shap",
    "matplotlib",
    "joblib",
    "opacus",
    "cryptography"
]

library_versions = {}

print("\n" + "=" * 75)
print("LIBRARY VERSION AUDIT")
print("=" * 75)

for package in packages:

    try:

        version = metadata.version(package)

    except metadata.PackageNotFoundError:

        version = "NOT INSTALLED"

    library_versions[package] = version

    print(f"{package:22s}: {version}")

# ------------------------------------------------------------
# 14. Save complete audit reports to Google Drive
# ------------------------------------------------------------

audit_report = {
    "cell": 2,
    "original_source": str(SOURCE_ROOT),
    "source_scripts": len(source_manifest),
    "reviewer_comments": [
        "Major 5",
        "Major 7",
        "Major 10",
        "Minor 26",
        "Minor 27"
    ],
    "raw_datasets": raw_inventory,
    "processed_datasets": processed_inventory,
    "processed_columns": processed_columns,
    "class_distributions": processed_distributions,
    "dimension_checks": dimension_checks,
    "removed_columns": removed_columns,
    "engineered_columns": engineered_columns,
    "identifier_related_features": identifier_features,
    "model_hyperparameters": model_hyperparameters,
    "library_versions": library_versions,
    "original_files_modified": False,
    "model_training_performed": False,
    "numerical_reproduction_performed": False
}

reports = {
    "cell02_complete_original_audit.json": audit_report,
    "cell02_original_source_hashes.json": source_manifest,
    "cell02_original_hyperparameters.json": model_hyperparameters,
    "cell02_library_versions.json": library_versions,
    "cell02_original_script_contents.json": script_contents
}

for filename, content in reports.items():

    with open(
        REPORT_DIR / filename,
        "w",
        encoding="utf-8"
    ) as f:

        json.dump(
            content,
            f,
            indent=4,
            default=str
        )

# Save the Cell 02 source code itself
ip = get_ipython()

if ip is not None:

    cell_source = ip.history_manager.input_hist_raw[-1]

    (
        CODE_DIR / "cell02_original_audit.py"
    ).write_text(
        cell_source,
        encoding="utf-8"
    )

# ------------------------------------------------------------
# 15. Verify original scripts were not modified
# ------------------------------------------------------------

for item in source_manifest:

    path = SOURCE_ROOT / item["file"]

    assert sha256_file(path) == item["sha256"], (
        f"Original source changed: {path}"
    )

print("\nOriginal source integrity: PASS")

# ------------------------------------------------------------
# 16. Save checkpoint
# ------------------------------------------------------------

checkpoint = {
    "cell": 2,
    "status": "COMPLETED",
    "source_scripts_audited": len(source_manifest),
    "raw_datasets_audited": len(raw_inventory),
    "processed_datasets_audited": len(processed_inventory),
    "removed_columns": removed_columns,
    "engineered_columns": engineered_columns,
    "identifier_related_features": identifier_features,
    "reviewer_comments": [
        "Major 5",
        "Major 7",
        "Major 10",
        "Minor 26",
        "Minor 27"
    ],
    "original_code_modified": False,
    "models_retrained": False,
    "next_cell": 3
}

with open(
    CHECKPOINT_DIR / "cell02_completed.json",
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        checkpoint,
        f,
        indent=4,
        default=str
    )

print("\n" + "=" * 75)
print("CELL 02 COMPLETED SUCCESSFULLY")
print("=" * 75)

print("Original source code preserved.")
print("Original preprocessing audited.")
print("Original dataset dimensions verified.")
print("Class distributions recorded.")
print("Feature transformations identified.")
print("Model hyperparameters extracted.")
print("Library versions recorded.")
print("Audit tables saved in Google Drive.")
print("Source code and checkpoint saved.")
print("No original files modified.")
print("No model retraining performed.")

print(
    "\nREADY FOR CELL 03/15: "
    "LEAKAGE-FREE PREPROCESSING AND FEATURE SELECTION"
)
```

```python
# ============================================================
# CELL 03/15
# LEAKAGE-FREE PREPROCESSING AND FEATURE SELECTION
#
# Reviewer Comments:
# Major 5, Major 10, Minor 27
# ============================================================

import gc
import json
import re
import hashlib
from pathlib import Path

import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split

# ------------------------------------------------------------
# 1. Restore Google Drive workspace
# ------------------------------------------------------------

PROJECT_ROOT = Path(
    "/content/drive/MyDrive/Narayan_Patil_Reviewer_Revision"
)

SOURCE_ROOT = PROJECT_ROOT / "01_Original_Project" / "SBN"
REVISED_DIR = PROJECT_ROOT / "02_Revised_Project"

CODE_DIR = REVISED_DIR / "code"
DATA_DIR = REVISED_DIR / "data"
MODEL_DIR = REVISED_DIR / "models"
TABLE_DIR = REVISED_DIR / "results" / "tables"
FIGURE_DIR = REVISED_DIR / "results" / "figures"
REPORT_DIR = REVISED_DIR / "reports"
CHECKPOINT_DIR = REVISED_DIR / "checkpoints"

assert (
    CHECKPOINT_DIR / "cell02_completed.json"
).exists(), "Complete Cell 02 first."

SEED = 42
TARGET = "PotentialFraud"
GROUP = "Provider"

MASTER_PATH = (
    SOURCE_ROOT
    / "data"
    / "processed"
    / "master_dataset.csv"
)

print("=" * 75)
print("CELL 03: LEAKAGE-FREE PREPROCESSING")
print("=" * 75)

# ------------------------------------------------------------
# 2. Load original verified master dataset
# ------------------------------------------------------------

master = pd.read_csv(
    MASTER_PATH,
    low_memory=False
)

assert master.shape == (558211, 55), (
    f"Unexpected master dataset shape: {master.shape}"
)

assert master[TARGET].notna().all()
assert master[GROUP].notna().all()

master[TARGET] = master[TARGET].astype(int)
master[GROUP] = master[GROUP].astype(str)

assert set(master[TARGET].unique()) == {0, 1}

master["_row_id"] = np.arange(
    len(master),
    dtype=np.int64
)

print("\nOriginal dataset shape:", (len(master), 55))
print("\nOriginal class distribution:")
print(master[TARGET].value_counts().to_string())

# ------------------------------------------------------------
# 3. Verify provider-level label consistency
# ------------------------------------------------------------

provider_label_counts = (
    master.groupby(GROUP)[TARGET].nunique()
)

assert provider_label_counts.max() == 1, (
    "Inconsistent fraud labels within providers."
)

providers = (
    master[[GROUP, TARGET]]
    .drop_duplicates()
    .reset_index(drop=True)
)

print("\nUnique providers:", len(providers))

print("\nProvider-level fraud distribution:")
print(providers[TARGET].value_counts().to_string())

assert len(providers) == 5410

# ------------------------------------------------------------
# 4. Provider-disjoint stratified partitioning
#
# Outer split:
# 80% development providers
# 20% independent test providers
#
# Inner split:
# 80% training providers
# 20% validation providers
#
# The validation set will be used for model selection.
# The final test set remains independent.
# ------------------------------------------------------------

def create_provider_partitions(
    provider_table,
    seed=42
):

    development, test = train_test_split(
        provider_table,
        test_size=0.20,
        random_state=seed,
        stratify=provider_table[TARGET]
    )

    train, validation = train_test_split(
        development,
        test_size=0.20,
        random_state=seed,
        stratify=development[TARGET]
    )

    return {
        "train": set(train[GROUP]),
        "validation": set(validation[GROUP]),
        "test": set(test[GROUP])
    }


provider_sets = create_provider_partitions(
    providers,
    seed=SEED
)

assert provider_sets["train"].isdisjoint(
    provider_sets["validation"]
)

assert provider_sets["train"].isdisjoint(
    provider_sets["test"]
)

assert provider_sets["validation"].isdisjoint(
    provider_sets["test"]
)

assert len(
    set.union(*provider_sets.values())
) == len(providers)

partitions = {
    name: master.loc[
        master[GROUP].isin(provider_ids)
    ].copy()
    for name, provider_ids in provider_sets.items()
}

assert sum(
    len(df) for df in partitions.values()
) == len(master)

partition_summary = []

print("\n" + "=" * 75)
print("PROVIDER-DISJOINT DATA PARTITIONS")
print("=" * 75)

for name, df in partitions.items():

    record = {
        "partition": name,
        "claims": len(df),
        "providers": df[GROUP].nunique(),
        "fraud_providers": df.loc[
            df[TARGET] == 1, GROUP
        ].nunique(),
        "nonfraud_providers": df.loc[
            df[TARGET] == 0, GROUP
        ].nunique(),
        "nonfraud_claims": int(
            (df[TARGET] == 0).sum()
        ),
        "fraud_claims": int(
            (df[TARGET] == 1).sum()
        ),
        "fraud_percentage": float(
            df[TARGET].mean() * 100
        )
    }

    partition_summary.append(record)

partition_summary_df = pd.DataFrame(
    partition_summary
)

print(
    partition_summary_df.to_string(index=False)
)

print("\nProvider overlap: ZERO")
print("SMOTE applied: NO")

# ------------------------------------------------------------
# 5. Preserve split membership for reproducibility
# ------------------------------------------------------------

split_manifest = pd.concat(
    [
        df[
            ["_row_id", "Provider",
             "ClaimID", TARGET]
        ].assign(partition=name)

        for name, df in partitions.items()
    ],
    ignore_index=True
)

assert split_manifest["_row_id"].is_unique

split_manifest.to_csv(
    DATA_DIR / "cell03_split_manifest.csv",
    index=False
)

partition_summary_df.to_csv(
    TABLE_DIR / "cell03_partition_summary.csv",
    index=False
)

provider_manifest = pd.concat(
    [
        providers.loc[
            providers[GROUP].isin(provider_ids)
        ].assign(partition=name)

        for name, provider_ids in provider_sets.items()
    ],
    ignore_index=True
)

provider_manifest.to_csv(
    DATA_DIR / "cell03_provider_split_manifest.csv",
    index=False
)

# ------------------------------------------------------------
# 6. Audit possible beneficiary overlap
#
# Provider-disjoint evaluation is the primary design.
# Beneficiary overlap is reported transparently.
# ------------------------------------------------------------

beneficiary_sets = {
    name: set(
        df["BeneID"].dropna().astype(str)
    )
    for name, df in partitions.items()
}

beneficiary_overlap = {
    "train_validation": len(
        beneficiary_sets["train"]
        & beneficiary_sets["validation"]
    ),
    "train_test": len(
        beneficiary_sets["train"]
        & beneficiary_sets["test"]
    ),
    "validation_test": len(
        beneficiary_sets["validation"]
        & beneficiary_sets["test"]
    )
}

print("\nBeneficiary overlap audit:")
print(beneficiary_overlap)

# ------------------------------------------------------------
# 7. Training-only missing-value threshold
# ------------------------------------------------------------

original_columns = [
    c for c in master.columns
    if c != "_row_id"
]

training_missing = (
    partitions["train"][original_columns]
    .isna()
    .mean()
)

protected_columns = [
    TARGET,
    GROUP,
    "ClaimID",
    "BeneID"
]

high_missing_columns = [
    c for c in training_missing.index
    if training_missing[c] > 0.70
    and c not in protected_columns
]

missing_audit_df = pd.DataFrame({
    "feature": training_missing.index,
    "training_missing_percentage":
        training_missing.values * 100,
    "removed_by_missing_threshold": [
        c in high_missing_columns
        for c in training_missing.index
    ]
})

missing_audit_df.to_csv(
    TABLE_DIR / "cell03_training_missing_values.csv",
    index=False
)

print("\n" + "=" * 75)
print("TRAINING-ONLY MISSING-VALUE AUDIT")
print("=" * 75)

print("\nColumns exceeding 70% missing values:")

for column in high_missing_columns:
    print(
        f" - {column}: "
        f"{training_missing[column]*100:.2f}%"
    )

# ------------------------------------------------------------
# 8. Preserve original feature engineering
#
# Compute clinical counts BEFORE dropping sparse
# diagnosis and procedure columns.
# ------------------------------------------------------------

ENGINEERED_FEATURES = [
    "DiagnosisCount",
    "ProcedureCount",
    "TotalClaimAmount",
    "ReimbursementRatio",
    "ChronicConditionCount"
]

def valid_medical_code_count(df, columns):

    result = np.zeros(
        len(df),
        dtype=np.int16
    )

    for col in columns:

        values = (
            df[col]
            .astype("string")
            .str.strip()
            .str.lower()
        )

        valid = (
            values.notna()
            & ~values.isin([
                "", "nan", "none",
                "unknown", "0", "0.0"
            ])
        )

        result += (
            valid.fillna(False)
            .to_numpy(dtype=np.int16)
        )

    return result


def deterministic_feature_engineering(df):

    df = df.copy()

    # Original date-derived features
    date_columns = [
        "ClaimStartDt",
        "ClaimEndDt",
        "DOB"
    ]

    for column in date_columns:

        df[column] = pd.to_datetime(
            df[column],
            errors="coerce"
        )

    df["ClaimDuration"] = (
        df["ClaimEndDt"]
        - df["ClaimStartDt"]
    ).dt.days

    df["PatientAge"] = (
        df["ClaimStartDt"]
        - df["DOB"]
    ).dt.days / 365.0

    # Invalid values become missing, then receive
    # training-derived imputation.
    df.loc[
        df["ClaimDuration"] < 0,
        "ClaimDuration"
    ] = np.nan

    df.loc[
        (df["PatientAge"] < 0)
        | (df["PatientAge"] > 120),
        "PatientAge"
    ] = np.nan

    # Retain original diagnosis-count concept
    diagnosis_cols = [
        c for c in df.columns
        if "DiagnosisCode" in c
    ]

    procedure_cols = [
        c for c in df.columns
        if "ProcedureCode" in c
    ]

    assert diagnosis_cols
    assert procedure_cols

    df["DiagnosisCount"] = (
        valid_medical_code_count(
            df, diagnosis_cols
        )
    )

    df["ProcedureCount"] = (
        valid_medical_code_count(
            df, procedure_cols
        )
    )

    # Original reimbursement feature
    ip_amount = pd.to_numeric(
        df["IPAnnualReimbursementAmt"],
        errors="coerce"
    )

    op_amount = pd.to_numeric(
        df["OPAnnualReimbursementAmt"],
        errors="coerce"
    )

    df["TotalClaimAmount"] = (
        ip_amount + op_amount
    )

    # Correct denominator according to manuscript
    # description: reimbursement per diagnosis,
    # with denominator bounded below by one.
    df["ReimbursementRatio"] = (
        df["TotalClaimAmount"]
        / df["DiagnosisCount"].clip(lower=1)
    )

    chronic_cols = [
        c for c in df.columns
        if c.startswith("ChronicCond")
    ]

    assert chronic_cols

    chronic = df[chronic_cols].apply(
        pd.to_numeric,
        errors="coerce"
    )

    df["ChronicConditionCount"] = (
        chronic.eq(1).sum(axis=1)
    )

    return df


# ------------------------------------------------------------
# 9. Apply identical deterministic transformations
# ------------------------------------------------------------

engineered = {
    name: deterministic_feature_engineering(df)
    for name, df in partitions.items()
}

print("\nFEATURE ENGINEERING COMPLETE")

print(
    engineered["train"][
        ENGINEERED_FEATURES
    ].describe().round(3).to_string()
)

assert (
    engineered["train"]["ProcedureCount"] > 0
).any(), "ProcedureCount is unexpectedly constant."

# ------------------------------------------------------------
# 10. Remove leakage-prone identifiers and raw dates
# ------------------------------------------------------------

identifier_columns = [
    "BeneID",
    "ClaimID",
    "Provider",
    "AttendingPhysician",
    "OperatingPhysician",
    "OtherPhysician"
]

raw_date_columns = [
    "ClaimStartDt",
    "ClaimEndDt",
    "AdmissionDt",
    "DischargeDt",
    "DOB",
    "DOD"
]

excluded_columns = sorted(
    set(
        high_missing_columns
        + identifier_columns
        + raw_date_columns
        + [TARGET, "_row_id"]
    )
)

feature_columns = [
    c for c in engineered["train"].columns
    if c not in excluded_columns
]

assert TARGET not in feature_columns
assert GROUP not in feature_columns

assert all(
    c not in feature_columns
    for c in identifier_columns
)

assert all(
    c in feature_columns
    for c in ENGINEERED_FEATURES
)

X_raw = {
    name: df[feature_columns].copy()
    for name, df in engineered.items()
}

y_partitions = {
    name: df[TARGET].to_numpy(
        dtype=np.int8
    )
    for name, df in engineered.items()
}

# ------------------------------------------------------------
# 11. Identify numerical and categorical features
# ------------------------------------------------------------

categorical_columns = []

for column in feature_columns:

    is_category = (
        not pd.api.types.is_numeric_dtype(
            X_raw["train"][column]
        )
        or "DiagnosisCode" in column
        or "ProcedureCode" in column
        or column in [
            "Gender",
            "Race",
            "State",
            "County",
            "RenalDiseaseIndicator"
        ]
        or column.startswith("ChronicCond")
    )

    if is_category:
        categorical_columns.append(column)

numerical_columns = [
    c for c in feature_columns
    if c not in categorical_columns
]

print("\n" + "=" * 75)
print("REVISED FEATURE INVENTORY")
print("=" * 75)

print("Total predictors:", len(feature_columns))
print("Numerical:", len(numerical_columns))
print("Categorical:", len(categorical_columns))

print("\nExcluded original identifiers:")

for column in identifier_columns:
    print(" -", column)

print("\nRetained predictors:")

for i, column in enumerate(
    feature_columns, start=1
):
    print(f"{i:02d}. {column}")

# ------------------------------------------------------------
# 12. Fit imputation parameters on training data ONLY
# ------------------------------------------------------------

numeric_medians = {}
categorical_modes = {}

for column in numerical_columns:

    values = pd.to_numeric(
        X_raw["train"][column],
        errors="coerce"
    ).replace(
        [np.inf, -np.inf],
        np.nan
    )

    median = values.median()

    numeric_medians[column] = (
        float(median)
        if pd.notna(median)
        else 0.0
    )

for column in categorical_columns:

    values = (
        X_raw["train"][column]
        .astype("string")
        .str.strip()
    )

    values = values.replace(
        ["", "nan", "None", "Unknown"],
        pd.NA
    )

    mode = values.mode(dropna=True)

    categorical_modes[column] = (
        str(mode.iloc[0])
        if not mode.empty
        else "__MISSING__"
    )

# ------------------------------------------------------------
# 13. Apply training-derived imputation
# ------------------------------------------------------------

def apply_fitted_imputation(
    frame,
    numeric_medians,
    categorical_modes
):

    frame = frame.copy()

    for column, median in numeric_medians.items():

        frame[column] = (
            pd.to_numeric(
                frame[column],
                errors="coerce"
            )
            .replace(
                [np.inf, -np.inf],
                np.nan
            )
            .fillna(median)
            .astype(np.float32)
        )

    for column, mode in categorical_modes.items():

        frame[column] = (
            frame[column]
            .astype("string")
            .str.strip()
            .replace(
                ["", "nan", "None", "Unknown"],
                pd.NA
            )
            .fillna(mode)
            .astype(str)
        )

    return frame


X_preprocessed = {
    name: apply_fitted_imputation(
        df,
        numeric_medians,
        categorical_modes
    )
    for name, df in X_raw.items()
}

for name, df in X_preprocessed.items():

    assert df.isna().sum().sum() == 0, (
        f"Missing values remain in {name}"
    )

    assert list(df.columns) == feature_columns

    assert len(df) == len(
        y_partitions[name]
    )

    assert np.isfinite(
        df[numerical_columns].to_numpy()
    ).all()

    print(
        f"\n{name.upper()} preprocessing: PASS"
    )
    print("Shape:", df.shape)
    print("Missing values: 0")

# ------------------------------------------------------------
# 14. Save corrected datasets to Google Drive
#
# Parquet format reduces file size and preserves types.
# No encoding, scaling or SMOTE is applied yet.
# ------------------------------------------------------------

saved_datasets = []

for name, df in X_preprocessed.items():

    output = df.copy()

    output[TARGET] = y_partitions[name]

    output["_row_id"] = engineered[name][
        "_row_id"
    ].to_numpy()

    path = DATA_DIR / f"cell03_{name}_unencoded.parquet"

    output.to_parquet(
        path,
        index=False
    )

    saved_datasets.append({
        "partition": name,
        "file": str(path),
        "rows": len(output),
        "predictors": len(feature_columns),
        "size_mb": round(
            path.stat().st_size / 1024**2, 3
        )
    })

    print(
        "\nSaved:",
        path.name,
        "|",
        round(
            path.stat().st_size / 1024**2, 2
        ),
        "MB"
    )

    del output

# ------------------------------------------------------------
# 15. Save feature transformation audit
# ------------------------------------------------------------

feature_audit = []

for column in original_columns:

    if column == TARGET:

        action = "target"

    elif column in identifier_columns:

        action = "excluded_identifier"

    elif column in raw_date_columns:

        action = "excluded_raw_date"

    elif column in high_missing_columns:

        action = "excluded_high_missingness"

    elif column in feature_columns:

        action = "retained"

    else:

        action = "excluded"

    feature_audit.append({
        "feature": column,
        "action": action,
        "training_missing_percentage": float(
            training_missing.get(column, 0) * 100
        )
    })

for column in [
    "ClaimDuration",
    "PatientAge"
] + ENGINEERED_FEATURES:

    feature_audit.append({
        "feature": column,
        "action": "engineered_and_retained",
        "training_missing_percentage": None
    })

pd.DataFrame(feature_audit).to_csv(
    TABLE_DIR / "cell03_feature_selection_audit.csv",
    index=False
)

# ------------------------------------------------------------
# 16. Save fitted preprocessing specification
# ------------------------------------------------------------

preprocessing_spec = {
    "random_seed": SEED,
    "target": TARGET,
    "grouping_variable": GROUP,
    "classification_unit": "claim",
    "label_unit": "provider",
    "split_strategy":
        "provider-disjoint stratified partitioning",
    "outer_test_fraction": 0.20,
    "inner_validation_fraction": 0.20,
    "missing_threshold": 0.70,
    "high_missing_columns": high_missing_columns,
    "excluded_identifiers": identifier_columns,
    "excluded_raw_dates": raw_date_columns,
    "feature_columns": feature_columns,
    "numerical_columns": numerical_columns,
    "categorical_columns": categorical_columns,
    "engineered_features": ENGINEERED_FEATURES,
    "numeric_medians": numeric_medians,
    "categorical_modes": categorical_modes,
    "beneficiary_overlap": beneficiary_overlap,
    "provider_overlap": 0,
    "smote_applied": False,
    "scaling_applied": False,
    "categorical_encoding_applied": False,
    "reimbursement_ratio_formula":
        "TotalClaimAmount / max(DiagnosisCount, 1)",
    "medical_code_counts":
        "computed before sparse-column removal",
    "feature_selection_method":
        "training missingness and domain-based "
        "identifier exclusion",
    "evaluation_scope":
        "retrospective provider-disjoint claim "
        "classification; not prospective validation",
    "saved_datasets": saved_datasets
}

with open(
    REPORT_DIR / "cell03_preprocessing_spec.json",
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        preprocessing_spec,
        f,
        indent=4,
        default=str
    )

# ------------------------------------------------------------
# 17. Verify original source integrity again
# ------------------------------------------------------------

with open(
    REPORT_DIR / "cell02_original_source_hashes.json",
    "r"
) as f:

    original_hashes = json.load(f)


def sha256_file(path):

    digest = hashlib.sha256()

    with open(path, "rb") as f:

        for chunk in iter(
            lambda: f.read(1024 * 1024),
            b""
        ):

            digest.update(chunk)

    return digest.hexdigest()


for item in original_hashes:

    path = SOURCE_ROOT / item["file"]

    assert (
        sha256_file(path) == item["sha256"]
    ), f"Original code modified: {path}"

print("\nOriginal source integrity: PASS")

# ------------------------------------------------------------
# 18. Save complete Cell 03 code
# ------------------------------------------------------------

ip = get_ipython()

if ip is not None:

    (
        CODE_DIR
        / "cell03_leakage_free_preprocessing.py"
    ).write_text(
        ip.history_manager.input_hist_raw[-1],
        encoding="utf-8"
    )

# ------------------------------------------------------------
# 19. Save checkpoint
# ------------------------------------------------------------

checkpoint = {
    "cell": 3,
    "status": "COMPLETED",
    "reviewer_comments": [
        "Major 5",
        "Major 10",
        "Minor 27"
    ],
    "original_claims": len(master),
    "original_providers": len(providers),
    "provider_overlap": 0,
    "retained_predictors": len(feature_columns),
    "removed_high_missing_columns":
        high_missing_columns,
    "partitions": partition_summary,
    "smote_applied": False,
    "original_code_modified": False,
    "models_retrained": False,
    "next_cell": 4
}

with open(
    CHECKPOINT_DIR / "cell03_completed.json",
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        checkpoint,
        f,
        indent=4,
        default=str
    )

# ------------------------------------------------------------
# 20. Release unnecessary memory
# ------------------------------------------------------------

del master
del partitions
del engineered
del X_raw
del X_preprocessed
del split_manifest

gc.collect()

print("\n" + "=" * 75)
print("CELL 03 COMPLETED SUCCESSFULLY")
print("=" * 75)

print("Original dataset retained.")
print("Provider-disjoint splitting completed.")
print("Provider overlap: ZERO.")
print("Training-only missing-value threshold fitted.")
print("Training-only imputation completed.")
print("Identifier predictors excluded.")
print("Original feature-engineering approach retained.")
print("Diagnosis and procedure counts corrected.")
print("Reimbursement ratio corrected.")
print("Unencoded datasets saved in Google Drive.")
print("Preprocessing specification saved.")
print("Complete revised cell code saved.")
print("Original source code unchanged.")
print("SMOTE not applied.")
print("No model retraining performed.")

print(
    "\nREADY FOR CELL 04/15: "
    "TRAINING-ONLY ENCODING, SCALING AND SMOTE"
)
```

```python
# ============================================================
# CELL 04/15 - RAM-EFFICIENT CORRECTED VERSION
# TRAINING-ONLY ENCODING, SCALING AND BATCHED SMOTENC
#
# Reviewer Comment: Major 10
# ============================================================

import os
import gc
import sys
import json
import time
import joblib
import hashlib
import subprocess
import importlib.metadata as metadata

from pathlib import Path
from google.colab import drive

import numpy as np
import pandas as pd
import psutil

from sklearn import set_config
from sklearn.preprocessing import OrdinalEncoder
from sklearn.preprocessing import StandardScaler
from threadpoolctl import threadpool_limits

try:
    from imblearn.over_sampling import SMOTENC
except ImportError:
    subprocess.check_call([
        sys.executable, "-m", "pip",
        "install", "-q", "imbalanced-learn"
    ])
    from imblearn.over_sampling import SMOTENC

set_config(working_memory=64)

START_TIME = time.perf_counter()

SEED = 42
TARGET = "PotentialFraud"
N_BATCHES = 48

# ------------------------------------------------------------
# 1. Reconnect Google Drive
# ------------------------------------------------------------

drive.mount("/content/drive")

PROJECT_ROOT = Path(
    "/content/drive/MyDrive/Narayan_Patil_Reviewer_Revision"
)

SOURCE_ROOT = PROJECT_ROOT / "01_Original_Project" / "SBN"
REVISED_DIR = PROJECT_ROOT / "02_Revised_Project"

CODE_DIR = REVISED_DIR / "code"
DATA_DIR = REVISED_DIR / "data"
MODEL_DIR = REVISED_DIR / "models"
TABLE_DIR = REVISED_DIR / "results" / "tables"
FIGURE_DIR = REVISED_DIR / "results" / "figures"
REPORT_DIR = REVISED_DIR / "reports"
CHECKPOINT_DIR = REVISED_DIR / "checkpoints"

for folder in [
    CODE_DIR, DATA_DIR, MODEL_DIR,
    TABLE_DIR, FIGURE_DIR, REPORT_DIR,
    CHECKPOINT_DIR
]:
    folder.mkdir(parents=True, exist_ok=True)

for cell in [1, 2, 3]:

    assert (
        CHECKPOINT_DIR
        / f"cell{cell:02d}_completed.json"
    ).exists(), f"Cell {cell} checkpoint missing."

print("=" * 75)
print("CELL 04: RAM-EFFICIENT PREPROCESSING")
print("=" * 75)

print("Cells 01-03 checkpoints: VERIFIED")

def show_ram():

    memory = psutil.virtual_memory()

    print(
        "RAM used:",
        round(memory.used / 1024**3, 2),
        "GB | Available:",
        round(memory.available / 1024**3, 2),
        "GB",
        flush=True
    )

show_ram()

# ------------------------------------------------------------
# 2. Restore verified Cell 03 specification
# ------------------------------------------------------------

with open(
    REPORT_DIR / "cell03_preprocessing_spec.json"
) as f:

    spec = json.load(f)

feature_columns = spec["feature_columns"]
numerical_columns = spec["numerical_columns"]
categorical_columns = spec["categorical_columns"]

categorical_indices = [
    feature_columns.index(c)
    for c in categorical_columns
]

numerical_indices = [
    feature_columns.index(c)
    for c in numerical_columns
]

assert len(feature_columns) == 34
assert len(numerical_columns) == 14
assert len(categorical_columns) == 20

print("\nPredictors:", len(feature_columns))
print("Numerical:", len(numerical_columns))
print("Categorical:", len(categorical_columns))

# ------------------------------------------------------------
# 3. Verify provider-disjoint partitions
# ------------------------------------------------------------

manifest = pd.read_csv(
    DATA_DIR / "cell03_split_manifest.csv",
    usecols=[
        "_row_id", "Provider",
        "PotentialFraud", "partition"
    ]
)

assert manifest["_row_id"].is_unique
assert len(manifest) == 558211

partition_names = [
    "train", "validation", "test"
]

expected_counts = {
    "train": [217674, 143370],
    "validation": [57803, 30589],
    "test": [69938, 38837]
}

provider_sets = {
    name: set(
        manifest.loc[
            manifest["partition"] == name,
            "Provider"
        ].astype(str)
    )
    for name in partition_names
}

for a, b in [
    ("train", "validation"),
    ("train", "test"),
    ("validation", "test")
]:

    assert provider_sets[a].isdisjoint(
        provider_sets[b]
    )

print("\nProvider overlap: ZERO")

# ------------------------------------------------------------
# 4. Fit encoder and scalers using training only
# ------------------------------------------------------------

train_df = pd.read_parquet(
    DATA_DIR / "cell03_train_unencoded.parquet"
)

assert len(train_df) == 361044

encoder = OrdinalEncoder(
    handle_unknown="use_encoded_value",
    unknown_value=-1,
    dtype=np.float32
)

encoder.fit(
    train_df[categorical_columns]
)

numeric_scaler = StandardScaler()

numeric_scaler.fit(
    train_df[numerical_columns]
    .to_numpy(dtype=np.float32)
)

category_scaler = StandardScaler()

category_scaler.fit(
    encoder.transform(
        train_df[categorical_columns]
    )
)

joblib.dump(
    encoder,
    MODEL_DIR / "cell04_ordinal_encoder.joblib"
)

joblib.dump(
    numeric_scaler,
    MODEL_DIR / "cell04_numeric_scaler.joblib"
)

joblib.dump(
    category_scaler,
    MODEL_DIR / "cell04_category_scaler.joblib"
)

del train_df
gc.collect()

print("\nTraining-only encoder fitted: PASS")
print("Training-only scalers fitted: PASS")

# ------------------------------------------------------------
# 5. Encode each partition independently
# ------------------------------------------------------------

encoding_audit = []

def encode_partition(df, name):

    X = np.empty(
        (len(df), len(feature_columns)),
        dtype=np.float32
    )

    X[:, numerical_indices] = (
        numeric_scaler.transform(
            df[numerical_columns]
            .to_numpy(dtype=np.float32)
        )
    )

    categorical_values = encoder.transform(
        df[categorical_columns]
    )

    X[:, categorical_indices] = (
        categorical_values
    )

    for j, col in enumerate(
        categorical_columns
    ):

        encoding_audit.append({
            "partition": name,
            "feature": col,
            "training_categories":
                len(encoder.categories_[j]),
            "unknown_category_count":
                int(
                    (categorical_values[:, j] == -1)
                    .sum()
                )
        })

    assert np.isfinite(X).all()

    return X


def finalize_model_matrix(X):

    result = X.copy()

    result[:, categorical_indices] = (
        category_scaler.transform(
            result[:, categorical_indices]
        ).astype(np.float32)
    )

    return result


distribution_records = []

def record_distribution(
    name, y, synthetic_count=0
):

    counts = np.bincount(y, minlength=2)

    distribution_records.append({
        "dataset": name,
        "non_fraud": int(counts[0]),
        "fraud": int(counts[1]),
        "total": len(y),
        "fraud_percentage":
            float(y.mean() * 100),
        "synthetic_records": synthetic_count
    })


feature_array = np.asarray(
    feature_columns,
    dtype=str
)

print("\nENCODING ORIGINAL PARTITIONS")
print("-" * 75)

for name in partition_names:

    df = pd.read_parquet(
        DATA_DIR
        / f"cell03_{name}_unencoded.parquet"
    )

    X = encode_partition(df, name)

    y = df[TARGET].to_numpy(
        dtype=np.int8
    )

    row_ids = df["_row_id"].to_numpy(
        dtype=np.int64
    )

    aligned = (
        manifest.loc[
            manifest["partition"] == name
        ]
        .set_index("_row_id")
        .loc[row_ids]
    )

    assert np.array_equal(
        aligned[TARGET].to_numpy(),
        y
    )

    assert np.bincount(
        y, minlength=2
    ).tolist() == expected_counts[name]

    # Preserve categorical codes before final scaling.
    np.savez_compressed(
        DATA_DIR / f"cell04_{name}_real.npz",
        X=X,
        y=y,
        row_id=row_ids,
        feature_names=feature_array
    )

    final_X = finalize_model_matrix(X)

    final_name = (
        "train_unbalanced"
        if name == "train"
        else name
    )

    np.savez_compressed(
        DATA_DIR
        / f"cell04_{final_name}_final.npz",
        X=final_X,
        y=y,
        row_id=row_ids,
        feature_names=feature_array
    )

    record_distribution(final_name, y)

    print(
        f"{name.upper()}: PASS |",
        X.shape,
        "| Class counts:",
        np.bincount(y).tolist(),
        flush=True
    )

    del df, X, y, row_ids, aligned, final_X

    gc.collect()

del manifest
del provider_sets

gc.collect()

pd.DataFrame(
    encoding_audit
).to_csv(
    TABLE_DIR / "cell04_encoding_audit.csv",
    index=False
)

print("\nOriginal partitions encoded and saved.")
show_ram()

# ------------------------------------------------------------
# 6. Load only training data for SMOTENC
# ------------------------------------------------------------

with np.load(
    DATA_DIR / "cell04_train_real.npz",
    allow_pickle=False
) as data:

    X_real = data["X"].astype(np.float32)
    y_real = data["y"].astype(np.int8)
    real_row_ids = data["row_id"]

assert X_real.shape == (361044, 34)

before_counts = np.bincount(
    y_real,
    minlength=2
)

assert before_counts.tolist() == [
    217674, 143370
]

synthetic_target = int(
    before_counts[0] - before_counts[1]
)

assert synthetic_target == 74304

print("\n" + "=" * 75)
print("MEMORY-BOUNDED TRAINING-ONLY SMOTENC")
print("=" * 75)

print("Original training records:", len(y_real))
print("Synthetic records required:", synthetic_target)
print("Number of batches:", N_BATCHES)

# ------------------------------------------------------------
# 7. Construct disjoint stratified batches
#
# Every original training record participates.
# No original records are discarded.
# ------------------------------------------------------------

rng = np.random.default_rng(SEED)

majority_indices = np.flatnonzero(
    y_real == 0
)

minority_indices = np.flatnonzero(
    y_real == 1
)

rng.shuffle(majority_indices)
rng.shuffle(minority_indices)

majority_batches = np.array_split(
    majority_indices,
    N_BATCHES
)

minority_batches = np.array_split(
    minority_indices,
    N_BATCHES
)

synthetic_X = np.empty(
    (synthetic_target, len(feature_columns)),
    dtype=np.float32
)

synthetic_y = np.ones(
    synthetic_target,
    dtype=np.int8
)

cursor = 0
batch_records = []

smote_start = time.perf_counter()

# ------------------------------------------------------------
# 8. Apply SMOTENC separately to each batch
#
# This avoids the full-dataset nearest-neighbor
# calculation that likely caused the RAM crash.
# ------------------------------------------------------------

with threadpool_limits(limits=2):

    for batch_no, (
        majority_batch,
        minority_batch
    ) in enumerate(
        zip(majority_batches, minority_batches),
        start=1
    ):

        batch_indices = np.concatenate([
            majority_batch,
            minority_batch
        ])

        rng.shuffle(batch_indices)

        X_batch = X_real[batch_indices]
        y_batch = y_real[batch_indices]

        n_original = len(y_batch)

        n_majority = len(majority_batch)
        n_minority = len(minority_batch)

        required = n_majority - n_minority

        assert required >= 0
        assert n_minority >= 6

        smote = SMOTENC(
            categorical_features=categorical_indices,
            sampling_strategy={
                1: n_majority
            },
            random_state=SEED + batch_no,
            k_neighbors=5
        )

        X_resampled, y_resampled = (
            smote.fit_resample(
                X_batch,
                y_batch
            )
        )

        X_resampled = np.asarray(
            X_resampled,
            dtype=np.float32
        )

        y_resampled = np.asarray(
            y_resampled,
            dtype=np.int8
        )

        assert np.array_equal(
            y_resampled[:n_original],
            y_batch
        )

        assert np.allclose(
            X_resampled[:n_original],
            X_batch,
            atol=1e-6
        )

        new_X = X_resampled[n_original:]
        new_y = y_resampled[n_original:]

        assert len(new_X) == required
        assert np.all(new_y == 1)
        assert np.isfinite(new_X).all()

        synthetic_X[
            cursor:cursor + required
        ] = new_X

        cursor += required

        batch_records.append({
            "batch": batch_no,
            "original_records": n_original,
            "original_nonfraud": n_majority,
            "original_fraud": n_minority,
            "synthetic_records": required,
            "balanced_records":
                len(y_resampled),
            "random_seed": SEED + batch_no
        })

        print(
            f"Batch {batch_no:02d}/{N_BATCHES}"
            f" | Synthetic: {required:,}"
            f" | Cumulative: {cursor:,}"
            f"/{synthetic_target:,}",
            flush=True
        )

        del (
            smote,
            X_batch,
            y_batch,
            X_resampled,
            y_resampled,
            new_X,
            new_y,
            batch_indices
        )

        if batch_no % 4 == 0:

            gc.collect()
            show_ram()

assert cursor == synthetic_target

smote_seconds = (
    time.perf_counter() - smote_start
)

print("\nBatched SMOTENC: COMPLETED")

pd.DataFrame(
    batch_records
).to_csv(
    TABLE_DIR / "cell04_smote_batch_audit.csv",
    index=False
)

# ------------------------------------------------------------
# 9. Combine all original and synthetic records
# ------------------------------------------------------------

X_balanced = np.concatenate([
    X_real,
    synthetic_X
], axis=0)

y_balanced = np.concatenate([
    y_real,
    synthetic_y
])

assert X_balanced.shape == (435348, 34)

assert np.bincount(
    y_balanced,
    minlength=2
).tolist() == [217674, 217674]

assert np.array_equal(
    X_balanced[:len(X_real)],
    X_real
)

assert np.array_equal(
    y_balanced[:len(y_real)],
    y_real
)

print("\nOriginal records retained: PASS")
print("Synthetic records:", len(synthetic_y))

# ------------------------------------------------------------
# 10. Verify categorical validity
# ------------------------------------------------------------

category_validity = []

for col, idx in zip(
    categorical_columns,
    categorical_indices
):

    original_codes = np.unique(
        X_real[:, idx]
    )

    synthetic_codes = np.unique(
        synthetic_X[:, idx]
    )

    valid = bool(
        np.isin(
            synthetic_codes,
            original_codes
        ).all()
    )

    assert valid, (
        f"Invalid category generated: {col}"
    )

    category_validity.append({
        "feature": col,
        "original_categories":
            len(original_codes),
        "synthetic_categories":
            len(synthetic_codes),
        "status": "PASS"
    })

pd.DataFrame(
    category_validity
).to_csv(
    TABLE_DIR
    / "cell04_smote_category_validity.csv",
    index=False
)

print("Synthetic categorical validity: PASS")

# ------------------------------------------------------------
# 11. Final scaling and saving balanced training set
# ------------------------------------------------------------

X_train = finalize_model_matrix(
    X_balanced
)

y_train = y_balanced

balanced_row_ids = np.concatenate([
    real_row_ids,
    np.full(
        synthetic_target,
        -1,
        dtype=np.int64
    )
])

synthetic_flags = np.concatenate([
    np.zeros(len(y_real), dtype=np.int8),
    np.ones(synthetic_target, dtype=np.int8)
])

np.savez_compressed(
    DATA_DIR / "cell04_train_balanced_final.npz",
    X=X_train,
    y=y_train,
    row_id=balanced_row_ids,
    is_synthetic=synthetic_flags,
    feature_names=feature_array
)

record_distribution(
    "train_balanced",
    y_train,
    synthetic_target
)

print("\nBalanced training dataset saved.")

del (
    X_balanced,
    synthetic_X,
    synthetic_y,
    X_real,
    X_train
)

gc.collect()

# ------------------------------------------------------------
# 12. Final validation and test verification
# ------------------------------------------------------------

saved_files = []

for name in [
    "train_balanced",
    "train_unbalanced",
    "validation",
    "test"
]:

    path = (
        DATA_DIR
        / f"cell04_{name}_final.npz"
    )

    with np.load(
        path,
        allow_pickle=False
    ) as data:

        X = data["X"]
        y = data["y"]
        ids = data["row_id"]

        assert X.shape[1] == 34
        assert np.isfinite(X).all()

        if name == "validation":

            assert len(y) == 88392

            assert np.bincount(
                y
            ).tolist() == [57803, 30589]

            assert np.all(ids >= 0)

        elif name == "test":

            assert len(y) == 108775

            assert np.bincount(
                y
            ).tolist() == [69938, 38837]

            assert np.all(ids >= 0)

        elif name == "train_balanced":

            assert len(y) == 435348

            assert (
                data["is_synthetic"].sum()
                == 74304
            )

        saved_files.append({
            "dataset": name,
            "file": str(path),
            "rows": len(y),
            "features": X.shape[1]
        })

        print(
            name,
            "| Shape:", X.shape,
            "| PASS"
        )

    del X, y, ids
    gc.collect()

# ------------------------------------------------------------
# 13. Save reviewer-ready distribution table
# ------------------------------------------------------------

distribution_df = pd.DataFrame(
    distribution_records
)

distribution_df.to_csv(
    TABLE_DIR
    / "cell04_final_class_distribution.csv",
    index=False
)

print("\nFINAL DATASET DISTRIBUTION")
print(
    distribution_df.to_string(index=False)
)

# ------------------------------------------------------------
# 14. Save complete preprocessing bundle
# ------------------------------------------------------------

preprocessing_bundle = {
    "feature_names": feature_columns,
    "numerical_columns": numerical_columns,
    "categorical_columns": categorical_columns,
    "numerical_indices": numerical_indices,
    "categorical_indices": categorical_indices,
    "ordinal_encoder": encoder,
    "numeric_scaler": numeric_scaler,
    "category_scaler": category_scaler,
    "cell03_spec": spec,
    "unknown_category_value": -1,
    "smote_method": "Batched SMOTENC",
    "smote_random_state": SEED,
    "smote_batches": N_BATCHES,
    "smote_k_neighbors": 5
}

joblib.dump(
    preprocessing_bundle,
    MODEL_DIR
    / "cell04_preprocessing_bundle.joblib"
)

# ------------------------------------------------------------
# 15. Verify original source code integrity
# ------------------------------------------------------------

def sha256_file(path):

    digest = hashlib.sha256()

    with open(path, "rb") as f:

        for chunk in iter(
            lambda: f.read(1024 * 1024), b""
        ):

            digest.update(chunk)

    return digest.hexdigest()


with open(
    REPORT_DIR / "cell02_original_source_hashes.json"
) as f:

    original_hashes = json.load(f)

for item in original_hashes:

    path = SOURCE_ROOT / item["file"]

    assert (
        sha256_file(path) == item["sha256"]
    )

print("\nOriginal source integrity: PASS")

# ------------------------------------------------------------
# 16. Save complete report
# ------------------------------------------------------------

report = {
    "cell": 4,
    "reviewer_comment": "Major 10",
    "status": "COMPLETED",
    "seed": SEED,
    "feature_count": 34,
    "numerical_features": 14,
    "categorical_features": 20,
    "encoder": "OrdinalEncoder",
    "encoder_fit_partition":
        "real training only",
    "scaling": "StandardScaler",
    "scaler_fit_partition":
        "real training only",
    "oversampling": "Batched SMOTENC",
    "oversampling_partition":
        "training only",
    "oversampling_batches": N_BATCHES,
    "smote_k_neighbors": 5,
    "smote_seconds": smote_seconds,
    "implementation_limitation":
        "Nearest neighbors are restricted to "
        "randomly partitioned stratified training "
        "batches for memory efficiency.",
    "real_training_records": 361044,
    "balanced_training_records": 435348,
    "synthetic_training_records": 74304,
    "validation_oversampled": False,
    "test_oversampled": False,
    "provider_overlap": 0,
    "original_records_discarded": 0,
    "original_code_modified": False,
    "dataset_distribution":
        distribution_records,
    "saved_files": saved_files,
    "imbalanced_learn_version":
        metadata.version("imbalanced-learn"),
    "scikit_learn_version":
        metadata.version("scikit-learn"),
    "elapsed_seconds":
        time.perf_counter() - START_TIME
}

with open(
    REPORT_DIR / "cell04_preprocessing_report.json",
    "w"
) as f:

    json.dump(report, f, indent=4)

# ------------------------------------------------------------
# 17. Save complete revised cell code
# ------------------------------------------------------------

ip = get_ipython()

if ip is not None:

    (
        CODE_DIR
        / "cell04_encoding_scaling_smote.py"
    ).write_text(
        ip.history_manager.input_hist_raw[-1],
        encoding="utf-8"
    )

# ------------------------------------------------------------
# 18. Save completion checkpoint
# ------------------------------------------------------------

checkpoint = {
    "cell": 4,
    "status": "COMPLETED",
    "reviewer_comment": "Major 10",
    "features": 34,
    "real_training_records": 361044,
    "balanced_training_records": 435348,
    "synthetic_training_records": 74304,
    "validation_records": 88392,
    "test_records": 108775,
    "smote_method": "Batched SMOTENC",
    "smote_batches": N_BATCHES,
    "provider_overlap": 0,
    "validation_smote": False,
    "test_smote": False,
    "original_code_modified": False,
    "next_cell": 5
}

with open(
    CHECKPOINT_DIR / "cell04_completed.json",
    "w"
) as f:

    json.dump(checkpoint, f, indent=4)

gc.collect()

print("\n" + "=" * 75)
print("CELL 04 COMPLETED SUCCESSFULLY")
print("=" * 75)

print("Training-only encoding: PASS")
print("Training-only scaling: PASS")
print("Batched SMOTENC: PASS")
print("All original training records retained.")
print("Synthetic categorical validity: PASS")
print("Validation distribution preserved: PASS")
print("Test distribution preserved: PASS")
print("Provider overlap: ZERO")
print("Original code unchanged.")
print("All datasets saved in Google Drive.")
print("Preprocessing objects saved.")
print("Complete revised code saved.")
print("Checkpoint saved.")

show_ram()

print(
    "\nTotal execution seconds:",
    round(
        time.perf_counter() - START_TIME,
        2
    )
)

print(
    "\nREADY FOR CELL 05/15: "
    "LIGHTGBM, CATBOOST AND XGBOOST RETRAINING"
)
```

```python
# ============================================================
# CELL 05/15 - CORRECTED REVIEWER-SPECIFIC VERSION
#
# Major 5  : Training and testing performance comparison
# Major 10 : Stratified splitting and training-only SMOTE
# Major 7  : Reproducibility
# Minor 26 : Original XGBoost hyperparameters
#
# NO NEW MODELS
# NO HYPERPARAMETER TUNING
# NO FEATURE ENGINEERING MODIFICATIONS
# NO EXTERNAL DATASET
# ============================================================

import os
import gc
import sys
import json
import time
import joblib
import hashlib
import subprocess
import platform
import importlib.metadata as metadata

from pathlib import Path

import numpy as np
import pandas as pd
import psutil

from google.colab import drive

from sklearn import set_config
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OrdinalEncoder
from sklearn.preprocessing import StandardScaler

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix
)

from imblearn.over_sampling import SMOTE
from threadpoolctl import threadpool_limits

import lightgbm as lgb
import xgboost as xgb

from catboost import CatBoostClassifier

# ------------------------------------------------------------
# 1. Restore workspace
# ------------------------------------------------------------

drive.mount("/content/drive")

START_TIME = time.perf_counter()

SEED = 42

np.random.seed(SEED)

set_config(working_memory=64)

PROJECT_ROOT = Path(
    "/content/drive/MyDrive/Narayan_Patil_Reviewer_Revision"
)

SOURCE_ROOT = (
    PROJECT_ROOT
    / "01_Original_Project"
    / "SBN"
)

# Separate workspace to preserve earlier results.
REVISED_DIR = (
    PROJECT_ROOT
    / "03_Reviewer_Only_Revision"
)

CODE_DIR = REVISED_DIR / "code"
DATA_DIR = REVISED_DIR / "data"
MODEL_DIR = REVISED_DIR / "models"
RESULT_DIR = REVISED_DIR / "results"
TABLE_DIR = RESULT_DIR / "tables"
PREDICTION_DIR = RESULT_DIR / "predictions"
REPORT_DIR = REVISED_DIR / "reports"
CHECKPOINT_DIR = REVISED_DIR / "checkpoints"

PREVIOUS_REPORT_DIR = (
    PROJECT_ROOT
    / "02_Revised_Project"
    / "reports"
)

for folder in [

    CODE_DIR,
    DATA_DIR,
    MODEL_DIR,
    TABLE_DIR,
    PREDICTION_DIR,
    REPORT_DIR,
    CHECKPOINT_DIR

]:

    folder.mkdir(
        parents=True,
        exist_ok=True
    )

print("=" * 75)
print("CELL 05: REVIEWER-ONLY CORRECTED IMPLEMENTATION")
print("=" * 75)

print("\nOriginal source:", SOURCE_ROOT)
print("New revision workspace:", REVISED_DIR)

print(
    "Available RAM:",
    round(
        psutil.virtual_memory().available
        / 1024**3,
        2
    ),
    "GB"
)

# ------------------------------------------------------------
# 2. Verify GPU
# ------------------------------------------------------------

gpu = subprocess.run(

    [
        "nvidia-smi",
        "--query-gpu=name,memory.total",
        "--format=csv,noheader"
    ],

    capture_output=True,
    text=True

)

assert gpu.returncode == 0, (
    "Enable T4 GPU before running this cell."
)

print("\nGPU:", gpu.stdout.strip())

# ------------------------------------------------------------
# 3. Load ORIGINAL featured dataset
#
# Preserve original feature engineering.
# Preserve all 42 original predictors.
# ------------------------------------------------------------

FEATURED_PATH = (
    SOURCE_ROOT
    / "data"
    / "processed"
    / "featured_dataset.csv"
)

assert FEATURED_PATH.exists()

df = pd.read_csv(
    FEATURED_PATH,
    low_memory=False
)

TARGET = "PotentialFraud"

assert df.shape == (558211, 43)

feature_names = [
    column
    for column in df.columns
    if column != TARGET
]

assert len(feature_names) == 42

assert df[TARGET].notna().all()

df[TARGET] = df[TARGET].astype(np.int8)

assert np.bincount(
    df[TARGET],
    minlength=2
).tolist() == [345415, 212796]

print("\nORIGINAL DATASET")

print("Records:", len(df))
print("Original predictors:", len(feature_names))

print("\nOriginal class distribution:")

print(
    df[TARGET].value_counts().to_string()
)

print("\nOriginal feature engineering: RETAINED")
print("Original Provider feature: RETAINED")
print("Additional feature removal: NONE")

# ------------------------------------------------------------
# 4. Original 80:20 stratified split
#
# Reviewer-required correction:
# Split BEFORE encoding, scaling and SMOTE.
# ------------------------------------------------------------

all_indices = np.arange(
    len(df),
    dtype=np.int64
)

train_indices, test_indices = train_test_split(

    all_indices,

    test_size=0.20,

    random_state=SEED,

    stratify=df[TARGET]

)

assert set(train_indices).isdisjoint(
    test_indices
)

assert len(train_indices) + len(test_indices) == len(df)

train_df = df.iloc[
    train_indices
].copy()

test_df = df.iloc[
    test_indices
].copy()

y_train_real = train_df[TARGET].to_numpy(
    dtype=np.int8
)

y_test = test_df[TARGET].to_numpy(
    dtype=np.int8
)

print("\nSTRATIFIED TRAIN-TEST SPLIT")

print("Training records:", len(train_df))
print("Testing records :", len(test_df))

print(
    "Training classes:",
    np.bincount(
        y_train_real,
        minlength=2
    ).tolist()
)

print(
    "Testing classes:",
    np.bincount(
        y_test,
        minlength=2
    ).tolist()
)

print("Original 80:20 ratio: RETAINED")
print("Row overlap: ZERO")
print("Separate validation set: NOT ADDED")

# Record provider overlap rather than concealing it.

train_providers = set(
    train_df["Provider"].astype(str)
)

test_providers = set(
    test_df["Provider"].astype(str)
)

provider_overlap = len(
    train_providers & test_providers
)

print(
    "\nProvider overlap:",
    provider_overlap
)

print(
    "Evaluation scope: "
    "stratified claim-wise classification."
)

# Save original split membership.

np.savez_compressed(

    DATA_DIR / "cell05_original_split_indices.npz",

    train_indices=train_indices,

    test_indices=test_indices

)

split_manifest = pd.DataFrame({

    "row_id": all_indices,

    "Provider": df["Provider"].astype(str),

    "PotentialFraud": df[TARGET]

})

split_manifest["partition"] = "test"

split_manifest.loc[
    train_indices,
    "partition"
] = "train"

split_manifest.to_csv(

    DATA_DIR / "cell05_split_manifest.csv",

    index=False

)

# ------------------------------------------------------------
# 5. Preserve original categorical encoding approach
#
# Ordinal integer encoding, fitted on training only.
# Unknown test categories receive -1.
# ------------------------------------------------------------

categorical_columns = (

    train_df[feature_names]

    .select_dtypes(
        include=["object", "string", "category"]
    )

    .columns.tolist()

)

numerical_columns = [

    column
    for column in feature_names
    if column not in categorical_columns

]

X_train_real = np.empty(

    (len(train_df), len(feature_names)),

    dtype=np.float32

)

X_test_real = np.empty(

    (len(test_df), len(feature_names)),

    dtype=np.float32

)

# Original numerical missing-value handling.

for column in numerical_columns:

    idx = feature_names.index(column)

    X_train_real[:, idx] = (

        pd.to_numeric(
            train_df[column],
            errors="coerce"
        )

        .fillna(0)

        .to_numpy(dtype=np.float32)

    )

    X_test_real[:, idx] = (

        pd.to_numeric(
            test_df[column],
            errors="coerce"
        )

        .fillna(0)

        .to_numpy(dtype=np.float32)

    )

# Training-only encoding.

encoder = OrdinalEncoder(

    handle_unknown="use_encoded_value",

    unknown_value=-1,

    dtype=np.float32

)

if categorical_columns:

    train_categories = (

        train_df[categorical_columns]

        .fillna("Unknown")

        .astype(str)

    )

    test_categories = (

        test_df[categorical_columns]

        .fillna("Unknown")

        .astype(str)

    )

    train_encoded = encoder.fit_transform(
        train_categories
    )

    test_encoded = encoder.transform(
        test_categories
    )

    for j, column in enumerate(
        categorical_columns
    ):

        idx = feature_names.index(column)

        X_train_real[:, idx] = (
            train_encoded[:, j]
        )

        X_test_real[:, idx] = (
            test_encoded[:, j]
        )

    del (
        train_categories,
        test_categories,
        train_encoded,
        test_encoded
    )

assert np.isfinite(X_train_real).all()
assert np.isfinite(X_test_real).all()

joblib.dump(

    encoder,

    MODEL_DIR / "cell05_original_encoder.joblib"

)

print("\nOriginal predictors:", len(feature_names))

print(
    "Training-only ordinal encoding: PASS"
)

# Save original real encoded records.

np.savez_compressed(

    DATA_DIR / "cell05_train_real_encoded.npz",

    X=X_train_real,

    y=y_train_real,

    row_id=train_indices,

    feature_names=np.asarray(
        feature_names,
        dtype=str
    )

)

del df, train_df, test_df, split_manifest

gc.collect()

# ------------------------------------------------------------
# 6. Apply ORIGINAL SMOTE
#
# Original algorithm: SMOTE
# Original seed: 42
# Original k_neighbors: 5
#
# Only required correction:
# SMOTE restricted to training records.
#
# No SMOTENC and no batch approximation.
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("ORIGINAL SMOTE - TRAINING ONLY")
print("=" * 75)

before_counts = np.bincount(

    y_train_real,

    minlength=2

)

print(
    "Before SMOTE:",
    before_counts.tolist()
)

smote = SMOTE(

    sampling_strategy="auto",

    random_state=SEED,

    k_neighbors=5

)

smote_start = time.perf_counter()

# Restrict temporary memory and parallel threads.

with threadpool_limits(limits=2):

    X_train_balanced, y_train_balanced = (

        smote.fit_resample(

            X_train_real,

            y_train_real

        )

    )

X_train_balanced = np.asarray(

    X_train_balanced,

    dtype=np.float32

)

y_train_balanced = np.asarray(

    y_train_balanced,

    dtype=np.int8

)

smote_seconds = (

    time.perf_counter() - smote_start

)

after_counts = np.bincount(

    y_train_balanced,

    minlength=2

)

synthetic_count = (

    len(y_train_balanced)

    - len(y_train_real)

)

assert after_counts[0] == after_counts[1]

assert np.array_equal(

    y_train_balanced[:len(y_train_real)],

    y_train_real

)

assert np.array_equal(

    X_train_balanced[:len(X_train_real)],

    X_train_real

)

print("\nAfter SMOTE:", after_counts.tolist())

print(
    "Synthetic training records:",
    synthetic_count
)

print(
    "SMOTE execution seconds:",
    round(smote_seconds, 2)
)

print("Testing SMOTE: NOT APPLIED")

del smote
gc.collect()

# ------------------------------------------------------------
# 7. Preserve original StandardScaler procedure
#
# Fit on balanced training set, as in the
# original source code.
# ------------------------------------------------------------

scaler = StandardScaler()

X_train = scaler.fit_transform(

    X_train_balanced

).astype(np.float32)

X_train_real_scaled = scaler.transform(

    X_train_real

).astype(np.float32)

X_test = scaler.transform(

    X_test_real

).astype(np.float32)

y_train = y_train_balanced.copy()

joblib.dump(

    scaler,

    MODEL_DIR / "cell05_original_scaler.joblib"

)

del X_train_balanced
del X_test_real
del X_train_real

gc.collect()

print("\nTraining-only scaler fitting: PASS")

# ------------------------------------------------------------
# 8. Save corrected datasets
# ------------------------------------------------------------

balanced_row_ids = np.concatenate([

    train_indices,

    np.full(
        synthetic_count,
        -1,
        dtype=np.int64
    )

])

synthetic_flags = (

    balanced_row_ids == -1

).astype(np.int8)

feature_array = np.asarray(

    feature_names,

    dtype=str

)

np.savez_compressed(

    DATA_DIR / "cell05_train_balanced.npz",

    X=X_train,

    y=y_train,

    row_id=balanced_row_ids,

    is_synthetic=synthetic_flags,

    feature_names=feature_array

)

np.savez_compressed(

    DATA_DIR / "cell05_train_real.npz",

    X=X_train_real_scaled,

    y=y_train_real,

    row_id=train_indices,

    feature_names=feature_array

)

np.savez_compressed(

    DATA_DIR / "cell05_test.npz",

    X=X_test,

    y=y_test,

    row_id=test_indices,

    feature_names=feature_array

)

assert X_train.shape[1] == 42
assert X_test.shape[1] == 42

print("\nFINAL DATASET SHAPES")

print("Original training:", X_train_real_scaled.shape)
print("Balanced training:", X_train.shape)
print("Testing:", X_test.shape)

# ------------------------------------------------------------
# 9. ORIGINAL model hyperparameters
#
# No tuning or architecture modifications.
# ------------------------------------------------------------

ORIGINAL_PARAMS = {

    "LightGBM": {

        "n_estimators": 300,

        "learning_rate": 0.05,

        "max_depth": 10,

        "num_leaves": 31,

        "subsample": 0.8,

        "colsample_bytree": 0.8,

        "random_state": SEED

    },

    "CatBoost": {

        "iterations": 400,

        "learning_rate": 0.05,

        "depth": 8,

        "l2_leaf_reg": 3,

        "random_state": SEED,

        "verbose": 100

    },

    "XGBoost": {

        "n_estimators": 400,

        "learning_rate": 0.05,

        "max_depth": 8,

        "subsample": 0.8,

        "colsample_bytree": 0.8,

        "eval_metric": "logloss",

        "random_state": SEED

    }

}

models = {

    "LightGBM": lgb.LGBMClassifier(

        **ORIGINAL_PARAMS["LightGBM"],

        device_type="cpu",

        n_jobs=2,

        verbosity=-1

    ),

    "CatBoost": CatBoostClassifier(

        **ORIGINAL_PARAMS["CatBoost"],

        task_type="GPU",

        devices="0",

        thread_count=2,

        allow_writing_files=False

    ),

    "XGBoost": xgb.XGBClassifier(

        **ORIGINAL_PARAMS["XGBoost"],

        tree_method="hist",

        device="cuda",

        n_jobs=2

    )

}

# ------------------------------------------------------------
# 10. Original evaluation metrics only
# ------------------------------------------------------------

def calculate_metrics(
    y_true,
    predictions,
    probabilities
):

    tn, fp, fn, tp = confusion_matrix(

        y_true,

        predictions,

        labels=[0, 1]

    ).ravel()

    return {

        "accuracy": float(

            accuracy_score(
                y_true,
                predictions
            )

        ),

        "precision": float(

            precision_score(
                y_true,
                predictions,
                zero_division=0
            )

        ),

        "recall": float(

            recall_score(
                y_true,
                predictions,
                zero_division=0
            )

        ),

        "f1": float(

            f1_score(
                y_true,
                predictions,
                zero_division=0
            )

        ),

        "roc_auc": float(

            roc_auc_score(
                y_true,
                probabilities
            )

        ),

        "tn": int(tn),

        "fp": int(fp),

        "fn": int(fn),

        "tp": int(tp)

    }

# ------------------------------------------------------------
# 11. Retrain ORIGINAL three classifiers
# ------------------------------------------------------------

metric_records = []
training_records = []

train_frame = pd.DataFrame(

    X_train,

    columns=feature_names

)

real_train_frame = pd.DataFrame(

    X_train_real_scaled,

    columns=feature_names

)

test_frame = pd.DataFrame(

    X_test,

    columns=feature_names

)

print("\n" + "=" * 75)
print("ORIGINAL MODEL RETRAINING")
print("=" * 75)

for model_name, model in models.items():

    print("\n" + "-" * 75)
    print("TRAINING:", model_name)
    print("-" * 75)

    model_start = time.perf_counter()

    model.fit(

        train_frame,

        y_train

    )

    training_seconds = (

        time.perf_counter()

        - model_start

    )

    training_records.append({

        "model": model_name,

        "seed": SEED,

        "device": (

            "CPU"

            if model_name == "LightGBM"

            else "GPU"

        ),

        "training_seconds": training_seconds,

        "training_records": len(y_train),

        "parameters": model.get_params()

    })

    # Major Comment 5:
    # Compare real training performance against testing.

    for partition, X, y, ids in [

        (

            "train_real",

            real_train_frame,

            y_train_real,

            train_indices

        ),

        (

            "test",

            test_frame,

            y_test,

            test_indices

        )

    ]:

        predictions = (

            model.predict(X)

            .reshape(-1)

            .astype(np.int8)

        )

        probabilities = (

            model.predict_proba(X)[:, 1]

        )

        metrics = calculate_metrics(

            y,

            predictions,

            probabilities

        )

        metric_records.append({

            "model": model_name,

            "seed": SEED,

            "partition": partition,

            "records": len(y),

            **metrics

        })

        print("\n", partition.upper())

        for key in [

            "accuracy",

            "precision",

            "recall",

            "f1",

            "roc_auc"

        ]:

            print(

                f"{key:15s}: "

                f"{metrics[key]:.6f}"

            )

        print(

            "Confusion matrix:",

            [

                [
                    metrics["tn"],
                    metrics["fp"]
                ],

                [
                    metrics["fn"],
                    metrics["tp"]
                ]

            ]

        )

        np.savez_compressed(

            PREDICTION_DIR

            / (

                f"cell05_{model_name.lower()}"

                f"_{partition}_seed42.npz"

            ),

            row_id=ids,

            y_true=y,

            prediction=predictions,

            probability=probabilities

        )

    # Save original architecture model.

    joblib.dump(

        model,

        MODEL_DIR

        / f"cell05_{model_name.lower()}_seed42.joblib"

    )

    if model_name == "XGBoost":

        model.save_model(

            MODEL_DIR

            / "cell05_xgboost_seed42.ubj"

        )

    print(

        "\nTraining seconds:",

        round(training_seconds, 2)

    )

    print("Model saved successfully.")

    # Save partial results after each model.

    pd.DataFrame(

        metric_records

    ).to_csv(

        TABLE_DIR / "cell05_partial_metrics.csv",

        index=False

    )

    gc.collect()

# ------------------------------------------------------------
# 12. Final original model comparison
# ------------------------------------------------------------

metrics_df = pd.DataFrame(

    metric_records

)

metrics_df.to_csv(

    TABLE_DIR / "cell05_all_model_metrics.csv",

    index=False

)

test_results = metrics_df.loc[

    metrics_df["partition"] == "test"

].copy()

test_results.to_csv(

    TABLE_DIR / "cell05_test_model_comparison.csv",

    index=False

)

print("\n" + "=" * 75)
print("FINAL MODEL COMPARISON")
print("=" * 75)

print(

    test_results[

        [

            "model",

            "accuracy",

            "precision",

            "recall",

            "f1",

            "roc_auc"

        ]

    ]

    .round(6)

    .to_string(index=False)

)

# ------------------------------------------------------------
# 13. Major Comment 5:
# Training-testing comparison
# ------------------------------------------------------------

gap_records = []

for model_name in models:

    subset = metrics_df.loc[

        metrics_df["model"] == model_name

    ].set_index("partition")

    for metric in [

        "accuracy",

        "precision",

        "recall",

        "f1",

        "roc_auc"

    ]:

        train_score = float(

            subset.loc["train_real", metric]

        )

        test_score = float(

            subset.loc["test", metric]

        )

        gap_records.append({

            "model": model_name,

            "metric": metric,

            "training": train_score,

            "testing": test_score,

            "train_test_gap":

                train_score - test_score

        })

gap_df = pd.DataFrame(gap_records)

gap_df.to_csv(

    TABLE_DIR / "cell05_generalization_gap.csv",

    index=False

)

print("\n" + "=" * 75)
print("TRAINING-TESTING PERFORMANCE COMPARISON")
print("=" * 75)

print(

    gap_df.round(6)

    .to_string(index=False)

)

# ------------------------------------------------------------
# 14. Save exact preprocessing distribution
# ------------------------------------------------------------

distribution_records = []

for partition, labels, synthetic in [

    ("train_real", y_train_real, 0),

    ("train_balanced", y_train, synthetic_count),

    ("test", y_test, 0)

]:

    counts = np.bincount(

        labels,

        minlength=2

    )

    distribution_records.append({

        "partition": partition,

        "nonfraud": int(counts[0]),

        "fraud": int(counts[1]),

        "total": len(labels),

        "synthetic_records": synthetic

    })

distribution_df = pd.DataFrame(

    distribution_records

)

distribution_df.to_csv(

    TABLE_DIR / "cell05_class_distribution.csv",

    index=False

)

print("\nFINAL CLASS DISTRIBUTION")

print(

    distribution_df.to_string(index=False)

)

# ------------------------------------------------------------
# 15. Verify original source integrity
# ------------------------------------------------------------

def sha256_file(path):

    digest = hashlib.sha256()

    with open(path, "rb") as f:

        for chunk in iter(

            lambda: f.read(1024 * 1024),

            b""

        ):

            digest.update(chunk)

    return digest.hexdigest()


with open(

    PREVIOUS_REPORT_DIR

    / "cell02_original_source_hashes.json"

) as f:

    original_hashes = json.load(f)

for item in original_hashes:

    assert (

        sha256_file(

            SOURCE_ROOT / item["file"]

        )

        == item["sha256"]

    )

print("\nOriginal source integrity: PASS")

# ------------------------------------------------------------
# 16. Reproducibility report
# ------------------------------------------------------------

library_versions = {

    package: metadata.version(package)

    for package in [

        "numpy",

        "pandas",

        "scikit-learn",

        "imbalanced-learn",

        "lightgbm",

        "catboost",

        "xgboost",

        "joblib"

    ]

}

report = {

    "cell": 5,

    "status": "COMPLETED",

    "revision_type": "Reviewer-only revision",

    "reviewer_comments": [

        "Major 5",

        "Major 7",

        "Major 10",

        "Minor 26"

    ],

    "random_seed": SEED,

    "dataset": str(FEATURED_PATH),

    "original_features": 42,

    "feature_names": feature_names,

    "original_feature_engineering_retained": True,

    "additional_feature_removal": False,

    "split_ratio": "80:20",

    "stratified": True,

    "split_unit": "claim",

    "provider_feature_retained": True,

    "provider_overlap": provider_overlap,

    "provider_independent_evaluation": False,

    "encoding_fit": "training only",

    "oversampling": "Original SMOTE",

    "smote_scope": "training only",

    "test_oversampled": False,

    "scaler_fit": "balanced training only",

    "smote_seconds": smote_seconds,

    "separate_validation_set": False,

    "original_hyperparameters": ORIGINAL_PARAMS,

    "hyperparameter_tuning": False,

    "new_model_added": False,

    "external_dataset_used": False,

    "original_source_modified": False,

    "dataset_distribution": distribution_records,

    "training_details": training_records,

    "python_version": sys.version,

    "operating_system": platform.platform(),

    "gpu": gpu.stdout.strip(),

    "logical_cpu_cores": os.cpu_count(),

    "ram_gb": round(

        psutil.virtual_memory().total / 1024**3,

        2

    ),

    "library_versions": library_versions,

    "limitations": [

        "Claims from the same provider may occur "
        "in both training and test sets.",

        "Provider identity is retained as an "
        "original predictor.",

        "The original featured dataset was "
        "constructed before splitting.",

        "Original standard SMOTE treats encoded "
        "categorical features as numeric."

    ],

    "elapsed_seconds":

        time.perf_counter() - START_TIME

}

with open(

    REPORT_DIR

    / "cell05_reviewer_only_report.json",

    "w"

) as f:

    json.dump(

        report,

        f,

        indent=4,

        default=str

    )

with open(

    REPORT_DIR

    / "cell05_original_hyperparameters.json",

    "w"

) as f:

    json.dump(

        ORIGINAL_PARAMS,

        f,

        indent=4

    )

# ------------------------------------------------------------
# 17. Save complete revised source code
# ------------------------------------------------------------

ip = get_ipython()

if ip is not None:

    (

        CODE_DIR

        / "cell05_reviewer_only_retraining.py"

    ).write_text(

        ip.history_manager.input_hist_raw[-1],

        encoding="utf-8"

    )

# ------------------------------------------------------------
# 18. Save checkpoint
# ------------------------------------------------------------

checkpoint = {

    "cell": 5,

    "status": "COMPLETED",

    "revision_workspace": str(REVISED_DIR),

    "original_predictors": 42,

    "train_real_records": len(y_train_real),

    "train_balanced_records": len(y_train),

    "test_records": len(y_test),

    "synthetic_training_records": synthetic_count,

    "original_models": [

        "LightGBM",

        "CatBoost",

        "XGBoost"

    ],

    "original_hyperparameters_retained": True,

    "original_feature_engineering_retained": True,

    "test_smote": False,

    "original_source_modified": False,

    "next_cell": 6

}

with open(

    CHECKPOINT_DIR / "cell05_completed.json",

    "w"

) as f:

    json.dump(

        checkpoint,

        f,

        indent=4

    )

gc.collect()

print("\n" + "=" * 75)
print("CELL 05 COMPLETED SUCCESSFULLY")
print("=" * 75)

print("Original 42 predictors: RETAINED")
print("Original feature engineering: RETAINED")
print("Original 80:20 split: RETAINED")
print("Stratified sampling: PASS")
print("Original SMOTE: TRAINING ONLY")
print("Original StandardScaler: RETAINED")
print("Original three classifiers: RETAINED")
print("Original hyperparameters: RETAINED")
print("Training-testing comparison: COMPLETED")
print("Hyperparameter tuning: NONE")
print("New feature engineering: NONE")
print("Additional dataset: NONE")
print("Original source integrity: PASS")
print("Models and results saved in Google Drive.")
print("Previous revision outputs preserved.")

print("\nRevision workspace:", REVISED_DIR)

print(

    "Execution seconds:",

    round(

        time.perf_counter() - START_TIME,

        2

    )

)

print("\nREADY FOR CELL 06/15")
```

```python
# ============================================================
# CELL 06/15
# REVIEWER-SPECIFIC BASELINE COMPARISON
# AND OVERFITTING ASSESSMENT
#
# Reviewer Comment: Major 5
#
# NO NEW MODEL TRAINING
# NO HYPERPARAMETER MODIFICATION
# NO DATASET MODIFICATION
# NO ADDITIONAL EXPERIMENTS
# ============================================================

import gc
import json
import time
import hashlib

from pathlib import Path

import numpy as np
import pandas as pd

from google.colab import drive

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix
)

START_TIME = time.perf_counter()

# ------------------------------------------------------------
# 1. Restore reviewer-only workspace
# ------------------------------------------------------------

drive.mount("/content/drive")

PROJECT_ROOT = Path(
    "/content/drive/MyDrive/Narayan_Patil_Reviewer_Revision"
)

SOURCE_ROOT = (
    PROJECT_ROOT
    / "01_Original_Project"
    / "SBN"
)

REVISED_DIR = (
    PROJECT_ROOT
    / "03_Reviewer_Only_Revision"
)

CODE_DIR = REVISED_DIR / "code"
DATA_DIR = REVISED_DIR / "data"
MODEL_DIR = REVISED_DIR / "models"
RESULT_DIR = REVISED_DIR / "results"
TABLE_DIR = RESULT_DIR / "tables"
PREDICTION_DIR = RESULT_DIR / "predictions"
REPORT_DIR = REVISED_DIR / "reports"
CHECKPOINT_DIR = REVISED_DIR / "checkpoints"

ORIGINAL_AUDIT_DIR = (
    PROJECT_ROOT
    / "02_Revised_Project"
    / "reports"
)

assert (
    CHECKPOINT_DIR / "cell05_completed.json"
).exists(), "Complete corrected Cell 05 first."

print("=" * 75)
print("CELL 06: REVIEWER-SPECIFIC BASELINE COMPARISON")
print("=" * 75)

print("\nWorkspace:", REVISED_DIR)

# ------------------------------------------------------------
# 2. Verify corrected Cell 05 checkpoint
# ------------------------------------------------------------

with open(
    CHECKPOINT_DIR / "cell05_completed.json"
) as f:

    previous_checkpoint = json.load(f)

assert previous_checkpoint["status"] == "COMPLETED"

assert previous_checkpoint["original_predictors"] == 42

assert previous_checkpoint["train_real_records"] == 446568

assert previous_checkpoint["train_balanced_records"] == 552664

assert previous_checkpoint["test_records"] == 111643

assert previous_checkpoint["test_smote"] is False

assert previous_checkpoint[
    "original_hyperparameters_retained"
] is True

assert previous_checkpoint[
    "original_feature_engineering_retained"
] is True

print("\nCorrected Cell 05 checkpoint: VERIFIED")

# ------------------------------------------------------------
# 3. Load existing Cell 05 results
# ------------------------------------------------------------

metrics_df = pd.read_csv(
    TABLE_DIR / "cell05_all_model_metrics.csv"
)

with open(
    REPORT_DIR / "cell05_reviewer_only_report.json"
) as f:

    previous_report = json.load(f)

with open(
    REPORT_DIR / "cell05_original_hyperparameters.json"
) as f:

    original_parameters = json.load(f)

MODEL_NAMES = [
    "LightGBM",
    "CatBoost",
    "XGBoost"
]

METRICS = [
    "accuracy",
    "precision",
    "recall",
    "f1",
    "roc_auc"
]

assert len(metrics_df) == 6

assert set(metrics_df["model"]) == set(MODEL_NAMES)

assert set(metrics_df["partition"]) == {
    "train_real",
    "test"
}

assert not metrics_df.duplicated(
    subset=["model", "partition"]
).any()

print("\nExisting model results: LOADED")
print("Original models:", MODEL_NAMES)
print("Original predictors: 42")

# ------------------------------------------------------------
# 4. Verify real training and testing labels
# ------------------------------------------------------------

reference_data = {}

dataset_files = {

    "train_real":
        "cell05_train_real.npz",

    "test":
        "cell05_test.npz"

}

for partition, filename in dataset_files.items():

    with np.load(
        DATA_DIR / filename,
        allow_pickle=False
    ) as data:

        reference_data[partition] = {

            "y": data["y"].copy(),

            "row_id": data["row_id"].copy(),

            "feature_names":
                data["feature_names"].tolist()

        }

assert len(
    reference_data["train_real"]["y"]
) == 446568

assert len(
    reference_data["test"]["y"]
) == 111643

assert len(
    reference_data["test"]["feature_names"]
) == 42

assert reference_data[
    "train_real"
]["feature_names"] == reference_data[
    "test"
]["feature_names"]

assert set(
    reference_data["train_real"]["row_id"]
).isdisjoint(
    reference_data["test"]["row_id"]
)

assert np.bincount(
    reference_data["train_real"]["y"],
    minlength=2
).tolist() == [276332, 170236]

assert np.bincount(
    reference_data["test"]["y"],
    minlength=2
).tolist() == [69083, 42560]

print("\nTraining records: 446,568")
print("Testing records : 111,643")
print("Feature alignment: PASS")
print("Row overlap: ZERO")
print("Test class distribution: VERIFIED")

# ------------------------------------------------------------
# 5. Independently recalculate saved performance metrics
#
# No model retraining.
# Use predictions already generated in Cell 05.
# ------------------------------------------------------------

def calculate_metrics(
    y_true,
    predictions,
    probabilities
):

    tn, fp, fn, tp = confusion_matrix(
        y_true,
        predictions,
        labels=[0, 1]
    ).ravel()

    return {

        "accuracy": float(
            accuracy_score(
                y_true,
                predictions
            )
        ),

        "precision": float(
            precision_score(
                y_true,
                predictions,
                zero_division=0
            )
        ),

        "recall": float(
            recall_score(
                y_true,
                predictions,
                zero_division=0
            )
        ),

        "f1": float(
            f1_score(
                y_true,
                predictions,
                zero_division=0
            )
        ),

        "roc_auc": float(
            roc_auc_score(
                y_true,
                probabilities
            )
        ),

        "tn": int(tn),
        "fp": int(fp),
        "fn": int(fn),
        "tp": int(tp)

    }


verification_records = []

print("\n" + "=" * 75)
print("SAVED PERFORMANCE VERIFICATION")
print("=" * 75)

for model_name in MODEL_NAMES:

    for partition in [
        "train_real",
        "test"
    ]:

        prediction_path = (

            PREDICTION_DIR

            / (
                f"cell05_{model_name.lower()}"
                f"_{partition}_seed42.npz"
            )

        )

        assert prediction_path.exists()

        with np.load(
            prediction_path,
            allow_pickle=False
        ) as data:

            y_true = data["y_true"]

            predictions = data["prediction"]

            probabilities = data["probability"]

            row_ids = data["row_id"]

            assert np.array_equal(
                y_true,
                reference_data[partition]["y"]
            )

            assert np.array_equal(
                row_ids,
                reference_data[partition]["row_id"]
            )

            assert np.isfinite(
                probabilities
            ).all()

            assert (
                (probabilities >= 0)
                & (probabilities <= 1)
            ).all()

            calculated = calculate_metrics(
                y_true,
                predictions,
                probabilities
            )

        stored = metrics_df.loc[

            (
                metrics_df["model"]
                == model_name
            )

            & (
                metrics_df["partition"]
                == partition
            )

        ].iloc[0]

        for metric in METRICS:

            assert np.isclose(

                calculated[metric],

                float(stored[metric]),

                atol=1e-8,

                rtol=1e-8

            ), (
                f"Metric mismatch: "
                f"{model_name}, "
                f"{partition}, "
                f"{metric}"
            )

        for count in [
            "tn", "fp", "fn", "tp"
        ]:

            assert calculated[count] == int(
                stored[count]
            )

        verification_records.append({

            "model": model_name,

            "partition": partition,

            "records": len(y_true),

            **calculated,

            "verification": "PASS"

        })

        print(
            f"PASS | {model_name:10s}"
            f" | {partition:12s}"
            f" | Accuracy: "
            f"{calculated['accuracy']:.6f}"
            f" | ROC-AUC: "
            f"{calculated['roc_auc']:.6f}"
        )

verification_df = pd.DataFrame(
    verification_records
)

verification_df.to_csv(

    TABLE_DIR
    / "cell06_verified_model_metrics.csv",

    index=False

)

# ------------------------------------------------------------
# 6. Reviewer-required training-testing comparison
# ------------------------------------------------------------

comparison_records = []

for model_name in MODEL_NAMES:

    subset = verification_df.loc[

        verification_df["model"] == model_name

    ].set_index("partition")

    record = {
        "model": model_name
    }

    for metric in METRICS:

        train_value = float(
            subset.loc["train_real", metric]
        )

        test_value = float(
            subset.loc["test", metric]
        )

        record[
            f"training_{metric}"
        ] = train_value

        record[
            f"testing_{metric}"
        ] = test_value

        record[
            f"{metric}_gap_pp"
        ] = (

            train_value - test_value

        ) * 100

    comparison_records.append(record)

comparison_df = pd.DataFrame(
    comparison_records
)

comparison_df.to_csv(

    TABLE_DIR
    / "cell06_training_testing_comparison.csv",

    index=False

)

print("\n" + "=" * 75)
print("TRAINING-TESTING PERFORMANCE GAPS")
print("=" * 75)

gap_columns = [

    "model",

    "accuracy_gap_pp",

    "precision_gap_pp",

    "recall_gap_pp",

    "f1_gap_pp",

    "roc_auc_gap_pp"

]

print(

    comparison_df[gap_columns]

    .round(6)

    .to_string(index=False)

)

print(
    "\nAll performance gaps are expressed "
    "in percentage points."
)

# ------------------------------------------------------------
# 7. Generate reviewer-ready performance table
# ------------------------------------------------------------

publication_table = verification_df[

    [

        "model",

        "partition",

        "accuracy",

        "precision",

        "recall",

        "f1",

        "roc_auc"

    ]

].copy()

publication_table["partition"] = (

    publication_table["partition"]

    .replace({

        "train_real": "Training",

        "test": "Testing"

    })

)

for metric in METRICS:

    publication_table[metric] = (

        publication_table[metric] * 100

    ).round(2)

publication_table = publication_table.rename(

    columns={

        "model": "Model",

        "partition": "Dataset",

        "accuracy": "Accuracy (%)",

        "precision": "Precision (%)",

        "recall": "Recall (%)",

        "f1": "F1-score (%)",

        "roc_auc": "ROC-AUC (%)"

    }

)

publication_table.to_csv(

    TABLE_DIR
    / "cell06_manuscript_performance_table.csv",

    index=False

)

print("\n" + "=" * 75)
print("MANUSCRIPT PERFORMANCE TABLE")
print("=" * 75)

print(
    publication_table.to_string(
        index=False
    )
)

# ------------------------------------------------------------
# 8. Preserve original framework classifier
# ------------------------------------------------------------

test_results = verification_df.loc[

    verification_df["partition"] == "test"

].copy()

xgb_result = test_results.loc[

    test_results["model"] == "XGBoost"

].iloc[0]

xgb_comparison = comparison_df.loc[

    comparison_df["model"] == "XGBoost"

].iloc[0]

print("\n" + "=" * 75)
print("ORIGINAL XGBOOST FRAMEWORK")
print("=" * 75)

print("Framework classifier: XGBoost")

print(
    "Testing accuracy:",
    round(
        xgb_result["accuracy"] * 100,
        4
    ),
    "%"
)

print(
    "Testing ROC-AUC:",
    round(
        xgb_result["roc_auc"],
        6
    )
)

print(
    "Accuracy gap:",
    round(
        xgb_comparison["accuracy_gap_pp"],
        6
    ),
    "percentage points"
)

print(
    "ROC-AUC gap:",
    round(
        xgb_comparison["roc_auc_gap_pp"],
        6
    ),
    "percentage points"
)

# ------------------------------------------------------------
# 9. Major Comment 5:
# Document FraudNetX comparison limitations
#
# No fabricated external training scores.
# No unsupported overfitting claim.
# ------------------------------------------------------------

fraudnetx_assessment = {

    "reviewer_comment": "Major 5",

    "external_baseline": "FraudNetX",

    "accuracy_reported_in_original_manuscript":
        0.9991,

    "external_accuracy_independently_verified":
        False,

    "external_training_metrics_available":
        False,

    "external_baseline_reproduced":
        False,

    "external_overfitting_established":
        False,

    "architecture_matched_external_comparison":
        False,

    "original_framework_classifier":
        "XGBoost",

    "internal_comparison_models":
        MODEL_NAMES,

    "internal_comparison_conditions": {

        "same_dataset": True,

        "same_train_test_split": True,

        "same_preprocessing": True,

        "same_oversampling_scope": True,

        "same_evaluation_metrics": True

    },

    "revision_action": (

        "Remove the unsupported assertion that "
        "FraudNetX is overfitted. Explain the "
        "differences in model architecture, "
        "federated learning configuration and "
        "evaluation protocols. Treat published "
        "FraudNetX results as contextual rather "
        "than a controlled head-to-head comparison."

    ),

    "remaining_requirement": (

        "Clarify the federated learning type "
        "after implementation and include the "
        "comparison limitations in the manuscript."

    )

}

with open(

    REPORT_DIR
    / "cell06_fraudnetx_assessment.json",

    "w"

) as f:

    json.dump(

        fraudnetx_assessment,

        f,

        indent=4

    )

print("\n" + "=" * 75)
print("FRAUDNETX COMPARISON ASSESSMENT")
print("=" * 75)

print("Unsupported overfitting claim: REMOVE")
print("External reproduction: NOT PERFORMED")
print("Internal fair comparison: VERIFIED")
print("Architecture differences: DOCUMENTED")

# ------------------------------------------------------------
# 10. Record evaluation limitations
#
# Preserve the original claim-wise methodology.
# Do not claim provider-independent generalization.
# ------------------------------------------------------------

provider_overlap = int(
    previous_report["provider_overlap"]
)

limitations = {

    "split_type":
        "Stratified claim-wise 80:20",

    "provider_overlap":
        provider_overlap,

    "provider_identifier_retained":
        True,

    "provider_independent_evaluation":
        False,

    "test_oversampled":
        False,

    "interpretation": (

        "The small training-testing performance "
        "gaps describe generalization under the "
        "original claim-wise evaluation protocol. "
        "Shared providers and retained provider "
        "identifiers may inflate predictive "
        "performance. These results do not establish "
        "generalization to unseen providers or "
        "prove the absence of overfitting."

    )

}

with open(

    REPORT_DIR
    / "cell06_evaluation_limitations.json",

    "w"

) as f:

    json.dump(

        limitations,

        f,

        indent=4

    )

print("\nEVALUATION SCOPE")

print("Claim-wise stratified evaluation")
print("Provider overlap:", provider_overlap)
print("Test oversampling: NONE")

# ------------------------------------------------------------
# 11. Verify original source integrity
# ------------------------------------------------------------

def sha256_file(path):

    digest = hashlib.sha256()

    with open(path, "rb") as f:

        for chunk in iter(

            lambda: f.read(1024 * 1024),

            b""

        ):

            digest.update(chunk)

    return digest.hexdigest()


with open(

    ORIGINAL_AUDIT_DIR
    / "cell02_original_source_hashes.json"

) as f:

    original_hashes = json.load(f)

for item in original_hashes:

    path = SOURCE_ROOT / item["file"]

    assert path.exists()

    assert (

        sha256_file(path)

        == item["sha256"]

    ), f"Original source modified: {path}"

print("\nOriginal source integrity: PASS")

# ------------------------------------------------------------
# 12. Verify trained models remain available
# ------------------------------------------------------------

model_inventory = []

for model_name in MODEL_NAMES:

    path = (

        MODEL_DIR

        / f"cell05_{model_name.lower()}_seed42.joblib"

    )

    assert path.exists()

    model_inventory.append({

        "model": model_name,

        "path": str(path),

        "size_bytes": path.stat().st_size,

        "sha256": sha256_file(path)

    })

print("All three trained models: AVAILABLE")

# ------------------------------------------------------------
# 13. Save complete reviewer-specific report
# ------------------------------------------------------------

report = {

    "cell": 6,

    "status": "COMPLETED",

    "reviewer_comment": "Major 5",

    "revision_type": "Reviewer-only",

    "original_predictors": 42,

    "original_models": MODEL_NAMES,

    "original_framework_classifier":
        "XGBoost",

    "original_hyperparameters":
        original_parameters,

    "training_testing_comparison":
        comparison_records,

    "prediction_verification":
        "PASS",

    "fraudnetx_assessment":
        fraudnetx_assessment,

    "evaluation_limitations":
        limitations,

    "model_inventory":
        model_inventory,

    "new_models_trained": 0,

    "hyperparameters_modified":
        False,

    "dataset_modified":
        False,

    "feature_engineering_modified":
        False,

    "original_source_modified":
        False,

    "additional_dataset_used":
        False,

    "statistical_testing":

        "Pending Major Comment 6",

    "elapsed_seconds":

        time.perf_counter() - START_TIME

}

with open(

    REPORT_DIR
    / "cell06_baseline_analysis_report.json",

    "w"

) as f:

    json.dump(

        report,

        f,

        indent=4,

        default=str

    )

# ------------------------------------------------------------
# 14. Save complete Cell 06 code
# ------------------------------------------------------------

ip = get_ipython()

if ip is not None:

    (

        CODE_DIR

        / "cell06_reviewer_only_baseline_analysis.py"

    ).write_text(

        ip.history_manager.input_hist_raw[-1],

        encoding="utf-8"

    )

# ------------------------------------------------------------
# 15. Save completion checkpoint
# ------------------------------------------------------------

checkpoint = {

    "cell": 6,

    "status": "COMPLETED",

    "reviewer_comment": "Major 5",

    "revision_workspace":
        str(REVISED_DIR),

    "original_predictors": 42,

    "internal_baseline_comparison":
        "COMPLETED",

    "training_testing_comparison":
        "COMPLETED",

    "prediction_verification":
        "PASS",

    "fraudnetx_overfitting_claim_supported":
        False,

    "fraudnetx_reproduced":
        False,

    "original_framework_classifier":
        "XGBoost",

    "new_models_trained": 0,

    "original_code_modified":
        False,

    "next_cell": 7

}

with open(

    CHECKPOINT_DIR
    / "cell06_completed.json",

    "w"

) as f:

    json.dump(

        checkpoint,

        f,

        indent=4

    )

gc.collect()

print("\n" + "=" * 75)
print("CELL 06 COMPLETED SUCCESSFULLY")
print("=" * 75)

print("Existing numerical results: VERIFIED")
print("Training-testing comparison: COMPLETED")
print("Internal baseline comparison: VERIFIED")
print("FraudNetX limitations: DOCUMENTED")
print("Original 42 predictors: RETAINED")
print("Original hyperparameters: RETAINED")
print("Original source integrity: PASS")
print("New model training: NONE")
print("Dataset modification: NONE")
print("Additional experiments: NONE")
print("All reports and tables saved in Google Drive.")
print("Complete revised code saved.")
print("Checkpoint saved.")

print(

    "\nTotal execution seconds:",

    round(

        time.perf_counter() - START_TIME,

        2

    )

)

print(

    "\nREADY FOR CELL 07/15: "
    "FEDERATED LEARNING IMPLEMENTATION"

)
```

```python
# ============================================================
# CELL 07/15
# MULTI-NODE HORIZONTAL FEDERATED XGBOOST
#
# Reviewer Comments:
# Major 1: Multi-node federated experiments
# Major 8: Federated aggregation
# Major 5: Federated learning type
#
# ORIGINAL 42 PREDICTORS RETAINED
# ORIGINAL XGBOOST HYPERPARAMETERS RETAINED
# NO ADDITIONAL DATASET
# NO HYPERPARAMETER OPTIMIZATION
# ============================================================

import gc
import json
import time
import hashlib
import subprocess

from pathlib import Path

import numpy as np
import pandas as pd
import xgboost as xgb
import psutil

from google.colab import drive

from sklearn.model_selection import StratifiedKFold

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix
)

from scipy.special import expit

# ------------------------------------------------------------
# 1. Restore reviewer-only workspace
# ------------------------------------------------------------

drive.mount("/content/drive")

START_TIME = time.perf_counter()

SEED = 42

CLIENT_COUNTS = [3, 5, 10]

COMMUNICATION_ROUNDS = 10

LOCAL_BOOSTING_ROUNDS = 40

PROJECT_ROOT = Path(
    "/content/drive/MyDrive/Narayan_Patil_Reviewer_Revision"
)

SOURCE_ROOT = (
    PROJECT_ROOT
    / "01_Original_Project"
    / "SBN"
)

REVISED_DIR = (
    PROJECT_ROOT
    / "03_Reviewer_Only_Revision"
)

CODE_DIR = REVISED_DIR / "code"
DATA_DIR = REVISED_DIR / "data"
MODEL_DIR = REVISED_DIR / "models"
RESULT_DIR = REVISED_DIR / "results"
TABLE_DIR = RESULT_DIR / "tables"
PREDICTION_DIR = RESULT_DIR / "predictions"
FIGURE_DIR = RESULT_DIR / "figures"
REPORT_DIR = REVISED_DIR / "reports"
CHECKPOINT_DIR = REVISED_DIR / "checkpoints"

FL_MODEL_DIR = MODEL_DIR / "federated"

for folder in [
    CODE_DIR,
    DATA_DIR,
    MODEL_DIR,
    TABLE_DIR,
    PREDICTION_DIR,
    FIGURE_DIR,
    REPORT_DIR,
    CHECKPOINT_DIR,
    FL_MODEL_DIR
]:

    folder.mkdir(
        parents=True,
        exist_ok=True
    )

assert (
    CHECKPOINT_DIR / "cell06_completed.json"
).exists(), "Complete Cell 06 first."

print("=" * 75)
print("CELL 07: MULTI-NODE FEDERATED XGBOOST")
print("=" * 75)

print("\nWorkspace:", REVISED_DIR)

# ------------------------------------------------------------
# 2. Verify GPU availability
# ------------------------------------------------------------

gpu_result = subprocess.run(

    [
        "nvidia-smi",
        "--query-gpu=name,memory.total",
        "--format=csv,noheader"
    ],

    capture_output=True,
    text=True

)

assert gpu_result.returncode == 0, (
    "Enable Tesla T4 GPU."
)

print("\nGPU:", gpu_result.stdout.strip())

print(
    "Available RAM:",
    round(
        psutil.virtual_memory().available
        / 1024**3,
        2
    ),
    "GB"
)

# ------------------------------------------------------------
# 3. Restore original XGBoost hyperparameters
# ------------------------------------------------------------

with open(
    REPORT_DIR
    / "cell05_original_hyperparameters.json"
) as f:

    original_params = json.load(f)["XGBoost"]

assert original_params["n_estimators"] == 400

assert original_params["learning_rate"] == 0.05

assert original_params["max_depth"] == 8

assert (
    LOCAL_BOOSTING_ROUNDS
    * COMMUNICATION_ROUNDS
    == original_params["n_estimators"]
)

print("\nOriginal XGBoost parameters:")

for key, value in original_params.items():

    print(f"{key}: {value}")

# XGBoost native training API.

FL_PARAMS = {

    "objective": "binary:logistic",

    "eta": original_params["learning_rate"],

    "max_depth": original_params["max_depth"],

    "subsample": original_params["subsample"],

    "colsample_bytree":
        original_params["colsample_bytree"],

    "eval_metric":
        original_params["eval_metric"],

    "seed": SEED,

    "tree_method": "hist",

    "device": "cuda",

    "nthread": 2,

    # Explicit zero initial log-odds.
    "base_score": 0.5,

    "verbosity": 0

}

# ------------------------------------------------------------
# 4. Load existing real training and test datasets
#
# Synthetic records are not distributed among clients.
# No preprocessing or feature engineering is changed.
# ------------------------------------------------------------

def load_dataset(filename):

    with np.load(

        DATA_DIR / filename,

        allow_pickle=False

    ) as data:

        X = data["X"].astype(
            np.float32
        )

        y = data["y"].astype(
            np.int8
        )

        row_ids = data["row_id"]

        names = data[
            "feature_names"
        ].tolist()

    assert np.isfinite(X).all()

    assert np.all(row_ids >= 0)

    return X, y, row_ids, names


X_train, y_train, train_ids, feature_names = (
    load_dataset("cell05_train_real.npz")
)

X_test, y_test, test_ids, test_names = (
    load_dataset("cell05_test.npz")
)

assert X_train.shape == (446568, 42)

assert X_test.shape == (111643, 42)

assert feature_names == test_names

assert set(train_ids).isdisjoint(
    test_ids
)

assert np.bincount(
    y_train
).tolist() == [276332, 170236]

assert np.bincount(
    y_test
).tolist() == [69083, 42560]

print("\nFEDERATED DATASET")

print("Real training records:", len(y_train))

print("Testing records:", len(y_test))

print("Original predictors:", len(feature_names))

print("Synthetic client records: 0")

print("Feature alignment: PASS")

print("Training-testing row overlap: ZERO")

# ------------------------------------------------------------
# 5. Horizontal client partitioning
#
# Each simulated client has different training rows
# but the same original feature space.
#
# Stratified, approximately IID partitioning.
# ------------------------------------------------------------

def create_clients(
    X,
    y,
    n_clients,
    seed=42
):

    splitter = StratifiedKFold(

        n_splits=n_clients,

        shuffle=True,

        random_state=seed

    )

    client_indices = []

    for _, local_indices in splitter.split(
        X, y
    ):

        client_indices.append(
            local_indices
        )

    combined = np.concatenate(
        client_indices
    )

    assert len(combined) == len(y)

    assert len(
        np.unique(combined)
    ) == len(y)

    assert np.array_equal(

        np.sort(combined),

        np.arange(len(y))

    )

    return client_indices

# ------------------------------------------------------------
# 6. Shared original evaluation metrics
# ------------------------------------------------------------

def calculate_metrics(
    y_true,
    probability
):

    probability = np.asarray(
        probability
    ).reshape(-1)

    assert np.isfinite(
        probability
    ).all()

    prediction = (
        probability >= 0.5
    ).astype(np.int8)

    tn, fp, fn, tp = confusion_matrix(

        y_true,

        prediction,

        labels=[0, 1]

    ).ravel()

    return {

        "accuracy": float(
            accuracy_score(
                y_true, prediction
            )
        ),

        "precision": float(
            precision_score(
                y_true,
                prediction,
                zero_division=0
            )
        ),

        "recall": float(
            recall_score(
                y_true,
                prediction,
                zero_division=0
            )
        ),

        "f1": float(
            f1_score(
                y_true,
                prediction,
                zero_division=0
            )
        ),

        "roc_auc": float(
            roc_auc_score(
                y_true,
                probability
            )
        ),

        "tn": int(tn),
        "fp": int(fp),
        "fn": int(fn),
        "tp": int(tp)

    }, prediction

# ------------------------------------------------------------
# 7. Local tree-update prediction
#
# A zero base margin isolates the additive tree
# contribution from the previous global prediction.
# ------------------------------------------------------------

def predict_tree_update(
    booster,
    dmatrix
):

    dmatrix.set_base_margin(

        np.zeros(
            dmatrix.num_row(),
            dtype=np.float32
        )

    )

    update = booster.predict(

        dmatrix,

        output_margin=True

    ).astype(np.float32)

    assert np.isfinite(
        update
    ).all()

    return update

# ------------------------------------------------------------
# 8. Federated training configuration
#
# Global additive update:
#
# F_(r+1)(x) = F_r(x)
#              + sum_k (n_k/N) * U_(k,r)(x)
#
# U_(k,r) is the local XGBoost tree update.
#
# Local training uses current global margins.
# The server aggregates serialized tree updates.
#
# This is tree-ensemble aggregation, NOT
# neural-network parameter averaging (FedAvg).
# ------------------------------------------------------------

all_round_records = []

all_client_records = []

all_summary_records = []

all_communication_records = []

centralized_metrics = pd.read_csv(

    TABLE_DIR
    / "cell05_all_model_metrics.csv"

)

centralized_xgb = centralized_metrics.loc[

    (
        centralized_metrics["model"]
        == "XGBoost"
    )

    & (

        centralized_metrics["partition"]
        == "test"

    )

].iloc[0]

print("\nExisting centralized XGBoost:")

print(
    "Accuracy:",
    round(
        centralized_xgb["accuracy"],
        6
    )
)

print(
    "ROC-AUC:",
    round(
        centralized_xgb["roc_auc"],
        6
    )
)

print(
    "Note: Cell 05 centralized model used SMOTE."
)

print(
    "Matched real-data centralized comparison "
    "will be completed in Cell 08."
)

# ------------------------------------------------------------
# 9. Execute 3, 5 and 10-client experiments
# ------------------------------------------------------------

for n_clients in CLIENT_COUNTS:

    configuration_start = time.perf_counter()

    print("\n" + "=" * 75)

    print(
        f"FEDERATED CONFIGURATION: "
        f"{n_clients} CLIENTS"
    )

    print("=" * 75)

    config_dir = (
        FL_MODEL_DIR
        / f"clients_{n_clients}"
    )

    config_dir.mkdir(
        parents=True,
        exist_ok=True
    )

    client_indices = create_clients(

        X_train,

        y_train,

        n_clients,

        seed=SEED

    )

    client_sizes = np.asarray(

        [
            len(indices)
            for indices in client_indices
        ],

        dtype=np.int64

    )

    client_weights = (

        client_sizes
        / client_sizes.sum()

    )

    assert np.isclose(

        client_weights.sum(),

        1.0

    )

    local_matrices = []

    local_margins = []

    split_records = []

    # --------------------------------------------------------
    # Initialize private client partitions
    # --------------------------------------------------------

    for client_id, indices in enumerate(

        client_indices,

        start=1

    ):

        local_X = X_train[indices]

        local_y = y_train[indices]

        local_matrix = xgb.DMatrix(

            local_X,

            label=local_y,

            feature_names=feature_names

        )

        local_matrices.append(
            local_matrix
        )

        local_margins.append(

            np.zeros(

                len(indices),

                dtype=np.float32

            )

        )

        counts = np.bincount(

            local_y,

            minlength=2

        )

        record = {

            "n_clients": n_clients,

            "client_id": client_id,

            "records": len(indices),

            "nonfraud": int(counts[0]),

            "fraud": int(counts[1]),

            "weight": float(
                client_weights[client_id - 1]
            )

        }

        all_client_records.append(record)

        split_records.append(

            pd.DataFrame({

                "row_id": train_ids[indices],

                "client_id": client_id

            })

        )

        print(

            f"Client {client_id:02d}"
            f" | Records: {len(indices):,}"
            f" | Non-fraud: {counts[0]:,}"
            f" | Fraud: {counts[1]:,}"
            f" | Weight: "
            f"{client_weights[client_id - 1]:.6f}"

        )

        del local_X, local_y

    pd.concat(

        split_records,

        ignore_index=True

    ).to_csv(

        DATA_DIR
        / f"cell07_client_partition_{n_clients}.csv",

        index=False

    )

    del split_records
    gc.collect()

    # Independent evaluation matrix.

    test_matrix = xgb.DMatrix(

        X_test,

        feature_names=feature_names

    )

    global_test_margin = np.zeros(

        len(y_test),

        dtype=np.float32

    )

    config_round_records = []

    ensemble_manifest = []

    total_upload_bytes = 0

    total_download_bytes = 0

    total_local_training_seconds = 0.0

    total_aggregation_seconds = 0.0

    total_tree_count = 0

    # --------------------------------------------------------
    # Execute 10 communication rounds
    # --------------------------------------------------------

    for round_no in range(

        1,

        COMMUNICATION_ROUNDS + 1

    ):

        round_start = time.perf_counter()

        round_dir = (

            config_dir
            / f"round_{round_no:02d}"

        )

        round_dir.mkdir(

            parents=True,

            exist_ok=True

        )

        local_payloads = []

        round_training_seconds = 0.0

        round_upload_bytes = 0

        round_download_bytes = 0

        round_tree_count = 0

        # ----------------------------------------------------
        # Clients train independently from the same
        # previous global model margins.
        # ----------------------------------------------------

        for client_index in range(n_clients):

            client_id = client_index + 1

            local_matrix = local_matrices[
                client_index
            ]

            local_matrix.set_base_margin(

                local_margins[
                    client_index
                ]

            )

            local_start = time.perf_counter()

            booster = xgb.train(

                params=FL_PARAMS.copy(),

                dtrain=local_matrix,

                num_boost_round=
                    LOCAL_BOOSTING_ROUNDS,

                verbose_eval=False

            )

            local_seconds = (

                time.perf_counter()

                - local_start

            )

            round_training_seconds += (
                local_seconds
            )

            assert (

                booster.num_boosted_rounds()

                == LOCAL_BOOSTING_ROUNDS

            )

            round_tree_count += (
                booster.num_boosted_rounds()
            )

            # Simulate client-to-server transmission
            # using actual serialized model bytes.

            payload = bytes(

                booster.save_raw(
                    raw_format="ubj"
                )

            )

            round_upload_bytes += len(
                payload
            )

            local_payloads.append(
                payload
            )

            model_path = (

                round_dir

                / f"client_{client_id:02d}.ubj"

            )

            model_path.write_bytes(
                payload
            )

            ensemble_manifest.append({

                "round": round_no,

                "client_id": client_id,

                "weight": float(
                    client_weights[
                        client_index
                    ]
                ),

                "model_file": str(
                    model_path.relative_to(
                        config_dir
                    )
                ),

                "serialized_bytes":
                    len(payload),

                "trees":
                    LOCAL_BOOSTING_ROUNDS

            })

            del booster

        # ----------------------------------------------------
        # 10. Server aggregation and simulated broadcast
        #
        # Each client receives the weighted tree ensemble.
        # Client-side global margins are updated locally.
        # ----------------------------------------------------

        aggregation_start = (
            time.perf_counter()
        )

        broadcast_metadata = json.dumps({

            "round": round_no,

            "weights":
                client_weights.tolist(),

            "payload_lengths": [
                len(payload)
                for payload in local_payloads
            ]

        }).encode("utf-8")

        broadcast_bytes = (

            sum(
                len(payload)
                for payload in local_payloads
            )

            + len(broadcast_metadata)

        )

        # Full round ensemble sent to every client.

        round_download_bytes = (
            n_clients * broadcast_bytes
        )

        received_models = []

        for payload in local_payloads:

            received = xgb.Booster()

            received.load_model(
                bytearray(payload)
            )

            # CPU inference avoids CUDA mismatch
            # during DMatrix prediction.

            received.set_param({

                "device": "cpu",

                "nthread": 2

            })

            received_models.append(
                received
            )

        # Update global model margin on each
        # client using its own local records.

        for target_client in range(n_clients):

            target_matrix = local_matrices[
                target_client
            ]

            weighted_update = np.zeros(

                target_matrix.num_row(),

                dtype=np.float32

            )

            for source_client, received in enumerate(

                received_models

            ):

                update = predict_tree_update(

                    received,

                    target_matrix

                )

                weighted_update += (

                    client_weights[
                        source_client
                    ]

                    * update

                )

            local_margins[target_client] += (
                weighted_update
            )

            assert np.isfinite(

                local_margins[
                    target_client
                ]

            ).all()

        # Separate evaluation process applies the
        # same aggregated model to held-out claims.

        test_update = np.zeros(

            len(y_test),

            dtype=np.float32

        )

        for source_client, received in enumerate(

            received_models

        ):

            test_update += (

                client_weights[
                    source_client
                ]

                * predict_tree_update(

                    received,

                    test_matrix

                )

            )

        global_test_margin += test_update

        aggregation_seconds = (

            time.perf_counter()

            - aggregation_start

        )

        # ----------------------------------------------------
        # 11. Measure convergence after each round
        #
        # Fixed configuration; no test-driven tuning.
        # ----------------------------------------------------

        probability = expit(

            global_test_margin

        ).astype(np.float64)

        metrics, predictions = (
            calculate_metrics(

                y_test,

                probability

            )
        )

        total_upload_bytes += (
            round_upload_bytes
        )

        total_download_bytes += (
            round_download_bytes
        )

        total_local_training_seconds += (
            round_training_seconds
        )

        total_aggregation_seconds += (
            aggregation_seconds
        )

        total_tree_count += (
            round_tree_count
        )

        round_elapsed = (

            time.perf_counter()

            - round_start

        )

        round_record = {

            "n_clients": n_clients,

            "round": round_no,

            "seed": SEED,

            **metrics,

            "local_training_seconds":
                round_training_seconds,

            "aggregation_seconds":
                aggregation_seconds,

            "round_elapsed_seconds":
                round_elapsed,

            "upload_bytes":
                round_upload_bytes,

            "download_bytes":
                round_download_bytes,

            "total_communication_bytes":

                round_upload_bytes
                + round_download_bytes,

            "trees_added":
                round_tree_count,

            "cumulative_trees":
                total_tree_count

        }

        all_round_records.append(
            round_record
        )

        config_round_records.append(
            round_record
        )

        all_communication_records.append({

            "n_clients": n_clients,

            "round": round_no,

            "upload_bytes":
                round_upload_bytes,

            "download_bytes":
                round_download_bytes,

            "total_bytes":

                round_upload_bytes
                + round_download_bytes,

            "uploaded_tree_count":
                round_tree_count,

            "broadcast_tree_count":
                n_clients * round_tree_count

        })

        print(

            f"\nRound {round_no:02d}"
            f"/{COMMUNICATION_ROUNDS}"

            f" | Accuracy: "
            f"{metrics['accuracy']:.6f}"

            f" | Recall: "
            f"{metrics['recall']:.6f}"

            f" | F1: "
            f"{metrics['f1']:.6f}"

            f" | AUC: "
            f"{metrics['roc_auc']:.6f}"

            f" | Communication: "
            f"{(round_upload_bytes + round_download_bytes)/1024**2:.3f} MiB"

            f" | Time: "
            f"{round_elapsed:.2f}s",

            flush=True

        )

        # Save results after every round.

        pd.DataFrame(

            all_round_records

        ).to_csv(

            TABLE_DIR
            / "cell07_round_metrics_partial.csv",

            index=False

        )

        pd.DataFrame(

            all_communication_records

        ).to_csv(

            TABLE_DIR
            / "cell07_communication_partial.csv",

            index=False

        )

        with open(

            config_dir
            / "ensemble_manifest.json",

            "w"

        ) as f:

            json.dump(

                ensemble_manifest,

                f,

                indent=4

            )

        del (
            local_payloads,
            received_models,
            test_update
        )

        gc.collect()

    # --------------------------------------------------------
    # 12. Save final predictions for this configuration
    # --------------------------------------------------------

    assert len(
        config_round_records
    ) == COMMUNICATION_ROUNDS

    np.savez_compressed(

        PREDICTION_DIR
        / f"cell07_fl_{n_clients}_clients_seed42.npz",

        row_id=test_ids,

        y_true=y_test,

        prediction=predictions,

        probability=probability,

        margin=global_test_margin

    )

    config_elapsed = (

        time.perf_counter()

        - configuration_start

    )

    summary_record = {

        "n_clients": n_clients,

        "seed": SEED,

        "rounds":
            COMMUNICATION_ROUNDS,

        "training_records":
            len(y_train),

        "test_records":
            len(y_test),

        "features":
            len(feature_names),

        "local_iterations_per_client":

            LOCAL_BOOSTING_ROUNDS
            * COMMUNICATION_ROUNDS,

        "total_ensemble_trees":
            total_tree_count,

        **metrics,

        "total_local_training_seconds":
            total_local_training_seconds,

        "total_aggregation_seconds":
            total_aggregation_seconds,

        "total_elapsed_seconds":
            config_elapsed,

        "total_upload_bytes":
            total_upload_bytes,

        "total_download_bytes":
            total_download_bytes,

        "total_communication_bytes":

            total_upload_bytes
            + total_download_bytes

    }

    all_summary_records.append(
        summary_record
    )

    pd.DataFrame(

        all_summary_records

    ).to_csv(

        TABLE_DIR
        / "cell07_summary_partial.csv",

        index=False

    )

    print("\nFINAL CONFIGURATION RESULT")

    print("Clients:", n_clients)

    print(
        "Accuracy:",
        round(metrics["accuracy"], 6)
    )

    print(
        "Recall:",
        round(metrics["recall"], 6)
    )

    print(
        "ROC-AUC:",
        round(metrics["roc_auc"], 6)
    )

    print(
        "Total execution seconds:",
        round(config_elapsed, 2)
    )

    print(
        "Total communication MiB:",
        round(
            (
                total_upload_bytes
                + total_download_bytes
            ) / 1024**2,
            3
        )
    )

    print(
        "Federated ensemble saved:",
        config_dir
    )

    del (
        local_matrices,
        local_margins,
        test_matrix,
        client_indices,
        global_test_margin
    )

    gc.collect()

# ------------------------------------------------------------
# 13. Consolidate final federated experiment results
# ------------------------------------------------------------

round_df = pd.DataFrame(
    all_round_records
)

summary_df = pd.DataFrame(
    all_summary_records
)

client_df = pd.DataFrame(
    all_client_records
)

communication_df = pd.DataFrame(
    all_communication_records
)

assert len(round_df) == 30

assert len(summary_df) == 3

assert len(client_df) == 18

assert len(communication_df) == 30

summary_df[
    "communication_mib"
] = (

    summary_df["total_communication_bytes"]

    / 1024**2

)

round_df.to_csv(

    TABLE_DIR
    / "cell07_federated_round_metrics.csv",

    index=False

)

summary_df.to_csv(

    TABLE_DIR
    / "cell07_federated_summary.csv",

    index=False

)

client_df.to_csv(

    TABLE_DIR
    / "cell07_client_distribution.csv",

    index=False

)

communication_df.to_csv(

    TABLE_DIR
    / "cell07_communication_overhead.csv",

    index=False

)

print("\n" + "=" * 75)
print("FINAL FEDERATED MODEL COMPARISON")
print("=" * 75)

print(

    summary_df[

        [

            "n_clients",

            "rounds",

            "accuracy",

            "precision",

            "recall",

            "f1",

            "roc_auc",

            "total_elapsed_seconds",

            "communication_mib"

        ]

    ]

    .round(6)

    .to_string(index=False)

)

# ------------------------------------------------------------
# 14. Reviewer-required convergence curves
# ------------------------------------------------------------

import matplotlib.pyplot as plt

for metric, ylabel in [

    ("accuracy", "Accuracy (%)"),

    ("recall", "Recall (%)")

]:

    plt.figure(
        figsize=(7, 4.5)
    )

    markers = {

        3: "o",

        5: "s",

        10: "^"

    }

    line_styles = {

        3: "-",

        5: "--",

        10: "-."

    }

    for n_clients in CLIENT_COUNTS:

        subset = round_df.loc[

            round_df["n_clients"]
            == n_clients

        ]

        plt.plot(

            subset["round"],

            subset[metric] * 100,

            color="black",

            marker=markers[n_clients],

            linestyle=line_styles[n_clients],

            linewidth=1.5,

            markersize=4,

            label=f"{n_clients} clients"

        )

    plt.xlabel(
        "Communication round"
    )

    plt.ylabel(ylabel)

    plt.xticks(

        range(

            1,

            COMMUNICATION_ROUNDS + 1

        )

    )

    plt.legend()

    plt.grid(

        alpha=0.25,

        linestyle=":"

    )

    plt.tight_layout()

    plt.savefig(

        FIGURE_DIR

        / f"cell07_federated_{metric}_convergence.png",

        dpi=600,

        bbox_inches="tight"

    )

    plt.show()

    plt.close()

# ------------------------------------------------------------
# 15. Save aggregation implementation details
# ------------------------------------------------------------

aggregation_report = {

    "federated_type":
        "Horizontal federated learning",

    "partitioning_strategy":

        "Stratified approximately IID "
        "claim-level partitioning",

    "client_configurations":
        CLIENT_COUNTS,

    "communication_rounds":
        COMMUNICATION_ROUNDS,

    "local_boosting_rounds":
        LOCAL_BOOSTING_ROUNDS,

    "total_local_boosting_rounds_per_client":
        400,

    "aggregation_method":

        "Sample-size-weighted additive "
        "XGBoost tree updates",

    "aggregation_formula":

        "F_(r+1)(x) = F_r(x) + "
        "sum_k[(n_k/N) * U_(k,r)(x)]",

    "aggregation_unit":
        "Additive tree-model functions",

    "standard_neural_network_fedavg":
        False,

    "global_margin_used_for_local_training":
        True,

    "model_parameters_transmitted":
        "Serialized XGBoost tree updates",

    "communication_measurement":

        "Actual serialized model payload bytes "
        "plus broadcast weight metadata",

    "physical_network_used":
        False,

    "execution_mode":

        "Sequential in-process simulation "
        "of independent clients",

    "real_training_records":
        len(y_train),

    "synthetic_client_records":
        0,

    "test_records":
        len(y_test),

    "features":
        feature_names,

    "preprocessing":

        "Reused centralized preprocessing "
        "from corrected Cell 05",

    "differential_privacy_enabled":
        False,

    "blockchain_enabled":
        False,

    "test_metrics_used_for_tuning":
        False,

    "limitations": [

        "Client records are simulated partitions "
        "of a single public dataset.",

        "Preprocessing was fitted centrally.",

        "Provider identity remains an original "
        "predictor and providers overlap between "
        "training and testing.",

        "Communication measurements exclude "
        "physical network protocol overhead "
        "and real network latency.",

        "Global ensemble tree count increases "
        "with the number of clients.",

        "The current experiment does not "
        "establish differential privacy."

    ]

}

with open(

    REPORT_DIR
    / "cell07_federated_aggregation_report.json",

    "w"

) as f:

    json.dump(

        aggregation_report,

        f,

        indent=4

    )

# ------------------------------------------------------------
# 16. Verify original source integrity
# ------------------------------------------------------------

def sha256_file(path):

    digest = hashlib.sha256()

    with open(path, "rb") as f:

        for chunk in iter(

            lambda: f.read(1024 * 1024),

            b""

        ):

            digest.update(chunk)

    return digest.hexdigest()


with open(

    PROJECT_ROOT

    / "02_Revised_Project"

    / "reports"

    / "cell02_original_source_hashes.json"

) as f:

    original_hashes = json.load(f)

for item in original_hashes:

    assert (

        sha256_file(

            SOURCE_ROOT / item["file"]

        )

        == item["sha256"]

    )

print("\nOriginal source integrity: PASS")

# ------------------------------------------------------------
# 17. Save complete reproducibility report
# ------------------------------------------------------------

report = {

    "cell": 7,

    "status": "COMPLETED",

    "reviewer_comments": [

        "Major 1",

        "Major 5",

        "Major 8"

    ],

    "seed": SEED,

    "gpu": gpu_result.stdout.strip(),

    "original_predictors": 42,

    "original_xgboost_parameters":
        original_params,

    "effective_federated_parameters":
        FL_PARAMS,

    "aggregation_details":
        aggregation_report,

    "federated_results":
        all_summary_records,

    "new_dataset_used":
        False,

    "feature_engineering_modified":
        False,

    "hyperparameter_tuning":
        False,

    "original_source_modified":
        False,

    "matched_centralized_comparison":

        "Pending Cell 08",

    "elapsed_seconds":

        time.perf_counter()
        - START_TIME

}

with open(

    REPORT_DIR
    / "cell07_federated_experiment_report.json",

    "w"

) as f:

    json.dump(

        report,

        f,

        indent=4,

        default=str

    )

# ------------------------------------------------------------
# 18. Save complete Cell 07 source code
# ------------------------------------------------------------

ip = get_ipython()

if ip is not None:

    (

        CODE_DIR

        / "cell07_federated_xgboost.py"

    ).write_text(

        ip.history_manager.input_hist_raw[-1],

        encoding="utf-8"

    )

# ------------------------------------------------------------
# 19. Save checkpoint
# ------------------------------------------------------------

checkpoint = {

    "cell": 7,

    "status": "COMPLETED",

    "reviewer_comments": [

        "Major 1",

        "Major 5",

        "Major 8"

    ],

    "federated_type":
        "Horizontal",

    "clients":
        CLIENT_COUNTS,

    "communication_rounds":
        COMMUNICATION_ROUNDS,

    "completed_rounds":
        len(round_df),

    "original_predictors": 42,

    "original_framework_classifier":
        "XGBoost",

    "aggregation":

        "Weighted additive tree updates",

    "convergence_curves_saved":
        True,

    "communication_bytes_measured":
        True,

    "matched_centralized_comparison_completed":
        False,

    "differential_privacy_completed":
        False,

    "original_source_modified":
        False,

    "next_cell": 8

}

with open(

    CHECKPOINT_DIR
    / "cell07_completed.json",

    "w"

) as f:

    json.dump(

        checkpoint,

        f,

        indent=4

    )

gc.collect()

print("\n" + "=" * 75)
print("CELL 07 COMPLETED SUCCESSFULLY")
print("=" * 75)

print("Horizontal federated learning: IMPLEMENTED")
print("3-client experiment: COMPLETED")
print("5-client experiment: COMPLETED")
print("10-client experiment: COMPLETED")
print("Communication rounds: 30 total")
print("Weighted tree aggregation: IMPLEMENTED")
print("Accuracy convergence curve: SAVED")
print("Recall convergence curve: SAVED")
print("Communication overhead: MEASURED")
print("Training time: MEASURED")
print("Original 42 predictors: RETAINED")
print("Original source integrity: PASS")
print("All federated models saved in Google Drive.")
print("Complete source code saved.")
print("Checkpoint saved.")

print(

    "\nTotal execution seconds:",

    round(

        time.perf_counter()
        - START_TIME,

        2

    )

)

print(

    "\nREADY FOR CELL 08/15: "
    "CENTRALIZED VS FEDERATED COMPARISON"

)
```

```python
# ============================================================
# CELL 08/15
# CENTRALIZED VS FEDERATED XGBOOST COMPARISON
#
# Reviewer Comments:
# Major 1: Centralized comparison and training time
# Major 5: Fair experimental comparison
#
# NO NEW CLASSIFIER
# NO FEATURE ENGINEERING
# NO HYPERPARAMETER TUNING
# NO ADDITIONAL DATASET
# ============================================================

import gc
import json
import time
import hashlib

from pathlib import Path

import numpy as np
import pandas as pd
import xgboost as xgb
import matplotlib.pyplot as plt

from google.colab import drive

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix
)

START_TIME = time.perf_counter()

SEED = 42

# ------------------------------------------------------------
# 1. Restore existing reviewer-only workspace
# ------------------------------------------------------------

drive.mount("/content/drive")

PROJECT_ROOT = Path(
    "/content/drive/MyDrive/Narayan_Patil_Reviewer_Revision"
)

SOURCE_ROOT = (
    PROJECT_ROOT
    / "01_Original_Project"
    / "SBN"
)

REVISED_DIR = (
    PROJECT_ROOT
    / "03_Reviewer_Only_Revision"
)

CODE_DIR = REVISED_DIR / "code"
DATA_DIR = REVISED_DIR / "data"
MODEL_DIR = REVISED_DIR / "models"
RESULT_DIR = REVISED_DIR / "results"
TABLE_DIR = RESULT_DIR / "tables"
PREDICTION_DIR = RESULT_DIR / "predictions"
FIGURE_DIR = RESULT_DIR / "figures"
REPORT_DIR = REVISED_DIR / "reports"
CHECKPOINT_DIR = REVISED_DIR / "checkpoints"

assert (
    CHECKPOINT_DIR / "cell07_completed.json"
).exists(), "Complete Cell 07 first."

print("=" * 75)
print("CELL 08: CENTRALIZED VS FEDERATED COMPARISON")
print("=" * 75)

print("\nWorkspace:", REVISED_DIR)

# ------------------------------------------------------------
# 2. Verify previous federated experiment
# ------------------------------------------------------------

with open(
    CHECKPOINT_DIR / "cell07_completed.json"
) as f:

    checkpoint07 = json.load(f)

assert checkpoint07["status"] == "COMPLETED"

assert checkpoint07["completed_rounds"] == 30

assert checkpoint07["original_predictors"] == 42

assert checkpoint07["clients"] == [3, 5, 10]

with open(
    REPORT_DIR / "cell07_federated_experiment_report.json"
) as f:

    fl_report = json.load(f)

fl_summary = pd.read_csv(
    TABLE_DIR / "cell07_federated_summary.csv"
)

fl_rounds = pd.read_csv(
    TABLE_DIR / "cell07_federated_round_metrics.csv"
)

assert len(fl_summary) == 3
assert len(fl_rounds) == 30

print("\nCell 07 checkpoint: VERIFIED")
print("Federated configurations: 3, 5, 10")
print("Completed communication rounds: 30")

# ------------------------------------------------------------
# 3. Restore the exact federated training parameters
# ------------------------------------------------------------

PARAMS = fl_report[
    "effective_federated_parameters"
].copy()

ORIGINAL_PARAMS = fl_report[
    "original_xgboost_parameters"
]

N_ESTIMATORS = ORIGINAL_PARAMS[
    "n_estimators"
]

assert N_ESTIMATORS == 400
assert PARAMS["eta"] == 0.05
assert PARAMS["max_depth"] == 8

print("\nCENTRALIZED XGBOOST PARAMETERS")

for key, value in PARAMS.items():

    print(f"{key}: {value}")

print("Boosting rounds:", N_ESTIMATORS)

print("Hyperparameter tuning: NONE")

# ------------------------------------------------------------
# 4. Load the SAME real datasets used in Cell 07
# ------------------------------------------------------------

def load_dataset(filename):

    with np.load(
        DATA_DIR / filename,
        allow_pickle=False
    ) as data:

        X = data["X"].astype(
            np.float32,
            copy=False
        )

        y = data["y"].astype(
            np.int8,
            copy=False
        )

        ids = data["row_id"].copy()

        names = data[
            "feature_names"
        ].tolist()

    assert np.isfinite(X).all()

    assert np.all(ids >= 0)

    assert len(X) == len(y)

    return X, y, ids, names


X_train, y_train, train_ids, feature_names = (
    load_dataset("cell05_train_real.npz")
)

X_test, y_test, test_ids, test_names = (
    load_dataset("cell05_test.npz")
)

assert X_train.shape == (446568, 42)

assert X_test.shape == (111643, 42)

assert feature_names == test_names

assert set(train_ids).isdisjoint(
    test_ids
)

assert np.bincount(
    y_train,
    minlength=2
).tolist() == [276332, 170236]

assert np.bincount(
    y_test,
    minlength=2
).tolist() == [69083, 42560]

print("\nMATCHED DATASET VERIFICATION")

print("Training:", X_train.shape)
print("Testing :", X_test.shape)
print("Original predictors: 42")
print("Synthetic training records: 0")
print("Row overlap: ZERO")
print("Dataset alignment: PASS")

# ------------------------------------------------------------
# 5. Verify federated predictions
# ------------------------------------------------------------

for n_clients in [3, 5, 10]:

    path = (

        PREDICTION_DIR

        / f"cell07_fl_{n_clients}_clients_seed42.npz"

    )

    assert path.exists()

    with np.load(
        path,
        allow_pickle=False
    ) as data:

        assert np.array_equal(

            data["row_id"],

            test_ids

        )

        assert np.array_equal(

            data["y_true"],

            y_test

        )

        assert np.isfinite(

            data["probability"]

        ).all()

    print(
        f"{n_clients}-client test alignment: PASS"
    )

# ------------------------------------------------------------
# 6. Original evaluation metrics
# ------------------------------------------------------------

def evaluate(
    y_true,
    probabilities
):

    probabilities = np.asarray(
        probabilities
    ).reshape(-1)

    assert len(y_true) == len(probabilities)

    assert np.isfinite(probabilities).all()

    assert (
        (probabilities >= 0)
        & (probabilities <= 1)
    ).all()

    predictions = (

        probabilities >= 0.5

    ).astype(np.int8)

    tn, fp, fn, tp = confusion_matrix(

        y_true,

        predictions,

        labels=[0, 1]

    ).ravel()

    metrics = {

        "accuracy": float(
            accuracy_score(
                y_true,
                predictions
            )
        ),

        "precision": float(
            precision_score(
                y_true,
                predictions,
                zero_division=0
            )
        ),

        "recall": float(
            recall_score(
                y_true,
                predictions,
                zero_division=0
            )
        ),

        "f1": float(
            f1_score(
                y_true,
                predictions,
                zero_division=0
            )
        ),

        "roc_auc": float(
            roc_auc_score(
                y_true,
                probabilities
            )
        ),

        "tn": int(tn),

        "fp": int(fp),

        "fn": int(fn),

        "tp": int(tp)

    }

    return metrics, predictions

# ------------------------------------------------------------
# 7. Train matched centralized XGBoost
#
# Same real training records as Cell 07.
# Same feature representation.
# Same base score and core hyperparameters.
# 400 centralized boosting rounds.
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("MATCHED CENTRALIZED XGBOOST TRAINING")
print("=" * 75)

central_start = time.perf_counter()

dtrain = xgb.DMatrix(

    X_train,

    label=y_train,

    feature_names=feature_names

)

dtest = xgb.DMatrix(

    X_test,

    feature_names=feature_names

)

training_start = time.perf_counter()

central_model = xgb.train(

    params=PARAMS.copy(),

    dtrain=dtrain,

    num_boost_round=N_ESTIMATORS,

    verbose_eval=False

)

central_training_seconds = (

    time.perf_counter()

    - training_start

)

assert (

    central_model.num_boosted_rounds()

    == N_ESTIMATORS

)

print("\nCentralized model training: COMPLETED")

print(

    "Training seconds:",

    round(
        central_training_seconds,
        4
    )

)

print("Centralized trees:", N_ESTIMATORS)

# ------------------------------------------------------------
# 8. Evaluate centralized model
# ------------------------------------------------------------

central_model.set_param({

    "device": "cpu",

    "nthread": 2

})

central_prediction_start = time.perf_counter()

central_test_probability = central_model.predict(

    dtest

)

central_metrics, central_predictions = (

    evaluate(

        y_test,

        central_test_probability

    )

)

central_prediction_seconds = (

    time.perf_counter()

    - central_prediction_start

)

central_train_probability = central_model.predict(

    dtrain

)

central_train_metrics, central_train_predictions = (

    evaluate(

        y_train,

        central_train_probability

    )

)

central_elapsed_seconds = (

    time.perf_counter()

    - central_start

)

print("\nCENTRALIZED TRAINING RESULTS")

for metric in [

    "accuracy",

    "precision",

    "recall",

    "f1",

    "roc_auc"

]:

    print(

        f"{metric:15s}: "

        f"{central_train_metrics[metric]:.6f}"

    )

print("\nCENTRALIZED TESTING RESULTS")

for metric in [

    "accuracy",

    "precision",

    "recall",

    "f1",

    "roc_auc"

]:

    print(

        f"{metric:15s}: "

        f"{central_metrics[metric]:.6f}"

    )

print(

    "\nConfusion matrix:",

    [

        [

            central_metrics["tn"],

            central_metrics["fp"]

        ],

        [

            central_metrics["fn"],

            central_metrics["tp"]

        ]

    ]

)

# ------------------------------------------------------------
# 9. Save centralized model and predictions
# ------------------------------------------------------------

central_model_path = (

    MODEL_DIR

    / "cell08_centralized_xgboost_real_seed42.ubj"

)

central_model.save_model(
    central_model_path
)

np.savez_compressed(

    PREDICTION_DIR

    / "cell08_centralized_real_test_seed42.npz",

    row_id=test_ids,

    y_true=y_test,

    probability=central_test_probability,

    prediction=central_predictions

)

np.savez_compressed(

    PREDICTION_DIR

    / "cell08_centralized_real_train_seed42.npz",

    row_id=train_ids,

    y_true=y_train,

    probability=central_train_probability,

    prediction=central_train_predictions

)

pd.DataFrame([

    {

        "model": "Centralized XGBoost",

        "partition": "train_real",

        "seed": SEED,

        "records": len(y_train),

        **central_train_metrics

    },

    {

        "model": "Centralized XGBoost",

        "partition": "test",

        "seed": SEED,

        "records": len(y_test),

        **central_metrics

    }

]).to_csv(

    TABLE_DIR

    / "cell08_centralized_train_test_metrics.csv",

    index=False

)

print("\nCentralized model and predictions saved.")

# ------------------------------------------------------------
# 10. Consolidate centralized and federated results
# ------------------------------------------------------------

comparison_records = [{

    "configuration": "Centralized XGBoost",

    "clients": 0,

    "communication_rounds": 0,

    "training_records": len(y_train),

    "test_records": len(y_test),

    "total_trees": N_ESTIMATORS,

    **central_metrics,

    "local_model_training_seconds":
        central_training_seconds,

    "aggregation_seconds": 0.0,

    "execution_seconds":
        central_elapsed_seconds,

    "communication_bytes": 0,

    "communication_mib": 0.0

}]

# Verify the exact final federated metrics
# using existing saved predictions.

for n_clients in [3, 5, 10]:

    row = fl_summary.loc[

        fl_summary["n_clients"] == n_clients

    ].iloc[0]

    with np.load(

        PREDICTION_DIR

        / f"cell07_fl_{n_clients}_clients_seed42.npz",

        allow_pickle=False

    ) as data:

        verified_metrics, _ = evaluate(

            y_test,

            data["probability"]

        )

    for metric in [

        "accuracy",

        "precision",

        "recall",

        "f1",

        "roc_auc"

    ]:

        assert np.isclose(

            verified_metrics[metric],

            float(row[metric]),

            atol=1e-8

        ), f"Federated metric mismatch: {n_clients}"

    comparison_records.append({

        "configuration":
            f"Federated ({n_clients} clients)",

        "clients": n_clients,

        "communication_rounds":
            int(row["rounds"]),

        "training_records":
            int(row["training_records"]),

        "test_records":
            int(row["test_records"]),

        "total_trees":
            int(row["total_ensemble_trees"]),

        **verified_metrics,

        "local_model_training_seconds":
            float(
                row["total_local_training_seconds"]
            ),

        "aggregation_seconds":
            float(
                row["total_aggregation_seconds"]
            ),

        "execution_seconds":
            float(
                row["total_elapsed_seconds"]
            ),

        "communication_bytes":
            int(
                row["total_communication_bytes"]
            ),

        "communication_mib":
            float(
                row["communication_mib"]
            )

    })

comparison_df = pd.DataFrame(
    comparison_records
)

comparison_df.to_csv(

    TABLE_DIR

    / "cell08_centralized_federated_comparison.csv",

    index=False

)

print("\n" + "=" * 75)
print("CENTRALIZED VS FEDERATED PERFORMANCE")
print("=" * 75)

print(

    comparison_df[

        [

            "configuration",

            "accuracy",

            "precision",

            "recall",

            "f1",

            "roc_auc",

            "total_trees"

        ]

    ]

    .round(6)

    .to_string(index=False)

)

# ------------------------------------------------------------
# 11. Reviewer-required training time and communication
# ------------------------------------------------------------

time_communication_df = comparison_df[[

    "configuration",

    "local_model_training_seconds",

    "aggregation_seconds",

    "execution_seconds",

    "communication_bytes",

    "communication_mib"

]].copy()

time_communication_df.to_csv(

    TABLE_DIR

    / "cell08_time_communication_comparison.csv",

    index=False

)

print("\n" + "=" * 75)
print("TRAINING TIME AND COMMUNICATION COMPARISON")
print("=" * 75)

print(

    time_communication_df

    .round(4)

    .to_string(index=False)

)

# ------------------------------------------------------------
# 12. Calculate differences relative to centralized model
# ------------------------------------------------------------

difference_records = []

for _, row in comparison_df.loc[

    comparison_df["clients"] > 0

].iterrows():

    record = {

        "clients": int(row["clients"]),

        "configuration": row["configuration"]

    }

    for metric in [

        "accuracy",

        "precision",

        "recall",

        "f1",

        "roc_auc"

    ]:

        record[f"{metric}_difference_pp"] = (

            row[metric]

            - central_metrics[metric]

        ) * 100

    record["training_time_ratio"] = (

        row["local_model_training_seconds"]

        / central_training_seconds

    )

    record["communication_mib"] = (

        row["communication_mib"]

    )

    difference_records.append(
        record
    )

difference_df = pd.DataFrame(
    difference_records
)

difference_df.to_csv(

    TABLE_DIR

    / "cell08_federated_centralized_differences.csv",

    index=False

)

print("\nDIFFERENCE RELATIVE TO CENTRALIZED MODEL")

print(

    difference_df

    .round(6)

    .to_string(index=False)

)

print(

    "\nPositive performance differences indicate "

    "higher federated scores."

)

# ------------------------------------------------------------
# 13. Preserve original Cell 05 centralized results
#
# SMOTE-based results are reported separately.
# ------------------------------------------------------------

cell05_metrics = pd.read_csv(

    TABLE_DIR

    / "cell05_all_model_metrics.csv"

)

cell05_xgb = cell05_metrics.loc[

    (cell05_metrics["model"] == "XGBoost")

    &

    (cell05_metrics["partition"] == "test")

].iloc[0]

centralized_reference_df = pd.DataFrame([

    {

        "configuration":
            "Cell 05 centralized with SMOTE",

        "training_records": 552664,

        "synthetic_records": 106096,

        "accuracy":
            float(cell05_xgb["accuracy"]),

        "precision":
            float(cell05_xgb["precision"]),

        "recall":
            float(cell05_xgb["recall"]),

        "f1":
            float(cell05_xgb["f1"]),

        "roc_auc":
            float(cell05_xgb["roc_auc"])

    },

    {

        "configuration":
            "Cell 08 centralized without SMOTE",

        "training_records": 446568,

        "synthetic_records": 0,

        **{

            metric: central_metrics[metric]

            for metric in [

                "accuracy",

                "precision",

                "recall",

                "f1",

                "roc_auc"

            ]

        }

    }

])

centralized_reference_df.to_csv(

    TABLE_DIR

    / "cell08_centralized_reference_conditions.csv",

    index=False

)

print("\nCENTRALIZED REFERENCE CONDITIONS")

print(

    centralized_reference_df

    .round(6)

    .to_string(index=False)

)

print("\nCell 05 original results: UNCHANGED")

# ------------------------------------------------------------
# 14. Required comparative trend figures
#
# Accuracy and recall for 3, 5, 10 nodes,
# with centralized reference.
# ------------------------------------------------------------

for metric, ylabel in [

    ("accuracy", "Accuracy (%)"),

    ("recall", "Recall (%)")

]:

    plt.figure(
        figsize=(7, 4.5)
    )

    markers = {

        3: "o",

        5: "s",

        10: "^"

    }

    linestyles = {

        3: "-",

        5: "--",

        10: "-."

    }

    for n_clients in [3, 5, 10]:

        subset = fl_rounds.loc[

            fl_rounds["n_clients"] == n_clients

        ].sort_values("round")

        plt.plot(

            subset["round"],

            subset[metric] * 100,

            color="black",

            linestyle=linestyles[n_clients],

            marker=markers[n_clients],

            markersize=4,

            linewidth=1.5,

            label=f"{n_clients} clients"

        )

    plt.axhline(

        central_metrics[metric] * 100,

        color="gray",

        linestyle=":",

        linewidth=2,

        label="Centralized final reference"

    )

    plt.xlabel(
        "Communication round"
    )

    plt.ylabel(ylabel)

    plt.xticks(
        range(1, 11)
    )

    plt.grid(
        alpha=0.25,
        linestyle=":"
    )

    plt.legend()

    plt.tight_layout()

    plt.savefig(

        FIGURE_DIR

        / f"cell08_centralized_federated_{metric}.png",

        dpi=600,

        bbox_inches="tight"

    )

    plt.show()

    plt.close()

print("\nRequired comparison figures: SAVED")

# ------------------------------------------------------------
# 15. Record comparison limitations
# ------------------------------------------------------------

comparison_conditions = {

    "same_real_training_records": True,

    "same_test_records": True,

    "same_feature_representation": True,

    "same_core_xgboost_hyperparameters": True,

    "same_initial_base_score": True,

    "same_evaluation_metrics": True,

    "same_classification_threshold": True,

    "centralized_total_trees": 400,

    "federated_total_trees": {

        "3_clients": 1200,

        "5_clients": 2000,

        "10_clients": 4000

    },

    "identical_global_model_capacity": False,

    "physical_network_simulated": False,

    "network_latency_included": False,

    "limitations": [

        "Federated ensembles contain more trees "

        "than the centralized model. Therefore, "

        "performance differences cannot be attributed "

        "exclusively to federated learning.",

        "Client execution is sequential within "

        "a single Colab runtime, rather than "

        "parallel execution across physical nodes.",

        "Communication overhead represents "

        "serialized model payload size, not "

        "measured network traffic.",

        "Zero centralized communication refers "

        "only to inter-client model transmission.",

        "Execution times include different "

        "evaluation and persistence operations; "

        "model-training time is reported separately.",

        "Provider overlap remains in the original "

        "claim-wise evaluation protocol.",

        "Global preprocessing was performed centrally.",

        "Results are from one experimental seed; "

        "statistical significance is not established."

    ]

}

with open(

    REPORT_DIR

    / "cell08_comparison_conditions.json",

    "w"

) as f:

    json.dump(

        comparison_conditions,

        f,

        indent=4

    )

# ------------------------------------------------------------
# 16. Verify original source integrity
# ------------------------------------------------------------

def sha256_file(path):

    digest = hashlib.sha256()

    with open(path, "rb") as f:

        for chunk in iter(

            lambda: f.read(1024 * 1024),

            b""

        ):

            digest.update(chunk)

    return digest.hexdigest()


with open(

    PROJECT_ROOT

    / "02_Revised_Project"

    / "reports"

    / "cell02_original_source_hashes.json"

) as f:

    original_hashes = json.load(f)

for item in original_hashes:

    assert (

        sha256_file(

            SOURCE_ROOT / item["file"]

        )

        == item["sha256"]

    )

print("\nOriginal source integrity: PASS")

# ------------------------------------------------------------
# 17. Save complete Cell 08 report
# ------------------------------------------------------------

report = {

    "cell": 8,

    "status": "COMPLETED",

    "reviewer_comments": [

        "Major 1",

        "Major 5"

    ],

    "seed": SEED,

    "original_predictors": 42,

    "centralized_parameters": PARAMS,

    "centralized_model_path":
        str(central_model_path),

    "centralized_training_records":
        len(y_train),

    "centralized_testing_records":
        len(y_test),

    "centralized_training_metrics":
        central_train_metrics,

    "centralized_testing_metrics":
        central_metrics,

    "centralized_training_seconds":
        central_training_seconds,

    "centralized_prediction_seconds":
        central_prediction_seconds,

    "comparison_results":
        comparison_records,

    "performance_differences":
        difference_records,

    "comparison_conditions":
        comparison_conditions,

    "original_cell05_results_modified": False,

    "original_cell07_results_modified": False,

    "additional_classifier_introduced": False,

    "hyperparameter_tuning": False,

    "feature_engineering_modified": False,

    "additional_dataset_used": False,

    "statistical_testing_completed": False,

    "original_source_modified": False,

    "elapsed_seconds":

        time.perf_counter() - START_TIME

}

with open(

    REPORT_DIR

    / "cell08_centralized_federated_report.json",

    "w"

) as f:

    json.dump(

        report,

        f,

        indent=4,

        default=str

    )

# ------------------------------------------------------------
# 18. Save complete source code
# ------------------------------------------------------------

ip = get_ipython()

if ip is not None:

    (

        CODE_DIR

        / "cell08_centralized_federated_comparison.py"

    ).write_text(

        ip.history_manager.input_hist_raw[-1],

        encoding="utf-8"

    )

# ------------------------------------------------------------
# 19. Save completion checkpoint
# ------------------------------------------------------------

checkpoint = {

    "cell": 8,

    "status": "COMPLETED",

    "reviewer_comments": [

        "Major 1",

        "Major 5"

    ],

    "centralized_reference_completed": True,

    "federated_results_verified": True,

    "same_training_data_comparison": True,

    "same_test_data_comparison": True,

    "identical_global_tree_count": False,

    "training_time_comparison_completed": True,

    "communication_comparison_completed": True,

    "comparison_figures_saved": True,

    "original_cell05_results_modified": False,

    "original_cell07_results_modified": False,

    "original_source_modified": False,

    "next_cell": 9

}

with open(

    CHECKPOINT_DIR

    / "cell08_completed.json",

    "w"

) as f:

    json.dump(

        checkpoint,

        f,

        indent=4

    )

del central_model, dtrain, dtest

gc.collect()

print("\n" + "=" * 75)
print("CELL 08 COMPLETED SUCCESSFULLY")
print("=" * 75)

print("Centralized XGBoost reference: COMPLETED")
print("Same real training dataset: VERIFIED")
print("Same testing dataset: VERIFIED")
print("Federated numerical results: VERIFIED")
print("Training time comparison: COMPLETED")
print("Communication comparison: COMPLETED")
print("Accuracy and recall trend figures: SAVED")
print("Comparison limitations: DOCUMENTED")
print("Original 42 predictors: RETAINED")
print("Original Cell 05 results: UNCHANGED")
print("Original Cell 07 results: UNCHANGED")
print("Original source integrity: PASS")
print("Hyperparameter tuning: NONE")
print("Additional dataset: NONE")
print("Complete code and results saved in Google Drive.")
print("Checkpoint saved.")

print(

    "\nTotal execution seconds:",

    round(

        time.perf_counter()

        - START_TIME,

        2

    )

)

print(

    "\nREADY FOR CELL 09/15: "

    "DIFFERENTIAL PRIVACY IMPLEMENTATION"

)
```

```python
# ============================================================
# CELL 09/15 - COMPLETE CORRECTED VERSION
# DIFFERENTIAL PRIVACY IMPLEMENTATION AND ACCOUNTING
#
# Reviewer Comment: Major 9
#
# MERGED CELL 09 + 09A + 09B
#
# XGBoost model slicing implemented directly
# No source recovery or AST-based execution
# Previous failed round included in privacy accounting
#
# Original 42 predictors retained
# Original XGBoost hyperparameters retained
# No external dataset
# No hyperparameter optimization
# Previous Cells 01-08 unchanged
# ============================================================

import os
import gc
import json
import time
import secrets
import hashlib
import joblib
import subprocess

from pathlib import Path

import numpy as np
import pandas as pd
import xgboost as xgb

from google.colab import drive
from scipy.special import expit

from sklearn.preprocessing import (
    OrdinalEncoder,
    StandardScaler
)

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix
)

drive.mount("/content/drive")

START_TIME = time.perf_counter()

SEED = 42

TARGET_EPSILON = 1.0
TARGET_DELTA = 1e-5

N_CLIENTS = 3
N_ROUNDS = 10
TREES_PER_ROUND = 40
N_TREES = 400

PUBLIC_REFERENCE_FRACTION = 0.10

LEARNING_RATE = 0.05
HESSIAN_FLOOR = 10.0

# One private round was already disclosed
# during the previous failed Cell 09 execution.

FAILED_ATTEMPT_ROUNDS = 1

# ------------------------------------------------------------
# 1. Restore existing reviewer-only workspace
# ------------------------------------------------------------

PROJECT_ROOT = Path(
    "/content/drive/MyDrive/Narayan_Patil_Reviewer_Revision"
)

SOURCE_ROOT = (
    PROJECT_ROOT
    / "01_Original_Project"
    / "SBN"
)

REVISED_DIR = (
    PROJECT_ROOT
    / "03_Reviewer_Only_Revision"
)

CODE_DIR = REVISED_DIR / "code"
DATA_DIR = REVISED_DIR / "data"
MODEL_DIR = REVISED_DIR / "models"
RESULT_DIR = REVISED_DIR / "results"
TABLE_DIR = RESULT_DIR / "tables"
PREDICTION_DIR = RESULT_DIR / "predictions"
REPORT_DIR = REVISED_DIR / "reports"
CHECKPOINT_DIR = REVISED_DIR / "checkpoints"

DP_DIR = MODEL_DIR / "differential_privacy"

for folder in [

    CODE_DIR,
    DATA_DIR,
    MODEL_DIR,
    TABLE_DIR,
    PREDICTION_DIR,
    REPORT_DIR,
    CHECKPOINT_DIR,
    DP_DIR

]:

    folder.mkdir(
        parents=True,
        exist_ok=True
    )

assert (
    CHECKPOINT_DIR / "cell08_completed.json"
).exists(), "Complete Cell 08 first."

assert not (
    CHECKPOINT_DIR / "cell09_completed.json"
).exists(), "Cell 09 is already completed."

# Prevent an accidental duplicate privacy release.

RELEASE_LEDGER = (
    REPORT_DIR
    / "cell09_merged_privacy_ledger.json"
)

assert not RELEASE_LEDGER.exists(), (
    "A merged Cell 09 release has already started. "
    "Do not repeat private training without "
    "checking the privacy ledger."
)

assert not (
    REPORT_DIR
    / "cell09_retry_privacy_accounting.json"
).exists(), (
    "A previous retry may have started. "
    "Check cumulative privacy accounting first."
)

print("=" * 75)
print("CELL 09: COMPLETE DIFFERENTIAL PRIVACY IMPLEMENTATION")
print("=" * 75)

print("\nWorkspace:", REVISED_DIR)

# ------------------------------------------------------------
# 2. Verify GPU and previous checkpoint
# ------------------------------------------------------------

gpu_result = subprocess.run(

    [
        "nvidia-smi",
        "--query-gpu=name,memory.total",
        "--format=csv,noheader"
    ],

    capture_output=True,
    text=True

)

assert gpu_result.returncode == 0, (
    "Enable Tesla T4 GPU."
)

print("\nGPU:", gpu_result.stdout.strip())

with open(
    CHECKPOINT_DIR / "cell08_completed.json"
) as f:

    checkpoint08 = json.load(f)

assert checkpoint08["status"] == "COMPLETED"

print("Cell 08 checkpoint: VERIFIED")
print("XGBoost version:", xgb.__version__)

# ------------------------------------------------------------
# 3. Formal Gaussian privacy accounting
#
# Gaussian mechanism:
#
# rho = sensitivity^2 / (2 * sigma^2)
#
# Composition:
#
# rho_total = sum of individual rho values
#
# Conversion:
#
# epsilon = rho + 2 * sqrt(rho * log(1/delta))
# ------------------------------------------------------------

def epsilon_from_rho(rho, delta):

    return float(

        rho

        + 2 * np.sqrt(
            rho * np.log(1.0 / delta)
        )

    )


def rho_from_epsilon(epsilon, delta):

    L = np.log(1.0 / delta)

    return float(

        (
            np.sqrt(L + epsilon)
            - np.sqrt(L)
        ) ** 2

    )


# Logistic gradient bound: |g| <= 1
# Logistic Hessian bound: 0 <= h <= 0.25
#
# Each record contributes to one leaf per tree.
#
# Replace-one adjacency sensitivity:
# 2 * sqrt(TREES_PER_ROUND * (1 + 0.25^2))

CONTRIBUTION_BOUND = float(

    np.sqrt(

        TREES_PER_ROUND

        * (1.0 + 0.25**2)

    )

)

SENSITIVITY = 2.0 * CONTRIBUTION_BOUND

RHO_TOTAL = rho_from_epsilon(

    TARGET_EPSILON,

    TARGET_DELTA

)

ORIGINAL_SIGMA = float(

    SENSITIVITY

    / np.sqrt(
        2.0 * RHO_TOTAL / N_ROUNDS
    )

)

FAILED_ATTEMPT_RHO = float(

    FAILED_ATTEMPT_ROUNDS

    * SENSITIVITY**2

    / (2.0 * ORIGINAL_SIGMA**2)

)

REMAINING_RHO = (

    RHO_TOTAL

    - FAILED_ATTEMPT_RHO

)

assert REMAINING_RHO > 0

# Corrected noise accounts for previously
# disclosed Round 01.

SIGMA = float(

    SENSITIVITY

    / np.sqrt(
        2.0 * REMAINING_RHO / N_ROUNDS
    )

)

RETRY_SIGMA = SIGMA

RHO_PER_ROUND = (

    SENSITIVITY**2

    / (2.0 * SIGMA**2)

)

VERIFIED_RHO = (

    N_ROUNDS

    * RHO_PER_ROUND

)

COMBINED_RHO = (

    FAILED_ATTEMPT_RHO

    + VERIFIED_RHO

)

VERIFIED_EPSILON = epsilon_from_rho(

    VERIFIED_RHO,

    TARGET_DELTA

)

COMBINED_EPSILON = epsilon_from_rho(

    COMBINED_RHO,

    TARGET_DELTA

)

assert COMBINED_EPSILON <= (

    TARGET_EPSILON + 1e-10

)

print("\n" + "=" * 75)
print("FORMAL PRIVACY BUDGET ACCOUNTING")
print("=" * 75)

print("Target epsilon:", TARGET_EPSILON)
print("Target delta:", TARGET_DELTA)

print(
    "Record contribution bound:",
    round(CONTRIBUTION_BOUND, 8)
)

print(
    "L2 sensitivity:",
    round(SENSITIVITY, 8)
)

print(
    "Original Gaussian sigma:",
    round(ORIGINAL_SIGMA, 8)
)

print(
    "Previously disclosed rounds:",
    FAILED_ATTEMPT_ROUNDS
)

print(
    "Previously consumed rho:",
    round(FAILED_ATTEMPT_RHO, 10)
)

print(
    "Remaining rho:",
    round(REMAINING_RHO, 10)
)

print(
    "Corrected Gaussian sigma:",
    round(SIGMA, 8)
)

print(
    "Retry epsilon:",
    round(VERIFIED_EPSILON, 10)
)

print(
    "Combined accounted epsilon:",
    round(COMBINED_EPSILON, 10)
)

print("Theoretical privacy accounting: PASS")

# ------------------------------------------------------------
# 4. Restore original XGBoost hyperparameters
# ------------------------------------------------------------

with open(

    REPORT_DIR
    / "cell07_federated_experiment_report.json"

) as f:

    fl_report = json.load(f)

PARAMS = fl_report[
    "effective_federated_parameters"
].copy()

PARAMS.update({

    "seed": SEED,

    "device": "cuda",

    "verbosity": 0

})

assert PARAMS["max_depth"] == 8
assert PARAMS["eta"] == 0.05

assert N_TREES == (
    N_ROUNDS * TREES_PER_ROUND
)

print("\nOriginal XGBoost hyperparameters: RETAINED")

# ------------------------------------------------------------
# 5. Restore fixed training and testing membership
# ------------------------------------------------------------

with np.load(

    DATA_DIR / "cell05_train_real.npz",

    allow_pickle=False

) as data:

    train_ids = data["row_id"].copy()

    feature_names = data[
        "feature_names"
    ].tolist()

with np.load(

    DATA_DIR / "cell05_test.npz",

    allow_pickle=False

) as data:

    test_ids = data["row_id"].copy()

assert len(train_ids) == 446568

assert len(test_ids) == 111643

assert len(feature_names) == 42

assert set(train_ids).isdisjoint(
    test_ids
)

FEATURED_PATH = (

    SOURCE_ROOT

    / "data"

    / "processed"

    / "featured_dataset.csv"

)

TARGET = "PotentialFraud"

assert FEATURED_PATH.exists()

df = pd.read_csv(

    FEATURED_PATH,

    dtype=str,

    low_memory=False

)

assert df.shape == (558211, 43)

# ------------------------------------------------------------
# 6. Reproduce the previous public/protected partition
#
# Identical fixed random seed and row membership
# used in the previous Cell 09.
#
# No protected label-based client allocation.
# ------------------------------------------------------------

rng = np.random.default_rng(SEED)

permuted_ids = rng.permutation(

    np.sort(train_ids)

)

n_public = int(

    len(train_ids)

    * PUBLIC_REFERENCE_FRACTION

)

public_ids = permuted_ids[:n_public]

protected_ids = permuted_ids[n_public:]

assert len(public_ids) == 44656

assert len(protected_ids) == 401912

assert set(public_ids).isdisjoint(
    protected_ids
)

assert set(public_ids).isdisjoint(
    test_ids
)

assert set(protected_ids).isdisjoint(
    test_ids
)

# Verify reference membership if it was saved
# during the previous failed execution.

previous_reference_path = (

    DP_DIR

    / "cell09_public_reference_indices.npz"

)

if previous_reference_path.exists():

    with np.load(

        previous_reference_path,

        allow_pickle=False

    ) as old_reference:

        assert np.array_equal(

            old_reference["public_ids"],

            public_ids

        ), "Public reference membership changed."

    print("\nPrevious reference membership: VERIFIED")

print("\nDATA PARTITION")

print(
    "Original training records:",
    len(train_ids)
)

print(
    "Designated public reference records:",
    len(public_ids)
)

print(
    "Protected training records:",
    len(protected_ids)
)

print(
    "Testing records:",
    len(test_ids)
)

print("Original predictors:", len(feature_names))
print("Partition overlap: ZERO")
print("External dataset: NONE")

# ------------------------------------------------------------
# 7. Public-reference preprocessing
#
# Encoder and scaler are fitted exclusively
# on the designated public reference subset.
# ------------------------------------------------------------

public_df = df.iloc[
    public_ids
].copy()

numerical_columns = []

categorical_columns = []

for col in feature_names:

    values = public_df[col].dropna()

    numeric_values = pd.to_numeric(

        values,

        errors="coerce"

    )

    if (

        len(values) > 0

        and numeric_values.notna().all()

    ):

        numerical_columns.append(col)

    else:

        categorical_columns.append(col)

numeric_indices = [

    feature_names.index(col)

    for col in numerical_columns

]

category_indices = [

    feature_names.index(col)

    for col in categorical_columns

]

encoder = OrdinalEncoder(

    handle_unknown="use_encoded_value",

    unknown_value=-1,

    dtype=np.float32

)

if categorical_columns:

    encoder.fit(

        public_df[categorical_columns]

        .fillna("Unknown")

        .astype(str)

    )


def encode_frame(frame):

    X = np.empty(

        (len(frame), len(feature_names)),

        dtype=np.float32

    )

    for col, idx in zip(

        numerical_columns,

        numeric_indices

    ):

        X[:, idx] = (

            pd.to_numeric(

                frame[col],

                errors="coerce"

            )

            .replace(

                [np.inf, -np.inf],

                np.nan

            )

            .fillna(0)

            .to_numpy(dtype=np.float32)

        )

    if categorical_columns:

        X[:, category_indices] = (

            encoder.transform(

                frame[categorical_columns]

                .fillna("Unknown")

                .astype(str)

            )

        )

    assert np.isfinite(X).all()

    return X


X_public_raw = encode_frame(
    public_df
)

scaler = StandardScaler()

scaler.fit(
    X_public_raw
)

X_public = scaler.transform(

    X_public_raw

).astype(np.float32)

y_public = public_df[TARGET].astype(

    np.int8

).to_numpy()

protected_df = df.iloc[
    protected_ids
]

X_protected = scaler.transform(

    encode_frame(protected_df)

).astype(np.float32)

y_protected = protected_df[TARGET].astype(

    np.int8

).to_numpy()

test_df = df.iloc[
    test_ids
]

X_test = scaler.transform(

    encode_frame(test_df)

).astype(np.float32)

y_test = test_df[TARGET].astype(

    np.int8

).to_numpy()

assert X_public.shape == (44656, 42)

assert X_protected.shape == (401912, 42)

assert X_test.shape == (111643, 42)

for labels in [

    y_public,

    y_protected,

    y_test

]:

    assert set(np.unique(labels)) == {0, 1}

joblib.dump({

    "encoder": encoder,

    "scaler": scaler,

    "feature_names": feature_names,

    "numerical_columns":
        numerical_columns,

    "categorical_columns":
        categorical_columns,

    "numeric_indices":
        numeric_indices,

    "category_indices":
        category_indices,

    "fit_scope":
        "Designated public reference only"

}, DP_DIR / "cell09_public_preprocessing.joblib")

np.savez_compressed(

    DP_DIR

    / "cell09_public_reference_indices.npz",

    public_ids=public_ids

)

del (

    df,

    public_df,

    protected_df,

    test_df,

    X_public_raw

)

gc.collect()

print("\nPublic-reference preprocessing: PASS")
print("Original 42 predictors: RETAINED")

# ------------------------------------------------------------
# 8. Restore previously trained public reference model
#
# Avoid unnecessary retraining if the public model
# was successfully saved during the failed attempt.
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("PUBLIC REFERENCE XGBOOST MODEL")
print("=" * 75)

public_model_path = (

    DP_DIR

    / "cell09_public_xgboost.ubj"

)

public_training_seconds = None

if public_model_path.exists():

    public_model = xgb.Booster()

    public_model.load_model(

        public_model_path

    )

    print(
        "Previously trained public model: RESTORED"
    )

else:

    public_start = time.perf_counter()

    dpublic = xgb.DMatrix(

        X_public,

        label=y_public,

        feature_names=feature_names

    )

    public_model = xgb.train(

        params=PARAMS.copy(),

        dtrain=dpublic,

        num_boost_round=N_TREES,

        verbose_eval=False

    )

    public_training_seconds = (

        time.perf_counter()

        - public_start

    )

    public_model.save_model(

        public_model_path

    )

    del dpublic

    print(
        "Public reference model: TRAINED"
    )

assert (

    public_model.num_boosted_rounds()

    == N_TREES

)

assert public_model.feature_names == feature_names

public_model.set_param({

    "device": "cpu",

    "nthread": 2

})

print("Public model trees:", N_TREES)

del X_public, y_public

gc.collect()

# ------------------------------------------------------------
# 9. Extract original public tree leaf structures
# ------------------------------------------------------------

def collect_leaf_ids(node):

    if "leaf" in node:

        return [

            int(node["nodeid"])

        ]

    result = []

    for child in node.get(
        "children", []
    ):

        result.extend(

            collect_leaf_ids(child)

        )

    return result


tree_dumps = public_model.get_dump(

    dump_format="json"

)

leaf_catalogue = [

    np.asarray(

        sorted(

            collect_leaf_ids(

                json.loads(tree)

            )

        ),

        dtype=np.int32

    )

    for tree in tree_dumps

]

assert len(leaf_catalogue) == N_TREES

print("\nFixed tree leaf structures: VERIFIED")

# ------------------------------------------------------------
# 10. CORRECTION 1:
# Direct XGBoost model slicing
#
# Avoid unsupported pred_leaf iteration_range
# with a non-zero starting tree.
#
# Every sliced model contains exactly 40 trees.
# ------------------------------------------------------------

TREE_SLICES = []

for round_index in range(N_ROUNDS):

    start = (

        round_index

        * TREES_PER_ROUND

    )

    end = (

        start

        + TREES_PER_ROUND

    )

    sliced_model = public_model[
        start:end
    ]

    sliced_model.set_param({

        "device": "cpu",

        "nthread": 2

    })

    assert (

        sliced_model.num_boosted_rounds()

        == TREES_PER_ROUND

    )

    TREE_SLICES.append(

        sliced_model

    )

assert len(TREE_SLICES) == N_ROUNDS

print("\nXGBoost model slicing: IMPLEMENTED")
print("Number of slices:", len(TREE_SLICES))
print("Trees per slice:", TREES_PER_ROUND)

# ------------------------------------------------------------
# 11. Prepare protected horizontal client partitions
# ------------------------------------------------------------

client_indices = np.array_split(

    np.arange(
        len(protected_ids)
    ),

    N_CLIENTS

)

client_matrices = []

client_labels = []

client_initial_margins = []

for client_index, indices in enumerate(

    client_indices,

    start=1

):

    local_matrix = xgb.DMatrix(

        X_protected[indices],

        feature_names=feature_names

    )

    client_matrices.append(

        local_matrix

    )

    client_labels.append(

        y_protected[indices].astype(

            np.float64

        )

    )

    client_initial_margins.append(

        public_model.predict(

            local_matrix,

            output_margin=True

        ).astype(np.float64)

    )

    print(

        f"Client {client_index:02d}"

        f" | Protected records: "

        f"{len(indices):,}"

    )

assert sum(

    len(indices)

    for indices in client_indices

) == len(protected_ids)

dtest = xgb.DMatrix(

    X_test,

    feature_names=feature_names

)

initial_test_margin = public_model.predict(

    dtest,

    output_margin=True

).astype(np.float64)

print("\nProtected client allocation: VERIFIED")

# ------------------------------------------------------------
# 12. Validate model slicing BEFORE private releases
#
# Use a small fixed test sample for technical
# compatibility verification.
# No private gradient statistics are released here.
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("XGBOOST COMPATIBILITY VALIDATION")
print("=" * 75)

check_matrix = xgb.DMatrix(

    X_test[:256],

    feature_names=feature_names

)

reference_leaves = public_model.predict(

    check_matrix,

    pred_leaf=True

).reshape(

    256,

    N_TREES

).astype(np.int32)

for round_index, model_slice in enumerate(

    TREE_SLICES

):

    start = (

        round_index

        * TREES_PER_ROUND

    )

    end = (

        start

        + TREES_PER_ROUND

    )

    predicted_leaves = model_slice.predict(

        check_matrix,

        pred_leaf=True

    ).reshape(

        256,

        TREES_PER_ROUND

    ).astype(np.int32)

    assert np.array_equal(

        predicted_leaves,

        reference_leaves[:, start:end]

    ), (

        f"Tree slice mismatch: "

        f"{round_index + 1}"

    )

    for j in range(TREES_PER_ROUND):

        assert np.isin(

            predicted_leaves[:, j],

            leaf_catalogue[start + j]

        ).all()

    print(

        f"Slice {round_index + 1:02d}: PASS"

    )

del (

    check_matrix,

    reference_leaves,

    predicted_leaves

)

gc.collect()

print("\nAll ten model slices: VERIFIED")
print("Previous XGBoostError: RESOLVED")
print("Source recovery dependency: NONE")
print("Previous StopIteration issue: REMOVED")

# ------------------------------------------------------------
# 13. Original evaluation metrics
# ------------------------------------------------------------

def evaluate(y_true, margin):

    probability = expit(margin)

    prediction = (

        probability >= 0.5

    ).astype(np.int8)

    tn, fp, fn, tp = confusion_matrix(

        y_true,

        prediction,

        labels=[0, 1]

    ).ravel()

    metrics = {

        "accuracy": float(

            accuracy_score(

                y_true,

                prediction

            )

        ),

        "precision": float(

            precision_score(

                y_true,

                prediction,

                zero_division=0

            )

        ),

        "recall": float(

            recall_score(

                y_true,

                prediction,

                zero_division=0

            )

        ),

        "f1": float(

            f1_score(

                y_true,

                prediction,

                zero_division=0

            )

        ),

        "roc_auc": float(

            roc_auc_score(

                y_true,

                probability

            )

        ),

        "tn": int(tn),

        "fp": int(fp),

        "fn": int(fn),

        "tp": int(tp)

    }

    return (

        metrics,

        probability,

        prediction

    )

# ------------------------------------------------------------
# 14. Gaussian random sampling
#
# OS-backed random source.
# Experimental seed is not used for privacy noise.
# Noise samples and seeds are not persisted.
# ------------------------------------------------------------

secure_random = secrets.SystemRandom()


def gaussian_noise(size, sigma):

    return np.fromiter(

        (

            secure_random.normalvariate(

                0.0,

                sigma

            )

            for _ in range(size)

        ),

        dtype=np.float64,

        count=size

    )

# ------------------------------------------------------------
# 15. Prepare release accounting
#
# Reserve the privacy cost before starting.
# Prevent accidental unaccounted duplicate releases.
# ------------------------------------------------------------

ledger = {

    "status": "BUDGET_RESERVED",

    "failed_attempt_rounds":
        FAILED_ATTEMPT_ROUNDS,

    "failed_attempt_rho":
        FAILED_ATTEMPT_RHO,

    "reserved_retry_rho":
        VERIFIED_RHO,

    "combined_reserved_rho":
        COMBINED_RHO,

    "combined_reserved_epsilon":
        COMBINED_EPSILON,

    "delta":
        TARGET_DELTA,

    "sigma":
        SIGMA,

    "completed_retry_rounds": 0

}

with open(
    RELEASE_LEDGER,
    "x"
) as f:

    json.dump(

        ledger,

        f,

        indent=4

    )

print("\nPrivacy release budget: RESERVED")

# ------------------------------------------------------------
# 16. Corrected DP federated refinement function
#
# Fixed public XGBoost tree structures.
# Bounded private gradient/Hessian statistics.
# Gaussian noise before release.
# Weighted leaf correction.
#
# CORRECTION:
# Use TREE_SLICES[round_no - 1].predict()
# for both client and test leaf prediction.
# ------------------------------------------------------------

def run_private_refinement(

    sigma,

    run_name

):

    run_dir = DP_DIR / run_name

    run_dir.mkdir(

        parents=True,

        exist_ok=True

    )

    if (

        run_dir / "final_report.json"

    ).exists():

        raise RuntimeError(

            "Private release already exists."

        )

    run_start = time.perf_counter()

    local_margins = [

        margin.copy()

        for margin in client_initial_margins

    ]

    test_margin = (

        initial_test_margin.copy()

    )

    round_records = []

    saved_updates = {}

    rho_per_round = float(

        SENSITIVITY**2

        / (2.0 * sigma**2)

    )

    for round_no in range(

        1,

        N_ROUNDS + 1

    ):

        round_start = time.perf_counter()

        start_tree = (

            round_no - 1

        ) * TREES_PER_ROUND

        end_tree = (

            start_tree

            + TREES_PER_ROUND

        )

        # Direct model slicing.
        # No incompatible iteration_range.

        current_model = TREE_SLICES[
            round_no - 1
        ]

        current_catalogue = (

            leaf_catalogue[
                start_tree:end_tree
            ]

        )

        leaf_counts = [

            len(leaves)

            for leaves in current_catalogue

        ]

        offsets = np.concatenate([

            [0],

            np.cumsum(leaf_counts)

        ]).astype(np.int64)

        total_leaves = int(
            offsets[-1]
        )

        aggregate_g = np.zeros(

            total_leaves,

            dtype=np.float64

        )

        aggregate_h = np.zeros(

            total_leaves,

            dtype=np.float64

        )

        client_leaf_positions = []

        # ----------------------------------------------------
        # A. Compute bounded private client statistics
        # ----------------------------------------------------

        for client_no in range(N_CLIENTS):

            matrix = client_matrices[
                client_no
            ]

            probability = expit(

                local_margins[client_no]

            )

            gradient = np.clip(

                probability

                - client_labels[client_no],

                -1.0,

                1.0

            )

            hessian = np.clip(

                probability

                * (1.0 - probability),

                0.0,

                0.25

            )

            # FIXED LEAF PREDICTION

            leaves = current_model.predict(

                matrix,

                pred_leaf=True

            ).astype(np.int32)

            leaves = leaves.reshape(

                len(gradient),

                TREES_PER_ROUND

            )

            positions = np.empty(

                leaves.shape,

                dtype=np.int32

            )

            for j, catalogue in enumerate(

                current_catalogue

            ):

                local_position = np.searchsorted(

                    catalogue,

                    leaves[:, j]

                )

                assert np.all(

                    local_position

                    < len(catalogue)

                )

                assert np.array_equal(

                    catalogue[local_position],

                    leaves[:, j]

                )

                positions[:, j] = (

                    local_position

                    + offsets[j]

                )

                start = offsets[j]

                end = offsets[j + 1]

                aggregate_g[start:end] += (

                    np.bincount(

                        local_position,

                        weights=gradient,

                        minlength=len(catalogue)

                    )

                )

                aggregate_h[start:end] += (

                    np.bincount(

                        local_position,

                        weights=hessian,

                        minlength=len(catalogue)

                    )

                )

            client_leaf_positions.append(

                positions

            )

            del (

                gradient,

                hessian,

                leaves,

                probability

            )

        # ----------------------------------------------------
        # B. Gaussian perturbation
        # ----------------------------------------------------

        noisy_aggregate = np.concatenate([

            aggregate_g,

            aggregate_h

        ])

        noisy_aggregate += gaussian_noise(

            size=len(noisy_aggregate),

            sigma=sigma

        )

        noisy_g = noisy_aggregate[
            :total_leaves
        ]

        noisy_h = noisy_aggregate[
            total_leaves:
        ]

        # Deterministic post-processing.
        # Does not consume additional DP budget.

        denominator = np.maximum(

            noisy_h + 1.0,

            HESSIAN_FLOOR

        )

        leaf_update = (

            -LEARNING_RATE

            * np.clip(

                noisy_g / denominator,

                -1.0,

                1.0

            )

        )

        assert np.isfinite(

            leaf_update

        ).all()

        # ----------------------------------------------------
        # C. Apply privatized client updates
        # ----------------------------------------------------

        for client_no in range(N_CLIENTS):

            local_margins[client_no] += (

                leaf_update[

                    client_leaf_positions[
                        client_no
                    ]

                ].sum(axis=1)

            )

        # ----------------------------------------------------
        # D. Apply identical updates to test predictions
        #
        # FIXED TEST LEAF PREDICTION
        # ----------------------------------------------------

        test_leaves = current_model.predict(

            dtest,

            pred_leaf=True

        ).astype(np.int32)

        test_leaves = test_leaves.reshape(

            len(y_test),

            TREES_PER_ROUND

        )

        for j, catalogue in enumerate(

            current_catalogue

        ):

            position = np.searchsorted(

                catalogue,

                test_leaves[:, j]

            )

            assert np.all(

                position

                < len(catalogue)

            )

            assert np.array_equal(

                catalogue[position],

                test_leaves[:, j]

            )

            test_margin += leaf_update[

                position + offsets[j]

            ]

        # ----------------------------------------------------
        # E. Save DP-derived updates only
        # ----------------------------------------------------

        saved_updates[

            f"round_{round_no:02d}_updates"

        ] = leaf_update.copy()

        saved_updates[

            f"round_{round_no:02d}_offsets"

        ] = offsets.copy()

        cumulative_rho = (

            round_no

            * rho_per_round

        )

        accounted_epsilon = epsilon_from_rho(

            cumulative_rho,

            TARGET_DELTA

        )

        combined_round_rho = (

            FAILED_ATTEMPT_RHO

            + cumulative_rho

        )

        combined_round_epsilon = epsilon_from_rho(

            combined_round_rho,

            TARGET_DELTA

        )

        metrics, _, _ = evaluate(

            y_test,

            test_margin

        )

        record = {

            "round": round_no,

            "clients": N_CLIENTS,

            "sigma": float(sigma),

            "l2_sensitivity": SENSITIVITY,

            "cumulative_rho":
                float(cumulative_rho),

            "epsilon":
                accounted_epsilon,

            "combined_rho":
                float(combined_round_rho),

            "combined_epsilon":
                combined_round_epsilon,

            "delta":
                TARGET_DELTA,

            **metrics,

            "round_seconds":

                time.perf_counter()

                - round_start

        }

        round_records.append(record)

        # Record each completed round before
        # displaying the private-model results.

        if run_name == "epsilon_1_retry_seed42":

            ledger[
                "completed_retry_rounds"
            ] = round_no

            ledger[
                "accounted_combined_rho"
            ] = float(combined_round_rho)

            ledger[
                "accounted_combined_epsilon"
            ] = combined_round_epsilon

            ledger[
                "status"
            ] = "IN_PROGRESS"

            temporary_ledger = (

                RELEASE_LEDGER.with_suffix(

                    ".tmp"

                )

            )

            with open(

                temporary_ledger,

                "w"

            ) as f:

                json.dump(

                    ledger,

                    f,

                    indent=4

                )

            os.replace(

                temporary_ledger,

                RELEASE_LEDGER

            )

        print(

            f"Round {round_no:02d}/{N_ROUNDS}"

            f" | Combined epsilon: "

            f"{combined_round_epsilon:.6f}"

            f" | Accuracy: "

            f"{metrics['accuracy']:.6f}"

            f" | Recall: "

            f"{metrics['recall']:.6f}"

            f" | AUC: "

            f"{metrics['roc_auc']:.6f}",

            flush=True

        )

        # Never persist raw unprotected aggregates.

        del (

            aggregate_g,

            aggregate_h,

            noisy_aggregate,

            noisy_g,

            noisy_h,

            client_leaf_positions,

            test_leaves

        )

        gc.collect()

    # --------------------------------------------------------
    # F. Final evaluation and standalone privacy accounting
    # --------------------------------------------------------

    total_rho = float(

        N_ROUNDS * rho_per_round

    )

    final_epsilon = epsilon_from_rho(

        total_rho,

        TARGET_DELTA

    )

    (

        final_metrics,

        final_probability,

        final_prediction

    ) = evaluate(

        y_test,

        test_margin

    )

    np.savez_compressed(

        run_dir

        / "private_leaf_updates.npz",

        **saved_updates

    )

    pd.DataFrame(

        round_records

    ).to_csv(

        run_dir / "round_metrics.csv",

        index=False

    )

    np.savez_compressed(

        run_dir / "test_predictions.npz",

        probability=final_probability,

        prediction=final_prediction

    )

    final_report = {

        "run_name": run_name,

        "epsilon": final_epsilon,

        "delta": TARGET_DELTA,

        "sigma": float(sigma),

        "rho": total_rho,

        "l2_sensitivity": SENSITIVITY,

        "rounds": N_ROUNDS,

        "clients": N_CLIENTS,

        "protected_records":
            len(protected_ids),

        "public_reference_records":
            len(public_ids),

        "test_records":
            len(test_ids),

        "features":
            len(feature_names),

        **final_metrics,

        "execution_seconds":

            time.perf_counter()

            - run_start

    }

    with open(

        run_dir / "final_report.json",

        "w"

    ) as f:

        json.dump(

            final_report,

            f,

            indent=4

        )

    return (

        final_report,

        round_records

    )

# ------------------------------------------------------------
# 17. Execute corrected ten-round private refinement
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("DP FEDERATED XGBOOST - CORRECTED EXECUTION")
print("=" * 75)

print(
    "All compatibility checks completed."
)

print(
    "Starting ten private communication rounds."
)

dp_result, dp_round_records = (

    run_private_refinement(

        sigma=SIGMA,

        run_name="epsilon_1_retry_seed42"

    )

)

assert len(dp_round_records) == N_ROUNDS

assert dp_result["rounds"] == 10

# ------------------------------------------------------------
# 18. Verify combined privacy budget
# ------------------------------------------------------------

actual_combined_rho = (

    FAILED_ATTEMPT_RHO

    + dp_result["rho"]

)

actual_combined_epsilon = epsilon_from_rho(

    actual_combined_rho,

    TARGET_DELTA

)

assert actual_combined_epsilon <= (

    TARGET_EPSILON + 1e-10

)

dp_result["failed_attempt_rho"] = (

    FAILED_ATTEMPT_RHO

)

dp_result["combined_rho"] = (

    actual_combined_rho

)

dp_result["combined_epsilon"] = (

    actual_combined_epsilon

)

dp_result["combined_delta"] = (

    TARGET_DELTA

)

ledger["status"] = "COMPLETED"

ledger["completed_retry_rounds"] = N_ROUNDS

ledger["actual_combined_rho"] = (

    actual_combined_rho

)

ledger["actual_combined_epsilon"] = (

    actual_combined_epsilon

)

with open(

    RELEASE_LEDGER,

    "w"

) as f:

    json.dump(

        ledger,

        f,

        indent=4

    )

print("\n" + "=" * 75)
print("FINAL DIFFERENTIAL PRIVACY RESULTS")
print("=" * 75)

for metric in [

    "accuracy",

    "precision",

    "recall",

    "f1",

    "roc_auc"

]:

    print(

        f"{metric:15s}: "

        f"{dp_result[metric]:.6f}"

    )

print(

    "\nConfusion matrix:",

    [

        [

            dp_result["tn"],

            dp_result["fp"]

        ],

        [

            dp_result["fn"],

            dp_result["tp"]

        ]

    ]

)

print(

    "\nRetry epsilon:",

    round(
        dp_result["epsilon"],
        10
    )

)

print(

    "Combined epsilon:",

    round(
        actual_combined_epsilon,
        10
    )

)

print("Delta:", TARGET_DELTA)

print(

    "Corrected Gaussian sigma:",

    round(SIGMA, 8)

)

print("Combined theoretical privacy accounting: PASS")

# ------------------------------------------------------------
# 19. Save reviewer-ready result tables
# ------------------------------------------------------------

pd.DataFrame(

    dp_round_records

).to_csv(

    TABLE_DIR

    / "cell09_dp_round_metrics.csv",

    index=False

)

pd.DataFrame([

    dp_result

]).to_csv(

    TABLE_DIR

    / "cell09_dp_final_results.csv",

    index=False

)

privacy_accounting_table = pd.DataFrame([

    {

        "stage": "Previous disclosed round",

        "rounds": FAILED_ATTEMPT_ROUNDS,

        "sigma": ORIGINAL_SIGMA,

        "rho": FAILED_ATTEMPT_RHO,

        "epsilon": epsilon_from_rho(

            FAILED_ATTEMPT_RHO,

            TARGET_DELTA

        ),

        "delta": TARGET_DELTA

    },

    {

        "stage": "Corrected ten-round experiment",

        "rounds": N_ROUNDS,

        "sigma": SIGMA,

        "rho": dp_result["rho"],

        "epsilon": dp_result["epsilon"],

        "delta": TARGET_DELTA

    },

    {

        "stage": "Combined privacy accounting",

        "rounds":

            FAILED_ATTEMPT_ROUNDS

            + N_ROUNDS,

        "sigma": None,

        "rho":
            actual_combined_rho,

        "epsilon":
            actual_combined_epsilon,

        "delta":
            TARGET_DELTA

    }

])

privacy_accounting_table.to_csv(

    TABLE_DIR

    / "cell09_privacy_accounting_table.csv",

    index=False

)

print("\nPRIVACY ACCOUNTING TABLE")

print(

    privacy_accounting_table

    .round(8)

    .to_string(index=False)

)

# ------------------------------------------------------------
# 20. Save complete privacy verification report
# ------------------------------------------------------------

privacy_report = {

    "cell": 9,

    "status": "COMPLETED",

    "reviewer_comment": "Major 9",

    "merged_correction_version": True,

    "mechanism":

        "Gaussian perturbation of aggregated "

        "gradient and Hessian statistics",

    "accountant": "zCDP",

    "privacy_unit":
        "Protected claim record",

    "adjacency":
        "Replace-one protected record",

    "target_epsilon":
        TARGET_EPSILON,

    "target_delta":
        TARGET_DELTA,

    "accounted_epsilon":
        actual_combined_epsilon,

    "retry_epsilon":
        dp_result["epsilon"],

    "delta":
        TARGET_DELTA,

    "sigma":
        SIGMA,

    "original_sigma":
        ORIGINAL_SIGMA,

    "sensitivity":
        SENSITIVITY,

    "rho_total":
        actual_combined_rho,

    "failed_attempt_rounds":
        FAILED_ATTEMPT_ROUNDS,

    "failed_attempt_rho":
        FAILED_ATTEMPT_RHO,

    "retry_rho":
        dp_result["rho"],

    "communication_rounds":
        N_ROUNDS,

    "clients":
        N_CLIENTS,

    "original_predictors":
        42,

    "original_tree_count":
        N_TREES,

    "original_xgboost_parameters":
        PARAMS,

    "public_reference_records":
        len(public_ids),

    "protected_training_records":
        len(protected_ids),

    "test_records":
        len(test_ids),

    "private_gradient_bound":
        1.0,

    "private_hessian_bound":
        0.25,

    "fixed_public_tree_structures":
        True,

    "public_preprocessing_only":
        True,

    "trusted_coordinator_assumed":
        True,

    "secure_aggregation_implemented":
        False,

    "noise_applied_before_release":
        True,

    "raw_private_aggregates_saved":
        False,

    "privacy_noise_seed_published":
        False,

    "xgboost_model_slicing":
        True,

    "all_tree_slices_verified":
        True,

    "source_recovery_required":
        False,

    "xgboost_compatibility_error_corrected":
        True,

    "stopiteration_error_corrected":
        True,

    "public_training_seconds":
        public_training_seconds,

    "dp_results":
        dp_result,

    "external_dataset_used":
        False,

    "hyperparameter_tuning":
        False,

    "original_cell05_results_modified":
        False,

    "original_cell07_results_modified":
        False,

    "original_cell08_results_modified":
        False,

    "privacy_utility_tradeoff":
        "Pending Cell 10",

    "limitations": [

        "The theoretical privacy guarantee is "

        "conditional on fixed reference data, "

        "feature schema, record membership "

        "and test data.",

        "The designated reference subset is "

        "excluded from the protected population.",

        "Privacy applies to individual claim "

        "rows, not entire patients or providers.",

        "The trusted coordinator can access "

        "unperturbed aggregate statistics.",

        "The mechanism is not cryptographic "

        "secure aggregation or local DP.",

        "Previously executed non-private models "

        "are not covered by this guarantee.",

        "The DP branch uses fixed reference tree "

        "structures rather than the exact Cell 07 "

        "federated training procedure.",

        "Floating-point Gaussian sampling is a "

        "research implementation, not a certified "

        "production privacy mechanism.",

        "Additional releases on these protected "

        "records require further privacy accounting."

    ]

}

with open(

    REPORT_DIR

    / "cell09_privacy_verification.json",

    "w"

) as f:

    json.dump(

        privacy_report,

        f,

        indent=4

    )

# Compatibility with subsequent reporting cells.

retry_accounting = {

    "status": "COMPLETED",

    "failed_attempt_rounds":
        FAILED_ATTEMPT_ROUNDS,

    "original_sigma":
        ORIGINAL_SIGMA,

    "failed_attempt_rho":
        FAILED_ATTEMPT_RHO,

    "remaining_rho":
        REMAINING_RHO,

    "retry_sigma":
        SIGMA,

    "retry_rho":
        dp_result["rho"],

    "retry_epsilon":
        dp_result["epsilon"],

    "combined_rho":
        actual_combined_rho,

    "combined_epsilon":
        actual_combined_epsilon,

    "delta":
        TARGET_DELTA,

    "correction":
        "Direct XGBoost model slicing",

    "source_recovery_required":
        False

}

with open(

    REPORT_DIR

    / "cell09_retry_privacy_accounting.json",

    "w"

) as f:

    json.dump(

        retry_accounting,

        f,

        indent=4

    )

# ------------------------------------------------------------
# 21. Verify original source integrity
# ------------------------------------------------------------

def sha256_file(path):

    digest = hashlib.sha256()

    with open(path, "rb") as f:

        for chunk in iter(

            lambda: f.read(
                1024 * 1024
            ),

            b""

        ):

            digest.update(chunk)

    return digest.hexdigest()


with open(

    PROJECT_ROOT

    / "02_Revised_Project"

    / "reports"

    / "cell02_original_source_hashes.json"

) as f:

    original_hashes = json.load(f)

for item in original_hashes:

    original_path = (

        SOURCE_ROOT

        / item["file"]

    )

    assert original_path.exists()

    assert (

        sha256_file(original_path)

        == item["sha256"]

    ), f"Original source modified: {original_path}"

print("\nOriginal source integrity: PASS")

# ------------------------------------------------------------
# 22. Save merged source code
#
# Colab history is used ONLY for source archival.
# No function extraction or execution from history.
# Failure to archive does not repeat private training.
# ------------------------------------------------------------

source_saved = False

try:

    ip = get_ipython()

    if ip is not None:

        current_source = ip.user_ns["In"][-1]

        if current_source.strip():

            (

                CODE_DIR

                / "cell09_differential_privacy.py"

            ).write_text(

                current_source,

                encoding="utf-8"

            )

            source_saved = True

except Exception as exc:

    print(

        "Source archival warning:",

        str(exc)

    )

print(

    "Complete merged source saved:",

    source_saved

)

# ------------------------------------------------------------
# 23. Save final completion checkpoint
# ------------------------------------------------------------

checkpoint = {

    "cell": 9,

    "status": "COMPLETED",

    "reviewer_comment": "Major 9",

    "merged_corrected_version": True,

    "completed_rounds": N_ROUNDS,

    "previous_disclosed_rounds":
        FAILED_ATTEMPT_ROUNDS,

    "target_epsilon":
        TARGET_EPSILON,

    "accounted_epsilon":
        actual_combined_epsilon,

    "retry_epsilon":
        dp_result["epsilon"],

    "delta":
        TARGET_DELTA,

    "sigma":
        SIGMA,

    "rho":
        actual_combined_rho,

    "privacy_budget_verified":
        True,

    "privacy_scope":
        "Protected claim records only",

    "private_training_completed":
        True,

    "original_predictors":
        42,

    "xgboost_slicing_verified":
        True,

    "compatibility_error_corrected":
        True,

    "source_recovery_required":
        False,

    "original_source_modified":
        False,

    "previous_results_modified":
        False,

    "merged_source_saved":
        source_saved,

    "privacy_utility_tradeoff_completed":
        False,

    "dp_run_name":
        "epsilon_1_retry_seed42",

    "next_cell":
        10

}

with open(

    CHECKPOINT_DIR

    / "cell09_completed.json",

    "w"

) as f:

    json.dump(

        checkpoint,

        f,

        indent=4

    )

gc.collect()

print("\n" + "=" * 75)
print("CELL 09 COMPLETED SUCCESSFULLY")
print("=" * 75)

print("Gaussian mechanism: IMPLEMENTED")
print("Gradient sensitivity: BOUNDED")
print("Hessian sensitivity: BOUNDED")
print("XGBoost model slicing: VERIFIED")
print("All ten model slices: PASS")
print("Ten private rounds: COMPLETED")
print("Previous XGBoostError: CORRECTED")
print("Previous StopIteration: CORRECTED")
print("Source recovery dependency: NONE")
print("Combined privacy budget: PASS")
print("Original 42 predictors: RETAINED")
print("Original source integrity: PASS")
print("Previous Cells 01-08: UNCHANGED")
print("Hyperparameter tuning: NONE")
print("External dataset: NONE")
print("DP model updates saved.")
print("Privacy verification report saved.")
print("Reviewer-ready tables saved.")
print("Checkpoint saved.")

print(

    "\nTotal execution seconds:",

    round(

        time.perf_counter()

        - START_TIME,

        2

    )

)

print(

    "\nREADY FOR CELL 10/15: "

    "PRIVACY-UTILITY TRADEOFF ANALYSIS"

)
```

```python
# ============================================================
# CELL 10/15 - PRIVACY-UTILITY TRADE-OFF ANALYSIS
#
# Reviewer Comment: Major 9
#
# Evaluate different Gaussian noise levels.
# Preserve verified Cell 09 results.
# Explicit cumulative privacy accounting.
#
# NO NEW CLASSIFIER
# NO HYPERPARAMETER TUNING
# NO FEATURE MODIFICATION
# NO EXTERNAL DATASET
# ============================================================

import os
import gc
import json
import time
import hashlib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from pathlib import Path
from scipy.special import expit

START_TIME = time.perf_counter()

print("=" * 75)
print("CELL 10: PRIVACY-UTILITY TRADE-OFF ANALYSIS")
print("=" * 75)

# ------------------------------------------------------------
# 1. Restore existing workspace
# ------------------------------------------------------------

PROJECT_ROOT = Path(
    "/content/drive/MyDrive/Narayan_Patil_Reviewer_Revision"
)

SOURCE_ROOT = (
    PROJECT_ROOT / "01_Original_Project" / "SBN"
)

REVISED_DIR = (
    PROJECT_ROOT / "03_Reviewer_Only_Revision"
)

CODE_DIR = REVISED_DIR / "code"
MODEL_DIR = REVISED_DIR / "models"
DATA_DIR = REVISED_DIR / "data"
RESULT_DIR = REVISED_DIR / "results"
TABLE_DIR = RESULT_DIR / "tables"
FIGURE_DIR = RESULT_DIR / "figures"
PREDICTION_DIR = RESULT_DIR / "predictions"
REPORT_DIR = REVISED_DIR / "reports"
CHECKPOINT_DIR = REVISED_DIR / "checkpoints"

DP_DIR = MODEL_DIR / "differential_privacy"

TRADEOFF_DIR = DP_DIR / "privacy_utility_tradeoff"

TRADEOFF_DIR.mkdir(
    parents=True,
    exist_ok=True
)

FIGURE_DIR.mkdir(
    parents=True,
    exist_ok=True
)

assert (
    CHECKPOINT_DIR / "cell09_completed.json"
).exists(), "Complete Cell 09 first."

assert not (
    CHECKPOINT_DIR / "cell10_completed.json"
).exists(), "Cell 10 already completed."

TRADEOFF_LEDGER = (
    REPORT_DIR / "cell10_privacy_release_ledger.json"
)

assert not TRADEOFF_LEDGER.exists(), (
    "A previous Cell 10 release has started. "
    "Do not rerun without checking privacy accounting."
)

print("\nWorkspace:", REVISED_DIR)

# ------------------------------------------------------------
# 2. Verify Cell 09 checkpoint
# ------------------------------------------------------------

with open(
    CHECKPOINT_DIR / "cell09_completed.json"
) as f:

    checkpoint09 = json.load(f)

with open(
    REPORT_DIR / "cell09_privacy_verification.json"
) as f:

    privacy09 = json.load(f)

assert checkpoint09["status"] == "COMPLETED"

assert checkpoint09["completed_rounds"] == 10

assert checkpoint09["original_predictors"] == 42

assert np.isclose(
    checkpoint09["accounted_epsilon"],
    1.0
)

assert np.isclose(
    checkpoint09["delta"],
    1e-5
)

reference_result = pd.read_csv(
    TABLE_DIR / "cell09_dp_final_results.csv"
).iloc[0].to_dict()

REFERENCE_SIGMA = float(
    checkpoint09["sigma"]
)

PREVIOUS_RHO = float(
    checkpoint09["rho"]
)

TARGET_DELTA = float(
    checkpoint09["delta"]
)

print("\nCell 09 checkpoint: VERIFIED")
print("Original DP results: PRESERVED")
print("Reference sigma:", REFERENCE_SIGMA)
print("Previously accounted epsilon: 1.0")
print("Delta:", TARGET_DELTA)

# ------------------------------------------------------------
# 3. Verify existing runtime objects
# ------------------------------------------------------------

required_objects = [

    "TREE_SLICES",
    "leaf_catalogue",
    "client_matrices",
    "client_labels",
    "client_initial_margins",
    "initial_test_margin",
    "dtest",
    "y_test",
    "test_ids",
    "protected_ids",
    "public_ids",
    "feature_names",
    "N_ROUNDS",
    "N_CLIENTS",
    "TREES_PER_ROUND",
    "SENSITIVITY",
    "LEARNING_RATE",
    "HESSIAN_FLOOR",
    "gaussian_noise",
    "evaluate"

]

missing_objects = [

    name for name in required_objects

    if name not in globals()

]

assert not missing_objects, (
    f"Missing runtime objects: {missing_objects}. "
    "Do not rerun Cell 09 automatically."
)

assert len(TREE_SLICES) == 10
assert len(leaf_catalogue) == 400
assert len(feature_names) == 42

assert N_CLIENTS == 3
assert N_ROUNDS == 10

assert len(protected_ids) == 401912
assert len(public_ids) == 44656
assert len(y_test) == 111643

assert all(

    model.num_boosted_rounds() == 40

    for model in TREE_SLICES

)

print("\nExisting runtime objects: VERIFIED")
print("Fixed XGBoost structures: VERIFIED")
print("Original 42 predictors: RETAINED")
print("Protected clients: 3")
print("Protected records: 401,912")
print("Testing records: 111,643")

# ------------------------------------------------------------
# 4. Privacy accountant
# ------------------------------------------------------------

def epsilon_from_rho(rho, delta):

    return float(

        rho

        + 2.0 * np.sqrt(

            rho * np.log(1.0 / delta)

        )

    )


def privacy_cost(sigma, rounds=N_ROUNDS):

    rho = float(

        rounds

        * SENSITIVITY**2

        / (2.0 * sigma**2)

    )

    epsilon = epsilon_from_rho(

        rho,

        TARGET_DELTA

    )

    return rho, epsilon

# ------------------------------------------------------------
# 5. Define reviewer-required noise configurations
# ------------------------------------------------------------

NOISE_CONFIGS = [

    {

        "configuration": "Lower noise",

        "run_name": "sigma_half",

        "sigma": REFERENCE_SIGMA * 0.5,

        "noise_multiplier": 0.5

    },

    {

        "configuration": "Higher noise",

        "run_name": "sigma_double",

        "sigma": REFERENCE_SIGMA * 2.0,

        "noise_multiplier": 2.0

    }

]

planned_records = []

for config in NOISE_CONFIGS:

    rho, epsilon = privacy_cost(

        config["sigma"]

    )

    config["rho"] = rho

    config["standalone_epsilon"] = epsilon

    planned_records.append({

        "configuration":
            config["configuration"],

        "sigma":
            config["sigma"],

        "noise_multiplier":
            config["noise_multiplier"],

        "standalone_epsilon":
            epsilon,

        "rho":
            rho,

        "delta":
            TARGET_DELTA

    })

additional_rho = sum(

    config["rho"]

    for config in NOISE_CONFIGS

)

planned_combined_rho = (

    PREVIOUS_RHO

    + additional_rho

)

planned_combined_epsilon = epsilon_from_rho(

    planned_combined_rho,

    TARGET_DELTA

)

print("\n" + "=" * 75)
print("PLANNED PRIVACY-UTILITY EXPERIMENTS")
print("=" * 75)

print(

    pd.DataFrame(planned_records)

    .round(8)

    .to_string(index=False)

)

print(

    "\nPreviously consumed rho:",

    round(PREVIOUS_RHO, 10)

)

print(

    "Additional planned rho:",

    round(additional_rho, 10)

)

print(

    "Combined planned epsilon:",

    round(planned_combined_epsilon, 10)

)

print(

    "Note: The combined research release budget "

    "exceeds epsilon=1.0."

)

print(

    "The original Cell 09 accounting "

    "will remain unchanged."

)

# ------------------------------------------------------------
# 6. Protect original Cell 09 saved results
# ------------------------------------------------------------

def sha256_file(path):

    digest = hashlib.sha256()

    with open(path, "rb") as f:

        for chunk in iter(

            lambda: f.read(1024 * 1024),

            b""

        ):

            digest.update(chunk)

    return digest.hexdigest()


protected_files = [

    TABLE_DIR / "cell09_dp_final_results.csv",

    TABLE_DIR / "cell09_dp_round_metrics.csv",

    REPORT_DIR / "cell09_privacy_verification.json",

    CHECKPOINT_DIR / "cell09_completed.json"

]

original_hashes09 = {

    str(path): sha256_file(path)

    for path in protected_files

}

print("\nCell 09 result integrity: RECORDED")

# ------------------------------------------------------------
# 7. Validate model slices before new releases
# ------------------------------------------------------------

for round_index, model_slice in enumerate(
    TREE_SLICES
):

    leaves = model_slice.predict(

        dtest,

        pred_leaf=True

    )

    assert leaves.size == (

        len(y_test)

        * TREES_PER_ROUND

    )

    del leaves

print("All ten tree slices: PASS")

gc.collect()

# ------------------------------------------------------------
# 8. Reserve cumulative privacy budget
# ------------------------------------------------------------

ledger10 = {

    "status": "BUDGET_RESERVED",

    "previous_rho": PREVIOUS_RHO,

    "previous_epsilon": 1.0,

    "reserved_additional_rho":
        additional_rho,

    "reserved_combined_rho":
        planned_combined_rho,

    "reserved_combined_epsilon":
        planned_combined_epsilon,

    "delta": TARGET_DELTA,

    "completed_additional_rho": 0.0,

    "completed_runs": [],

    "completed_rounds": []

}

def save_tradeoff_ledger():

    temp_path = TRADEOFF_LEDGER.with_suffix(
        ".tmp"
    )

    with open(temp_path, "w") as f:

        json.dump(

            ledger10,

            f,

            indent=4

        )

    os.replace(

        temp_path,

        TRADEOFF_LEDGER

    )


with open(
    TRADEOFF_LEDGER,
    "x"
) as f:

    json.dump(
        ledger10,
        f,
        indent=4
    )

print("\nAdditional privacy budget: RESERVED")

# ------------------------------------------------------------
# 9. Reuse the Cell 09 private refinement mechanism
#
# Same fixed trees, gradients, Hessians, learning
# rate, clipping rules and protected records.
#
# Only Gaussian sigma changes.
# ------------------------------------------------------------

def run_tradeoff_experiment(config):

    sigma = float(config["sigma"])

    run_name = config["run_name"]

    run_dir = TRADEOFF_DIR / run_name

    assert not run_dir.exists(), (
        f"Run directory already exists: {run_dir}"
    )

    run_dir.mkdir(
        parents=True
    )

    run_start = time.perf_counter()

    local_margins = [

        margin.copy()

        for margin in client_initial_margins

    ]

    test_margin = initial_test_margin.copy()

    round_records = []

    saved_updates = {}

    rho_per_round = float(

        SENSITIVITY**2

        / (2.0 * sigma**2)

    )

    for round_no in range(
        1,
        N_ROUNDS + 1
    ):

        round_start = time.perf_counter()

        start_tree = (

            round_no - 1

        ) * TREES_PER_ROUND

        end_tree = (

            start_tree

            + TREES_PER_ROUND

        )

        current_model = TREE_SLICES[
            round_no - 1
        ]

        catalogue = leaf_catalogue[
            start_tree:end_tree
        ]

        offsets = np.concatenate([

            [0],

            np.cumsum([

                len(leaves)

                for leaves in catalogue

            ])

        ]).astype(np.int64)

        total_leaves = int(
            offsets[-1]
        )

        aggregate_g = np.zeros(

            total_leaves,

            dtype=np.float64

        )

        aggregate_h = np.zeros(

            total_leaves,

            dtype=np.float64

        )

        client_positions = []

        # ----------------------------------------------------
        # A. Bounded client gradient/Hessian statistics
        # ----------------------------------------------------

        for client_no in range(N_CLIENTS):

            matrix = client_matrices[
                client_no
            ]

            probability = expit(

                local_margins[client_no]

            )

            gradient = np.clip(

                probability

                - client_labels[client_no],

                -1.0,

                1.0

            )

            hessian = np.clip(

                probability

                * (1.0 - probability),

                0.0,

                0.25

            )

            leaves = current_model.predict(

                matrix,

                pred_leaf=True

            ).astype(np.int32)

            leaves = leaves.reshape(

                len(gradient),

                TREES_PER_ROUND

            )

            positions = np.empty(

                leaves.shape,

                dtype=np.int32

            )

            for j, tree_leaves in enumerate(
                catalogue
            ):

                position = np.searchsorted(

                    tree_leaves,

                    leaves[:, j]

                )

                assert np.all(

                    position < len(tree_leaves)

                )

                assert np.array_equal(

                    tree_leaves[position],

                    leaves[:, j]

                )

                positions[:, j] = (

                    position

                    + offsets[j]

                )

                start = offsets[j]
                end = offsets[j + 1]

                aggregate_g[start:end] += (

                    np.bincount(

                        position,

                        weights=gradient,

                        minlength=len(tree_leaves)

                    )

                )

                aggregate_h[start:end] += (

                    np.bincount(

                        position,

                        weights=hessian,

                        minlength=len(tree_leaves)

                    )

                )

            client_positions.append(
                positions
            )

            del (

                gradient,

                hessian,

                probability,

                leaves

            )

        # ----------------------------------------------------
        # B. Gaussian perturbation
        # ----------------------------------------------------

        noisy_aggregate = np.concatenate([

            aggregate_g,

            aggregate_h

        ])

        noisy_aggregate += gaussian_noise(

            size=len(noisy_aggregate),

            sigma=sigma

        )

        noisy_g = noisy_aggregate[
            :total_leaves
        ]

        noisy_h = noisy_aggregate[
            total_leaves:
        ]

        denominator = np.maximum(

            noisy_h + 1.0,

            HESSIAN_FLOOR

        )

        leaf_update = (

            -LEARNING_RATE

            * np.clip(

                noisy_g / denominator,

                -1.0,

                1.0

            )

        )

        assert np.isfinite(
            leaf_update
        ).all()

        # ----------------------------------------------------
        # C. Apply privatized updates to clients
        # ----------------------------------------------------

        for client_no in range(N_CLIENTS):

            local_margins[client_no] += (

                leaf_update[

                    client_positions[client_no]

                ].sum(axis=1)

            )

        # ----------------------------------------------------
        # D. Apply identical updates to held-out test data
        # ----------------------------------------------------

        test_leaves = current_model.predict(

            dtest,

            pred_leaf=True

        ).astype(np.int32)

        test_leaves = test_leaves.reshape(

            len(y_test),

            TREES_PER_ROUND

        )

        for j, tree_leaves in enumerate(
            catalogue
        ):

            position = np.searchsorted(

                tree_leaves,

                test_leaves[:, j]

            )

            assert np.all(

                position < len(tree_leaves)

            )

            assert np.array_equal(

                tree_leaves[position],

                test_leaves[:, j]

            )

            test_margin += leaf_update[

                position + offsets[j]

            ]

        saved_updates[

            f"round_{round_no:02d}_updates"

        ] = leaf_update.copy()

        saved_updates[

            f"round_{round_no:02d}_offsets"

        ] = offsets.copy()

        # ----------------------------------------------------
        # E. Update privacy ledger BEFORE displaying results
        # ----------------------------------------------------

        ledger10["completed_additional_rho"] += (
            rho_per_round
        )

        cumulative_rho = (

            PREVIOUS_RHO

            + ledger10["completed_additional_rho"]

        )

        cumulative_epsilon = epsilon_from_rho(

            cumulative_rho,

            TARGET_DELTA

        )

        standalone_epsilon = epsilon_from_rho(

            round_no * rho_per_round,

            TARGET_DELTA

        )

        ledger10["status"] = "IN_PROGRESS"

        ledger10["completed_rounds"].append({

            "run": run_name,

            "round": round_no,

            "rho": rho_per_round,

            "cumulative_epsilon":
                cumulative_epsilon

        })

        save_tradeoff_ledger()

        # ----------------------------------------------------
        # F. Evaluate utility
        # ----------------------------------------------------

        metrics, _, _ = evaluate(

            y_test,

            test_margin

        )

        record = {

            "configuration":
                config["configuration"],

            "round":
                round_no,

            "sigma":
                sigma,

            "standalone_epsilon":
                standalone_epsilon,

            "cumulative_release_epsilon":
                cumulative_epsilon,

            "delta":
                TARGET_DELTA,

            **metrics,

            "round_seconds":

                time.perf_counter()

                - round_start

        }

        round_records.append(record)

        print(

            f"Round {round_no:02d}/{N_ROUNDS}"

            f" | Epsilon: "
            f"{standalone_epsilon:.6f}"

            f" | Accuracy: "
            f"{metrics['accuracy']:.6f}"

            f" | Recall: "
            f"{metrics['recall']:.6f}"

            f" | AUC: "
            f"{metrics['roc_auc']:.6f}",

            flush=True

        )

        pd.DataFrame(

            round_records

        ).to_csv(

            run_dir / "round_metrics.csv",

            index=False

        )

        # No raw private aggregate statistics
        # are saved to Google Drive.

        del (

            aggregate_g,

            aggregate_h,

            noisy_aggregate,

            noisy_g,

            noisy_h,

            client_positions,

            test_leaves

        )

        gc.collect()

    # --------------------------------------------------------
    # G. Save final experiment results
    # --------------------------------------------------------

    metrics, probability, prediction = evaluate(

        y_test,

        test_margin

    )

    rho, epsilon = privacy_cost(sigma)

    np.savez_compressed(

        run_dir / "private_leaf_updates.npz",

        **saved_updates

    )

    np.savez_compressed(

        run_dir / "test_predictions.npz",

        row_id=test_ids,

        probability=probability,

        prediction=prediction

    )

    final_result = {

        "configuration":
            config["configuration"],

        "run_name":
            run_name,

        "sigma":
            sigma,

        "noise_multiplier":
            config["noise_multiplier"],

        "rho":
            rho,

        "standalone_epsilon":
            epsilon,

        "delta":
            TARGET_DELTA,

        "rounds":
            N_ROUNDS,

        "clients":
            N_CLIENTS,

        "protected_records":
            len(protected_ids),

        "test_records":
            len(y_test),

        **metrics,

        "execution_seconds":

            time.perf_counter()

            - run_start

    }

    with open(

        run_dir / "final_report.json",

        "w"

    ) as f:

        json.dump(

            final_result,

            f,

            indent=4

        )

    ledger10["completed_runs"].append(

        run_name

    )

    save_tradeoff_ledger()

    return final_result, round_records

# ------------------------------------------------------------
# 10. Execute reviewer-required noise experiments
# ------------------------------------------------------------

tradeoff_results = []

tradeoff_round_records = []

for config in NOISE_CONFIGS:

    print("\n" + "=" * 75)

    print(

        "PRIVACY CONFIGURATION:",

        config["configuration"]

    )

    print("=" * 75)

    print("Sigma:", config["sigma"])

    print(

        "Standalone epsilon:",

        config["standalone_epsilon"]

    )

    result, records = (

        run_tradeoff_experiment(config)

    )

    tradeoff_results.append(result)

    tradeoff_round_records.extend(records)

    print("\nFINAL CONFIGURATION RESULTS")

    for key in [

        "accuracy",

        "precision",

        "recall",

        "f1",

        "roc_auc"

    ]:

        print(

            f"{key:15s}: "

            f"{result[key]:.6f}"

        )

    pd.DataFrame(

        tradeoff_results

    ).to_csv(

        TABLE_DIR
        / "cell10_tradeoff_partial_results.csv",

        index=False

    )

# ------------------------------------------------------------
# 11. Add existing Cell 09 reference results
#
# No reference retraining.
# ------------------------------------------------------------

reference_row = {

    "configuration":
        "Reference noise",

    "run_name":
        "cell09_existing_reference",

    "sigma":
        REFERENCE_SIGMA,

    "noise_multiplier":
        1.0,

    "rho":
        float(reference_result["rho"]),

    "standalone_epsilon":
        float(reference_result["epsilon"]),

    "delta":
        TARGET_DELTA,

    "rounds":
        10,

    "clients":
        3,

    "protected_records":
        401912,

    "test_records":
        111643,

    "accuracy":
        float(reference_result["accuracy"]),

    "precision":
        float(reference_result["precision"]),

    "recall":
        float(reference_result["recall"]),

    "f1":
        float(reference_result["f1"]),

    "roc_auc":
        float(reference_result["roc_auc"]),

    "execution_seconds":
        float(reference_result["execution_seconds"])

}

all_results = [

    reference_row,

    *tradeoff_results

]

tradeoff_df = pd.DataFrame(

    all_results

).sort_values(

    "sigma"

).reset_index(

    drop=True

)

assert len(tradeoff_df) == 3

tradeoff_df.to_csv(

    TABLE_DIR

    / "cell10_privacy_utility_comparison.csv",

    index=False

)

# ------------------------------------------------------------
# 12. Consolidate round-wise results
# ------------------------------------------------------------

reference_rounds = pd.read_csv(

    TABLE_DIR / "cell09_dp_round_metrics.csv"

)

reference_rounds["configuration"] = (
    "Reference noise"
)

reference_rounds["standalone_epsilon"] = (
    reference_rounds["epsilon"]
)

reference_rounds["cumulative_release_epsilon"] = (
    reference_rounds["combined_epsilon"]
)

new_rounds = pd.DataFrame(
    tradeoff_round_records
)

round_df = pd.concat([

    reference_rounds,

    new_rounds

], ignore_index=True)

assert len(round_df) == 30

round_df.to_csv(

    TABLE_DIR

    / "cell10_privacy_utility_round_metrics.csv",

    index=False

)

# ------------------------------------------------------------
# 13. Generate final comparison table
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("FINAL PRIVACY-UTILITY TRADE-OFF RESULTS")
print("=" * 75)

display_columns = [

    "configuration",

    "sigma",

    "standalone_epsilon",

    "accuracy",

    "precision",

    "recall",

    "f1",

    "roc_auc"

]

print(

    tradeoff_df[display_columns]

    .round(6)

    .to_string(index=False)

)

# Manuscript-ready percentage table.

publication_table = tradeoff_df[
    display_columns
].copy()

for metric in [

    "accuracy",

    "precision",

    "recall",

    "f1"

]:

    publication_table[metric] = (

        publication_table[metric]

        * 100

    ).round(2)

publication_table = publication_table.rename(

    columns={

        "configuration": "Configuration",

        "sigma": "Gaussian sigma",

        "standalone_epsilon":
            "Standalone epsilon",

        "accuracy": "Accuracy (%)",

        "precision": "Precision (%)",

        "recall": "Recall (%)",

        "f1": "F1-score (%)",

        "roc_auc": "ROC-AUC"

    }

)

publication_table.to_csv(

    TABLE_DIR

    / "cell10_manuscript_privacy_table.csv",

    index=False

)

print("\nMANUSCRIPT-READY TABLE")

print(

    publication_table

    .round(6)

    .to_string(index=False)

)

# ------------------------------------------------------------
# 14. Reviewer-required noise versus accuracy figure
# ------------------------------------------------------------

plt.figure(
    figsize=(7, 4.5)
)

plt.plot(

    tradeoff_df["sigma"],

    tradeoff_df["accuracy"] * 100,

    marker="o",

    color="black",

    linewidth=1.5,

    markersize=6

)

for _, row in tradeoff_df.iterrows():

    plt.annotate(

        f"{row['accuracy'] * 100:.2f}%",

        (

            row["sigma"],

            row["accuracy"] * 100

        ),

        textcoords="offset points",

        xytext=(0, 10),

        ha="center",

        fontsize=9

    )

plt.xlabel(
    "Gaussian noise standard deviation (sigma)"
)

plt.ylabel(
    "Accuracy (%)"
)

plt.xticks(

    tradeoff_df["sigma"],

    [

        f"{value:.2f}"

        for value in tradeoff_df["sigma"]

    ]

)

plt.grid(
    alpha=0.25,
    linestyle=":"
)

plt.margins(
    y=0.2
)

plt.tight_layout()

plt.savefig(

    FIGURE_DIR

    / "cell10_gaussian_noise_accuracy.png",

    dpi=600,

    bbox_inches="tight"

)

plt.show()
plt.close()

print("\nNoise-accuracy figure: SAVED")

# ------------------------------------------------------------
# 15. Privacy budget versus accuracy figure
# ------------------------------------------------------------

plot_df = tradeoff_df.sort_values(

    "standalone_epsilon"

)

plt.figure(
    figsize=(7, 4.5)
)

plt.plot(

    plot_df["standalone_epsilon"],

    plot_df["accuracy"] * 100,

    marker="s",

    color="black",

    linewidth=1.5

)

plt.xlabel(
    "Standalone privacy budget (epsilon)"
)

plt.ylabel(
    "Accuracy (%)"
)

plt.grid(

    alpha=0.25,

    linestyle=":"

)

plt.tight_layout()

plt.savefig(

    FIGURE_DIR

    / "cell10_epsilon_accuracy_tradeoff.png",

    dpi=600,

    bbox_inches="tight"

)

plt.show()
plt.close()

print("Privacy-utility figure: SAVED")

# ------------------------------------------------------------
# 16. Final cumulative privacy accounting
# ------------------------------------------------------------

assert len(
    ledger10["completed_runs"]
) == 2

assert len(
    ledger10["completed_rounds"]
) == 20

actual_additional_rho = float(

    ledger10["completed_additional_rho"]

)

assert np.isclose(

    actual_additional_rho,

    additional_rho,

    atol=1e-12

)

final_combined_rho = (

    PREVIOUS_RHO

    + actual_additional_rho

)

final_combined_epsilon = epsilon_from_rho(

    final_combined_rho,

    TARGET_DELTA

)

assert np.isclose(

    final_combined_epsilon,

    planned_combined_epsilon,

    atol=1e-10

)

ledger10["status"] = "COMPLETED"

ledger10["final_combined_rho"] = (
    final_combined_rho
)

ledger10["final_combined_epsilon"] = (
    final_combined_epsilon
)

save_tradeoff_ledger()

accounting_records = [

    {

        "release": "Cell 09 including failed round",

        "rho": PREVIOUS_RHO,

        "epsilon": 1.0,

        "delta": TARGET_DELTA

    }

]

for result in tradeoff_results:

    accounting_records.append({

        "release": result["configuration"],

        "rho": result["rho"],

        "epsilon":
            result["standalone_epsilon"],

        "delta": TARGET_DELTA

    })

accounting_records.append({

    "release": "Combined Cell 09 and Cell 10",

    "rho": final_combined_rho,

    "epsilon": final_combined_epsilon,

    "delta": TARGET_DELTA

})

accounting_df = pd.DataFrame(

    accounting_records

)

accounting_df.to_csv(

    TABLE_DIR

    / "cell10_cumulative_privacy_accounting.csv",

    index=False

)

print("\n" + "=" * 75)
print("CUMULATIVE PRIVACY ACCOUNTING")
print("=" * 75)

print(

    accounting_df

    .round(8)

    .to_string(index=False)

)

print(

    "\nFinal combined theoretical epsilon:",

    round(final_combined_epsilon, 10)

)

print("Delta:", TARGET_DELTA)

print("Cumulative accounting: PASS")

print(

    "The combined budget exceeds epsilon=1."

)

print(

    "Standalone and cumulative privacy budgets "

    "must be reported separately."

)

# ------------------------------------------------------------
# 17. Verify original Cell 09 results unchanged
# ------------------------------------------------------------

for path_string, original_hash in (

    original_hashes09.items()

):

    assert (

        sha256_file(Path(path_string))

        == original_hash

    ), f"Cell 09 result modified: {path_string}"

print("\nOriginal Cell 09 results: UNCHANGED")

# ------------------------------------------------------------
# 18. Verify original project source integrity
# ------------------------------------------------------------

with open(

    PROJECT_ROOT

    / "02_Revised_Project"

    / "reports"

    / "cell02_original_source_hashes.json"

) as f:

    original_source_hashes = json.load(f)

for item in original_source_hashes:

    source_path = SOURCE_ROOT / item["file"]

    assert source_path.exists()

    assert (

        sha256_file(source_path)

        == item["sha256"]

    )

print("Original source integrity: PASS")

# ------------------------------------------------------------
# 19. Save reviewer-specific verification report
# ------------------------------------------------------------

report10 = {

    "cell": 10,

    "status": "COMPLETED",

    "reviewer_comment": "Major 9",

    "experiment":
        "Gaussian noise versus predictive utility",

    "original_predictors":
        42,

    "noise_configurations":
        3,

    "new_noise_experiments":
        2,

    "communication_rounds_per_run":
        10,

    "additional_private_rounds":
        20,

    "reference_sigma":
        REFERENCE_SIGMA,

    "delta":
        TARGET_DELTA,

    "previous_rho":
        PREVIOUS_RHO,

    "additional_rho":
        actual_additional_rho,

    "combined_rho":
        final_combined_rho,

    "combined_epsilon":
        final_combined_epsilon,

    "cell09_accounted_epsilon":
        1.0,

    "cell09_standalone_epsilon":
        float(reference_result["epsilon"]),

    "sensitivity":
        SENSITIVITY,

    "results":
        all_results,

    "same_protected_records":
        True,

    "same_tree_structures":
        True,

    "same_test_records":
        True,

    "same_refinement_parameters":
        True,

    "only_sigma_varied":
        True,

    "original_cell09_results_modified":
        False,

    "original_source_modified":
        False,

    "hyperparameter_tuning":
        False,

    "external_dataset_used":
        False,

    "limitations": [

        "Privacy accounting is conditional on "
        "fixed public reference data, schema, "
        "record membership and test data.",

        "The privacy unit is an individual "
        "protected claim row.",

        "The trusted coordinator can access "
        "unperturbed aggregate statistics.",

        "Standalone epsilon values do not "
        "represent the cumulative privacy cost "
        "of all released experiments.",

        "The complete research release budget "
        "exceeds epsilon=1.",

        "Previously trained non-private models "
        "are not covered by this accounting.",

        "The floating-point Gaussian sampler "
        "is not a certified production mechanism.",

        "Each noise configuration uses one "
        "stochastic realization. Statistical "
        "significance has not been established."

    ],

    "elapsed_seconds":

        time.perf_counter()

        - START_TIME

}

with open(

    REPORT_DIR

    / "cell10_privacy_utility_report.json",

    "w"

) as f:

    json.dump(

        report10,

        f,

        indent=4,

        default=str

    )

# ------------------------------------------------------------
# 20. Save complete source code
# ------------------------------------------------------------

source_saved = False

try:

    ip = get_ipython()

    if ip is not None:

        current_source = ip.user_ns["In"][-1]

        if current_source.strip():

            (

                CODE_DIR

                / "cell10_privacy_utility_tradeoff.py"

            ).write_text(

                current_source,

                encoding="utf-8"

            )

            source_saved = True

except Exception as exc:

    print(

        "Source archival warning:",

        str(exc)

    )

# ------------------------------------------------------------
# 21. Save completion checkpoint
# ------------------------------------------------------------

checkpoint10 = {

    "cell": 10,

    "status": "COMPLETED",

    "reviewer_comment": "Major 9",

    "noise_configurations_completed": 3,

    "additional_private_rounds": 20,

    "privacy_utility_tradeoff_completed": True,

    "cumulative_privacy_accounting_completed": True,

    "combined_epsilon":
        final_combined_epsilon,

    "delta":
        TARGET_DELTA,

    "original_cell09_results_modified":
        False,

    "original_predictors":
        42,

    "original_source_modified":
        False,

    "hyperparameter_tuning":
        False,

    "source_saved":
        source_saved,

    "next_cell":
        11

}

with open(

    CHECKPOINT_DIR

    / "cell10_completed.json",

    "w"

) as f:

    json.dump(

        checkpoint10,

        f,

        indent=4

    )

gc.collect()

print("\n" + "=" * 75)
print("CELL 10 COMPLETED SUCCESSFULLY")
print("=" * 75)

print("Gaussian noise configurations: 3")
print("Additional noise experiments: 2")
print("Privacy-utility comparison: COMPLETED")
print("Noise-accuracy figure: SAVED")
print("Privacy-utility figure: SAVED")
print("Cumulative privacy accounting: PASS")
print("Original Cell 09 results: UNCHANGED")
print("Original 42 predictors: RETAINED")
print("Original source integrity: PASS")
print("Hyperparameter tuning: NONE")
print("Additional dataset: NONE")
print("Results and models saved in Google Drive.")
print("Checkpoint saved.")

print(

    "\nTotal execution seconds:",

    round(

        time.perf_counter()

        - START_TIME,

        2

    )

)

print(

    "\nREADY FOR CELL 11/15: "

    "BLOCKCHAIN CONSENSUS AND SCALABILITY"

)
```

```python
# ============================================================
# CELL 11/15
# BLOCKCHAIN CONSENSUS AND SCALABILITY
#
# Reviewer Comments:
# Major 2  : Blockchain scalability and consensus metrics
# Minor 29 : Consensus and Merkle root implementation
#
# NO NEW DATASET
# NO SYNTHETIC FRAUD RECORDS
# NO CLASSIFIER RETRAINING
# NO MODIFICATION TO PREVIOUS RESULTS
# ============================================================

import os
import gc
import json
import time
import hashlib
import secrets
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from pathlib import Path

from google.colab import drive

from cryptography.hazmat.primitives.asymmetric.ed25519 import (
    Ed25519PrivateKey,
    Ed25519PublicKey
)

from cryptography.hazmat.primitives import serialization
from cryptography.exceptions import InvalidSignature

drive.mount("/content/drive")

START_TIME = time.perf_counter()

SEED = 42

SCALES = [1000, 10000, 100000]

TRANSACTIONS_PER_BLOCK = 100

N_VALIDATORS = 3

QUORUM = 2

# ------------------------------------------------------------
# 1. Restore existing reviewer-only workspace
# ------------------------------------------------------------

PROJECT_ROOT = Path(
    "/content/drive/MyDrive/Narayan_Patil_Reviewer_Revision"
)

SOURCE_ROOT = (
    PROJECT_ROOT / "01_Original_Project" / "SBN"
)

REVISED_DIR = (
    PROJECT_ROOT / "03_Reviewer_Only_Revision"
)

CODE_DIR = REVISED_DIR / "code"
DATA_DIR = REVISED_DIR / "data"
MODEL_DIR = REVISED_DIR / "models"
RESULT_DIR = REVISED_DIR / "results"
TABLE_DIR = RESULT_DIR / "tables"
FIGURE_DIR = RESULT_DIR / "figures"
REPORT_DIR = REVISED_DIR / "reports"
CHECKPOINT_DIR = REVISED_DIR / "checkpoints"

BLOCKCHAIN_DIR = RESULT_DIR / "blockchain"

for folder in [
    CODE_DIR,
    TABLE_DIR,
    FIGURE_DIR,
    REPORT_DIR,
    CHECKPOINT_DIR,
    BLOCKCHAIN_DIR
]:

    folder.mkdir(
        parents=True,
        exist_ok=True
    )

assert (
    CHECKPOINT_DIR / "cell10_completed.json"
).exists(), "Complete Cell 10 first."

assert not (
    CHECKPOINT_DIR / "cell11_completed.json"
).exists(), "Cell 11 is already completed."

print("=" * 75)
print("CELL 11: BLOCKCHAIN CONSENSUS AND SCALABILITY")
print("=" * 75)

print("\nWorkspace:", REVISED_DIR)

# ------------------------------------------------------------
# 2. Verify previous checkpoint
# ------------------------------------------------------------

with open(
    CHECKPOINT_DIR / "cell10_completed.json"
) as f:

    checkpoint10 = json.load(f)

assert checkpoint10["status"] == "COMPLETED"

assert checkpoint10[
    "privacy_utility_tradeoff_completed"
] is True

print("\nCell 10 checkpoint: VERIFIED")
print("Previous model results: PRESERVED")

# ------------------------------------------------------------
# 3. Blockchain configuration
# ------------------------------------------------------------

print("\nBLOCKCHAIN CONFIGURATION")

print("Consensus: PoA-style signed authority voting")
print("Authorized validators:", N_VALIDATORS)
print("Required signatures:", QUORUM)
print("Topology: Simulated fully connected")
print("Hash algorithm: SHA-256")
print("Digital signature: Ed25519")
print("Transactions per block:", TRANSACTIONS_PER_BLOCK)
print("Scalability levels:", SCALES)
print("Execution: CPU-based local simulation")

# ------------------------------------------------------------
# 4. Utility functions
# ------------------------------------------------------------

def canonical_bytes(obj):

    return json.dumps(

        obj,

        sort_keys=True,

        separators=(",", ":"),

        ensure_ascii=False,

        allow_nan=False

    ).encode("utf-8")


def sha256_bytes(value):

    return hashlib.sha256(value).hexdigest()


def sha256_file(path):

    digest = hashlib.sha256()

    with open(path, "rb") as f:

        for chunk in iter(

            lambda: f.read(1024 * 1024),

            b""

        ):

            digest.update(chunk)

    return digest.hexdigest()

# ------------------------------------------------------------
# 5. Protect existing experimental results
# ------------------------------------------------------------

previous_files = [

    TABLE_DIR / "cell05_all_model_metrics.csv",

    TABLE_DIR / "cell07_federated_summary.csv",

    TABLE_DIR / "cell08_centralized_federated_comparison.csv",

    TABLE_DIR / "cell09_dp_final_results.csv",

    TABLE_DIR / "cell10_privacy_utility_comparison.csv"

]

previous_hashes = {

    str(path): sha256_file(path)

    for path in previous_files

}

print("\nPrevious result integrity: RECORDED")

# ------------------------------------------------------------
# 6. Load real fraud-labelled records
#
# Use original dataset only.
#
# Ground-truth fraud-labelled records are used as
# blockchain benchmarking workloads.
#
# This is NOT an assertion that every record was
# correctly detected by the classification model.
# ------------------------------------------------------------

FEATURED_PATH = (

    SOURCE_ROOT
    / "data"
    / "processed"
    / "featured_dataset.csv"

)

assert FEATURED_PATH.exists()

df = pd.read_csv(

    FEATURED_PATH,

    low_memory=False

)

assert df.shape == (558211, 43)

TARGET = "PotentialFraud"

df[TARGET] = pd.to_numeric(

    df[TARGET],

    errors="raise"

)

fraud_indices = np.flatnonzero(

    df[TARGET].to_numpy() == 1

)

assert len(fraud_indices) == 212796

assert len(fraud_indices) >= max(SCALES)

rng = np.random.default_rng(SEED)

selected_indices = rng.choice(

    fraud_indices,

    size=max(SCALES),

    replace=False

)

assert len(
    np.unique(selected_indices)
) == 100000

selected_fraud_df = df.iloc[

    selected_indices

].copy()

print("\nORIGINAL FRAUD WORKLOAD")

print("Total original records:", len(df))

print(
    "Available fraud-labelled records:",
    len(fraud_indices)
)

print(
    "Selected unique fraud records:",
    len(selected_fraud_df)
)

print("Synthetic fraud records: 0")
print("External dataset: NONE")

# ------------------------------------------------------------
# 7. Prepare evidence commitments
#
# Store salted SHA-256 commitments instead of raw
# beneficiary, provider or medical attributes.
#
# Commitment opening secrets are retained in memory
# during this benchmark and are not published.
# ------------------------------------------------------------

transaction_start = time.perf_counter()

transactions = []

commitment_nonces = []

evidence_payloads = []

for transaction_id, (row_id, record) in enumerate(

    zip(

        selected_indices,

        selected_fraud_df.itertuples(
            index=False,
            name=None
        )

    ),

    start=1

):

    # Canonical representation of the source record.

    evidence = canonical_bytes({

        "source_row": int(row_id),

        "columns": [

            str(column)

            for column in selected_fraud_df.columns

        ],

        "record": [

            None if pd.isna(value)
            else str(value)

            for value in record

        ]

    })

    # Random secret nonce prevents straightforward
    # guessing of low-entropy record contents.

    nonce = secrets.token_bytes(32)

    evidence_hash = sha256_bytes(

        b"EVIDENCE_V1"

        + nonce

        + evidence

    )

    transaction = {

        "transaction_id":
            transaction_id,

        "evidence_commitment":
            evidence_hash,

        "evidence_type":
            "fraud_labelled_claim"

    }

    transactions.append(transaction)

    commitment_nonces.append(nonce)

    evidence_payloads.append(evidence)

assert len(transactions) == 100000

assert len({

    tx["evidence_commitment"]

    for tx in transactions

}) == 100000

evidence_preparation_seconds = (

    time.perf_counter()

    - transaction_start

)

# Verify commitment construction.

for i in [0, 49999, 99999]:

    assert sha256_bytes(

        b"EVIDENCE_V1"

        + commitment_nonces[i]

        + evidence_payloads[i]

    ) == transactions[i]["evidence_commitment"]

del df
del selected_fraud_df

gc.collect()

print("\nEvidence commitments: VERIFIED")
print("Unique evidence commitments: 100,000")

print(

    "Evidence preparation seconds:",

    round(evidence_preparation_seconds, 4)

)

print("Raw medical records stored on-chain: NO")

# ------------------------------------------------------------
# 8. SHA-256 Merkle tree implementation
#
# Domain separation:
# Leaf hash     = SHA256(0x00 || transaction)
# Internal hash = SHA256(0x01 || left || right)
#
# Duplicate final node for odd-sized levels.
# ------------------------------------------------------------

def transaction_leaf_hash(transaction):

    return hashlib.sha256(

        b"\x00"

        + canonical_bytes(transaction)

    ).digest()


def merkle_root(transaction_list):

    if not transaction_list:

        return hashlib.sha256(
            b""
        ).hexdigest()

    level = [

        transaction_leaf_hash(transaction)

        for transaction in transaction_list

    ]

    while len(level) > 1:

        if len(level) % 2 == 1:

            level.append(level[-1])

        next_level = []

        for i in range(
            0,
            len(level),
            2
        ):

            parent = hashlib.sha256(

                b"\x01"

                + level[i]

                + level[i + 1]

            ).digest()

            next_level.append(parent)

        level = next_level

    return level[0].hex()

# ------------------------------------------------------------
# 9. Merkle inclusion proof
# ------------------------------------------------------------

def merkle_proof(transaction_list, index):

    assert 0 <= index < len(transaction_list)

    level = [

        transaction_leaf_hash(transaction)

        for transaction in transaction_list

    ]

    current_index = index

    proof = []

    while len(level) > 1:

        if len(level) % 2:

            level.append(level[-1])

        sibling_index = current_index ^ 1

        proof.append({

            "sibling_hash":
                level[sibling_index].hex(),

            "sibling_side":

                "left"

                if sibling_index < current_index

                else "right"

        })

        level = [

            hashlib.sha256(

                b"\x01"

                + level[i]

                + level[i + 1]

            ).digest()

            for i in range(
                0,
                len(level),
                2
            )

        ]

        current_index //= 2

    return proof


def verify_merkle_proof(
    transaction,
    proof,
    expected_root
):

    current_hash = transaction_leaf_hash(
        transaction
    )

    for item in proof:

        sibling = bytes.fromhex(

            item["sibling_hash"]

        )

        if item["sibling_side"] == "left":

            current_hash = hashlib.sha256(

                b"\x01"

                + sibling

                + current_hash

            ).digest()

        else:

            current_hash = hashlib.sha256(

                b"\x01"

                + current_hash

                + sibling

            ).digest()

    return (

        current_hash.hex()

        == expected_root

    )

# ------------------------------------------------------------
# 10. Verify Merkle root and inclusion proof
# ------------------------------------------------------------

sample_transactions = transactions[:100]

sample_root = merkle_root(
    sample_transactions
)

sample_proof = merkle_proof(

    sample_transactions,

    0

)

assert verify_merkle_proof(

    sample_transactions[0],

    sample_proof,

    sample_root

)

tampered_transaction = dict(
    sample_transactions[0]
)

tampered_transaction[
    "evidence_commitment"
] = "f" * 64

assert not verify_merkle_proof(

    tampered_transaction,

    sample_proof,

    sample_root

)

print("\nMERKLE TREE VERIFICATION")

print("SHA-256 Merkle root: PASS")
print("Valid inclusion proof: PASS")
print("Tampered transaction rejected: PASS")

# ------------------------------------------------------------
# 11. Initialize authorized PoA validator nodes
#
# Independent Ed25519 key pairs.
# Private validator keys are not published.
# ------------------------------------------------------------

validator_private_keys = {

    f"validator_{i + 1}":

        Ed25519PrivateKey.generate()

    for i in range(N_VALIDATORS)

}

validator_public_keys = {

    validator_id:

        private_key.public_key()

    for validator_id, private_key

    in validator_private_keys.items()

}

validator_ids = list(
    validator_public_keys
)

public_key_registry = {

    validator_id:

        public_key.public_bytes(

            encoding=serialization.Encoding.Raw,

            format=serialization.PublicFormat.Raw

        ).hex()

    for validator_id, public_key

    in validator_public_keys.items()

}

with open(

    BLOCKCHAIN_DIR / "validator_public_keys.json",

    "w"

) as f:

    json.dump(

        public_key_registry,

        f,

        indent=4

    )

print("\nPOA VALIDATOR REGISTRY")

for validator_id in validator_ids:

    print(
        validator_id,
        "| Authorized: YES"
    )

print("Private validator keys published: NO")

# ------------------------------------------------------------
# 12. Block construction
# ------------------------------------------------------------

GENESIS_HASH = "0" * 64


def create_block(
    height,
    previous_hash,
    block_transactions
):

    proposer_id = validator_ids[

        (height - 1) % N_VALIDATORS

    ]

    root = merkle_root(
        block_transactions
    )

    header = {

        "height":
            int(height),

        "previous_hash":
            previous_hash,

        "merkle_root":
            root,

        "transaction_count":
            len(block_transactions),

        "timestamp_ns":
            time.time_ns(),

        "proposer":
            proposer_id

    }

    block_hash = sha256_bytes(

        canonical_bytes(header)

    )

    return {

        "header":
            header,

        "block_hash":
            block_hash,

        "transactions":
            block_transactions,

        "signatures":
            []

    }

# ------------------------------------------------------------
# 13. Independent authority validation
# ------------------------------------------------------------

def validate_candidate(
    block,
    expected_previous_hash,
    expected_height
):

    header = block["header"]

    if header["height"] != expected_height:

        return False

    if (

        header["previous_hash"]

        != expected_previous_hash

    ):

        return False

    if header["proposer"] not in validator_ids:

        return False

    expected_proposer = validator_ids[

        (expected_height - 1) % N_VALIDATORS

    ]

    if header["proposer"] != expected_proposer:

        return False

    if not (

        0 < len(block["transactions"])

        <= TRANSACTIONS_PER_BLOCK

    ):

        return False

    if (

        header["transaction_count"]

        != len(block["transactions"])

    ):

        return False

    if (

        header["merkle_root"]

        != merkle_root(block["transactions"])

    ):

        return False

    calculated_hash = sha256_bytes(

        canonical_bytes(header)

    )

    return (

        calculated_hash

        == block["block_hash"]

    )

# ------------------------------------------------------------
# 14. Consensus signature verification
#
# Count unique authorized signatures only.
# ------------------------------------------------------------

def verify_consensus(block):

    signed_message = (

        b"POA_BLOCK_V1"

        + bytes.fromhex(

            block["block_hash"]

        )

    )

    valid_signers = set()

    for vote in block["signatures"]:

        validator_id = vote["validator"]

        if validator_id not in validator_public_keys:

            continue

        if validator_id in valid_signers:

            continue

        try:

            validator_public_keys[
                validator_id
            ].verify(

                bytes.fromhex(
                    vote["signature"]
                ),

                signed_message

            )

            valid_signers.add(
                validator_id
            )

        except (
            InvalidSignature,
            ValueError
        ):

            continue

    return (

        len(valid_signers) >= QUORUM

        and

        block["header"]["proposer"] in valid_signers

    )

# ------------------------------------------------------------
# 15. PoA-style consensus simulation
#
# Rotating authorized proposer.
# All validators independently validate the block.
# Each validator maintains its accepted chain tip.
# At least 2 unique authorized signatures required.
#
# Single-process simulated message delivery.
# No artificial network delays.
# ------------------------------------------------------------

def reach_consensus(
    block,
    validator_states
):

    votes = []

    signed_message = (

        b"POA_BLOCK_V1"

        + bytes.fromhex(

            block["block_hash"]

        )

    )

    for validator_id in validator_ids:

        state = validator_states[
            validator_id
        ]

        valid = validate_candidate(

            block,

            state["tip_hash"],

            state["height"] + 1

        )

        if not valid:

            continue

        signature = validator_private_keys[
            validator_id
        ].sign(

            signed_message

        )

        votes.append({

            "validator":
                validator_id,

            "signature":
                signature.hex()

        })

    block["signatures"] = votes

    if not verify_consensus(block):

        return False

    # Replicate committed header state.

    for validator_id in validator_ids:

        validator_states[
            validator_id
        ] = {

            "height":
                block["header"]["height"],

            "tip_hash":
                block["block_hash"]

        }

    return True

# ------------------------------------------------------------
# 16. Full blockchain integrity verification
# ------------------------------------------------------------

def verify_chain(chain):

    expected_previous_hash = GENESIS_HASH

    for expected_height, block in enumerate(

        chain,

        start=1

    ):

        if not validate_candidate(

            block,

            expected_previous_hash,

            expected_height

        ):

            return False

        if not verify_consensus(block):

            return False

        expected_previous_hash = (

            block["block_hash"]

        )

    return True

# ------------------------------------------------------------
# 17. Execute scalability experiments
# ------------------------------------------------------------

summary_records = []

block_records = []

for scale in SCALES:

    print("\n" + "=" * 75)

    print(
        f"BLOCKCHAIN EXPERIMENT: {scale:,} RECORDS"
    )

    print("=" * 75)

    scale_transactions = transactions[:scale]

    assert len(scale_transactions) == scale

    validator_states = {

        validator_id: {

            "height": 0,

            "tip_hash": GENESIS_HASH

        }

        for validator_id in validator_ids

    }

    chain = []

    block_latencies_ms = []

    consensus_latencies_ms = []

    finalized_latencies_ms = []

    block_sizes = []

    previous_hash = GENESIS_HASH

    benchmark_start = time.perf_counter()

    for start in range(

        0,

        scale,

        TRANSACTIONS_PER_BLOCK

    ):

        height = (

            start // TRANSACTIONS_PER_BLOCK

            + 1

        )

        block_transactions = scale_transactions[

            start:

            start + TRANSACTIONS_PER_BLOCK

        ]

        # A. Block generation latency.

        generation_start = time.perf_counter_ns()

        block = create_block(

            height,

            previous_hash,

            block_transactions

        )

        generation_end = time.perf_counter_ns()

        generation_ms = (

            generation_end

            - generation_start

        ) / 1e6

        # B. Consensus latency.

        consensus_start = time.perf_counter_ns()

        accepted = reach_consensus(

            block,

            validator_states

        )

        consensus_end = time.perf_counter_ns()

        consensus_ms = (

            consensus_end

            - consensus_start

        ) / 1e6

        assert accepted, (

            f"Consensus failed at block {height}"

        )

        # C. Commit accepted block.

        chain.append(block)

        previous_hash = block["block_hash"]

        finalized_ms = (

            time.perf_counter_ns()

            - generation_start

        ) / 1e6

        # D. Actual canonical serialized block size.

        serialized_block = canonical_bytes(

            block

        )

        block_size = len(
            serialized_block
        )

        block_latencies_ms.append(
            generation_ms
        )

        consensus_latencies_ms.append(
            consensus_ms
        )

        finalized_latencies_ms.append(
            finalized_ms
        )

        block_sizes.append(
            block_size
        )

        block_records.append({

            "records":
                scale,

            "block":
                height,

            "transactions":
                len(block_transactions),

            "block_generation_ms":
                generation_ms,

            "consensus_ms":
                consensus_ms,

            "finalization_ms":
                finalized_ms,

            "storage_bytes":
                block_size,

            "valid_signatures":
                len(block["signatures"])

        })

        if (

            height % 100 == 0

            or

            start + TRANSACTIONS_PER_BLOCK >= scale

        ):

            print(

                f"Blocks committed: {height:,}"

                f" | Transactions: "

                f"{min(start + TRANSACTIONS_PER_BLOCK, scale):,}",

                flush=True

            )

    benchmark_seconds = (

        time.perf_counter()

        - benchmark_start

    )

    # --------------------------------------------------------
    # 18. Full chain verification
    # --------------------------------------------------------

    verification_start = time.perf_counter()

    integrity_pass = verify_chain(chain)

    verification_seconds = (

        time.perf_counter()

        - verification_start

    )

    assert integrity_pass

    committed_transactions = sum(

        len(block["transactions"])

        for block in chain

    )

    assert committed_transactions == scale

    assert len(chain) == (

        scale // TRANSACTIONS_PER_BLOCK

    )

    assert all(

        state["height"] == len(chain)

        and

        state["tip_hash"] == previous_hash

        for state in validator_states.values()

    )

    # --------------------------------------------------------
    # 19. Tampering rejection verification
    #
    # Modify one evidence commitment without updating
    # the Merkle root or signatures.
    # --------------------------------------------------------

    original_block = chain[0]

    tampered_block = {

        **original_block,

        "transactions": [

            dict(transaction)

            for transaction

            in original_block["transactions"]

        ]

    }

    tampered_block["transactions"][0][

        "evidence_commitment"

    ] = "a" * 64

    tampered_chain = [

        tampered_block,

        *chain[1:]

    ]

    tamper_rejected = not verify_chain(

        tampered_chain

    )

    assert tamper_rejected

    # --------------------------------------------------------
    # 20. Verify invalid consensus rejection
    #
    # Duplicate signatures cannot satisfy quorum.
    # --------------------------------------------------------

    insufficient_quorum_block = {

        **chain[0],

        "signatures": [

            chain[0]["signatures"][0],

            chain[0]["signatures"][0]

        ]

    }

    assert not verify_consensus(

        insufficient_quorum_block

    )

    # --------------------------------------------------------
    # 21. Persist complete blockchain
    #
    # JSONL uses one canonical block per line.
    # Storage figures include newline delimiters.
    # --------------------------------------------------------

    chain_path = (

        BLOCKCHAIN_DIR

        / f"cell11_chain_{scale}.jsonl"

    )

    with open(

        chain_path,

        "wb"

    ) as f:

        for block in chain:

            f.write(

                canonical_bytes(block)

                + b"\n"

            )

    storage_bytes = chain_path.stat().st_size

    assert storage_bytes == (

        sum(block_sizes)

        + len(chain)

    )

    # Reload serialized blockchain and verify.

    with open(

        chain_path,

        "r",

        encoding="utf-8"

    ) as f:

        reloaded_chain = [

            json.loads(line)

            for line in f

        ]

    assert verify_chain(
        reloaded_chain
    )

    # --------------------------------------------------------
    # 22. Reviewer-required performance metrics
    # --------------------------------------------------------

    tps = (

        committed_transactions

        / benchmark_seconds

    )

    average_storage_bytes = (

        storage_bytes

        / len(chain)

    )

    storage_gb_per_million = (

        storage_bytes

        / committed_transactions

        * 1_000_000

        / 1_000_000_000

    )

    # Three-replica storage is an extrapolation,
    # not a separately measured disk deployment.

    replicated_storage_gb_per_million = (

        N_VALIDATORS

        * storage_gb_per_million

    )

    result = {

        "fraud_records":
            scale,

        "blocks":
            len(chain),

        "transactions_per_block":
            TRANSACTIONS_PER_BLOCK,

        "validators":
            N_VALIDATORS,

        "quorum":
            QUORUM,

        "mean_block_generation_ms":

            float(
                np.mean(block_latencies_ms)
            ),

        "mean_consensus_ms":

            float(
                np.mean(consensus_latencies_ms)
            ),

        "mean_finalization_ms":

            float(
                np.mean(finalized_latencies_ms)
            ),

        "p95_finalization_ms":

            float(

                np.percentile(
                    finalized_latencies_ms,
                    95
                )

            ),

        "throughput_tps":
            float(tps),

        "total_processing_seconds":
            float(benchmark_seconds),

        "chain_verification_seconds":
            float(verification_seconds),

        "total_storage_bytes":
            storage_bytes,

        "mean_storage_bytes_per_block":
            average_storage_bytes,

        "storage_gb_per_million_transactions":
            storage_gb_per_million,

        "estimated_3_replica_gb_per_million":
            replicated_storage_gb_per_million,

        "chain_integrity":
            integrity_pass,

        "tampering_rejected":
            tamper_rejected,

        "invalid_quorum_rejected":
            True,

        "validator_tip_consistency":
            True,

        "chain_file":
            str(chain_path)

    }

    summary_records.append(
        result
    )

    print("\nFINAL SCALE RESULTS")

    print(
        "Records:",
        scale
    )

    print(
        "Blocks:",
        len(chain)
    )

    print(

        "Block generation (ms/block):",

        round(

            result["mean_block_generation_ms"],

            6

        )

    )

    print(

        "Consensus latency (ms/block):",

        round(

            result["mean_consensus_ms"],

            6

        )

    )

    print(

        "Finalization latency (ms/block):",

        round(

            result["mean_finalization_ms"],

            6

        )

    )

    print(

        "Throughput (TPS):",

        round(tps, 4)

    )

    print(

        "Storage (bytes/block):",

        round(average_storage_bytes, 4)

    )

    print(

        "Storage (GB/million transactions):",

        round(

            storage_gb_per_million,

            6

        )

    )

    print("Chain integrity: PASS")
    print("Tampering rejection: PASS")
    print("Invalid quorum rejection: PASS")
    print("Validator chain tips: CONSISTENT")
    print("Serialized chain reload: PASS")

    # Save partial results immediately.

    pd.DataFrame(

        summary_records

    ).to_csv(

        TABLE_DIR

        / "cell11_blockchain_partial_results.csv",

        index=False

    )

    del (

        chain,

        tampered_chain,

        tampered_block,

        reloaded_chain

    )

    gc.collect()

# ------------------------------------------------------------
# 23. Consolidate blockchain performance results
# ------------------------------------------------------------

summary_df = pd.DataFrame(

    summary_records

)

block_df = pd.DataFrame(

    block_records

)

assert len(summary_df) == 3

assert len(block_df) == 1110

summary_df.to_csv(

    TABLE_DIR

    / "cell11_blockchain_scalability.csv",

    index=False

)

block_df.to_csv(

    TABLE_DIR

    / "cell11_block_level_metrics.csv",

    index=False

)

print("\n" + "=" * 75)
print("FINAL BLOCKCHAIN SCALABILITY COMPARISON")
print("=" * 75)

display_columns = [

    "fraud_records",

    "blocks",

    "mean_block_generation_ms",

    "mean_consensus_ms",

    "throughput_tps",

    "mean_storage_bytes_per_block",

    "storage_gb_per_million_transactions"

]

print(

    summary_df[display_columns]

    .round(6)

    .to_string(index=False)

)

# ------------------------------------------------------------
# 24. Save manuscript-ready comparison table
# ------------------------------------------------------------

publication_df = summary_df[

    display_columns

].copy()

publication_df = publication_df.rename(

    columns={

        "fraud_records":
            "Fraud records",

        "blocks":
            "Blocks",

        "mean_block_generation_ms":
            "Block generation (ms/block)",

        "mean_consensus_ms":
            "Consensus (ms/block)",

        "throughput_tps":
            "Throughput (TPS)",

        "mean_storage_bytes_per_block":
            "Storage (bytes/block)",

        "storage_gb_per_million_transactions":
            "Storage (GB/million transactions)"

    }

)

publication_df.to_csv(

    TABLE_DIR

    / "cell11_manuscript_blockchain_table.csv",

    index=False

)

print("\nMANUSCRIPT-READY BLOCKCHAIN TABLE")

print(

    publication_df.round(4)

    .to_string(index=False)

)

# ------------------------------------------------------------
# 25. Generate reviewer-required scalability figures
# ------------------------------------------------------------

figure_metrics = [

    (

        "throughput_tps",

        "Transaction throughput (TPS)",

        "throughput"

    ),

    (

        "mean_consensus_ms",

        "Mean consensus latency (ms/block)",

        "consensus_latency"

    ),

    (

        "mean_block_generation_ms",

        "Block generation latency (ms/block)",

        "block_generation"

    ),

    (

        "storage_gb_per_million_transactions",

        "Storage (GB/million transactions)",

        "storage"

    )

]

for metric, ylabel, filename in figure_metrics:

    plt.figure(
        figsize=(7, 4.5)
    )

    plt.plot(

        summary_df["fraud_records"],

        summary_df[metric],

        color="black",

        marker="o",

        linewidth=1.5,

        markersize=6

    )

    plt.xscale("log")

    plt.xticks(

        SCALES,

        ["1K", "10K", "100K"]

    )

    plt.xlabel(
        "Number of fraud-labelled records"
    )

    plt.ylabel(ylabel)

    plt.grid(

        alpha=0.25,

        linestyle=":"

    )

    plt.tight_layout()

    plt.savefig(

        FIGURE_DIR

        / f"cell11_blockchain_{filename}.png",

        dpi=600,

        bbox_inches="tight"

    )

    plt.show()

    plt.close()

print("\nBlockchain scalability figures: SAVED")

# ------------------------------------------------------------
# 26. Save consensus and Merkle technical details
# ------------------------------------------------------------

technical_report = {

    "consensus":

        "PoA-style authenticated authority voting",

    "validators":
        N_VALIDATORS,

    "quorum":
        QUORUM,

    "proposer_selection":
        "Deterministic round-robin",

    "topology":
        "Simulated fully connected",

    "signature_algorithm":
        "Ed25519",

    "hash_algorithm":
        "SHA-256",

    "merkle_leaf":

        "SHA256(0x00 || canonical transaction)",

    "merkle_internal_node":

        "SHA256(0x01 || left hash || right hash)",

    "odd_merkle_level":
        "Duplicate final node",

    "genesis_previous_hash":
        GENESIS_HASH,

    "block_header_fields": [

        "height",

        "previous_hash",

        "merkle_root",

        "transaction_count",

        "timestamp_ns",

        "proposer"

    ],

    "transaction_fields": [

        "transaction_id",

        "evidence_commitment",

        "evidence_type"

    ],

    "evidence_commitment":

        "SHA256(domain || secret nonce || canonical evidence)",

    "raw_medical_records_on_chain":
        False,

    "commitment_secrets_published":
        False,

    "merkle_inclusion_proof_verified":
        True,

    "tampered_transaction_rejected":
        True,

    "duplicate_signature_quorum_rejected":
        True,

    "physical_distributed_network":
        False,

    "artificial_network_delay":
        False,

    "full_byzantine_fault_tolerance_tested":
        False,

    "protocol_justification":

        "Known healthcare and insurance organizations "
        "can operate as authorized validator nodes. "
        "The simulation uses signed authority voting "
        "without computational mining."

}

with open(

    REPORT_DIR

    / "cell11_blockchain_technical_report.json",

    "w"

) as f:

    json.dump(

        technical_report,

        f,

        indent=4

    )

# ------------------------------------------------------------
# 27. Verify previous results unchanged
# ------------------------------------------------------------

for path_string, expected_hash in (

    previous_hashes.items()

):

    assert sha256_file(

        Path(path_string)

    ) == expected_hash

print("\nPrevious Cells 05-10 results: UNCHANGED")

# ------------------------------------------------------------
# 28. Verify original source integrity
# ------------------------------------------------------------

with open(

    PROJECT_ROOT

    / "02_Revised_Project"

    / "reports"

    / "cell02_original_source_hashes.json"

) as f:

    original_source_hashes = json.load(f)

for item in original_source_hashes:

    source_path = (

        SOURCE_ROOT

        / item["file"]

    )

    assert source_path.exists()

    assert (

        sha256_file(source_path)

        == item["sha256"]

    )

print("Original source integrity: PASS")

# ------------------------------------------------------------
# 29. Save complete blockchain experiment report
# ------------------------------------------------------------

report11 = {

    "cell": 11,

    "status": "COMPLETED",

    "reviewer_comments": [

        "Major 2",

        "Minor 29"

    ],

    "random_seed":
        SEED,

    "original_dataset":
        str(FEATURED_PATH),

    "original_fraud_records":
        212796,

    "maximum_unique_fraud_workload":
        100000,

    "synthetic_fraud_records":
        0,

    "external_dataset_used":
        False,

    "scalability_levels":
        SCALES,

    "transactions_per_block":
        TRANSACTIONS_PER_BLOCK,

    "evidence_preparation_seconds":
        evidence_preparation_seconds,

    "technical_configuration":
        technical_report,

    "results":
        summary_records,

    "classifier_retrained":
        False,

    "previous_results_modified":
        False,

    "original_source_modified":
        False,

    "limitations": [

        "The blockchain is a single-process "
        "simulation with three logical validators.",

        "Measured throughput and latency exclude "
        "physical network transmission, real "
        "deployment delays and disk persistence.",

        "Evidence preparation and full-chain "
        "verification are timed separately.",

        "Storage is measured from canonical JSONL "
        "serialization, not a production blockchain "
        "database.",

        "Three-validator replicated storage is "
        "estimated from single-chain storage.",

        "The PoA-style quorum mechanism does not "
        "establish formal Byzantine fault tolerance.",

        "Fraud workloads use ground-truth labelled "
        "records, not exclusively model-detected cases.",

        "The benchmark demonstrates cryptographic "
        "integrity and local scalability, not "
        "production deployment performance.",

        "Commitment opening nonces are not persisted; "
        "long-term external verification of the "
        "underlying evidence requires a separate "
        "authorized evidence retention mechanism.",

        "This experiment does not extend the "
        "differential privacy guarantee to raw "
        "blockchain evidence or the entire framework."

    ],

    "elapsed_seconds":

        time.perf_counter()

        - START_TIME

}

with open(

    REPORT_DIR

    / "cell11_blockchain_experiment_report.json",

    "w"

) as f:

    json.dump(

        report11,

        f,

        indent=4,

        default=str

    )

# ------------------------------------------------------------
# 30. Save complete Cell 11 source code
# ------------------------------------------------------------

source_saved = False

try:

    ip = get_ipython()

    if ip is not None:

        current_source = ip.user_ns["In"][-1]

        if current_source.strip():

            (

                CODE_DIR

                / "cell11_blockchain_scalability.py"

            ).write_text(

                current_source,

                encoding="utf-8"

            )

            source_saved = True

except Exception as exc:

    print(

        "Source archival warning:",

        str(exc)

    )

# ------------------------------------------------------------
# 31. Save completion checkpoint
# ------------------------------------------------------------

checkpoint11 = {

    "cell": 11,

    "status": "COMPLETED",

    "reviewer_comments": [

        "Major 2",

        "Minor 29"

    ],

    "consensus":
        "PoA-style",

    "validators":
        N_VALIDATORS,

    "quorum":
        QUORUM,

    "scalability_levels":
        SCALES,

    "total_blocks_benchmarked":
        len(block_df),

    "block_generation_measured":
        True,

    "consensus_latency_measured":
        True,

    "throughput_measured":
        True,

    "storage_measured":
        True,

    "merkle_verification_completed":
        True,

    "tampering_rejected":
        True,

    "invalid_quorum_rejected":
        True,

    "synthetic_fraud_records":
        0,

    "previous_results_modified":
        False,

    "original_source_modified":
        False,

    "source_saved":
        source_saved,

    "next_cell":
        12

}

with open(

    CHECKPOINT_DIR

    / "cell11_completed.json",

    "w"

) as f:

    json.dump(

        checkpoint11,

        f,

        indent=4

    )

# Remove temporary plaintext evidence copies.

del evidence_payloads
del commitment_nonces

gc.collect()

print("\n" + "=" * 75)
print("CELL 11 COMPLETED SUCCESSFULLY")
print("=" * 75)

print("PoA-style consensus: IMPLEMENTED")
print("Authorized validators: 3")
print("Consensus signatures required: 2")
print("SHA-256 block hashing: VERIFIED")
print("Merkle root verification: PASS")
print("Merkle inclusion proof: PASS")
print("Tampering rejection: PASS")
print("Invalid quorum rejection: PASS")
print("1K fraud workload: COMPLETED")
print("10K fraud workload: COMPLETED")
print("100K fraud workload: COMPLETED")
print("Block generation latency: MEASURED")
print("Consensus latency: MEASURED")
print("Transaction throughput: MEASURED")
print("Storage overhead: MEASURED")
print("Blockchain scalability figures: SAVED")
print("Synthetic fraud records: NONE")
print("Previous results: UNCHANGED")
print("Original source integrity: PASS")
print("Complete code and results saved.")
print("Checkpoint saved.")

print(

    "\nTotal execution seconds:",

    round(

        time.perf_counter()

        - START_TIME,

        2

    )

)

print(

    "\nREADY FOR CELL 12/15: "

    "COMPONENT ABLATION ANALYSIS"

)
```

```python
# ============================================================
# CELL 12/15 - COMPONENT ABLATION ANALYSIS
#
# Reviewer Comment: Major 4
#
# A. Complete framework
# B. Without Blockchain
# C. Without SHAP
# D. Without Differential Privacy
# E. Without Feature Engineering
#
# No external dataset
# No hyperparameter optimization
# Previous results preserved
# ============================================================

import os
import gc
import json
import time
import hashlib
import secrets
import joblib

from pathlib import Path

import numpy as np
import pandas as pd
import xgboost as xgb
import matplotlib.pyplot as plt

from scipy.special import expit

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

START_TIME = time.perf_counter()
SEED = 42

print("=" * 75)
print("CELL 12: COMPONENT ABLATION ANALYSIS")
print("=" * 75)

# ------------------------------------------------------------
# 1. Restore workspace
# ------------------------------------------------------------

PROJECT_ROOT = Path(
    "/content/drive/MyDrive/Narayan_Patil_Reviewer_Revision"
)

SOURCE_ROOT = PROJECT_ROOT / "01_Original_Project" / "SBN"

REVISED_DIR = PROJECT_ROOT / "03_Reviewer_Only_Revision"

CODE_DIR = REVISED_DIR / "code"
DATA_DIR = REVISED_DIR / "data"
MODEL_DIR = REVISED_DIR / "models"
RESULT_DIR = REVISED_DIR / "results"
TABLE_DIR = RESULT_DIR / "tables"
FIGURE_DIR = RESULT_DIR / "figures"
REPORT_DIR = REVISED_DIR / "reports"
CHECKPOINT_DIR = REVISED_DIR / "checkpoints"
PREDICTION_DIR = RESULT_DIR / "predictions"

DP_DIR = MODEL_DIR / "differential_privacy"
ABLATION_DIR = MODEL_DIR / "ablation"

ABLATION_DIR.mkdir(parents=True, exist_ok=True)

assert (
    CHECKPOINT_DIR / "cell11_completed.json"
).exists(), "Complete Cell 11 first."

assert not (
    CHECKPOINT_DIR / "cell12_completed.json"
).exists(), "Cell 12 already completed."

LEDGER12 = REPORT_DIR / "cell12_privacy_ledger.json"

assert not LEDGER12.exists(), (
    "An ablation privacy release has already started. "
    "Check the ledger before rerunning."
)

with open(CHECKPOINT_DIR / "cell11_completed.json") as f:
    checkpoint11 = json.load(f)

assert checkpoint11["status"] == "COMPLETED"

print("\nWorkspace:", REVISED_DIR)
print("Cell 11 checkpoint: VERIFIED")

# Save complete source before private execution.

try:
    (CODE_DIR / "cell12_component_ablation.py").write_text(
        get_ipython().user_ns["In"][-1],
        encoding="utf-8"
    )
except Exception as exc:
    print("Source archival warning:", exc)

# ------------------------------------------------------------
# 2. Preserve existing experimental results
# ------------------------------------------------------------

def sha256_file(path):

    digest = hashlib.sha256()

    with open(path, "rb") as f:
        for chunk in iter(lambda: f.read(1024 * 1024), b""):
            digest.update(chunk)

    return digest.hexdigest()


previous_files = [

    TABLE_DIR / "cell05_all_model_metrics.csv",

    TABLE_DIR / "cell07_federated_summary.csv",

    TABLE_DIR / "cell08_centralized_federated_comparison.csv",

    TABLE_DIR / "cell09_dp_final_results.csv",

    TABLE_DIR / "cell10_privacy_utility_comparison.csv",

    TABLE_DIR / "cell11_blockchain_scalability.csv"

]

previous_hashes = {
    str(path): sha256_file(path)
    for path in previous_files
}

print("Previous experimental result hashes: RECORDED")

# ------------------------------------------------------------
# 3. Verify Cell 09 and Cell 10 privacy records
# ------------------------------------------------------------

with open(CHECKPOINT_DIR / "cell09_completed.json") as f:
    checkpoint09 = json.load(f)

with open(REPORT_DIR / "cell10_privacy_utility_report.json") as f:
    report10 = json.load(f)

REFERENCE_SIGMA = float(checkpoint09["sigma"])

PREVIOUS_RHO = float(report10["combined_rho"])

TARGET_DELTA = float(checkpoint09["delta"])

N_CLIENTS = 3
N_ROUNDS = 10
TREES_PER_ROUND = 40
N_TREES = 400

LEARNING_RATE = 0.05
HESSIAN_FLOOR = 10.0

SENSITIVITY = float(
    2 * np.sqrt(TREES_PER_ROUND * (1 + 0.25**2))
)

def epsilon_from_rho(rho, delta):

    return float(
        rho + 2 * np.sqrt(rho * np.log(1 / delta))
    )


ADDITIONAL_RHO = float(
    N_ROUNDS * SENSITIVITY**2
    / (2 * REFERENCE_SIGMA**2)
)

PLANNED_RHO = PREVIOUS_RHO + ADDITIONAL_RHO

PLANNED_EPSILON = epsilon_from_rho(
    PLANNED_RHO, TARGET_DELTA
)

print("\nPRIVACY ACCOUNTING")

print("Reference sigma:", REFERENCE_SIGMA)

print(
    "Previous cumulative epsilon:",
    epsilon_from_rho(PREVIOUS_RHO, TARGET_DELTA)
)

print(
    "Additional DP ablation rho:",
    ADDITIONAL_RHO
)

print(
    "Planned cumulative DP-only epsilon:",
    PLANNED_EPSILON
)

print(
    "Non-private ablation outputs are not covered "
    "by this privacy accounting."
)

# ------------------------------------------------------------
# 4. Restore exact Cell 09 reference data
#
# No changes to original dataset membership.
# ------------------------------------------------------------

with np.load(
    DATA_DIR / "cell05_train_real.npz",
    allow_pickle=False
) as data:

    train_ids = data["row_id"].copy()
    feature_names = data["feature_names"].tolist()

with np.load(
    DATA_DIR / "cell05_test.npz",
    allow_pickle=False
) as data:

    test_ids = data["row_id"].copy()

with np.load(
    DP_DIR / "cell09_public_reference_indices.npz",
    allow_pickle=False
) as data:

    public_ids = data["public_ids"].copy()

rng = np.random.default_rng(SEED)

permuted_train = rng.permutation(np.sort(train_ids))

assert np.array_equal(
    public_ids,
    permuted_train[:44656]
)

protected_ids = permuted_train[44656:]

assert len(feature_names) == 42
assert len(protected_ids) == 401912
assert len(test_ids) == 111643

assert set(public_ids).isdisjoint(protected_ids)
assert set(protected_ids).isdisjoint(test_ids)

preprocessor = joblib.load(
    DP_DIR / "cell09_public_preprocessing.joblib"
)

assert preprocessor["feature_names"] == feature_names

FEATURED_PATH = (
    SOURCE_ROOT / "data" / "processed"
    / "featured_dataset.csv"
)

df12 = pd.read_csv(
    FEATURED_PATH,
    dtype=str,
    low_memory=False
)

assert df12.shape == (558211, 43)


def encode_saved(frame):

    X = np.empty(
        (len(frame), len(feature_names)),
        dtype=np.float32
    )

    for col, idx in zip(
        preprocessor["numerical_columns"],
        preprocessor["numeric_indices"]
    ):

        X[:, idx] = (
            pd.to_numeric(
                frame[col],
                errors="coerce"
            )
            .replace([np.inf, -np.inf], np.nan)
            .fillna(0)
            .to_numpy(dtype=np.float32)
        )

    category_columns = preprocessor[
        "categorical_columns"
    ]

    if category_columns:

        X[:, preprocessor["category_indices"]] = (
            preprocessor["encoder"].transform(
                frame[category_columns]
                .fillna("Unknown")
                .astype(str)
            )
        )

    X = preprocessor["scaler"].transform(X).astype(
        np.float32
    )

    assert np.isfinite(X).all()

    return X


X_public12 = encode_saved(df12.iloc[public_ids])

X_protected12 = encode_saved(
    df12.iloc[protected_ids]
)

X_test12 = encode_saved(df12.iloc[test_ids])

y_public12 = (
    df12.iloc[public_ids]["PotentialFraud"]
    .astype(np.int8)
    .to_numpy()
)

y_protected12 = (
    df12.iloc[protected_ids]["PotentialFraud"]
    .astype(np.int8)
    .to_numpy()
)

y_test12 = (
    df12.iloc[test_ids]["PotentialFraud"]
    .astype(np.int8)
    .to_numpy()
)

del df12
gc.collect()

print("\nDATASET VERIFICATION")

print("Reference records:", len(y_public12))
print("Protected records:", len(y_protected12))
print("Testing records:", len(y_test12))
print("Original predictors:", len(feature_names))

# ------------------------------------------------------------
# 5. Identify only the original engineered features
#
# Exclude derived age, duration and five original
# engineered features in the requested ablation.
# ------------------------------------------------------------

ENGINEERED_FEATURES = [

    "ClaimDuration",
    "PatientAge",
    "DiagnosisCount",
    "ProcedureCount",
    "TotalClaimAmount",
    "ReimbursementRatio",
    "ChronicConditionCount"

]


def normalize_name(name):

    return "".join(
        char.lower()
        for char in name
        if char.isalnum()
    )


name_lookup = {
    normalize_name(name): name
    for name in feature_names
}

missing_engineered = [

    name for name in ENGINEERED_FEATURES

    if normalize_name(name) not in name_lookup

]

assert not missing_engineered, (
    "Verify original engineered feature names: "
    + str(missing_engineered)
)

removed_features = [

    name_lookup[normalize_name(name)]

    for name in ENGINEERED_FEATURES

]

retained_features = [

    name for name in feature_names

    if name not in removed_features

]

retained_indices = [

    feature_names.index(name)

    for name in retained_features

]

assert len(retained_features) == 35

print("\nFEATURE ENGINEERING ABLATION")

print("Original predictors:", len(feature_names))
print("Removed engineered predictors:")

for name in removed_features:
    print(" -", name)

print("Retained predictors:", len(retained_features))

print(
    "Feature removal applies ONLY to the "
    "reviewer-requested ablation variant."
)

# ------------------------------------------------------------
# 6. Restore complete framework reference model
# ------------------------------------------------------------

reference_model = xgb.Booster()

reference_model.load_model(
    DP_DIR / "cell09_public_xgboost.ubj"
)

reference_model.set_param({
    "device": "cpu",
    "nthread": 2
})

assert reference_model.num_boosted_rounds() == 400

assert reference_model.feature_names == feature_names

REFERENCE_RUN_DIR = (
    DP_DIR / checkpoint09["dp_run_name"]
)

with np.load(
    REFERENCE_RUN_DIR / "private_leaf_updates.npz",
    allow_pickle=False
) as data:

    reference_updates = {
        key: data[key].copy()
        for key in data.files
    }

with np.load(
    REFERENCE_RUN_DIR / "test_predictions.npz",
    allow_pickle=False
) as data:

    reference_probability = data["probability"].copy()

    reference_prediction = data["prediction"].copy()

assert len(reference_probability) == len(y_test12)

assert np.isfinite(reference_probability).all()

print("\nOriginal Cell 09 DP model: RESTORED")
print("Original Cell 09 predictions: RESTORED")

# ------------------------------------------------------------
# 7. Original XGBoost parameters
# ------------------------------------------------------------

with open(
    REPORT_DIR / "cell07_federated_experiment_report.json"
) as f:

    fl_report = json.load(f)

PARAMS12 = fl_report[
    "effective_federated_parameters"
].copy()

PARAMS12.update({
    "seed": SEED,
    "device": "cuda",
    "verbosity": 0
})

assert PARAMS12["max_depth"] == 8
assert PARAMS12["eta"] == 0.05

print("Original XGBoost hyperparameters: RETAINED")

# ------------------------------------------------------------
# 8. Evaluation metrics
# ------------------------------------------------------------

def evaluate_probability(y, probability):

    probability = np.asarray(
        probability
    ).reshape(-1)

    prediction = (
        probability >= 0.5
    ).astype(np.int8)

    return {

        "accuracy": float(
            accuracy_score(y, prediction)
        ),

        "precision": float(
            precision_score(
                y, prediction, zero_division=0
            )
        ),

        "recall": float(
            recall_score(
                y, prediction, zero_division=0
            )
        ),

        "f1": float(
            f1_score(
                y, prediction, zero_division=0
            )
        ),

        "roc_auc": float(
            roc_auc_score(y, probability)
        )

    }


reference_metrics = evaluate_probability(
    y_test12,
    reference_probability
)

stored_reference = pd.read_csv(
    TABLE_DIR / "cell09_dp_final_results.csv"
).iloc[0]

for metric, value in reference_metrics.items():

    assert np.isclose(
        value,
        float(stored_reference[metric]),
        atol=1e-8
    )

print("Original DP performance: VERIFIED")

# ------------------------------------------------------------
# 9. Fixed tree catalogue and model slicing
# ------------------------------------------------------------

def collect_leaf_ids(node):

    if "leaf" in node:
        return [int(node["nodeid"])]

    result = []

    for child in node.get("children", []):

        result.extend(collect_leaf_ids(child))

    return result


def prepare_tree_structure(model):

    catalogue = [

        np.asarray(

            sorted(collect_leaf_ids(json.loads(tree))),

            dtype=np.int32

        )

        for tree in model.get_dump(dump_format="json")

    ]

    assert len(catalogue) == N_TREES

    slices = []

    for r in range(N_ROUNDS):

        start = r * TREES_PER_ROUND

        end = start + TREES_PER_ROUND

        model_slice = model[start:end]

        model_slice.set_param({
            "device": "cpu",
            "nthread": 2
        })

        assert (
            model_slice.num_boosted_rounds()
            == TREES_PER_ROUND
        )

        slices.append(model_slice)

    return catalogue, slices


reference_catalogue, reference_slices = (
    prepare_tree_structure(reference_model)
)

# ------------------------------------------------------------
# 10. Reconstruct refined model for exact TreeSHAP
#
# Original model remains unchanged.
# Update only leaf prediction values in an in-memory
# copy of the public reference tree model.
# ------------------------------------------------------------

def reconstruct_refined_model(
    base_model,
    catalogue,
    saved_updates
):

    model_json = json.loads(
        bytes(
            base_model.save_raw(raw_format="json")
        ).decode("utf-8")
    )

    trees = model_json[
        "learner"
    ][
        "gradient_booster"
    ][
        "model"
    ][
        "trees"
    ]

    assert len(trees) == N_TREES

    for round_no in range(1, N_ROUNDS + 1):

        updates = saved_updates[
            f"round_{round_no:02d}_updates"
        ]

        offsets = saved_updates[
            f"round_{round_no:02d}_offsets"
        ]

        start_tree = (
            round_no - 1
        ) * TREES_PER_ROUND

        for j in range(TREES_PER_ROUND):

            tree_index = start_tree + j

            tree = trees[tree_index]

            leaf_ids = catalogue[tree_index]

            correction = updates[
                offsets[j]:offsets[j + 1]
            ]

            assert len(correction) == len(leaf_ids)

            for leaf_id, delta in zip(
                leaf_ids,
                correction
            ):

                leaf_id = int(leaf_id)

                assert (
                    tree["left_children"][leaf_id] == -1
                )

                tree["split_conditions"][leaf_id] = float(
                    tree["split_conditions"][leaf_id]
                    + delta
                )

    refined = xgb.Booster()

    refined.load_model(
        bytearray(
            json.dumps(
                model_json,
                separators=(",", ":")
            ).encode("utf-8")
        )
    )

    refined.set_param({
        "device": "cpu",
        "nthread": 2
    })

    return refined


full_model12 = reconstruct_refined_model(

    reference_model,

    reference_catalogue,

    reference_updates

)

full_test_matrix = xgb.DMatrix(
    X_test12,
    feature_names=feature_names
)

reconstructed_probability = full_model12.predict(
    full_test_matrix
)

assert np.allclose(

    reconstructed_probability,

    reference_probability,

    atol=2e-5,

    rtol=2e-5

), "Refined model reconstruction mismatch."

print("\nComplete framework model reconstruction: PASS")

# ------------------------------------------------------------
# 11. Verify blockchain and SHAP functions
# ------------------------------------------------------------

required_blockchain_functions = [

    "create_block",
    "reach_consensus",
    "verify_chain",
    "canonical_bytes",
    "sha256_bytes",
    "validator_ids",
    "GENESIS_HASH"

]

missing = [

    name for name in required_blockchain_functions

    if name not in globals()

]

assert not missing, (
    f"Missing Cell 11 runtime objects: {missing}"
)

# A fixed 100-record test audit subset is used for
# actual SHAP and blockchain component measurements.

AUDIT_SIZE = 100

audit_indices = np.random.default_rng(SEED).choice(

    len(y_test12),

    size=AUDIT_SIZE,

    replace=False

)

audit_indices = np.sort(audit_indices)

print("Component audit records:", AUDIT_SIZE)

# ------------------------------------------------------------
# 12. Actual TreeSHAP execution
# ------------------------------------------------------------

def execute_shap(
    model,
    X,
    names,
    variant
):

    audit_matrix = xgb.DMatrix(

        X[audit_indices],

        feature_names=names

    )

    started = time.perf_counter()

    shap_values = model.predict(

        audit_matrix,

        pred_contribs=True

    )

    elapsed = time.perf_counter() - started

    raw_margin = model.predict(

        audit_matrix,

        output_margin=True

    )

    assert shap_values.shape == (

        AUDIT_SIZE,

        len(names) + 1

    )

    assert np.allclose(

        shap_values.sum(axis=1),

        raw_margin,

        atol=1e-3,

        rtol=1e-4

    )

    np.savez_compressed(

        ABLATION_DIR / f"{variant}_shap_values.npz",

        row_id=test_ids[audit_indices],

        shap_values=shap_values[:, :-1],

        base_value=shap_values[:, -1],

        feature_names=np.asarray(names, dtype=str)

    )

    return {

        "shap_enabled": True,

        "explained_records": AUDIT_SIZE,

        "shap_seconds": float(elapsed),

        "shap_additivity_verified": True

    }

# ------------------------------------------------------------
# 13. Actual blockchain component execution
#
# Use commitments derived from held-out predictions.
# Do not commit protected raw training records.
# ------------------------------------------------------------

def execute_blockchain(
    probabilities,
    variant
):

    audit_transactions = []

    started = time.perf_counter()

    for index, test_position in enumerate(
        audit_indices,
        start=1
    ):

        evidence = canonical_bytes({

            "prediction_probability":
                float(probabilities[test_position]),

            "predicted_label":
                int(probabilities[test_position] >= 0.5)

        })

        commitment = sha256_bytes(

            b"EVIDENCE_V1"

            + secrets.token_bytes(32)

            + evidence

        )

        audit_transactions.append({

            "transaction_id": index,

            "evidence_commitment": commitment,

            "evidence_type": "prediction_evidence"

        })

    block = create_block(

        1,

        GENESIS_HASH,

        audit_transactions

    )

    states = {

        validator: {

            "height": 0,

            "tip_hash": GENESIS_HASH

        }

        for validator in validator_ids

    }

    assert reach_consensus(block, states)

    assert verify_chain([block])

    elapsed = time.perf_counter() - started

    block_path = (
        ABLATION_DIR / f"{variant}_audit_block.json"
    )

    block_path.write_bytes(canonical_bytes(block))

    return {

        "blockchain_enabled": True,

        "committed_records": AUDIT_SIZE,

        "blockchain_seconds": float(elapsed),

        "blockchain_storage_bytes":
            block_path.stat().st_size,

        "blockchain_integrity_verified": True

    }

# ------------------------------------------------------------
# 14. Common ablation refinement mechanism
#
# Identical fixed-tree mechanism from Cell 09.
# sigma = 0 for the requested non-private ablation.
# sigma = reference sigma for feature ablation.
# ------------------------------------------------------------

secure_random12 = secrets.SystemRandom()


def noise_vector(size, sigma):

    if sigma == 0:
        return np.zeros(size, dtype=np.float64)

    return np.fromiter(

        (
            secure_random12.normalvariate(0, sigma)

            for _ in range(size)
        ),

        dtype=np.float64,

        count=size

    )


def save_ledger12(data):

    temporary = LEDGER12.with_suffix(".tmp")

    with open(temporary, "w") as f:
        json.dump(data, f, indent=4)

    os.replace(temporary, LEDGER12)


def refine_ablation(
    base_model,
    X_private,
    X_testing,
    names,
    sigma,
    variant,
    ledger=None
):

    started = time.perf_counter()

    catalogue, slices = prepare_tree_structure(
        base_model
    )

    client_indices = np.array_split(

        np.arange(len(y_protected12)),

        N_CLIENTS

    )

    matrices = [

        xgb.DMatrix(
            X_private[idx],
            feature_names=names
        )

        for idx in client_indices

    ]

    labels = [
        y_protected12[idx]
        for idx in client_indices
    ]

    margins = [

        base_model.predict(
            matrix,
            output_margin=True
        ).astype(np.float64)

        for matrix in matrices

    ]

    test_matrix = xgb.DMatrix(
        X_testing,
        feature_names=names
    )

    test_margin = base_model.predict(

        test_matrix,

        output_margin=True

    ).astype(np.float64)

    saved_updates = {}

    round_records = []

    for r in range(N_ROUNDS):

        start_tree = r * TREES_PER_ROUND

        current_catalogue = catalogue[
            start_tree:start_tree + TREES_PER_ROUND
        ]

        offsets = np.concatenate([

            [0],

            np.cumsum([
                len(leaves)
                for leaves in current_catalogue
            ])

        ]).astype(np.int64)

        total_leaves = int(offsets[-1])

        G = np.zeros(total_leaves, dtype=np.float64)

        H = np.zeros(total_leaves, dtype=np.float64)

        positions_by_client = []

        for client_no, matrix in enumerate(matrices):

            probability = expit(margins[client_no])

            gradient = np.clip(
                probability - labels[client_no],
                -1,
                1
            )

            hessian = np.clip(
                probability * (1 - probability),
                0,
                0.25
            )

            leaves = slices[r].predict(

                matrix,

                pred_leaf=True

            ).astype(np.int32).reshape(

                len(gradient),

                TREES_PER_ROUND

            )

            positions = np.empty(
                leaves.shape,
                dtype=np.int32
            )

            for j, tree_leaves in enumerate(
                current_catalogue
            ):

                pos = np.searchsorted(
                    tree_leaves,
                    leaves[:, j]
                )

                assert np.all(pos < len(tree_leaves))

                assert np.array_equal(
                    tree_leaves[pos],
                    leaves[:, j]
                )

                positions[:, j] = pos + offsets[j]

                left = offsets[j]
                right = offsets[j + 1]

                G[left:right] += np.bincount(

                    pos,

                    weights=gradient,

                    minlength=len(tree_leaves)

                )

                H[left:right] += np.bincount(

                    pos,

                    weights=hessian,

                    minlength=len(tree_leaves)

                )

            positions_by_client.append(positions)

        # Same Gaussian mechanism as Cell 09.
        # Only sigma is zero in the no-DP variant.

        joint = np.concatenate([G, H])

        joint += noise_vector(len(joint), sigma)

        noisy_g = joint[:total_leaves]

        noisy_h = joint[total_leaves:]

        leaf_update = (

            -LEARNING_RATE

            * np.clip(

                noisy_g / np.maximum(
                    noisy_h + 1.0,
                    HESSIAN_FLOOR
                ),

                -1.0,

                1.0

            )

        )

        assert np.isfinite(leaf_update).all()

        for client_no in range(N_CLIENTS):

            margins[client_no] += (

                leaf_update[
                    positions_by_client[client_no]
                ].sum(axis=1)

            )

        test_leaves = slices[r].predict(

            test_matrix,

            pred_leaf=True

        ).astype(np.int32).reshape(

            len(y_test12),

            TREES_PER_ROUND

        )

        for j, tree_leaves in enumerate(
            current_catalogue
        ):

            pos = np.searchsorted(
                tree_leaves,
                test_leaves[:, j]
            )

            assert np.all(pos < len(tree_leaves))

            assert np.array_equal(
                tree_leaves[pos],
                test_leaves[:, j]
            )

            test_margin += leaf_update[
                pos + offsets[j]
            ]

        saved_updates[
            f"round_{r + 1:02d}_updates"
        ] = leaf_update.copy()

        saved_updates[
            f"round_{r + 1:02d}_offsets"
        ] = offsets.copy()

        if ledger is not None:

            ledger["completed_rounds"] = r + 1

            ledger["consumed_additional_rho"] = (

                (r + 1)

                * SENSITIVITY**2

                / (2 * sigma**2)

            )

            ledger["status"] = "IN_PROGRESS"

            save_ledger12(ledger)

        metrics = evaluate_probability(

            y_test12,

            expit(test_margin)

        )

        round_records.append({

            "variant": variant,

            "round": r + 1,

            "sigma": sigma,

            **metrics

        })

        print(

            f"{variant} | Round {r + 1:02d}/10"

            f" | Accuracy: {metrics['accuracy']:.6f}"

            f" | Recall: {metrics['recall']:.6f}"

            f" | AUC: {metrics['roc_auc']:.6f}",

            flush=True

        )

        del G, H, joint, positions_by_client
        gc.collect()

    probability = expit(test_margin)

    refined_model = reconstruct_refined_model(

        base_model,

        catalogue,

        saved_updates

    )

    assert np.allclose(

        refined_model.predict(test_matrix),

        probability,

        atol=2e-5,

        rtol=2e-5

    )

    refined_model.save_model(

        ABLATION_DIR / f"{variant}_model.ubj"

    )

    np.savez_compressed(

        ABLATION_DIR / f"{variant}_leaf_updates.npz",

        **saved_updates

    )

    pd.DataFrame(round_records).to_csv(

        TABLE_DIR / f"cell12_{variant}_round_metrics.csv",

        index=False

    )

    return (

        refined_model,

        probability,

        float(time.perf_counter() - started)

    )

# ------------------------------------------------------------
# 15. Complete framework and post-prediction ablations
#
# Preserve the actual verified Cell 09 predictions.
# SHAP and Blockchain are independently executed.
# ------------------------------------------------------------

ablation_results = []

full_shap = execute_shap(

    full_model12,

    X_test12,

    feature_names,

    "complete_framework"

)

full_blockchain = execute_blockchain(

    reference_probability,

    "complete_framework"

)


def make_result(
    configuration,
    probability,
    feature_count,
    dp_enabled,
    shap_info,
    blockchain_info,
    training_seconds=None
):

    return {

        "configuration": configuration,

        "features": feature_count,

        "federated_learning": True,

        "dp_enabled": dp_enabled,

        **evaluate_probability(
            y_test12,
            probability
        ),

        **shap_info,

        **blockchain_info,

        "training_seconds": training_seconds

    }


no_shap_info = {

    "shap_enabled": False,

    "explained_records": 0,

    "shap_seconds": 0.0,

    "shap_additivity_verified": None

}

no_blockchain_info = {

    "blockchain_enabled": False,

    "committed_records": 0,

    "blockchain_seconds": 0.0,

    "blockchain_storage_bytes": 0,

    "blockchain_integrity_verified": None

}

# A. Complete framework

ablation_results.append(

    make_result(

        "Complete framework",

        reference_probability,

        42,

        True,

        full_shap,

        full_blockchain,

        float(stored_reference["execution_seconds"])

    )

)

# B. Without Blockchain
# Reuse identical classifier and SHAP outputs.
# Blockchain is disabled.

ablation_results.append(

    make_result(

        "Without Blockchain",

        reference_probability.copy(),

        42,

        True,

        full_shap,

        no_blockchain_info

    )

)

# C. Without SHAP
# Execute blockchain but omit explanation calculation.

no_shap_blockchain = execute_blockchain(

    reference_probability,

    "without_shap"

)

ablation_results.append(

    make_result(

        "Without SHAP",

        reference_probability.copy(),

        42,

        True,

        no_shap_info,

        no_shap_blockchain

    )

)

for result in ablation_results[1:3]:

    for metric in [

        "accuracy",

        "precision",

        "recall",

        "f1",

        "roc_auc"

    ]:

        assert result[metric] == reference_metrics[metric]

print("\nCOMPLETE FRAMEWORK: VERIFIED")
print("Without Blockchain: COMPLETED")
print("Without SHAP: COMPLETED")

print(
    "Predictive metric invariance for post-prediction "
    "components: VERIFIED"
)

# ------------------------------------------------------------
# 16. Without Differential Privacy
#
# Identical public trees, features, clients,
# update formula and learning rate.
#
# Only Gaussian noise is disabled.
# This is a non-private comparator.
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("ABLATION: WITHOUT DIFFERENTIAL PRIVACY")
print("=" * 75)

no_dp_model, no_dp_probability, no_dp_seconds = (

    refine_ablation(

        reference_model,

        X_protected12,

        X_test12,

        feature_names,

        sigma=0.0,

        variant="without_dp"

    )

)

no_dp_shap = execute_shap(

    no_dp_model,

    X_test12,

    feature_names,

    "without_dp"

)

no_dp_blockchain = execute_blockchain(

    no_dp_probability,

    "without_dp"

)

ablation_results.append(

    make_result(

        "Without Differential Privacy",

        no_dp_probability,

        42,

        False,

        no_dp_shap,

        no_dp_blockchain,

        no_dp_seconds

    )

)

print("\nWithout DP: COMPLETED")

# ------------------------------------------------------------
# 17. Without Feature Engineering
#
# Remove only the seven identified derived features.
# Use identical reference/protected/test membership.
#
# Retrain fixed reference trees with original
# hyperparameters and reduced feature space.
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("ABLATION: WITHOUT FEATURE ENGINEERING")
print("=" * 75)

X_public_reduced = X_public12[:, retained_indices]

X_protected_reduced = X_protected12[:, retained_indices]

X_test_reduced = X_test12[:, retained_indices]

assert X_public_reduced.shape[1] == 35

assert X_protected_reduced.shape[1] == 35

assert X_test_reduced.shape[1] == 35

public_reduced_matrix = xgb.DMatrix(

    X_public_reduced,

    label=y_public12,

    feature_names=retained_features

)

public_reduced_model = xgb.train(

    params=PARAMS12.copy(),

    dtrain=public_reduced_matrix,

    num_boost_round=N_TREES,

    verbose_eval=False

)

public_reduced_model.set_param({

    "device": "cpu",

    "nthread": 2

})

public_reduced_model.save_model(

    ABLATION_DIR / "without_fe_public_model.ubj"

)

# Reserve DP budget before the new private release.

ledger12 = {

    "status": "BUDGET_RESERVED",

    "previous_rho": PREVIOUS_RHO,

    "reserved_additional_rho": ADDITIONAL_RHO,

    "planned_combined_rho": PLANNED_RHO,

    "planned_combined_epsilon": PLANNED_EPSILON,

    "sigma": REFERENCE_SIGMA,

    "delta": TARGET_DELTA,

    "completed_rounds": 0,

    "consumed_additional_rho": 0.0

}

with open(LEDGER12, "x") as f:

    json.dump(ledger12, f, indent=4)

no_fe_model, no_fe_probability, no_fe_seconds = (

    refine_ablation(

        public_reduced_model,

        X_protected_reduced,

        X_test_reduced,

        retained_features,

        sigma=REFERENCE_SIGMA,

        variant="without_feature_engineering",

        ledger=ledger12

    )

)

assert ledger12["completed_rounds"] == 10

assert np.isclose(

    ledger12["consumed_additional_rho"],

    ADDITIONAL_RHO,

    atol=1e-12

)

no_fe_shap = execute_shap(

    no_fe_model,

    X_test_reduced,

    retained_features,

    "without_feature_engineering"

)

no_fe_blockchain = execute_blockchain(

    no_fe_probability,

    "without_feature_engineering"

)

ablation_results.append(

    make_result(

        "Without Feature Engineering",

        no_fe_probability,

        35,

        True,

        no_fe_shap,

        no_fe_blockchain,

        no_fe_seconds

    )

)

ledger12["status"] = "COMPLETED"

ledger12["final_combined_rho"] = PLANNED_RHO

ledger12["final_combined_epsilon"] = PLANNED_EPSILON

save_ledger12(ledger12)

print("\nWithout feature engineering: COMPLETED")

# ------------------------------------------------------------
# 18. Save ablation predictions
# ------------------------------------------------------------

prediction_sets = {

    "complete_framework":
        reference_probability,

    "without_blockchain":
        reference_probability.copy(),

    "without_shap":
        reference_probability.copy(),

    "without_dp":
        no_dp_probability,

    "without_feature_engineering":
        no_fe_probability

}

for variant, probability in prediction_sets.items():

    np.savez_compressed(

        PREDICTION_DIR / f"cell12_{variant}_predictions.npz",

        row_id=test_ids,

        probability=probability,

        prediction=(probability >= 0.5).astype(np.int8)

    )

# ------------------------------------------------------------
# 19. Final reviewer-required ablation table
# ------------------------------------------------------------

ablation_df = pd.DataFrame(ablation_results)

assert len(ablation_df) == 5

ablation_df.to_csv(

    TABLE_DIR / "cell12_component_ablation.csv",

    index=False

)

display_columns = [

    "configuration",

    "features",

    "accuracy",

    "precision",

    "recall",

    "f1",

    "roc_auc"

]

print("\n" + "=" * 75)
print("FINAL COMPONENT ABLATION RESULTS")
print("=" * 75)

print(

    ablation_df[display_columns]

    .round(6)

    .to_string(index=False)

)

# ------------------------------------------------------------
# 20. Quantify contribution relative to full framework
# ------------------------------------------------------------

difference_records = []

for _, row in ablation_df.iloc[1:].iterrows():

    record = {

        "configuration": row["configuration"]

    }

    for metric in [

        "accuracy",

        "precision",

        "recall",

        "f1",

        "roc_auc"

    ]:

        record[f"{metric}_difference_pp"] = (

            float(row[metric])

            - reference_metrics[metric]

        ) * 100

    difference_records.append(record)

difference_df = pd.DataFrame(difference_records)

difference_df.to_csv(

    TABLE_DIR / "cell12_ablation_differences.csv",

    index=False

)

print("\nDIFFERENCE RELATIVE TO COMPLETE FRAMEWORK")

print(

    difference_df

    .round(6)

    .to_string(index=False)

)

# ------------------------------------------------------------
# 21. Report actual SHAP and Blockchain contributions
#
# Post-prediction components provide explanation
# and integrity rather than classification gains.
# ------------------------------------------------------------

component_columns = [

    "configuration",

    "shap_enabled",

    "explained_records",

    "shap_seconds",

    "blockchain_enabled",

    "committed_records",

    "blockchain_seconds",

    "blockchain_storage_bytes",

    "blockchain_integrity_verified"

]

component_df = ablation_df[component_columns].copy()

component_df.to_csv(

    TABLE_DIR / "cell12_component_operational_metrics.csv",

    index=False

)

print("\nPOST-PREDICTION COMPONENT CONTRIBUTIONS")

print(

    component_df

    .round(6)

    .to_string(index=False)

)

# ------------------------------------------------------------
# 22. Manuscript-ready ablation table
# ------------------------------------------------------------

publication_df = ablation_df[display_columns].copy()

for metric in [

    "accuracy",

    "precision",

    "recall",

    "f1"

]:

    publication_df[metric] = (

        publication_df[metric] * 100

    ).round(2)

publication_df = publication_df.rename(columns={

    "configuration": "Configuration",

    "features": "Predictors",

    "accuracy": "Accuracy (%)",

    "precision": "Precision (%)",

    "recall": "Recall (%)",

    "f1": "F1-score (%)",

    "roc_auc": "ROC-AUC"

})

publication_df.to_csv(

    TABLE_DIR / "cell12_manuscript_ablation_table.csv",

    index=False

)

print("\nMANUSCRIPT-READY ABLATION TABLE")

print(publication_df.to_string(index=False))

# ------------------------------------------------------------
# 23. Ablation comparison figure
# ------------------------------------------------------------

labels = [

    "Complete",

    "No Blockchain",

    "No SHAP",

    "No DP",

    "No FE"

]

metrics_to_plot = [

    ("accuracy", "Accuracy"),

    ("recall", "Recall"),

    ("f1", "F1-score"),

    ("roc_auc", "ROC-AUC")

]

positions = np.arange(len(labels))

width = 0.18

hatches = ["", "//", "xx", ".."]

plt.figure(figsize=(10, 5))

for i, (metric, label) in enumerate(metrics_to_plot):

    plt.bar(

        positions + (i - 1.5) * width,

        ablation_df[metric] * 100,

        width=width,

        color="white",

        edgecolor="black",

        hatch=hatches[i],

        label=label

    )

plt.xticks(

    positions,

    labels,

    rotation=15

)

plt.ylabel("Performance (%)")

plt.ylim(0, 100)

plt.legend(ncol=2)

plt.tight_layout()

plt.savefig(

    FIGURE_DIR / "cell12_component_ablation.png",

    dpi=600,

    bbox_inches="tight"

)

plt.show()
plt.close()

print("\nAblation comparison figure: SAVED")

# ------------------------------------------------------------
# 24. Verify previous results and original source
# ------------------------------------------------------------

for path_string, expected_hash in previous_hashes.items():

    assert sha256_file(Path(path_string)) == expected_hash

print("\nPrevious Cells 05-11 results: UNCHANGED")

with open(

    PROJECT_ROOT / "02_Revised_Project"

    / "reports"

    / "cell02_original_source_hashes.json"

) as f:

    original_source_hashes = json.load(f)

for item in original_source_hashes:

    path = SOURCE_ROOT / item["file"]

    assert path.exists()

    assert sha256_file(path) == item["sha256"]

print("Original source integrity: PASS")

# ------------------------------------------------------------
# 25. Save cumulative DP-only accounting
# ------------------------------------------------------------

privacy_accounting12 = {

    "previous_rho": PREVIOUS_RHO,

    "previous_epsilon":

        epsilon_from_rho(
            PREVIOUS_RHO,
            TARGET_DELTA
        ),

    "additional_feature_ablation_rho":

        ADDITIONAL_RHO,

    "feature_ablation_standalone_epsilon":

        epsilon_from_rho(
            ADDITIONAL_RHO,
            TARGET_DELTA
        ),

    "cumulative_dp_only_rho":

        PLANNED_RHO,

    "cumulative_dp_only_epsilon":

        PLANNED_EPSILON,

    "delta": TARGET_DELTA,

    "non_private_ablation_included_in_bound": False,

    "whole_framework_dp_claim": False

}

with open(

    REPORT_DIR / "cell12_privacy_accounting.json",

    "w"

) as f:

    json.dump(privacy_accounting12, f, indent=4)

print("\nCUMULATIVE DP-ONLY ACCOUNTING")

print(

    "Previous epsilon:",

    privacy_accounting12["previous_epsilon"]

)

print(

    "Updated epsilon:",

    privacy_accounting12["cumulative_dp_only_epsilon"]

)

print("Delta:", TARGET_DELTA)

print(

    "The non-private comparator is outside "

    "this privacy guarantee."

)

# ------------------------------------------------------------
# 26. Save reviewer-specific report
# ------------------------------------------------------------

report12 = {

    "cell": 12,

    "status": "COMPLETED",

    "reviewer_comment": "Major 4",

    "configurations": 5,

    "original_predictors": 42,

    "without_feature_engineering_predictors": 35,

    "removed_engineered_features": removed_features,

    "same_reference_records": True,

    "same_protected_records": True,

    "same_testing_records": True,

    "same_core_hyperparameters": True,

    "complete_framework_retrained": False,

    "post_prediction_metric_invariance_verified": True,

    "shap_method": "XGBoost exact TreeSHAP",

    "component_audit_records": AUDIT_SIZE,

    "results": ablation_results,

    "privacy_accounting": privacy_accounting12,

    "external_dataset_used": False,

    "hyperparameter_tuning": False,

    "previous_results_modified": False,

    "original_source_modified": False,

    "limitations": [

        "Blockchain and SHAP do not modify classifier "
        "predictions and therefore do not improve "
        "predictive metrics directly.",

        "SHAP and Blockchain operational measurements "
        "use a fixed 100-record held-out audit subset.",

        "Predictive metrics use the complete "
        "111643-record test set.",

        "The no-DP comparator uses the same fixed-tree "
        "refinement procedure without Gaussian noise.",

        "The no-feature-engineering ablation requires "
        "retraining the public reference trees.",

        "Feature ablation includes a new independent "
        "Gaussian noise realization; its observed "
        "difference is not a pure deterministic "
        "feature effect.",

        "The non-private ablation and earlier "
        "non-private models are not covered by "
        "the DP-only privacy accounting.",

        "Privacy accounting remains conditional on "
        "the assumptions established in Cell 09.",

        "Statistical significance remains pending."

    ],

    "elapsed_seconds":

        time.perf_counter() - START_TIME

}

with open(

    REPORT_DIR / "cell12_component_ablation_report.json",

    "w"

) as f:

    json.dump(

        report12,

        f,

        indent=4,

        default=str

    )

# ------------------------------------------------------------
# 27. Save completion checkpoint
# ------------------------------------------------------------

checkpoint12 = {

    "cell": 12,

    "status": "COMPLETED",

    "reviewer_comment": "Major 4",

    "ablation_configurations_completed": 5,

    "without_blockchain_completed": True,

    "without_shap_completed": True,

    "without_dp_completed": True,

    "without_feature_engineering_completed": True,

    "complete_framework_predictors": 42,

    "feature_ablation_predictors": 35,

    "predictive_metric_invariance_verified": True,

    "component_specific_utility_measured": True,

    "cumulative_dp_only_epsilon": PLANNED_EPSILON,

    "delta": TARGET_DELTA,

    "previous_results_modified": False,

    "original_source_modified": False,

    "external_dataset_used": False,

    "next_cell": 13

}

with open(

    CHECKPOINT_DIR / "cell12_completed.json",

    "w"

) as f:

    json.dump(checkpoint12, f, indent=4)

gc.collect()

print("\n" + "=" * 75)
print("CELL 12 COMPLETED SUCCESSFULLY")
print("=" * 75)

print("Complete framework: VERIFIED")
print("Without Blockchain: COMPLETED")
print("Without SHAP: COMPLETED")
print("Without Differential Privacy: COMPLETED")
print("Without Feature Engineering: COMPLETED")
print("All five ablation configurations: COMPLETED")
print("Predictive metrics: MEASURED")
print("SHAP component utility: MEASURED")
print("Blockchain component utility: MEASURED")
print("Cumulative DP-only accounting: COMPLETED")
print("Original framework predictors: 42")
print("Feature ablation predictors: 35")
print("Previous results: UNCHANGED")
print("Original source integrity: PASS")
print("Hyperparameter tuning: NONE")
print("Additional dataset: NONE")
print("Ablation models, tables and figure: SAVED")
print("Complete source code: SAVED")
print("Checkpoint saved.")

print(

    "\nTotal execution seconds:",

    round(time.perf_counter() - START_TIME, 2)

)

print(

    "\nREADY FOR CELL 13/15: "

    "FIVE INDEPENDENT RUNS AND STATISTICAL TESTING"

)
```

```python
# ============================================================
# CELL 13/15 - REPEATED RUNS AND STATISTICAL TESTING
#
# Reviewer Comment: Major 6
#
# Five runs: 42, 43, 44, 45, 46
# Preserve existing seed-42 results
#
# Mean, sample SD, 95% CI
# Paired t-tests with Holm correction
#
# NO HYPERPARAMETER TUNING
# NO ADDITIONAL CLASSIFIER
# NO EXTERNAL DATASET
# ============================================================

import os
import gc
import json
import time
import copy
import hashlib
import secrets
import joblib

from pathlib import Path

import numpy as np
import pandas as pd
import xgboost as xgb
import lightgbm as lgb

from catboost import CatBoostClassifier

from scipy.special import expit
from scipy.stats import t, ttest_rel

from sklearn.model_selection import StratifiedKFold

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

from google.colab import drive

drive.mount("/content/drive")

START_TIME = time.perf_counter()

SEEDS = [42, 43, 44, 45, 46]

ADDITIONAL_SEEDS = SEEDS[1:]

METRICS = [
    "accuracy",
    "precision",
    "recall",
    "f1",
    "roc_auc"
]

PROJECT_ROOT = Path(
    "/content/drive/MyDrive/Narayan_Patil_Reviewer_Revision"
)

SOURCE_ROOT = PROJECT_ROOT / "01_Original_Project" / "SBN"

REVISED_DIR = PROJECT_ROOT / "03_Reviewer_Only_Revision"

CODE_DIR = REVISED_DIR / "code"
DATA_DIR = REVISED_DIR / "data"
MODEL_DIR = REVISED_DIR / "models"
RESULT_DIR = REVISED_DIR / "results"
TABLE_DIR = RESULT_DIR / "tables"
FIGURE_DIR = RESULT_DIR / "figures"
REPORT_DIR = REVISED_DIR / "reports"
CHECKPOINT_DIR = REVISED_DIR / "checkpoints"
PREDICTION_DIR = RESULT_DIR / "predictions"

DP_DIR = MODEL_DIR / "differential_privacy"

REPEAT_DIR = MODEL_DIR / "repeated_runs"

REPEAT_DIR.mkdir(parents=True, exist_ok=True)

assert (
    CHECKPOINT_DIR / "cell12_completed.json"
).exists(), "Complete Cell 12 first."

assert not (
    CHECKPOINT_DIR / "cell13_completed.json"
).exists(), "Cell 13 already completed."

LEDGER13 = REPORT_DIR / "cell13_privacy_ledger.json"

assert not LEDGER13.exists(), (
    "Previous repeated-run privacy release detected. "
    "Inspect ledger before repeating this cell."
)

print("=" * 75)
print("CELL 13: REPEATED RUNS AND STATISTICAL TESTING")
print("=" * 75)

print("\nWorkspace:", REVISED_DIR)

# ------------------------------------------------------------
# 1. Verify previous checkpoint and runtime functions
# ------------------------------------------------------------

with open(
    CHECKPOINT_DIR / "cell12_completed.json"
) as f:

    checkpoint12 = json.load(f)

assert checkpoint12["status"] == "COMPLETED"

required_objects = [

    "X_public12",
    "X_protected12",
    "X_test12",
    "y_public12",
    "y_protected12",
    "y_test12",
    "reference_model",
    "public_reduced_model",
    "retained_indices",
    "retained_features",
    "feature_names",
    "prepare_tree_structure",
    "create_block",
    "reach_consensus",
    "verify_chain",
    "canonical_bytes",
    "validator_ids",
    "GENESIS_HASH",
    "transactions"

]

missing = [

    name for name in required_objects
    if name not in globals()

]

assert not missing, (
    f"Missing runtime objects: {missing}. "
    "Do not rerun completed privacy cells."
)

assert len(transactions) >= 100000

assert len(feature_names) == 42

assert len(retained_features) == 35

print("\nCell 12 checkpoint: VERIFIED")
print("Required runtime objects: AVAILABLE")
print("Original predictors: 42")
print("Repeated-run seeds:", SEEDS)

# ------------------------------------------------------------
# 2. Preserve previous results
# ------------------------------------------------------------

def sha256_file(path):

    digest = hashlib.sha256()

    with open(path, "rb") as f:

        for chunk in iter(
            lambda: f.read(1024 * 1024), b""
        ):

            digest.update(chunk)

    return digest.hexdigest()


previous_files = sorted(
    path for path in TABLE_DIR.glob("cell*.csv")
    if not path.name.startswith("cell13")
)

previous_hashes = {

    str(path): sha256_file(path)
    for path in previous_files

}

print(
    "Previous result files protected:",
    len(previous_hashes)
)

try:

    (CODE_DIR / "cell13_repeated_runs_statistics.py").write_text(

        get_ipython().user_ns["In"][-1],

        encoding="utf-8"

    )

except Exception as exc:

    print("Source archival warning:", exc)

# ------------------------------------------------------------
# 3. Load original datasets and parameters
# ------------------------------------------------------------

def load_npz_dataset(filename):

    with np.load(
        DATA_DIR / filename,
        allow_pickle=False
    ) as data:

        return (

            data["X"].astype(
                np.float32, copy=False
            ),

            data["y"].astype(
                np.int8, copy=False
            )

        )


X_balanced13, y_balanced13 = load_npz_dataset(
    "cell05_train_balanced.npz"
)

X_real13, y_real13 = load_npz_dataset(
    "cell05_train_real.npz"
)

X_test13, y_test13 = load_npz_dataset(
    "cell05_test.npz"
)

assert X_balanced13.shape == (552664, 42)

assert X_real13.shape == (446568, 42)

assert X_test13.shape == (111643, 42)

assert np.array_equal(
    y_test13,
    y_test12
)

with open(

    REPORT_DIR
    / "cell05_original_hyperparameters.json"

) as f:

    original_params13 = json.load(f)

with open(

    REPORT_DIR
    / "cell07_federated_experiment_report.json"

) as f:

    federated_report13 = json.load(f)

BASE_XGB_PARAMS13 = federated_report13[
    "effective_federated_parameters"
].copy()

print("\nFixed train-test datasets: VERIFIED")
print("Original hyperparameters: RESTORED")

# ------------------------------------------------------------
# 4. Evaluation function
# ------------------------------------------------------------

def score_probability(y, probability):

    probability = np.asarray(
        probability
    ).reshape(-1)

    assert len(probability) == len(y)

    assert np.isfinite(probability).all()

    prediction = (
        probability >= 0.5
    ).astype(np.int8)

    return {

        "accuracy": float(
            accuracy_score(y, prediction)
        ),

        "precision": float(
            precision_score(
                y, prediction, zero_division=0
            )
        ),

        "recall": float(
            recall_score(
                y, prediction, zero_division=0
            )
        ),

        "f1": float(
            f1_score(
                y, prediction, zero_division=0
            )
        ),

        "roc_auc": float(
            roc_auc_score(y, probability)
        )

    }

# ------------------------------------------------------------
# 5. Import existing seed-42 results
#
# Preserve original results; no seed-42 retraining.
# ------------------------------------------------------------

all_runs = []

blockchain_runs = []

def append_result(
    experiment,
    configuration,
    seed,
    metrics,
    **extra
):

    all_runs.append({

        "experiment": experiment,

        "configuration": configuration,

        "seed": int(seed),

        **{
            metric: float(metrics[metric])
            for metric in METRICS
        },

        **extra

    })


# Cell 05: original centralized classifiers.

cell05 = pd.read_csv(
    TABLE_DIR / "cell05_all_model_metrics.csv"
)

for _, row in cell05.loc[
    cell05["partition"] == "test"
].iterrows():

    append_result(

        "Centralized classifiers",

        row["model"],

        42,

        row

    )

# Cell 08: centralized real-data reference and FL.

cell08 = pd.read_csv(
    TABLE_DIR / "cell08_centralized_federated_comparison.csv"
)

for _, row in cell08.iterrows():

    append_result(

        "Centralized versus federated",

        row["configuration"],

        42,

        row,

        training_seconds=float(
            row["local_model_training_seconds"]
        ),

        aggregation_seconds=float(
            row["aggregation_seconds"]
        ),

        communication_mib=float(
            row["communication_mib"]
        )

    )

# Cell 10: privacy-utility comparison.

cell10 = pd.read_csv(
    TABLE_DIR / "cell10_privacy_utility_comparison.csv"
)

for _, row in cell10.iterrows():

    append_result(

        "Privacy utility",

        row["configuration"],

        42,

        row,

        sigma=float(row["sigma"]),

        standalone_epsilon=float(
            row["standalone_epsilon"]
        )

    )

# Cell 12: ablation comparison.

cell12 = pd.read_csv(
    TABLE_DIR / "cell12_component_ablation.csv"
)

for _, row in cell12.iterrows():

    append_result(

        "Component ablation",

        row["configuration"],

        42,

        row

    )

# Cell 11: blockchain scalability.

BLOCK_METRICS = [

    "mean_block_generation_ms",

    "mean_consensus_ms",

    "mean_finalization_ms",

    "throughput_tps",

    "mean_storage_bytes_per_block",

    "storage_gb_per_million_transactions"

]

cell11 = pd.read_csv(
    TABLE_DIR / "cell11_blockchain_scalability.csv"
)

for _, row in cell11.iterrows():

    blockchain_runs.append({

        "seed": 42,

        "fraud_records": int(row["fraud_records"]),

        **{
            metric: float(row[metric])
            for metric in BLOCK_METRICS
        }

    })

assert len(all_runs) == 15

assert len(blockchain_runs) == 3

print("\nExisting seed-42 results: IMPORTED")

print(
    "Predictive configurations:",
    len(all_runs)
)

print(
    "Blockchain configurations:",
    len(blockchain_runs)
)

# ------------------------------------------------------------
# 6. Additional DP privacy accounting
#
# Per additional seed:
# - Lower noise
# - Reference noise
# - Higher noise
# - Without feature engineering
#
# Full, no-SHAP and no-Blockchain variants reuse
# reference DP predictions, consuming no extra DP
# training budget.
# ------------------------------------------------------------

REFERENCE_SIGMA13 = float(

    cell10.loc[

        cell10["configuration"] == "Reference noise",

        "sigma"

    ].iloc[0]

)

DELTA13 = 1e-5

SENSITIVITY13 = float(
    2 * np.sqrt(40 * (1 + 0.25**2))
)

with open(
    REPORT_DIR / "cell12_privacy_accounting.json"
) as f:

    previous_privacy13 = json.load(f)

PREVIOUS_RHO13 = float(

    previous_privacy13["cumulative_dp_only_rho"]

)


def rho_for_sigma(sigma):

    return float(
        10 * SENSITIVITY13**2
        / (2 * sigma**2)
    )


def epsilon_for_rho(rho):

    return float(

        rho

        + 2 * np.sqrt(
            rho * np.log(1 / DELTA13)
        )

    )


NOISE_CONFIGS13 = {

    "Lower noise":
        REFERENCE_SIGMA13 * 0.5,

    "Reference noise":
        REFERENCE_SIGMA13,

    "Higher noise":
        REFERENCE_SIGMA13 * 2

}

rho_per_seed = (

    sum(
        rho_for_sigma(sigma)
        for sigma in NOISE_CONFIGS13.values()
    )

    + rho_for_sigma(REFERENCE_SIGMA13)

)

ADDITIONAL_RHO13 = (

    len(ADDITIONAL_SEEDS)

    * rho_per_seed

)

PLANNED_RHO13 = (

    PREVIOUS_RHO13

    + ADDITIONAL_RHO13

)

PLANNED_EPSILON13 = epsilon_for_rho(
    PLANNED_RHO13
)

print("\nADDITIONAL PRIVACY ACCOUNTING")

print(
    "Previous DP-only epsilon:",
    epsilon_for_rho(PREVIOUS_RHO13)
)

print(
    "Planned additional rho:",
    ADDITIONAL_RHO13
)

print(
    "Planned cumulative DP-only epsilon:",
    PLANNED_EPSILON13
)

print(
    "This does not protect the joint release "
    "of non-private comparator outputs."
)

ledger13 = {

    "status": "BUDGET_RESERVED",

    "previous_rho":
        PREVIOUS_RHO13,

    "reserved_additional_rho":
        ADDITIONAL_RHO13,

    "planned_combined_rho":
        PLANNED_RHO13,

    "planned_combined_epsilon":
        PLANNED_EPSILON13,

    "delta":
        DELTA13,

    "consumed_additional_rho":
        0.0,

    "completed_private_runs":
        []

}


def save_ledger13():

    temporary = LEDGER13.with_suffix(".tmp")

    with open(temporary, "w") as f:

        json.dump(
            ledger13, f, indent=4
        )

    os.replace(
        temporary, LEDGER13
    )


with open(LEDGER13, "x") as f:

    json.dump(
        ledger13, f, indent=4
    )

# ------------------------------------------------------------
# 7. Save partial results after each experiment
# ------------------------------------------------------------

def save_partial_results():

    pd.DataFrame(all_runs).to_csv(

        TABLE_DIR
        / "cell13_predictive_runs_partial.csv",

        index=False

    )

    pd.DataFrame(blockchain_runs).to_csv(

        TABLE_DIR
        / "cell13_blockchain_runs_partial.csv",

        index=False

    )

# ------------------------------------------------------------
# 8. Repeat original centralized classifiers
#
# Fixed Cell 05 balanced training data.
# Only model-training seed varies.
# ------------------------------------------------------------

def repeat_centralized_classifiers(seed):

    models = {}

    for name in [
        "LightGBM",
        "CatBoost",
        "XGBoost"
    ]:

        params = original_params13[name].copy()

        params["random_state"] = seed

        if name == "LightGBM":

            model = lgb.LGBMClassifier(

                **params,

                device_type="cpu",

                n_jobs=2,

                verbosity=-1

            )

        elif name == "CatBoost":

            params["verbose"] = False

            model = CatBoostClassifier(

                **params,

                task_type="GPU",

                devices="0",

                thread_count=2,

                allow_writing_files=False

            )

        else:

            model = xgb.XGBClassifier(

                **params,

                tree_method="hist",

                device="cuda",

                n_jobs=2

            )

        start = time.perf_counter()

        model.fit(
            X_balanced13,
            y_balanced13
        )

        training_seconds = (
            time.perf_counter() - start
        )

        probability = model.predict_proba(
            X_test13
        )[:, 1]

        metrics = score_probability(
            y_test13, probability
        )

        append_result(

            "Centralized classifiers",

            name,

            seed,

            metrics,

            training_seconds=training_seconds

        )

        joblib.dump(

            model,

            REPEAT_DIR
            / f"{name.lower()}_seed{seed}.joblib"

        )

        print(

            f"{name} | Seed {seed}"

            f" | Accuracy: {metrics['accuracy']:.6f}"

            f" | AUC: {metrics['roc_auc']:.6f}"

        )

        del model
        gc.collect()

        save_partial_results()

# ------------------------------------------------------------
# 9. Repeat matched centralized real-data reference
# ------------------------------------------------------------

def repeat_centralized_real(seed):

    params = BASE_XGB_PARAMS13.copy()

    params["seed"] = seed

    dtrain = xgb.DMatrix(

        X_real13,

        label=y_real13,

        feature_names=feature_names

    )

    dtest = xgb.DMatrix(

        X_test13,

        feature_names=feature_names

    )

    start = time.perf_counter()

    model = xgb.train(

        params,

        dtrain,

        num_boost_round=400,

        verbose_eval=False

    )

    training_seconds = (
        time.perf_counter() - start
    )

    model.set_param({
        "device": "cpu",
        "nthread": 2
    })

    metrics = score_probability(

        y_test13,

        model.predict(dtest)

    )

    append_result(

        "Centralized versus federated",

        "Centralized XGBoost",

        seed,

        metrics,

        training_seconds=training_seconds,

        aggregation_seconds=0.0,

        communication_mib=0.0

    )

    model.save_model(

        REPEAT_DIR
        / f"centralized_real_seed{seed}.ubj"

    )

    print(

        f"Centralized real | Seed {seed}"

        f" | Accuracy: {metrics['accuracy']:.6f}"

        f" | AUC: {metrics['roc_auc']:.6f}"

    )

    del model, dtrain, dtest

    gc.collect()

    save_partial_results()

# ------------------------------------------------------------
# 10. Repeat original federated XGBoost configuration
#
# Same 10 rounds and 40 local trees per round.
# Same weighted additive tree aggregation.
# Client partition and training seed vary.
# ------------------------------------------------------------

def repeat_federated(seed, n_clients):

    params = BASE_XGB_PARAMS13.copy()
    params["seed"] = seed

    splitter = StratifiedKFold(

        n_splits=n_clients,

        shuffle=True,

        random_state=seed

    )

    indices = [

        local_idx

        for _, local_idx in splitter.split(

            X_real13,

            y_real13

        )

    ]

    weights = np.array([

        len(idx)

        for idx in indices

    ], dtype=np.float64)

    weights /= weights.sum()

    matrices = [

        xgb.DMatrix(

            X_real13[idx],

            label=y_real13[idx],

            feature_names=feature_names

        )

        for idx in indices

    ]

    margins = [

        np.zeros(

            len(idx),

            dtype=np.float32

        )

        for idx in indices

    ]

    dtest = xgb.DMatrix(

        X_test13,

        feature_names=feature_names

    )

    test_margin = np.zeros(
        len(y_test13), dtype=np.float32
    )

    total_bytes = 0

    training_seconds = 0.0

    aggregation_seconds = 0.0

    run_dir = (

        REPEAT_DIR
        / f"fl_{n_clients}_clients_seed{seed}"

    )

    run_dir.mkdir(
        parents=True,
        exist_ok=False
    )

    manifest = []

    for r in range(1, 11):

        payloads = []

        for i, matrix in enumerate(matrices):

            matrix.set_base_margin(
                margins[i]
            )

            start = time.perf_counter()

            model = xgb.train(

                params.copy(),

                matrix,

                num_boost_round=40,

                verbose_eval=False

            )

            training_seconds += (
                time.perf_counter() - start
            )

            payload = bytes(

                model.save_raw(
                    raw_format="ubj"
                )

            )

            payloads.append(payload)

            filename = (
                f"round_{r:02d}_client_{i+1:02d}.ubj"
            )

            (run_dir / filename).write_bytes(
                payload
            )

            manifest.append({

                "round": r,

                "client": i + 1,

                "weight": float(weights[i]),

                "model_file": filename

            })

            del model

        aggregation_start = time.perf_counter()

        metadata = json.dumps({

            "round": r,

            "weights": weights.tolist(),

            "payload_lengths": [

                len(payload)
                for payload in payloads

            ]

        }).encode("utf-8")

        upload_bytes = sum(
            len(payload) for payload in payloads
        )

        total_bytes += (

            upload_bytes

            + n_clients * (
                upload_bytes + len(metadata)
            )

        )

        received_models = []

        for payload in payloads:

            model = xgb.Booster()

            model.load_model(
                bytearray(payload)
            )

            model.set_param({

                "device": "cpu",

                "nthread": 2

            })

            received_models.append(model)

        for i, matrix in enumerate(matrices):

            matrix.set_base_margin(

                np.zeros(
                    matrix.num_row(),
                    dtype=np.float32
                )

            )

            update = np.zeros(
                matrix.num_row(),
                dtype=np.float32
            )

            for weight, model in zip(

                weights,
                received_models

            ):

                update += (

                    weight

                    * model.predict(

                        matrix,

                        output_margin=True

                    )

                )

            margins[i] += update

        round_update = np.zeros(

            len(y_test13),

            dtype=np.float32

        )

        dtest.set_base_margin(

            np.zeros(

                len(y_test13),

                dtype=np.float32

            )

        )

        for weight, model in zip(

            weights,

            received_models

        ):

            round_update += (

                weight

                * model.predict(

                    dtest,

                    output_margin=True

                )

            )

        test_margin += round_update

        aggregation_seconds += (

            time.perf_counter()
            - aggregation_start

        )

        del payloads, received_models

        gc.collect()

    metrics = score_probability(

        y_test13,

        expit(test_margin)

    )

    append_result(

        "Centralized versus federated",

        f"Federated ({n_clients} clients)",

        seed,

        metrics,

        training_seconds=training_seconds,

        aggregation_seconds=aggregation_seconds,

        communication_mib=total_bytes / 1024**2

    )

    with open(
        run_dir / "ensemble_manifest.json", "w"
    ) as f:

        json.dump(
            manifest, f, indent=4
        )

    print(

        f"FL {n_clients} clients | Seed {seed}"

        f" | Accuracy: {metrics['accuracy']:.6f}"

        f" | AUC: {metrics['roc_auc']:.6f}"

    )

    del matrices, margins, dtest

    gc.collect()

    save_partial_results()

# ------------------------------------------------------------
# 11. Repeated fixed-tree refinement
#
# Preserve exact Cell 09/12 mechanism.
#
# Privacy randomness uses OS-backed independent noise.
# The run seed does not determine Gaussian noise.
# ------------------------------------------------------------

secure_rng13 = secrets.SystemRandom()


def repeat_fixed_tree(
    model,
    X_private,
    X_testing,
    names,
    sigma,
    seed,
    variant
):

    catalogue, slices = prepare_tree_structure(
        model
    )

    partition_rng = np.random.default_rng(seed)

    private_order = partition_rng.permutation(

        len(y_protected12)

    )

    partitions = np.array_split(

        private_order, 3

    )

    matrices = [

        xgb.DMatrix(

            X_private[idx],

            feature_names=names

        )

        for idx in partitions

    ]

    labels = [

        y_protected12[idx]

        for idx in partitions

    ]

    margins = [

        model.predict(

            matrix,

            output_margin=True

        ).astype(np.float64)

        for matrix in matrices

    ]

    test_matrix = xgb.DMatrix(

        X_testing,

        feature_names=names

    )

    test_margin = model.predict(

        test_matrix,

        output_margin=True

    ).astype(np.float64)

    private_run = sigma > 0

    updates = {}

    start_time = time.perf_counter()

    for r in range(10):

        tree_catalogue = catalogue[
            r * 40:(r + 1) * 40
        ]

        offsets = np.concatenate([

            [0],

            np.cumsum([

                len(leaves)

                for leaves in tree_catalogue

            ])

        ]).astype(np.int64)

        total_leaves = int(offsets[-1])

        G = np.zeros(
            total_leaves,
            dtype=np.float64
        )

        H = np.zeros(
            total_leaves,
            dtype=np.float64
        )

        positions = []

        for i, matrix in enumerate(matrices):

            probability = expit(
                margins[i]
            )

            gradient = np.clip(

                probability - labels[i],

                -1.0,

                1.0

            )

            hessian = np.clip(

                probability
                * (1 - probability),

                0,

                0.25

            )

            leaves = slices[r].predict(

                matrix,

                pred_leaf=True

            ).astype(np.int32).reshape(

                len(labels[i]),

                40

            )

            pos_matrix = np.empty(

                leaves.shape,

                dtype=np.int32

            )

            for j, leaf_ids in enumerate(
                tree_catalogue
            ):

                pos = np.searchsorted(

                    leaf_ids,

                    leaves[:, j]

                )

                assert np.all(
                    pos < len(leaf_ids)
                )

                assert np.array_equal(

                    leaf_ids[pos],

                    leaves[:, j]

                )

                pos_matrix[:, j] = (
                    pos + offsets[j]
                )

                a, b = offsets[j:j + 2]

                G[a:b] += np.bincount(

                    pos,

                    weights=gradient,

                    minlength=len(leaf_ids)

                )

                H[a:b] += np.bincount(

                    pos,

                    weights=hessian,

                    minlength=len(leaf_ids)

                )

            positions.append(pos_matrix)

        joint = np.concatenate([G, H])

        if private_run:

            joint += np.fromiter(

                (

                    secure_rng13.normalvariate(
                        0, sigma
                    )

                    for _ in range(len(joint))

                ),

                dtype=np.float64,

                count=len(joint)

            )

        update = (

            -0.05

            * np.clip(

                joint[:total_leaves]

                / np.maximum(

                    joint[total_leaves:] + 1.0,

                    10.0

                ),

                -1.0,

                1.0

            )

        )

        assert np.isfinite(update).all()

        if private_run:

            ledger13["consumed_additional_rho"] += (

                SENSITIVITY13**2

                / (2 * sigma**2)

            )

            ledger13["status"] = "IN_PROGRESS"

            ledger13["active_run"] = {

                "seed": seed,

                "variant": variant,

                "round": r + 1

            }

            save_ledger13()

        for i in range(3):

            margins[i] += (

                update[positions[i]]

                .sum(axis=1)

            )

        test_leaves = slices[r].predict(

            test_matrix,

            pred_leaf=True

        ).astype(np.int32).reshape(

            len(y_test12),

            40

        )

        for j, leaf_ids in enumerate(
            tree_catalogue
        ):

            pos = np.searchsorted(

                leaf_ids,

                test_leaves[:, j]

            )

            assert np.array_equal(

                leaf_ids[pos],

                test_leaves[:, j]

            )

            test_margin += update[
                pos + offsets[j]
            ]

        updates[
            f"round_{r + 1:02d}_updates"
        ] = update.copy()

        updates[
            f"round_{r + 1:02d}_offsets"
        ] = offsets.copy()

        del (
            positions,
            G,
            H,
            joint,
            test_leaves
        )

        gc.collect()

    probability = expit(test_margin)

    metrics = score_probability(

        y_test12,

        probability

    )

    np.savez_compressed(

        REPEAT_DIR
        / f"{variant}_seed{seed}_updates.npz",

        **updates

    )

    np.savez_compressed(

        REPEAT_DIR
        / f"{variant}_seed{seed}_predictions.npz",

        probability=probability,

        prediction=(
            probability >= 0.5
        ).astype(np.int8)

    )

    if private_run:

        ledger13[
            "completed_private_runs"
        ].append({

            "seed": seed,

            "variant": variant,

            "sigma": sigma,

            "rho": rho_for_sigma(sigma)

        })

        save_ledger13()

    print(

        f"{variant} | Seed {seed}"

        f" | Accuracy: {metrics['accuracy']:.6f}"

        f" | AUC: {metrics['roc_auc']:.6f}"

    )

    return metrics

# ------------------------------------------------------------
# 12. Repeat privacy and ablation experiments
#
# Reuse fixed public reference models.
# Same feature sets, data, and refinement settings.
# ------------------------------------------------------------

def repeat_privacy_and_ablation(seed):

    privacy_results = {}

    for configuration, sigma in (

        NOISE_CONFIGS13.items()

    ):

        variant = (

            configuration.lower()
            .replace(" ", "_")

        )

        metrics = repeat_fixed_tree(

            reference_model,

            X_protected12,

            X_test12,

            feature_names,

            sigma,

            seed,

            variant

        )

        privacy_results[
            configuration
        ] = metrics

        append_result(

            "Privacy utility",

            configuration,

            seed,

            metrics,

            sigma=sigma,

            standalone_epsilon=epsilon_for_rho(

                rho_for_sigma(sigma)

            )

        )

        save_partial_results()

    # Post-prediction component ablations.
    # No additional private model training.

    full_metrics = privacy_results[
        "Reference noise"
    ]

    for variant in [

        "Complete framework",

        "Without Blockchain",

        "Without SHAP"

    ]:

        append_result(

            "Component ablation",

            variant,

            seed,

            full_metrics

        )

    # Non-private fixed-tree comparator.

    no_dp_metrics = repeat_fixed_tree(

        reference_model,

        X_protected12,

        X_test12,

        feature_names,

        0.0,

        seed,

        "without_dp"

    )

    append_result(

        "Component ablation",

        "Without Differential Privacy",

        seed,

        no_dp_metrics

    )

    # Same reduced public tree structures from Cell 12.

    no_fe_metrics = repeat_fixed_tree(

        public_reduced_model,

        X_protected12[:, retained_indices],

        X_test12[:, retained_indices],

        retained_features,

        REFERENCE_SIGMA13,

        seed,

        "without_feature_engineering"

    )

    append_result(

        "Component ablation",

        "Without Feature Engineering",

        seed,

        no_fe_metrics

    )

    save_partial_results()

# ------------------------------------------------------------
# 13. Repeat blockchain scalability benchmark
#
# Same workload, validator protocol, signatures
# and block capacity.
#
# Fresh transaction ordering for additional runs.
# ------------------------------------------------------------

def repeat_blockchain(seed):

    rng = np.random.default_rng(seed)

    for scale in [1000, 10000, 100000]:

        order = rng.permutation(scale)

        workload = [

            transactions[i]
            for i in order

        ]

        validator_states = {

            validator: {

                "height": 0,

                "tip_hash": GENESIS_HASH

            }

            for validator in validator_ids

        }

        previous_hash = GENESIS_HASH

        chain = []

        generation_times = []

        consensus_times = []

        finalization_times = []

        block_sizes = []

        started = time.perf_counter()

        for start in range(0, scale, 100):

            height = start // 100 + 1

            generation_start = (
                time.perf_counter_ns()
            )

            block = create_block(

                height,

                previous_hash,

                workload[start:start + 100]

            )

            generation_times.append(

                (

                    time.perf_counter_ns()

                    - generation_start

                ) / 1e6

            )

            consensus_start = (
                time.perf_counter_ns()
            )

            assert reach_consensus(

                block,

                validator_states

            )

            consensus_times.append(

                (

                    time.perf_counter_ns()

                    - consensus_start

                ) / 1e6

            )

            chain.append(block)

            previous_hash = block["block_hash"]

            finalization_times.append(

                (

                    time.perf_counter_ns()

                    - generation_start

                ) / 1e6

            )

            block_sizes.append(

                len(canonical_bytes(block)) + 1

            )

        elapsed = time.perf_counter() - started

        assert verify_chain(chain)

        storage_bytes = sum(block_sizes)

        blockchain_runs.append({

            "seed": seed,

            "fraud_records": scale,

            "mean_block_generation_ms":

                float(np.mean(generation_times)),

            "mean_consensus_ms":

                float(np.mean(consensus_times)),

            "mean_finalization_ms":

                float(np.mean(finalization_times)),

            "throughput_tps":

                float(scale / elapsed),

            "mean_storage_bytes_per_block":

                float(storage_bytes / len(chain)),

            "storage_gb_per_million_transactions":

                float(storage_bytes / scale / 1000)

        })

        print(

            f"Blockchain | Seed {seed}"

            f" | Records: {scale:,}"

            f" | TPS: {scale / elapsed:.2f}"

        )

        del chain, workload

        gc.collect()

        save_partial_results()

# ------------------------------------------------------------
# 14. Execute the four required additional runs
# ------------------------------------------------------------

for seed in ADDITIONAL_SEEDS:

    print("\n" + "=" * 75)

    print("ADDITIONAL RUN - SEED:", seed)

    print("=" * 75)

    repeat_centralized_classifiers(seed)

    repeat_centralized_real(seed)

    for n_clients in [3, 5, 10]:

        repeat_federated(

            seed,

            n_clients

        )

    repeat_privacy_and_ablation(seed)

    repeat_blockchain(seed)

    print(f"\nSeed {seed}: COMPLETED")

    save_partial_results()

    gc.collect()

# ------------------------------------------------------------
# 15. Verify five runs per configuration
# ------------------------------------------------------------

runs_df = pd.DataFrame(all_runs)

blockchain_df = pd.DataFrame(blockchain_runs)

assert not runs_df.duplicated([

    "experiment",

    "configuration",

    "seed"

]).any()

for _, group in runs_df.groupby([

    "experiment",

    "configuration"

]):

    assert sorted(group["seed"]) == SEEDS

for _, group in blockchain_df.groupby(
    "fraud_records"
):

    assert sorted(group["seed"]) == SEEDS

assert len(runs_df) == 75

assert len(blockchain_df) == 15

runs_df.to_csv(

    TABLE_DIR
    / "cell13_all_predictive_runs.csv",

    index=False

)

blockchain_df.to_csv(

    TABLE_DIR
    / "cell13_all_blockchain_runs.csv",

    index=False

)

print("\nFive runs per predictive configuration: PASS")
print("Five runs per blockchain workload: PASS")

# ------------------------------------------------------------
# 16. Mean, sample SD and 95% confidence intervals
# ------------------------------------------------------------

def summarize_values(values):

    values = np.asarray(
        values, dtype=np.float64
    )

    values = values[np.isfinite(values)]

    n = len(values)

    if n < 2:

        return None

    mean = float(
        np.mean(values)
    )

    sd = float(
        np.std(values, ddof=1)
    )

    half_width = float(

        t.ppf(0.975, n - 1)

        * sd / np.sqrt(n)

    )

    return {

        "n_runs": n,

        "mean": mean,

        "sd": sd,

        "ci95_lower": mean - half_width,

        "ci95_upper": mean + half_width

    }


statistics_records = []

for (experiment, config), group in (

    runs_df.groupby([

        "experiment",

        "configuration"

    ])

):

    for metric in METRICS:

        statistics_records.append({

            "experiment": experiment,

            "configuration": config,

            "metric": metric,

            **summarize_values(
                group[metric]
            )

        })

statistics_df = pd.DataFrame(
    statistics_records
)

statistics_df.to_csv(

    TABLE_DIR
    / "cell13_predictive_mean_sd_ci.csv",

    index=False

)

block_statistics = []

for scale, group in blockchain_df.groupby(
    "fraud_records"
):

    for metric in BLOCK_METRICS:

        block_statistics.append({

            "fraud_records": int(scale),

            "metric": metric,

            **summarize_values(
                group[metric]
            )

        })

block_statistics_df = pd.DataFrame(
    block_statistics
)

block_statistics_df.to_csv(

    TABLE_DIR
    / "cell13_blockchain_mean_sd_ci.csv",

    index=False

)

# ------------------------------------------------------------
# 17. Manuscript-ready mean ± SD tables
# ------------------------------------------------------------

publication_records = []

for (experiment, config), group in (

    runs_df.groupby([

        "experiment",

        "configuration"

    ])

):

    record = {

        "Experiment": experiment,

        "Configuration": config,

        "Runs": len(group)

    }

    for metric in METRICS:

        mean = group[metric].mean()

        sd = group[metric].std(ddof=1)

        if metric == "roc_auc":

            record[metric] = (

                f"{mean:.6f} ± {sd:.6f}"

            )

        else:

            record[metric] = (

                f"{100 * mean:.2f} "

                f"± {100 * sd:.2f}"

            )

    publication_records.append(record)

publication_df = pd.DataFrame(
    publication_records
)

publication_df.to_csv(

    TABLE_DIR
    / "cell13_manuscript_mean_sd_table.csv",

    index=False

)

print("\n" + "=" * 75)
print("FINAL PREDICTIVE PERFORMANCE: MEAN ± SD")
print("=" * 75)

print(

    publication_df.to_string(index=False)

)

# Blockchain manuscript table.

block_publication_records = []

for scale, group in blockchain_df.groupby(
    "fraud_records"
):

    record = {

        "Fraud records": int(scale),

        "Runs": len(group)

    }

    for metric in BLOCK_METRICS:

        mean = group[metric].mean()

        sd = group[metric].std(ddof=1)

        record[metric] = (

            f"{mean:.6f} ± {sd:.6f}"

        )

    block_publication_records.append(record)

block_publication_df = pd.DataFrame(
    block_publication_records
)

block_publication_df.to_csv(

    TABLE_DIR
    / "cell13_manuscript_blockchain_mean_sd.csv",

    index=False

)

print("\nBLOCKCHAIN RESULTS: MEAN ± SD")

print(

    block_publication_df.to_string(index=False)

)

# ------------------------------------------------------------
# 18. Reviewer-required significance testing
#
# Predefined comparisons within experiment families.
# Paired by run index.
#
# Accuracy and ROC-AUC are the tested metrics.
# ------------------------------------------------------------

comparisons = [

    (

        "Centralized classifiers",

        "XGBoost",

        "LightGBM"

    ),

    (

        "Centralized classifiers",

        "XGBoost",

        "CatBoost"

    ),

    (

        "Centralized versus federated",

        "Federated (3 clients)",

        "Centralized XGBoost"

    ),

    (

        "Centralized versus federated",

        "Federated (5 clients)",

        "Centralized XGBoost"

    ),

    (

        "Centralized versus federated",

        "Federated (10 clients)",

        "Centralized XGBoost"

    ),

    (

        "Privacy utility",

        "Reference noise",

        "Lower noise"

    ),

    (

        "Privacy utility",

        "Reference noise",

        "Higher noise"

    ),

    (

        "Component ablation",

        "Complete framework",

        "Without Differential Privacy"

    ),

    (

        "Component ablation",

        "Complete framework",

        "Without Feature Engineering"

    )

]

test_records = []

for experiment, method_a, method_b in comparisons:

    group_a = runs_df.loc[

        (

            runs_df["experiment"]
            == experiment

        )

        &

        (

            runs_df["configuration"]
            == method_a

        )

    ].set_index("seed").loc[SEEDS]

    group_b = runs_df.loc[

        (

            runs_df["experiment"]
            == experiment

        )

        &

        (

            runs_df["configuration"]
            == method_b

        )

    ].set_index("seed").loc[SEEDS]

    for metric in ["accuracy", "roc_auc"]:

        a = group_a[metric].to_numpy()

        b = group_b[metric].to_numpy()

        differences = a - b

        summary = summarize_values(
            differences
        )

        sd_difference = np.std(
            differences, ddof=1
        )

        if np.isclose(
            sd_difference, 0,
            atol=1e-15,
            rtol=0
        ):

            statistic = None

            pvalue = (

                1.0

                if np.allclose(
                    differences, 0,
                    atol=1e-15,
                    rtol=0
                )

                else None

            )

            test_note = (
                "Degenerate paired differences"
            )

        else:

            result = ttest_rel(a, b)

            statistic = float(
                result.statistic
            )

            pvalue = float(
                result.pvalue
            )

            test_note = (
                "Exploratory paired t-test"
            )

        test_records.append({

            "experiment": experiment,

            "method_a": method_a,

            "method_b": method_b,

            "metric": metric,

            "n_pairs": len(SEEDS),

            "mean_difference_pp":
                float(np.mean(differences) * 100),

            "difference_ci95_lower_pp":

                summary["ci95_lower"] * 100,

            "difference_ci95_upper_pp":

                summary["ci95_upper"] * 100,

            "t_statistic": statistic,

            "p_value": pvalue,

            "test_note": test_note

        })

tests_df = pd.DataFrame(test_records)

# ------------------------------------------------------------
# 19. Holm multiple-testing correction
# ------------------------------------------------------------

def holm_adjust(pvalues):

    pvalues = np.asarray(
        pvalues,
        dtype=np.float64
    )

    n = len(pvalues)

    order = np.argsort(pvalues)

    sorted_p = pvalues[order]

    adjusted_sorted = np.maximum.accumulate(

        (n - np.arange(n)) * sorted_p

    )

    adjusted_sorted = np.minimum(

        adjusted_sorted,

        1.0

    )

    adjusted = np.empty(n)

    adjusted[order] = adjusted_sorted

    return adjusted


tests_df["holm_adjusted_p"] = np.nan

valid = tests_df["p_value"].notna()

tests_df.loc[

    valid,

    "holm_adjusted_p"

] = holm_adjust(

    tests_df.loc[valid, "p_value"]

)

tests_df["significant_at_0_05"] = (

    tests_df["holm_adjusted_p"] < 0.05

)

tests_df.to_csv(

    TABLE_DIR
    / "cell13_statistical_significance_tests.csv",

    index=False

)

print("\n" + "=" * 75)
print("PAIRED STATISTICAL TESTING")
print("=" * 75)

print(

    tests_df[[

        "experiment",

        "method_a",

        "method_b",

        "metric",

        "mean_difference_pp",

        "p_value",

        "holm_adjusted_p",

        "significant_at_0_05"

    ]].round(6).to_string(index=False)

)

# ------------------------------------------------------------
# 20. Final cumulative DP-only privacy accounting
# ------------------------------------------------------------

assert len(

    ledger13["completed_private_runs"]

) == 16

actual_additional_rho = float(

    ledger13["consumed_additional_rho"]

)

assert np.isclose(

    actual_additional_rho,

    ADDITIONAL_RHO13,

    atol=1e-10,

    rtol=1e-10

)

final_rho13 = (

    PREVIOUS_RHO13

    + actual_additional_rho

)

final_epsilon13 = epsilon_for_rho(
    final_rho13
)

ledger13["status"] = "COMPLETED"

ledger13["final_combined_rho"] = final_rho13

ledger13["final_combined_epsilon"] = (
    final_epsilon13
)

save_ledger13()

print("\nCUMULATIVE DP-ONLY ACCOUNTING")

print(

    "Previous epsilon:",

    epsilon_for_rho(PREVIOUS_RHO13)

)

print(

    "Updated epsilon:",

    final_epsilon13

)

print("Delta:", DELTA13)

print(

    "Non-private outputs remain outside "

    "the DP-only bound."

)

# ------------------------------------------------------------
# 21. Verify all previous results unchanged
# ------------------------------------------------------------

for path_string, expected_hash in (

    previous_hashes.items()

):

    assert sha256_file(

        Path(path_string)

    ) == expected_hash

print("\nPrevious experimental results: UNCHANGED")

# ------------------------------------------------------------
# 22. Original source integrity
# ------------------------------------------------------------

with open(

    PROJECT_ROOT

    / "02_Revised_Project"

    / "reports"

    / "cell02_original_source_hashes.json"

) as f:

    source_hashes = json.load(f)

for item in source_hashes:

    path = SOURCE_ROOT / item["file"]

    assert path.exists()

    assert sha256_file(path) == item["sha256"]

print("Original source integrity: PASS")

# ------------------------------------------------------------
# 23. Repeated-run and statistical report
# ------------------------------------------------------------

report13 = {

    "cell": 13,

    "status": "COMPLETED",

    "reviewer_comment": "Major 6",

    "seeds": SEEDS,

    "runs_per_configuration": 5,

    "predictive_configurations": 15,

    "predictive_run_records": len(runs_df),

    "blockchain_configurations": 3,

    "blockchain_run_records": len(blockchain_df),

    "seed42_results_reused": True,

    "fixed_train_test_split": True,

    "fixed_smote_training_data": True,

    "fixed_public_dp_tree_structures": True,

    "standard_deviation":
        "Sample SD with ddof=1",

    "confidence_interval":
        "Student t 95% interval across runs",

    "significance_test":
        "Exploratory paired t-test",

    "multiple_testing_correction":
        "Holm",

    "number_of_tests": len(tests_df),

    "additional_private_runs": 16,

    "previous_dp_only_rho":
        PREVIOUS_RHO13,

    "additional_dp_only_rho":
        actual_additional_rho,

    "cumulative_dp_only_rho":
        final_rho13,

    "cumulative_dp_only_epsilon":
        final_epsilon13,

    "delta": DELTA13,

    "original_predictors": 42,

    "hyperparameter_tuning": False,

    "external_dataset_used": False,

    "previous_results_modified": False,

    "original_source_modified": False,

    "limitations": [

        "The five runs use a fixed train-test split. "
        "They characterize stochastic algorithm "
        "variability rather than independent "
        "population samples.",

        "The original balanced training dataset "
        "is held fixed across model-training seeds.",

        "Provider overlap remains in the "
        "claim-wise evaluation protocol.",

        "Paired t-tests with five runs have "
        "limited power and depend on distributional "
        "assumptions about paired differences.",

        "The reported confidence intervals are "
        "conditional on the fixed dataset.",

        "DP runs reuse fixed public reference trees "
        "and vary independent Gaussian noise.",

        "Post-prediction component ablations "
        "reuse identical predictive outputs.",

        "SHAP and Blockchain operational ablation "
        "timings from Cell 12 are single-run "
        "measurements and not repeated here.",

        "Blockchain results represent local "
        "simulation rather than physical network "
        "performance.",

        "DP-only accounting excludes non-private "
        "models and cannot establish privacy "
        "of their joint release."

    ],

    "elapsed_seconds":

        time.perf_counter() - START_TIME

}

with open(

    REPORT_DIR
    / "cell13_statistical_analysis_report.json",

    "w"

) as f:

    json.dump(

        report13,

        f,

        indent=4,

        default=str

    )

# ------------------------------------------------------------
# 24. Save completion checkpoint
# ------------------------------------------------------------

checkpoint13 = {

    "cell": 13,

    "status": "COMPLETED",

    "reviewer_comment": "Major 6",

    "runs_per_configuration": 5,

    "seeds": SEEDS,

    "predictive_run_records": 75,

    "blockchain_run_records": 15,

    "mean_sd_completed": True,

    "confidence_intervals_completed": True,

    "statistical_tests_completed": True,

    "holm_correction_completed": True,

    "cumulative_dp_only_epsilon":
        final_epsilon13,

    "delta": DELTA13,

    "original_predictors": 42,

    "previous_results_modified": False,

    "original_source_modified": False,

    "next_cell": 14

}

with open(

    CHECKPOINT_DIR
    / "cell13_completed.json",

    "w"

) as f:

    json.dump(

        checkpoint13,

        f,

        indent=4

    )

gc.collect()

print("\n" + "=" * 75)
print("CELL 13 COMPLETED SUCCESSFULLY")
print("=" * 75)

print("Five runs per predictive configuration: PASS")
print("Five runs per blockchain workload: PASS")
print("Mean and standard deviation: COMPLETED")
print("95% confidence intervals: COMPLETED")
print("Paired significance testing: COMPLETED")
print("Holm multiple-testing correction: COMPLETED")
print("DP-only cumulative accounting: COMPLETED")
print("Original 42 predictors: RETAINED")
print("Previous results: UNCHANGED")
print("Original source integrity: PASS")
print("Hyperparameter tuning: NONE")
print("External dataset: NONE")
print("Statistical tables and models saved.")
print("Complete source code saved.")
print("Checkpoint saved.")

print(

    "\nTotal execution seconds:",

    round(time.perf_counter() - START_TIME, 2)

)

print(

    "\nREADY FOR CELL 14/15: "

    "SHAP INTERPRETABILITY AND FINAL RESULTS"

)
```

```python
# ============================================================
# CELL 14/15 - SHAP INTERPRETABILITY AND FINAL RESULTS
#
# Global and local SHAP explanations
# Final reviewer-ready statistical tables
# Experimental environment documentation
#
# NO MODEL RETRAINING
# NO ADDITIONAL PRIVATE TRAINING
# NO FEATURE MODIFICATION
# NO HYPERPARAMETER TUNING
# ============================================================

import os
import gc
import sys
import json
import time
import hashlib
import platform
import subprocess
import importlib.metadata

from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import xgboost as xgb
import psutil

from scipy.special import expit

START_TIME = time.perf_counter()

SEED14 = 42
SHAP_SAMPLE_SIZE = 2000
TOP_FEATURES = 15

print("=" * 75)
print("CELL 14: SHAP INTERPRETABILITY AND FINAL RESULTS")
print("=" * 75)

# ------------------------------------------------------------
# 1. Restore reviewer-only workspace
# ------------------------------------------------------------

PROJECT_ROOT = Path(
    "/content/drive/MyDrive/Narayan_Patil_Reviewer_Revision"
)

SOURCE_ROOT = PROJECT_ROOT / "01_Original_Project" / "SBN"
REVISED_DIR = PROJECT_ROOT / "03_Reviewer_Only_Revision"

CODE_DIR = REVISED_DIR / "code"
DATA_DIR = REVISED_DIR / "data"
MODEL_DIR = REVISED_DIR / "models"
RESULT_DIR = REVISED_DIR / "results"
TABLE_DIR = RESULT_DIR / "tables"
FIGURE_DIR = RESULT_DIR / "figures"
REPORT_DIR = REVISED_DIR / "reports"
CHECKPOINT_DIR = REVISED_DIR / "checkpoints"
PREDICTION_DIR = RESULT_DIR / "predictions"

DP_DIR = MODEL_DIR / "differential_privacy"
SHAP_DIR = RESULT_DIR / "shap"

FINAL_TABLE_DIR = TABLE_DIR / "final_manuscript"

for folder in [
    SHAP_DIR,
    FINAL_TABLE_DIR,
    FIGURE_DIR,
    REPORT_DIR
]:
    folder.mkdir(parents=True, exist_ok=True)

assert (
    CHECKPOINT_DIR / "cell13_completed.json"
).exists(), "Complete Cell 13 first."

assert not (
    CHECKPOINT_DIR / "cell14_completed.json"
).exists(), "Cell 14 already completed."

with open(
    CHECKPOINT_DIR / "cell13_completed.json"
) as f:
    checkpoint13 = json.load(f)

assert checkpoint13["status"] == "COMPLETED"
assert checkpoint13["runs_per_configuration"] == 5
assert checkpoint13["predictive_run_records"] == 75
assert checkpoint13["blockchain_run_records"] == 15

print("\nWorkspace:", REVISED_DIR)
print("Cell 13 checkpoint: VERIFIED")
print("Five-run statistical results: AVAILABLE")

# ------------------------------------------------------------
# 2. Verify existing runtime objects
# ------------------------------------------------------------

required_objects = [
    "full_model12",
    "X_test12",
    "y_test12",
    "feature_names",
    "test_ids"
]

missing = [
    name for name in required_objects
    if name not in globals()
]

assert not missing, (
    f"Missing runtime objects: {missing}. "
    "Restore saved objects instead of rerunning DP training."
)

assert X_test12.shape == (111643, 42)
assert len(y_test12) == 111643
assert len(test_ids) == 111643
assert len(feature_names) == 42
assert full_model12.num_boosted_rounds() == 400
assert full_model12.feature_names == feature_names

print("\nComplete framework model: AVAILABLE")
print("Original predictors: 42")
print("Testing records: 111,643")
print("Original tree count: 400")

# ------------------------------------------------------------
# 3. Preserve original experimental files
# ------------------------------------------------------------

def sha256_file(path):
    digest = hashlib.sha256()

    with open(path, "rb") as f:
        for chunk in iter(
            lambda: f.read(1024 * 1024), b""
        ):
            digest.update(chunk)

    return digest.hexdigest()


previous_files = sorted([
    path for path in TABLE_DIR.glob("cell*.csv")
    if not path.name.startswith("cell14")
])

previous_hashes14 = {
    str(path): sha256_file(path)
    for path in previous_files
}

print(
    "\nPrevious experimental files protected:",
    len(previous_hashes14)
)

# ------------------------------------------------------------
# 4. Verify reconstructed DP model
#
# Use exact Cell 09 model predictions.
# No retraining or private refinement.
# ------------------------------------------------------------

with open(
    CHECKPOINT_DIR / "cell09_completed.json"
) as f:
    checkpoint09 = json.load(f)

REFERENCE_RUN = (
    DP_DIR / checkpoint09["dp_run_name"]
)

with np.load(
    REFERENCE_RUN / "test_predictions.npz",
    allow_pickle=False
) as data:

    reference_probability14 = (
        data["probability"].copy()
    )

full_model12.set_param({
    "device": "cpu",
    "nthread": 2
})

dtest14 = xgb.DMatrix(
    X_test12,
    feature_names=feature_names
)

reconstructed_probability14 = (
    full_model12.predict(dtest14)
)

max_probability_difference = float(
    np.max(
        np.abs(
            reconstructed_probability14
            - reference_probability14
        )
    )
)

assert np.allclose(
    reconstructed_probability14,
    reference_probability14,
    atol=2e-5,
    rtol=2e-5
)

print("\nMODEL RECONSTRUCTION VERIFICATION")

print(
    "Maximum probability difference:",
    max_probability_difference
)

print("Original DP predictions: VERIFIED")

# Save an inference-ready reconstructed model.

FINAL_MODEL_PATH = (
    MODEL_DIR / "cell14_complete_framework_dp.ubj"
)

full_model12.save_model(FINAL_MODEL_PATH)

reloaded_model14 = xgb.Booster()

reloaded_model14.load_model(FINAL_MODEL_PATH)

reloaded_model14.set_param({
    "device": "cpu",
    "nthread": 2
})

assert np.allclose(
    reloaded_model14.predict(dtest14),
    reference_probability14,
    atol=2e-5,
    rtol=2e-5
)

print("Inference-ready DP model: SAVED AND VERIFIED")

# ------------------------------------------------------------
# 5. Select fixed held-out SHAP sample
#
# No protected training records used for explanations.
# ------------------------------------------------------------

rng14 = np.random.default_rng(SEED14)

shap_indices14 = np.sort(
    rng14.choice(
        len(y_test12),
        size=SHAP_SAMPLE_SIZE,
        replace=False
    )
)

assert len(
    np.unique(shap_indices14)
) == SHAP_SAMPLE_SIZE

X_shap14 = X_test12[shap_indices14]

y_shap14 = y_test12[shap_indices14]

dshap14 = xgb.DMatrix(
    X_shap14,
    feature_names=feature_names
)

print("\nSHAP EVALUATION SAMPLE")

print("Sample size:", SHAP_SAMPLE_SIZE)
print("Predictors:", len(feature_names))

print(
    "Non-fraud records:",
    int(np.sum(y_shap14 == 0))
)

print(
    "Fraud-labelled records:",
    int(np.sum(y_shap14 == 1))
)

# ------------------------------------------------------------
# 6. Calculate exact TreeSHAP contributions
#
# XGBoost native TreeSHAP implementation.
# SHAP contributions explain raw log-odds.
# ------------------------------------------------------------

shap_start = time.perf_counter()

shap_all14 = full_model12.predict(
    dshap14,
    pred_contribs=True,
    approx_contribs=False
)

shap_seconds14 = (
    time.perf_counter() - shap_start
)

assert shap_all14.shape == (
    SHAP_SAMPLE_SIZE, 43
)

assert np.isfinite(shap_all14).all()

shap_values14 = shap_all14[:, :-1]

base_values14 = shap_all14[:, -1]

raw_margins14 = full_model12.predict(
    dshap14,
    output_margin=True
)

shap_probability14 = full_model12.predict(
    dshap14
)

# ------------------------------------------------------------
# 7. Verify SHAP additivity
#
# Raw prediction margin =
# SHAP base value + sum of SHAP contributions
# ------------------------------------------------------------

reconstructed_margins14 = (
    base_values14
    + shap_values14.sum(axis=1)
)

additivity_errors14 = np.abs(
    reconstructed_margins14
    - raw_margins14
)

max_additivity_error = float(
    additivity_errors14.max()
)

mean_additivity_error = float(
    additivity_errors14.mean()
)

assert np.allclose(
    reconstructed_margins14,
    raw_margins14,
    atol=1e-3,
    rtol=1e-4
), "SHAP additivity verification failed."

assert np.allclose(
    expit(reconstructed_margins14),
    shap_probability14,
    atol=1e-4,
    rtol=1e-4
)

print("\nSHAP COMPUTATION")

print(
    "SHAP matrix:",
    shap_values14.shape
)

print(
    "Maximum additivity error:",
    max_additivity_error
)

print(
    "Mean additivity error:",
    mean_additivity_error
)

print(
    "SHAP computation seconds:",
    round(shap_seconds14, 4)
)

print("SHAP additivity: PASS")
print("Probability reconstruction: PASS")

# Store explanation artifacts in research workspace.
# These files contain held-out case information.
# They should not be publicly released without review.

np.savez_compressed(
    SHAP_DIR / "cell14_global_shap_values.npz",

    row_id=test_ids[shap_indices14],

    shap_values=shap_values14,

    base_values=base_values14,

    feature_names=np.asarray(
        feature_names,
        dtype=str
    ),

    probability=shap_probability14
)

# ------------------------------------------------------------
# 8. Global SHAP feature importance
# ------------------------------------------------------------

mean_abs_shap = np.mean(
    np.abs(shap_values14),
    axis=0
)

mean_signed_shap = np.mean(
    shap_values14,
    axis=0
)

shap_importance14 = pd.DataFrame({

    "feature": feature_names,

    "mean_absolute_shap":
        mean_abs_shap,

    "mean_signed_shap":
        mean_signed_shap

}).sort_values(
    "mean_absolute_shap",
    ascending=False
).reset_index(drop=True)

shap_importance14.insert(
    0,
    "rank",
    np.arange(1, 43)
)

total_importance14 = (
    shap_importance14[
        "mean_absolute_shap"
    ].sum()
)

assert total_importance14 > 0

shap_importance14[
    "relative_importance_percent"
] = (
    shap_importance14[
        "mean_absolute_shap"
    ]
    / total_importance14
    * 100
)

shap_importance14.to_csv(
    TABLE_DIR / "cell14_global_shap_importance.csv",
    index=False
)

print("\n" + "=" * 75)
print("GLOBAL SHAP FEATURE IMPORTANCE")
print("=" * 75)

print(
    shap_importance14.head(15)
    .round(6)
    .to_string(index=False)
)

# ------------------------------------------------------------
# 9. Global SHAP importance figure
#
# Top 15 features displayed.
# All 42 features retained in the numerical table.
# ------------------------------------------------------------

plot_importance14 = (
    shap_importance14.head(TOP_FEATURES)
    .sort_values("mean_absolute_shap")
)

plt.figure(figsize=(8, 6))

plt.barh(
    plot_importance14["feature"],
    plot_importance14["mean_absolute_shap"],
    color="gray",
    edgecolor="black",
    linewidth=0.7
)

plt.xlabel(
    "Mean absolute SHAP value (log-odds)"
)

plt.ylabel("Predictor")

plt.grid(
    axis="x",
    linestyle=":",
    alpha=0.3
)

plt.tight_layout()

plt.savefig(
    FIGURE_DIR / "cell14_global_shap_importance.png",
    dpi=600,
    bbox_inches="tight"
)

plt.show()
plt.close()

print("\nGlobal SHAP importance figure: SAVED")

# ------------------------------------------------------------
# 10. Signed SHAP distribution plot
#
# Black-and-white visualization.
# Avoid interpreting ordinal-encoded category codes
# as inherently ordered clinical quantities.
# ------------------------------------------------------------

top_names14 = (
    shap_importance14.head(TOP_FEATURES)
    ["feature"].tolist()
)

top_indices14 = [
    feature_names.index(name)
    for name in top_names14
]

plt.figure(figsize=(8, 6))

jitter_rng = np.random.default_rng(SEED14)

for i, feature_index in enumerate(
    top_indices14
):

    position = (
        TOP_FEATURES - 1 - i
    )

    jitter = jitter_rng.uniform(
        -0.22,
        0.22,
        SHAP_SAMPLE_SIZE
    )

    plt.scatter(
        shap_values14[:, feature_index],
        position + jitter,
        s=5,
        color="black",
        alpha=0.18,
        linewidths=0,
        rasterized=True
    )

plt.axvline(
    0,
    color="gray",
    linestyle="--",
    linewidth=1
)

plt.yticks(
    np.arange(TOP_FEATURES),
    top_names14[::-1]
)

plt.xlabel(
    "SHAP contribution to fraud log-odds"
)

plt.tight_layout()

plt.savefig(
    FIGURE_DIR / "cell14_shap_summary_distribution.png",
    dpi=600,
    bbox_inches="tight"
)

plt.show()
plt.close()

print("Signed SHAP distribution figure: SAVED")

# ------------------------------------------------------------
# 11. Local fraud prediction explanation
#
# Select the first correctly classified positive
# case in the fixed sample.
#
# Illustrative example, not a representative claim
# about all fraud cases.
# ------------------------------------------------------------

eligible_local14 = np.flatnonzero(
    (y_shap14 == 1)
    & (shap_probability14 >= 0.5)
)

assert len(eligible_local14) > 0

local_index14 = int(
    eligible_local14[0]
)

local_shap14 = (
    shap_values14[local_index14]
)

local_base14 = float(
    base_values14[local_index14]
)

local_margin14 = float(
    raw_margins14[local_index14]
)

local_probability14 = float(
    shap_probability14[local_index14]
)

local_df14 = pd.DataFrame({

    "feature": feature_names,

    "shap_contribution":
        local_shap14,

    "absolute_contribution":
        np.abs(local_shap14)

}).sort_values(
    "absolute_contribution",
    ascending=False
)

local_df14.to_csv(
    SHAP_DIR / "cell14_local_explanation.csv",
    index=False
)

print("\nLOCAL FRAUD EXPLANATION")

print(
    "Predicted fraud probability:",
    round(local_probability14, 6)
)

print(
    "SHAP base log-odds:",
    round(local_base14, 6)
)

print(
    "Final prediction log-odds:",
    round(local_margin14, 6)
)

print(
    local_df14.head(10)
    .round(6)
    .to_string(index=False)
)

# ------------------------------------------------------------
# 12. Local SHAP contribution figure
# ------------------------------------------------------------

local_plot14 = (
    local_df14.head(10)
    .sort_values("absolute_contribution")
)

local_colors14 = [
    "black" if value >= 0 else "white"
    for value in local_plot14[
        "shap_contribution"
    ]
]

plt.figure(figsize=(8, 5))

plt.barh(
    local_plot14["feature"],
    local_plot14["shap_contribution"],
    color=local_colors14,
    edgecolor="black",
    linewidth=0.8
)

plt.axvline(
    0,
    color="gray",
    linestyle="--"
)

plt.xlabel(
    "SHAP contribution (log-odds)"
)

plt.tight_layout()

plt.savefig(
    FIGURE_DIR / "cell14_local_shap_explanation.png",
    dpi=600,
    bbox_inches="tight"
)

plt.show()
plt.close()

print("Local SHAP explanation figure: SAVED")

# ------------------------------------------------------------
# 13. Restore verified repeated-run results
# ------------------------------------------------------------

runs14 = pd.read_csv(
    TABLE_DIR / "cell13_all_predictive_runs.csv"
)

block_runs14 = pd.read_csv(
    TABLE_DIR / "cell13_all_blockchain_runs.csv"
)

statistics14 = pd.read_csv(
    TABLE_DIR / "cell13_statistical_significance_tests.csv"
)

summary14 = pd.read_csv(
    TABLE_DIR / "cell13_manuscript_mean_sd_table.csv"
)

assert len(runs14) == 75
assert len(block_runs14) == 15
assert len(statistics14) == 18
assert len(summary14) == 15

assert (
    runs14.groupby([
        "experiment",
        "configuration"
    ]).size() == 5
).all()

assert (
    block_runs14.groupby(
        "fraud_records"
    ).size() == 5
).all()

print("\nFive-run predictive results: VERIFIED")
print("Five-run blockchain results: VERIFIED")
print("Statistical significance tests: VERIFIED")

# ------------------------------------------------------------
# 14. Final manuscript predictive tables
#
# Do not modify Cell 13 source tables.
# Export separate manuscript-ready copies.
# ------------------------------------------------------------

TABLE_GROUPS14 = {

    "Centralized classifiers":
        "final_centralized_classifier_results.csv",

    "Centralized versus federated":
        "final_centralized_federated_results.csv",

    "Privacy utility":
        "final_privacy_utility_results.csv",

    "Component ablation":
        "final_component_ablation_results.csv"
}

rename_columns14 = {

    "accuracy": "Accuracy (%)",

    "precision": "Precision (%)",

    "recall": "Recall (%)",

    "f1": "F1-score (%)",

    "roc_auc": "ROC-AUC"
}

for experiment, filename in TABLE_GROUPS14.items():

    table = summary14.loc[
        summary14["Experiment"] == experiment
    ].copy()

    assert len(table) > 0

    table = table.drop(
        columns=["Experiment"]
    ).rename(
        columns=rename_columns14
    )

    table.to_csv(
        FINAL_TABLE_DIR / filename,
        index=False
    )

    print("\n" + experiment.upper())

    print(table.to_string(index=False))

# ------------------------------------------------------------
# 15. Final blockchain mean ± SD table
# ------------------------------------------------------------

block_summary14 = pd.read_csv(
    TABLE_DIR / "cell13_manuscript_blockchain_mean_sd.csv"
)

assert len(block_summary14) == 3

block_summary14.to_csv(
    FINAL_TABLE_DIR / "final_blockchain_scalability.csv",
    index=False
)

print("\nBLOCKCHAIN SCALABILITY: MEAN ± SD")

print(
    block_summary14.to_string(index=False)
)

# ------------------------------------------------------------
# 16. Final training time and communication statistics
#
# Report mean ± SD across five runs.
# Only configurations with five valid measurements
# are accepted.
# ------------------------------------------------------------

fl_runs14 = runs14.loc[
    runs14["experiment"]
    == "Centralized versus federated"
].copy()

resource_records14 = []

for configuration, group in fl_runs14.groupby(
    "configuration"
):

    record = {
        "Configuration": configuration,
        "Runs": len(group)
    }

    for metric, label in [

        (
            "training_seconds",
            "Training time (s)"
        ),

        (
            "aggregation_seconds",
            "Aggregation time (s)"
        ),

        (
            "communication_mib",
            "Communication (MiB)"
        )

    ]:

        values = group[metric].dropna()

        assert len(values) == 5

        record[label] = (
            f"{values.mean():.4f} "
            f"± {values.std(ddof=1):.4f}"
        )

    resource_records14.append(record)

resource_table14 = pd.DataFrame(
    resource_records14
)

resource_table14.to_csv(
    FINAL_TABLE_DIR
    / "final_training_communication_mean_sd.csv",
    index=False
)

print("\nTRAINING AND COMMUNICATION: MEAN ± SD")

print(
    resource_table14.to_string(index=False)
)

# ------------------------------------------------------------
# 17. Export exact significance results
#
# Scientific notation avoids falsely reporting
# small p-values as exactly zero.
# ------------------------------------------------------------

publication_tests14 = statistics14.copy()

publication_tests14["p_value"] = (
    publication_tests14["p_value"].map(
        lambda value:
        f"{value:.6e}"
        if pd.notna(value)
        else "NA"
    )
)

publication_tests14["holm_adjusted_p"] = (
    publication_tests14[
        "holm_adjusted_p"
    ].map(
        lambda value:
        f"{value:.6e}"
        if pd.notna(value)
        else "NA"
    )
)

publication_tests14.to_csv(
    FINAL_TABLE_DIR
    / "final_statistical_significance.csv",
    index=False
)

print("\nStatistical significance table: SAVED")
print("Scientific p-value notation: APPLIED")

# ------------------------------------------------------------
# 18. Final privacy accounting table
#
# Preserve separate standalone and cumulative
# theoretical DP-only privacy budgets.
# ------------------------------------------------------------

with open(
    REPORT_DIR / "cell10_privacy_utility_report.json"
) as f:
    report10_14 = json.load(f)

with open(
    REPORT_DIR / "cell12_privacy_accounting.json"
) as f:
    report12_14 = json.load(f)

with open(
    REPORT_DIR / "cell13_statistical_analysis_report.json"
) as f:
    report13_14 = json.load(f)

privacy_summary14 = pd.DataFrame([

    {
        "stage": "Cell 09 reference model only",
        "epsilon": float(
            checkpoint09["retry_epsilon"]
        ),
        "delta": checkpoint09["delta"],
        "scope": "Standalone DP model training"
    },

    {
        "stage": "Cell 09 including failed attempt",
        "epsilon": float(
            checkpoint09["accounted_epsilon"]
        ),
        "delta": checkpoint09["delta"],
        "scope": "Cumulative DP-only releases"
    },

    {
        "stage": "Through Cell 10",
        "epsilon": float(
            report10_14["combined_epsilon"]
        ),
        "delta": 1e-5,
        "scope": "Cumulative DP-only releases"
    },

    {
        "stage": "Through Cell 12",
        "epsilon": float(
            report12_14[
                "cumulative_dp_only_epsilon"
            ]
        ),
        "delta": 1e-5,
        "scope": "Cumulative DP-only releases"
    },

    {
        "stage": "Through Cell 13",
        "epsilon": float(
            report13_14[
                "cumulative_dp_only_epsilon"
            ]
        ),
        "delta": 1e-5,
        "scope": "Cumulative DP-only releases"
    }

])

privacy_summary14.to_csv(
    FINAL_TABLE_DIR / "final_privacy_accounting.csv",
    index=False
)

print("\nFINAL PRIVACY ACCOUNTING")

print(
    privacy_summary14.round(8)
    .to_string(index=False)
)

print(
    "\nAdditional private training in Cell 14: NONE"
)

print(
    "No additional protected-training DP cost "
    "under the fixed-test-data assumption."
)

# ------------------------------------------------------------
# 19. Document hardware configuration
#
# Major Comment 7 reproducibility requirements.
# Capture observed current runtime hardware.
# ------------------------------------------------------------

def command_output14(command):

    try:
        result = subprocess.run(
            command,
            capture_output=True,
            text=True,
            check=False
        )

        if result.returncode == 0:
            return result.stdout.strip()

        return None

    except Exception:
        return None


cpu_model14 = None

try:

    with open("/proc/cpuinfo", "r") as f:

        for line in f:

            if line.lower().startswith(
                "model name"
            ):

                cpu_model14 = (
                    line.split(":", 1)[1].strip()
                )

                break

except Exception:
    pass

gpu_description14 = command_output14([

    "nvidia-smi",

    "--query-gpu=name,memory.total,driver_version",

    "--format=csv,noheader"

])

gpu_count_output14 = command_output14([

    "nvidia-smi",

    "--query-gpu=name",

    "--format=csv,noheader"

])

gpu_count14 = (
    len(gpu_count_output14.splitlines())
    if gpu_count_output14
    else None
)

hardware14 = {

    "platform": "Google Colab",

    "cpu_model": cpu_model14,

    "logical_cpu_count":
        psutil.cpu_count(logical=True),

    "physical_cpu_count":
        psutil.cpu_count(logical=False),

    "total_ram_gib":
        float(
            psutil.virtual_memory().total
            / 1024**3
        ),

    "gpu_count":
        gpu_count14,

    "gpu_description":
        gpu_description14,

    "operating_system":
        platform.platform(),

    "python_version":
        sys.version,

    "hardware_scope":
        "Observed Cell 14 runtime"

}

print("\n" + "=" * 75)
print("EXPERIMENTAL HARDWARE CONFIGURATION")
print("=" * 75)

for key, value in hardware14.items():

    print(f"{key}: {value}")

# ------------------------------------------------------------
# 20. Document software environment
# ------------------------------------------------------------

packages14 = [

    "numpy",

    "pandas",

    "scipy",

    "scikit-learn",

    "imbalanced-learn",

    "xgboost",

    "lightgbm",

    "catboost",

    "matplotlib",

    "joblib",

    "cryptography",

    "psutil"

]

software14 = {}

for package in packages14:

    try:

        software14[package] = (
            importlib.metadata.version(package)
        )

    except importlib.metadata.PackageNotFoundError:

        software14[package] = (
            "Not installed"
        )

print("\nSOFTWARE VERSIONS")

for package, version in software14.items():

    print(f"{package}: {version}")

# Full dependency snapshot.

pip_freeze14 = command_output14([

    sys.executable,

    "-m",

    "pip",

    "freeze"

])

if pip_freeze14:

    (
        REVISED_DIR / "requirements_full.txt"
    ).write_text(
        pip_freeze14 + "\n",
        encoding="utf-8"
    )

# Key package requirements.

key_requirements14 = [

    f"{package}=={version}"

    for package, version in software14.items()

    if version != "Not installed"
]

(
    REVISED_DIR / "requirements.txt"
).write_text(
    "\n".join(key_requirements14) + "\n",
    encoding="utf-8"
)

environment_report14 = {

    "hardware": hardware14,

    "software": software14,

    "original_seed": 42,

    "repeated_run_seeds": [
        42, 43, 44, 45, 46
    ],

    "privacy_randomness":

        "Independent OS-backed Gaussian randomness; "
        "noise seeds are not recorded or published",

    "shap_method":

        "Native XGBoost exact TreeSHAP",

    "external_dataset_used": False,

    "repository_url": None,

    "repository_status":

        "Local research code archived in Google Drive; "
        "public repository URL not yet supplied"

}

with open(
    REPORT_DIR / "cell14_experimental_environment.json",
    "w"
) as f:

    json.dump(
        environment_report14,
        f,
        indent=4
    )

print("\nHardware documentation: SAVED")
print("Software documentation: SAVED")
print("Dependency requirements: SAVED")

# ------------------------------------------------------------
# 21. Export SHAP manuscript-ready table
# ------------------------------------------------------------

shap_publication14 = (
    shap_importance14.head(15).copy()
)

shap_publication14 = shap_publication14.rename(

    columns={

        "rank": "Rank",

        "feature": "Predictor",

        "mean_absolute_shap":
            "Mean absolute SHAP",

        "mean_signed_shap":
            "Mean signed SHAP",

        "relative_importance_percent":
            "Relative importance (%)"

    }

)

shap_publication14.to_csv(
    FINAL_TABLE_DIR / "final_shap_feature_importance.csv",
    index=False
)

# ------------------------------------------------------------
# 22. Verify previous experimental results unchanged
# ------------------------------------------------------------

for path_string, expected_hash in (
    previous_hashes14.items()
):

    assert sha256_file(
        Path(path_string)
    ) == expected_hash

print("\nPrevious Cells 05-13 results: UNCHANGED")

# ------------------------------------------------------------
# 23. Verify original project source integrity
# ------------------------------------------------------------

with open(

    PROJECT_ROOT / "02_Revised_Project"

    / "reports"

    / "cell02_original_source_hashes.json"

) as f:

    source_hashes14 = json.load(f)

for item in source_hashes14:

    path = SOURCE_ROOT / item["file"]

    assert path.exists()

    assert sha256_file(path) == item["sha256"]

print("Original source integrity: PASS")

# ------------------------------------------------------------
# 24. Save consolidated manuscript results index
# ------------------------------------------------------------

final_tables14 = sorted(
    FINAL_TABLE_DIR.glob("*.csv")
)

final_figures14 = sorted(
    FIGURE_DIR.glob("*.png")
)

result_index14 = {

    "final_tables": [

        {
            "file": path.name,

            "path": str(path),

            "sha256": sha256_file(path)
        }

        for path in final_tables14

    ],

    "available_figures": [

        {
            "file": path.name,

            "path": str(path),

            "sha256": sha256_file(path)
        }

        for path in final_figures14

    ],

    "inference_ready_model": {

        "path": str(FINAL_MODEL_PATH),

        "sha256": sha256_file(
            FINAL_MODEL_PATH
        )

    },

    "preprocessing_model": str(

        DP_DIR
        / "cell09_public_preprocessing.joblib"

    )

}

with open(
    REPORT_DIR / "cell14_final_results_index.json",
    "w"
) as f:

    json.dump(
        result_index14,
        f,
        indent=4
    )

print(
    "\nFinal manuscript tables:",
    len(final_tables14)
)

print(
    "Available experimental figures:",
    len(final_figures14)
)

# ------------------------------------------------------------
# 25. Save complete SHAP verification report
# ------------------------------------------------------------

report14 = {

    "cell": 14,

    "status": "COMPLETED",

    "original_predictors": 42,

    "explained_model":

        "Seed-42 DP-enabled fixed-tree XGBoost "
        "complete framework",

    "model_retrained": False,

    "additional_private_training": False,

    "additional_protected_training_rho": 0.0,

    "cumulative_dp_only_epsilon": float(

        report13_14[
            "cumulative_dp_only_epsilon"
        ]

    ),

    "delta": 1e-5,

    "total_testing_records": 111643,

    "global_shap_sample_size":
        SHAP_SAMPLE_SIZE,

    "sample_seed": SEED14,

    "shap_method":

        "Native exact XGBoost TreeSHAP",

    "shap_output_scale":
        "Raw log-odds",

    "shap_additivity_verified": True,

    "maximum_shap_additivity_error":
        max_additivity_error,

    "mean_shap_additivity_error":
        mean_additivity_error,

    "maximum_model_probability_difference":
        max_probability_difference,

    "shap_computation_seconds":
        float(shap_seconds14),

    "top_15_features":

        shap_importance14.head(15)
        .to_dict(orient="records"),

    "local_example_probability":
        local_probability14,

    "five_run_tables_consolidated": True,

    "hardware_documented": True,

    "software_documented": True,

    "previous_results_modified": False,

    "original_source_modified": False,

    "external_dataset_used": False,

    "limitations": [

        "SHAP analysis explains the existing seed-42 "
        "reference model, not an average of five models.",

        "Global feature importance is estimated from "
        "2000 randomly selected held-out records.",

        "TreeSHAP uses reference-model tree-cover "
        "statistics and explains model contributions "
        "rather than causal relationships.",

        "The local explanation is illustrative and "
        "uses one correctly classified fraud-labelled case.",

        "Ordinal-encoded categorical values should "
        "not be interpreted as continuous clinical rankings.",

        "Provider overlap remains within the original "
        "claim-wise evaluation protocol.",

        "Saved explanation artifacts require separate "
        "review before public dissemination.",

        "The DP-only theoretical bound does not protect "
        "the joint release of non-private models.",

        "Hardware information describes the observed "
        "current runtime.",

        "A public repository URL remains to be supplied."

    ],

    "elapsed_seconds":

        time.perf_counter() - START_TIME

}

with open(
    REPORT_DIR / "cell14_shap_final_results_report.json",
    "w"
) as f:

    json.dump(
        report14,
        f,
        indent=4,
        default=str
    )

# ------------------------------------------------------------
# 26. Save complete Cell 14 source code
# ------------------------------------------------------------

source_saved14 = False

try:

    current_source14 = (
        get_ipython().user_ns["In"][-1]
    )

    if current_source14.strip():

        (
            CODE_DIR
            / "cell14_shap_final_results.py"
        ).write_text(

            current_source14,

            encoding="utf-8"

        )

        source_saved14 = True

except Exception as exc:

    print(
        "Source archival warning:",
        exc
    )

# ------------------------------------------------------------
# 27. Save completion checkpoint
# ------------------------------------------------------------

checkpoint14 = {

    "cell": 14,

    "status": "COMPLETED",

    "original_predictors": 42,

    "global_shap_completed": True,

    "local_shap_completed": True,

    "shap_additivity_verified": True,

    "shap_sample_records":
        SHAP_SAMPLE_SIZE,

    "inference_model_saved": True,

    "final_tables_consolidated": True,

    "hardware_documented": True,

    "software_documented": True,

    "additional_private_training": False,

    "additional_protected_training_rho": 0.0,

    "cumulative_dp_only_epsilon":

        float(
            report13_14[
                "cumulative_dp_only_epsilon"
            ]
        ),

    "delta": 1e-5,

    "previous_results_modified": False,

    "original_source_modified": False,

    "source_saved": source_saved14,

    "public_repository_url_available": False,

    "next_cell": 15

}

with open(
    CHECKPOINT_DIR / "cell14_completed.json",
    "w"
) as f:

    json.dump(
        checkpoint14,
        f,
        indent=4
    )

del (
    dtest14,
    dshap14,
    reloaded_model14
)

gc.collect()

print("\n" + "=" * 75)
print("CELL 14 COMPLETED SUCCESSFULLY")
print("=" * 75)

print("Complete framework model: VERIFIED")
print("Inference-ready DP model: SAVED")
print("Global SHAP explanations: COMPLETED")
print("Local SHAP explanation: COMPLETED")
print("SHAP additivity: PASS")
print("Original 42 predictors: RETAINED")
print("Five-run manuscript tables: CONSOLIDATED")
print("Blockchain statistical table: SAVED")
print("Significance testing table: SAVED")
print("Final privacy accounting: SAVED")
print("Hardware documentation: SAVED")
print("Software documentation: SAVED")
print("Dependency requirements: SAVED")
print("Additional private training: NONE")
print("Previous results: UNCHANGED")
print("Original source integrity: PASS")
print("Complete source code saved:", source_saved14)
print("Checkpoint saved.")

print(
    "\nTotal execution seconds:",
    round(time.perf_counter() - START_TIME, 2)
)

print(
    "\nREADY FOR CELL 15/15: "
    "FINAL CONSISTENCY AND REPRODUCIBILITY VERIFICATION"
)
```

```python
# ============================================================
# CELL 15/15
# FINAL CONSISTENCY AND REPRODUCIBILITY VERIFICATION
#
# FINAL RESEARCH AUDIT
# NO MODEL RETRAINING
# NO ADDITIONAL PRIVATE TRAINING
# NO EXPERIMENTAL RESULT MODIFICATION
# ============================================================

import os
import gc
import json
import time
import hashlib
import shutil
import zipfile
import platform
import subprocess

from pathlib import Path
from datetime import datetime, timezone

import numpy as np
import pandas as pd
import xgboost as xgb
import joblib

from scipy.stats import t, ttest_rel
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix
)

START_TIME = time.perf_counter()

print("=" * 75)
print("CELL 15: FINAL CONSISTENCY AND REPRODUCIBILITY AUDIT")
print("=" * 75)

# ------------------------------------------------------------
# 1. Restore final reviewer-only workspace
# ------------------------------------------------------------

PROJECT_ROOT = Path(
    "/content/drive/MyDrive/Narayan_Patil_Reviewer_Revision"
)

SOURCE_ROOT = PROJECT_ROOT / "01_Original_Project" / "SBN"

REVISED_DIR = PROJECT_ROOT / "03_Reviewer_Only_Revision"

CODE_DIR = REVISED_DIR / "code"
DATA_DIR = REVISED_DIR / "data"
MODEL_DIR = REVISED_DIR / "models"
RESULT_DIR = REVISED_DIR / "results"

TABLE_DIR = RESULT_DIR / "tables"
FIGURE_DIR = RESULT_DIR / "figures"
PREDICTION_DIR = RESULT_DIR / "predictions"

REPORT_DIR = REVISED_DIR / "reports"
CHECKPOINT_DIR = REVISED_DIR / "checkpoints"

DP_DIR = MODEL_DIR / "differential_privacy"

FINAL_TABLE_DIR = TABLE_DIR / "final_manuscript"

BACKUP_DIR = PROJECT_ROOT / "04_Final_Research_Backup"

BACKUP_DIR.mkdir(parents=True, exist_ok=True)

AUDIT_RECORDS = []
REPORTING_LIMITATIONS = []

print("\nWorkspace:", REVISED_DIR)
print("Backup directory:", BACKUP_DIR)

# ------------------------------------------------------------
# 2. General verification utilities
# ------------------------------------------------------------

def read_json(path):

    assert path.exists(), f"Missing file: {path}"

    with open(path, "r", encoding="utf-8") as f:
        return json.load(f)


def write_json(path, data):

    temporary = path.with_suffix(".tmp")

    with open(temporary, "w", encoding="utf-8") as f:
        json.dump(data, f, indent=4, default=str)

    os.replace(temporary, path)


def sha256_file(path):

    digest = hashlib.sha256()

    with open(path, "rb") as f:

        for chunk in iter(
            lambda: f.read(1024 * 1024), b""
        ):

            digest.update(chunk)

    return digest.hexdigest()


def audit_pass(requirement, details="Verified"):

    AUDIT_RECORDS.append({

        "requirement": requirement,

        "status": "PASS",

        "details": details

    })

    print(f"PASS | {requirement}")


def audit_note(requirement, details):

    AUDIT_RECORDS.append({

        "requirement": requirement,

        "status": "LIMITATION",

        "details": details

    })

    REPORTING_LIMITATIONS.append({

        "requirement": requirement,

        "details": details

    })

    print(f"NOTE | {requirement}: {details}")


def assert_close(actual, expected, name, atol=1e-8):

    assert np.isclose(
        float(actual),
        float(expected),
        atol=atol,
        rtol=1e-7
    ), (
        f"{name}: actual={actual}, expected={expected}"
    )


# ------------------------------------------------------------
# 3. Verify existing completion checkpoints
# ------------------------------------------------------------

print("\nCHECKPOINT VERIFICATION")

checkpoint_inventory = []

for path in sorted(CHECKPOINT_DIR.glob("*.json")):

    content = read_json(path)

    checkpoint_inventory.append({

        "file": path.name,

        "cell": content.get("cell"),

        "status": content.get("status"),

        "sha256": sha256_file(path)

    })

# Cells 05-14 are the authoritative reviewer-only
# experimental branch.

for cell in range(5, 15):

    path = CHECKPOINT_DIR / f"cell{cell:02d}_completed.json"

    content = read_json(path)

    assert content["status"] == "COMPLETED"

    print(f"Cell {cell:02d}: COMPLETED")

cp05 = read_json(CHECKPOINT_DIR / "cell05_completed.json")
cp07 = read_json(CHECKPOINT_DIR / "cell07_completed.json")
cp09 = read_json(CHECKPOINT_DIR / "cell09_completed.json")
cp10 = read_json(CHECKPOINT_DIR / "cell10_completed.json")
cp11 = read_json(CHECKPOINT_DIR / "cell11_completed.json")
cp12 = read_json(CHECKPOINT_DIR / "cell12_completed.json")
cp13 = read_json(CHECKPOINT_DIR / "cell13_completed.json")
cp14 = read_json(CHECKPOINT_DIR / "cell14_completed.json")

audit_pass(
    "Reviewer-only checkpoints",
    "All Cells 05-14 completed."
)

write_json(
    REPORT_DIR / "cell15_checkpoint_inventory.json",
    checkpoint_inventory
)

# ------------------------------------------------------------
# 4. Verify original source integrity
# ------------------------------------------------------------

print("\nORIGINAL SOURCE INTEGRITY")

original_source_hashes = read_json(

    PROJECT_ROOT
    / "02_Revised_Project"
    / "reports"
    / "cell02_original_source_hashes.json"

)

assert len(original_source_hashes) > 0

for item in original_source_hashes:

    path = SOURCE_ROOT / item["file"]

    assert path.exists(), f"Missing original source: {path}"

    assert sha256_file(path) == item["sha256"], (
        f"Original source changed: {path}"
    )

audit_pass(
    "Original source integrity",
    f"{len(original_source_hashes)} original files verified."
)

# ------------------------------------------------------------
# 5. Protect existing experimental outputs
# ------------------------------------------------------------

previous_files = [

    path for path in TABLE_DIR.rglob("*.csv")

    if not path.name.startswith("cell15")

]

previous_hashes15 = {

    str(path): sha256_file(path)

    for path in previous_files

}

print(
    "Existing result tables protected:",
    len(previous_hashes15)
)

# ------------------------------------------------------------
# 6. Verify dataset and split consistency
# ------------------------------------------------------------

print("\nDATASET AND PREPROCESSING AUDIT")

def read_dataset15(filename):

    with np.load(
        DATA_DIR / filename,
        allow_pickle=False
    ) as data:

        return {

            "shape": data["X"].shape,

            "y": data["y"].copy(),

            "row_id": data["row_id"].copy(),

            "feature_names":
                data["feature_names"].tolist()

        }


train15 = read_dataset15(
    "cell05_train_real.npz"
)

balanced15 = read_dataset15(
    "cell05_train_balanced.npz"
)

test15 = read_dataset15(
    "cell05_test.npz"
)

assert train15["shape"] == (446568, 42)
assert balanced15["shape"] == (552664, 42)
assert test15["shape"] == (111643, 42)

assert train15["feature_names"] == test15["feature_names"]

assert (
    train15["feature_names"]
    == balanced15["feature_names"]
)

assert len(np.unique(train15["row_id"])) == 446568

assert len(np.unique(test15["row_id"])) == 111643

assert set(train15["row_id"]).isdisjoint(
    test15["row_id"]
)

assert np.bincount(
    train15["y"],
    minlength=2
).tolist() == [276332, 170236]

assert np.bincount(
    test15["y"],
    minlength=2
).tolist() == [69083, 42560]

assert np.bincount(
    balanced15["y"],
    minlength=2
).tolist() == [276332, 276332]

with np.load(
    DATA_DIR / "cell05_train_balanced.npz",
    allow_pickle=False
) as data:

    synthetic_count = int(
        np.sum(data["is_synthetic"])
    )

assert synthetic_count == 106096

assert cp05["test_smote"] is False

dataset_summary15 = {

    "original_records": 558211,

    "original_predictors": 42,

    "real_training_records": 446568,

    "balanced_training_records": 552664,

    "synthetic_training_records": synthetic_count,

    "testing_records": 111643,

    "train_test_row_overlap": 0,

    "smote_applied_to_test": False,

    "training_class_counts": [276332, 170236],

    "testing_class_counts": [69083, 42560]

}

print(
    json.dumps(dataset_summary15, indent=4)
)

audit_pass(
    "Dataset dimensions and class counts"
)

audit_pass(
    "Training-testing row separation"
)

audit_pass(
    "Training-only SMOTE"
)

# Verify actual provider overlap.

split_manifest15 = pd.read_csv(

    DATA_DIR / "cell05_split_manifest.csv",

    usecols=["row_id", "Provider", "partition"],

    dtype={"Provider": str}

)

train_mask = split_manifest15["row_id"].isin(
    train15["row_id"]
)

test_mask = split_manifest15["row_id"].isin(
    test15["row_id"]
)

assert int(train_mask.sum()) == 446568
assert int(test_mask.sum()) == 111643

train_providers = set(

    split_manifest15.loc[
        train_mask, "Provider"
    ].dropna()

)

test_providers = set(

    split_manifest15.loc[
        test_mask, "Provider"
    ].dropna()

)

provider_overlap15 = len(
    train_providers & test_providers
)

dataset_summary15[
    "shared_providers"
] = provider_overlap15

print(
    "Shared training/testing providers:",
    provider_overlap15
)

audit_note(

    "Provider-level generalization",

    f"{provider_overlap15} providers overlap between "
    "training and testing. This is claim-wise evaluation, "
    "not unseen-provider validation."

)

write_json(

    REPORT_DIR / "cell15_dataset_verification.json",

    dataset_summary15

)

del split_manifest15
gc.collect()

# ------------------------------------------------------------
# 7. Verify saved baseline predictions
# ------------------------------------------------------------

print("\nCENTRALIZED PREDICTION AUDIT")

METRICS15 = [
    "accuracy",
    "precision",
    "recall",
    "f1",
    "roc_auc"
]


def calculate_metrics15(y, probability):

    probability = np.asarray(
        probability
    ).reshape(-1)

    prediction = (
        probability >= 0.5
    ).astype(np.int8)

    assert len(y) == len(probability)

    assert np.isfinite(probability).all()

    assert np.all(
        (probability >= 0)
        & (probability <= 1)
    )

    return {

        "accuracy": float(
            accuracy_score(y, prediction)
        ),

        "precision": float(
            precision_score(
                y, prediction, zero_division=0
            )
        ),

        "recall": float(
            recall_score(
                y, prediction, zero_division=0
            )
        ),

        "f1": float(
            f1_score(
                y, prediction, zero_division=0
            )
        ),

        "roc_auc": float(
            roc_auc_score(y, probability)
        )

    }


baseline_table15 = pd.read_csv(

    TABLE_DIR / "cell05_all_model_metrics.csv"

)

for model_name in [

    "LightGBM",
    "CatBoost",
    "XGBoost"

]:

    for partition, dataset in [

        ("train_real", train15),

        ("test", test15)

    ]:

        prediction_path = (

            PREDICTION_DIR

            / (
                f"cell05_{model_name.lower()}_"
                f"{partition}_seed42.npz"
            )

        )

        with np.load(
            prediction_path,
            allow_pickle=False
        ) as data:

            assert np.array_equal(

                data["row_id"],

                dataset["row_id"]

            )

            assert np.array_equal(

                data["y_true"],

                dataset["y"]

            )

            metrics = calculate_metrics15(

                data["y_true"],

                data["probability"]

            )

        stored = baseline_table15.loc[

            (baseline_table15["model"] == model_name)

            &

            (baseline_table15["partition"] == partition)

        ].iloc[0]

        for metric in METRICS15:

            assert_close(

                metrics[metric],

                stored[metric],

                f"{model_name}/{partition}/{metric}"

            )

        print(
            f"{model_name} | {partition}: VERIFIED"
        )

audit_pass(
    "Centralized prediction metrics"
)

# ------------------------------------------------------------
# 8. Verify federated learning results
# ------------------------------------------------------------

print("\nFEDERATED LEARNING AUDIT")

fl_summary15 = pd.read_csv(

    TABLE_DIR / "cell07_federated_summary.csv"

)

fl_rounds15 = pd.read_csv(

    TABLE_DIR / "cell07_federated_round_metrics.csv"

)

assert len(fl_rounds15) == 30

assert sorted(
    fl_summary15["n_clients"].tolist()
) == [3, 5, 10]

for clients in [3, 5, 10]:

    rounds = fl_rounds15.loc[
        fl_rounds15["n_clients"] == clients
    ]

    assert sorted(
        rounds["round"].tolist()
    ) == list(range(1, 11))

    stored = fl_summary15.loc[
        fl_summary15["n_clients"] == clients
    ].iloc[0]

    with np.load(

        PREDICTION_DIR

        / f"cell07_fl_{clients}_clients_seed42.npz",

        allow_pickle=False

    ) as data:

        assert np.array_equal(

            data["row_id"],

            test15["row_id"]

        )

        assert np.array_equal(

            data["y_true"],

            test15["y"]

        )

        metrics = calculate_metrics15(

            data["y_true"],

            data["probability"]

        )

    for metric in METRICS15:

        assert_close(

            metrics[metric],

            stored[metric],

            f"FL/{clients}/{metric}"

        )

    assert int(

        stored["total_ensemble_trees"]

    ) == clients * 400

    print(
        f"{clients} clients | 10 rounds: VERIFIED"
    )

audit_pass(
    "Federated configurations and prediction metrics"
)

audit_note(

    "Federated model capacity",

    "Centralized XGBoost contains 400 trees. "
    "The 3-, 5- and 10-client ensembles contain "
    "1200, 2000 and 4000 trees. Model capacity differs."

)

audit_note(

    "Federated execution environment",

    "Clients were simulated sequentially in one runtime. "
    "Communication measurements are serialized payload "
    "sizes rather than physical network traffic."

)

# ------------------------------------------------------------
# 9. Verify matched centralized comparison
# ------------------------------------------------------------

matched_table15 = pd.read_csv(

    TABLE_DIR

    / "cell08_centralized_federated_comparison.csv"

)

assert len(matched_table15) == 4

with np.load(

    PREDICTION_DIR

    / "cell08_centralized_real_test_seed42.npz",

    allow_pickle=False

) as data:

    assert np.array_equal(

        data["row_id"],

        test15["row_id"]

    )

    assert np.array_equal(

        data["y_true"],

        test15["y"]

    )

    central_metrics15 = calculate_metrics15(

        data["y_true"],

        data["probability"]

    )

central_stored15 = matched_table15.loc[

    matched_table15["configuration"]

    == "Centralized XGBoost"

].iloc[0]

for metric in METRICS15:

    assert_close(

        central_metrics15[metric],

        central_stored15[metric],

        f"Matched centralized/{metric}"

    )

audit_pass(
    "Matched centralized XGBoost comparison"
)

# ------------------------------------------------------------
# 10. Reconcile complete DP-only privacy accounting
# ------------------------------------------------------------

print("\nFINAL PRIVACY ACCOUNTING AUDIT")

privacy09 = read_json(

    REPORT_DIR / "cell09_privacy_verification.json"

)

privacy10 = read_json(

    REPORT_DIR / "cell10_privacy_utility_report.json"

)

privacy12 = read_json(

    REPORT_DIR / "cell12_privacy_accounting.json"

)

privacy13 = read_json(

    REPORT_DIR / "cell13_statistical_analysis_report.json"

)

delta15 = 1e-5

sensitivity15 = float(

    2 * np.sqrt(40 * (1 + 0.25**2))

)


def epsilon15(rho):

    return float(

        rho + 2 * np.sqrt(

            rho * np.log(1 / delta15)

        )

    )


rho09 = float(privacy09["rho_total"])

rho10 = float(privacy10["combined_rho"])

rho12 = float(

    privacy12["cumulative_dp_only_rho"]

)

rho13 = float(

    privacy13["cumulative_dp_only_rho"]

)

sigma15 = float(cp09["sigma"])

rho_reference15 = float(

    10 * sensitivity15**2

    / (2 * sigma15**2)

)

assert_close(

    epsilon15(rho_reference15),

    cp09["retry_epsilon"],

    "Reference standalone epsilon"

)

assert_close(

    epsilon15(rho09),

    1.0,

    "Cell 09 combined epsilon"

)

assert_close(

    rho10,

    rho09 + privacy10["additional_rho"],

    "Cell 10 cumulative rho"

)

assert_close(

    rho12,

    rho10

    + privacy12["additional_feature_ablation_rho"],

    "Cell 12 cumulative rho"

)

assert_close(

    rho13,

    rho12 + privacy13["additional_dp_only_rho"],

    "Cell 13 cumulative rho"

)

assert_close(

    epsilon15(rho13),

    cp13["cumulative_dp_only_epsilon"],

    "Final cumulative epsilon"

)

assert_close(

    cp14["additional_protected_training_rho"],

    0.0,

    "Cell 14 additional rho"

)

privacy_audit15 = pd.DataFrame([

    {

        "stage": "Reference model standalone",

        "rho": rho_reference15,

        "epsilon": epsilon15(rho_reference15),

        "delta": delta15

    },

    {

        "stage": "Through Cell 09",

        "rho": rho09,

        "epsilon": epsilon15(rho09),

        "delta": delta15

    },

    {

        "stage": "Through Cell 10",

        "rho": rho10,

        "epsilon": epsilon15(rho10),

        "delta": delta15

    },

    {

        "stage": "Through Cell 12",

        "rho": rho12,

        "epsilon": epsilon15(rho12),

        "delta": delta15

    },

    {

        "stage": "Through Cells 13-15",

        "rho": rho13,

        "epsilon": epsilon15(rho13),

        "delta": delta15

    }

])

privacy_audit15.to_csv(

    TABLE_DIR / "cell15_verified_privacy_accounting.csv",

    index=False

)

print(

    privacy_audit15.round(10)

    .to_string(index=False)

)

audit_pass(
    "Cumulative DP-only privacy accounting"
)

audit_note(

    "Privacy guarantee scope",

    "The theoretical bound is conditional on fixed "
    "reference data, feature schema, test data and "
    "protected claim-row membership. It does not cover "
    "patients, providers, non-private model releases, "
    "raw evidence selection or the complete pipeline."

)

# ------------------------------------------------------------
# 11. Verify complete framework reference predictions
# ------------------------------------------------------------

reference_dir15 = DP_DIR / cp09["dp_run_name"]

with np.load(

    reference_dir15 / "test_predictions.npz",

    allow_pickle=False

) as data:

    reference_probability15 = data[
        "probability"
    ].copy()

reference_metrics15 = calculate_metrics15(

    test15["y"],

    reference_probability15

)

stored_dp15 = pd.read_csv(

    TABLE_DIR / "cell09_dp_final_results.csv"

).iloc[0]

for metric in METRICS15:

    assert_close(

        reference_metrics15[metric],

        stored_dp15[metric],

        f"Reference DP/{metric}"

    )

audit_pass(
    "DP reference model metrics"
)

# ------------------------------------------------------------
# 12. Verify five ablation configurations
# ------------------------------------------------------------

print("\nCOMPONENT ABLATION AUDIT")

ablation15 = pd.read_csv(

    TABLE_DIR / "cell12_component_ablation.csv"

)

assert len(ablation15) == 5

ablation_names15 = {

    "Complete framework": "complete_framework",

    "Without Blockchain": "without_blockchain",

    "Without SHAP": "without_shap",

    "Without Differential Privacy": "without_dp",

    "Without Feature Engineering":
        "without_feature_engineering"

}

for configuration, filename in ablation_names15.items():

    stored = ablation15.loc[

        ablation15["configuration"] == configuration

    ].iloc[0]

    with np.load(

        PREDICTION_DIR

        / f"cell12_{filename}_predictions.npz",

        allow_pickle=False

    ) as data:

        assert np.array_equal(

            data["row_id"],

            test15["row_id"]

        )

        probability = data["probability"]

        metrics = calculate_metrics15(

            test15["y"],

            probability

        )

        if configuration in [

            "Complete framework",

            "Without Blockchain",

            "Without SHAP"

        ]:

            assert np.array_equal(

                probability,

                reference_probability15

            )

        for metric in METRICS15:

            assert_close(

                metrics[metric],

                stored[metric],

                f"Ablation/{configuration}/{metric}"

            )

    expected_features = (

        35 if configuration

        == "Without Feature Engineering"

        else 42

    )

    assert int(stored["features"]) == expected_features

    print(f"{configuration}: VERIFIED")

audit_pass(
    "Five component ablation configurations"
)

audit_pass(
    "Post-prediction metric invariance"
)

# ------------------------------------------------------------
# 13. Verify saved blockchain cryptographic integrity
#
# No new evidence selection or new blockchain training.
# Only verify existing saved transactions and signatures.
# ------------------------------------------------------------

print("\nBLOCKCHAIN INTEGRITY AUDIT")

from cryptography.hazmat.primitives.asymmetric.ed25519 import (
    Ed25519PublicKey
)

from cryptography.exceptions import InvalidSignature

BLOCKCHAIN_DIR15 = RESULT_DIR / "blockchain"

registry15 = read_json(

    BLOCKCHAIN_DIR15 / "validator_public_keys.json"

)

keys15 = {

    validator: Ed25519PublicKey.from_public_bytes(

        bytes.fromhex(key)

    )

    for validator, key in registry15.items()

}

assert len(keys15) == 3

validators15 = list(keys15)


def canonical15(obj):

    return json.dumps(

        obj,

        sort_keys=True,

        separators=(",", ":"),

        ensure_ascii=False,

        allow_nan=False

    ).encode("utf-8")


def merkle15(records):

    hashes = [

        hashlib.sha256(

            b"\x00" + canonical15(record)

        ).digest()

        for record in records

    ]

    assert len(hashes) > 0

    while len(hashes) > 1:

        if len(hashes) % 2:

            hashes.append(hashes[-1])

        hashes = [

            hashlib.sha256(

                b"\x01"

                + hashes[i]

                + hashes[i + 1]

            ).digest()

            for i in range(0, len(hashes), 2)

        ]

    return hashes[0].hex()


blockchain_table15 = pd.read_csv(

    TABLE_DIR / "cell11_blockchain_scalability.csv"

)

for scale in [1000, 10000, 100000]:

    chain_path = (

        BLOCKCHAIN_DIR15

        / f"cell11_chain_{scale}.jsonl"

    )

    previous_hash = "0" * 64

    block_count = 0

    transaction_count = 0

    transaction_ids = set()

    with open(

        chain_path,

        "r",

        encoding="utf-8"

    ) as f:

        for line in f:

            block = json.loads(line)

            header = block["header"]

            block_count += 1

            assert header["height"] == block_count

            assert (

                header["previous_hash"]

                == previous_hash

            )

            assert header["proposer"] == validators15[

                (block_count - 1) % 3

            ]

            records = block["transactions"]

            assert 0 < len(records) <= 100

            assert (

                header["transaction_count"]

                == len(records)

            )

            assert header["merkle_root"] == merkle15(

                records

            )

            expected_hash = hashlib.sha256(

                canonical15(header)

            ).hexdigest()

            assert expected_hash == block["block_hash"]

            signed_message = (

                b"POA_BLOCK_V1"

                + bytes.fromhex(expected_hash)

            )

            signers = set()

            for vote in block["signatures"]:

                validator = vote["validator"]

                if validator not in keys15:

                    continue

                try:

                    keys15[validator].verify(

                        bytes.fromhex(

                            vote["signature"]

                        ),

                        signed_message

                    )

                    signers.add(validator)

                except (

                    InvalidSignature,

                    ValueError

                ):

                    continue

            assert len(signers) >= 2

            assert header["proposer"] in signers

            for transaction in records:

                tx_id = transaction["transaction_id"]

                assert tx_id not in transaction_ids

                transaction_ids.add(tx_id)

                assert len(

                    bytes.fromhex(

                        transaction["evidence_commitment"]

                    )

                ) == 32

            transaction_count += len(records)

            previous_hash = expected_hash

    assert transaction_count == scale

    assert block_count == scale // 100

    stored = blockchain_table15.loc[

        blockchain_table15["fraud_records"] == scale

    ].iloc[0]

    assert int(stored["total_storage_bytes"]) == (

        chain_path.stat().st_size

    )

    assert_close(

        chain_path.stat().st_size / block_count,

        stored["mean_storage_bytes_per_block"],

        f"Blockchain storage/{scale}"

    )

    print(

        f"{scale:,} transactions"

        f" | {block_count:,} blocks"

        " | Hashes, Merkle roots and signatures: PASS"

    )

audit_pass(
    "Saved blockchain integrity at all scales"
)

audit_note(

    "Blockchain scalability scope",

    "Throughput and latency describe single-process "
    "PoA-style simulation. No physical network or "
    "production blockchain deployment was tested."

)

# ------------------------------------------------------------
# 14. Verify all repeated-run statistics
# ------------------------------------------------------------

print("\nFIVE-RUN STATISTICAL AUDIT")

runs15 = pd.read_csv(

    TABLE_DIR / "cell13_all_predictive_runs.csv"

)

block_runs15 = pd.read_csv(

    TABLE_DIR / "cell13_all_blockchain_runs.csv"

)

statistics15 = pd.read_csv(

    TABLE_DIR / "cell13_predictive_mean_sd_ci.csv"

)

block_statistics15 = pd.read_csv(

    TABLE_DIR / "cell13_blockchain_mean_sd_ci.csv"

)

assert len(runs15) == 75

assert len(block_runs15) == 15

assert not runs15.duplicated([

    "experiment",

    "configuration",

    "seed"

]).any()

assert not block_runs15.duplicated([

    "fraud_records",

    "seed"

]).any()

expected_seeds15 = [42, 43, 44, 45, 46]

for _, group in runs15.groupby([

    "experiment",

    "configuration"

]):

    assert sorted(

        group["seed"].tolist()

    ) == expected_seeds15

for _, group in block_runs15.groupby(

    "fraud_records"

):

    assert sorted(

        group["seed"].tolist()

    ) == expected_seeds15

assert len(statistics15) == 75

assert len(block_statistics15) == 18


def verify_summary15(values, stored, name):

    values = np.asarray(

        values,

        dtype=np.float64

    )

    assert len(values) == 5

    mean = np.mean(values)

    sd = np.std(values, ddof=1)

    half_width = (

        t.ppf(0.975, 4)

        * sd / np.sqrt(5)

    )

    assert_close(mean, stored["mean"], name + "/mean")

    assert_close(sd, stored["sd"], name + "/SD")

    assert_close(

        mean - half_width,

        stored["ci95_lower"],

        name + "/CI lower"

    )

    assert_close(

        mean + half_width,

        stored["ci95_upper"],

        name + "/CI upper"

    )


for _, row in statistics15.iterrows():

    subset = runs15.loc[

        (runs15["experiment"] == row["experiment"])

        &

        (

            runs15["configuration"]

            == row["configuration"]

        )

    ]

    verify_summary15(

        subset[row["metric"]].to_numpy(),

        row,

        row["configuration"] + "/" + row["metric"]

    )

for _, row in block_statistics15.iterrows():

    subset = block_runs15.loc[

        block_runs15["fraud_records"]

        == row["fraud_records"]

    ]

    verify_summary15(

        subset[row["metric"]].to_numpy(),

        row,

        str(row["fraud_records"]) + "/" + row["metric"]

    )

audit_pass(
    "Five runs per experimental configuration"
)

audit_pass(
    "Predictive mean, SD and confidence intervals"
)

audit_pass(
    "Blockchain mean, SD and confidence intervals"
)

audit_note(

    "Repeated-run inference scope",

    "Runs use a fixed dataset and test split. "
    "The no-DP refinement is deterministic under "
    "its fixed structures and therefore has zero "
    "observed SD. Statistical inference concerns "
    "algorithm variability, not independent populations."

)

# ------------------------------------------------------------
# 15. Independently verify significance tests
# ------------------------------------------------------------

print("\nSIGNIFICANCE TESTING AUDIT")

tests15 = pd.read_csv(

    TABLE_DIR / "cell13_statistical_significance_tests.csv"

)

assert len(tests15) == 18

recomputed_p = []

for _, row in tests15.iterrows():

    values = []

    for method in [

        row["method_a"],

        row["method_b"]

    ]:

        group = runs15.loc[

            (

                runs15["experiment"]

                == row["experiment"]

            )

            &

            (runs15["configuration"] == method)

        ].set_index("seed").loc[expected_seeds15]

        values.append(

            group[row["metric"]].to_numpy()

        )

    differences = values[0] - values[1]

    assert_close(

        differences.mean() * 100,

        row["mean_difference_pp"],

        "Paired mean difference"

    )

    if np.isclose(

        np.std(differences, ddof=1),

        0,

        atol=1e-15,

        rtol=0

    ):

        p_value = (

            1.0 if np.allclose(

                differences,

                0,

                atol=1e-15,

                rtol=0

            )

            else np.nan

        )

    else:

        result = ttest_rel(values[0], values[1])

        p_value = float(result.pvalue)

        assert_close(

            result.statistic,

            row["t_statistic"],

            "Paired t statistic"

        )

    if np.isnan(p_value):

        assert pd.isna(row["p_value"])

    else:

        assert np.isclose(

            p_value,

            row["p_value"],

            atol=1e-14,

            rtol=1e-7

        )

    recomputed_p.append(p_value)

# Independently reconstruct Holm correction.

p_values = np.asarray(recomputed_p)

valid = np.isfinite(p_values)

valid_p = p_values[valid]

order = np.argsort(valid_p)

sorted_p = valid_p[order]

adjusted_sorted = np.minimum(

    np.maximum.accumulate(

        (len(valid_p) - np.arange(len(valid_p)))

        * sorted_p

    ),

    1.0

)

adjusted = np.empty(len(valid_p))

adjusted[order] = adjusted_sorted

assert np.allclose(

    adjusted,

    tests15.loc[valid, "holm_adjusted_p"],

    atol=1e-14,

    rtol=1e-7

)

assert np.array_equal(

    adjusted < 0.05,

    tests15.loc[

        valid,

        "significant_at_0_05"

    ].to_numpy(dtype=bool)

)

audit_pass(
    "All 18 paired significance tests"
)

audit_pass(
    "Holm multiple-comparison correction"
)

# ------------------------------------------------------------
# 16. Verify saved SHAP and reconstructed model
# ------------------------------------------------------------

print("\nSHAP AND MODEL RECONSTRUCTION AUDIT")

shap_report15 = read_json(

    REPORT_DIR / "cell14_shap_final_results_report.json"

)

assert shap_report15[
    "shap_additivity_verified"
] is True

assert shap_report15[
    "global_shap_sample_size"
] == 2000

assert shap_report15[
    "original_predictors"
] == 42

assert shap_report15[
    "maximum_shap_additivity_error"
] < 1e-3

assert shap_report15[
    "maximum_model_probability_difference"
] < 2e-5

shap_values_path15 = (

    RESULT_DIR / "shap"

    / "cell14_global_shap_values.npz"

)

with np.load(

    shap_values_path15,

    allow_pickle=False

) as data:

    shap_values15 = data["shap_values"].copy()

    shap_features15 = data[
        "feature_names"
    ].tolist()

    shap_ids15 = data["row_id"].copy()

    shap_base15 = data["base_values"].copy()

    shap_probability15 = data["probability"].copy()

assert shap_values15.shape == (2000, 42)

assert np.isfinite(shap_values15).all()

assert len(np.unique(shap_ids15)) == 2000

assert set(shap_ids15).issubset(

    test15["row_id"]

)

assert shap_features15 == test15["feature_names"]

from scipy.special import expit

assert np.allclose(

    expit(

        shap_base15 + shap_values15.sum(axis=1)

    ),

    shap_probability15,

    atol=1e-4,

    rtol=1e-4

)

shap_importance15 = pd.read_csv(

    TABLE_DIR / "cell14_global_shap_importance.csv"

)

assert len(shap_importance15) == 42

calculated_importance15 = np.mean(

    np.abs(shap_values15),

    axis=0

)

for feature, value in zip(

    shap_features15,

    calculated_importance15

):

    stored = shap_importance15.loc[

        shap_importance15["feature"] == feature,

        "mean_absolute_shap"

    ].iloc[0]

    assert_close(

        value,

        stored,

        "SHAP importance/" + feature

    )

audit_pass(
    "SHAP matrix and numerical additivity"
)

audit_pass(
    "All 42 SHAP importance values"
)

# Load saved model without retraining.

final_model_path15 = (

    MODEL_DIR / "cell14_complete_framework_dp.ubj"

)

final_model15 = xgb.Booster()

final_model15.load_model(

    final_model_path15

)

assert final_model15.num_boosted_rounds() == 400

assert final_model15.feature_names == test15["feature_names"]

# Reconstruct the held-out feature representation
# using the saved reference-only preprocessing.

preprocessor15 = joblib.load(

    DP_DIR / "cell09_public_preprocessing.joblib"

)

feature_names15 = test15["feature_names"]

assert preprocessor15["feature_names"] == feature_names15

featured_path15 = (

    SOURCE_ROOT

    / "data"

    / "processed"

    / "featured_dataset.csv"

)

source_data15 = pd.read_csv(

    featured_path15,

    dtype=str,

    low_memory=False

)

test_frame15 = source_data15.iloc[

    test15["row_id"]

]

X_final15 = np.empty(

    (111643, 42),

    dtype=np.float32

)

for column, index in zip(

    preprocessor15["numerical_columns"],

    preprocessor15["numeric_indices"]

):

    X_final15[:, index] = (

        pd.to_numeric(

            test_frame15[column],

            errors="coerce"

        )

        .replace([np.inf, -np.inf], np.nan)

        .fillna(0)

        .to_numpy(dtype=np.float32)

    )

if preprocessor15["categorical_columns"]:

    X_final15[:, preprocessor15["category_indices"]] = (

        preprocessor15["encoder"].transform(

            test_frame15[

                preprocessor15["categorical_columns"]

            ]

            .fillna("Unknown")

            .astype(str)

        )

    )

X_final15 = preprocessor15["scaler"].transform(

    X_final15

).astype(np.float32)

assert np.isfinite(X_final15).all()

assert np.array_equal(

    test_frame15["PotentialFraud"].astype(

        np.int8

    ).to_numpy(),

    test15["y"]

)

final_model15.set_param({

    "device": "cpu",

    "nthread": 2

})

final_matrix15 = xgb.DMatrix(

    X_final15,

    feature_names=feature_names15

)

restored_probability15 = final_model15.predict(

    final_matrix15

)

assert np.allclose(

    restored_probability15,

    reference_probability15,

    atol=2e-5,

    rtol=2e-5

)

restored_difference15 = float(

    np.max(

        np.abs(

            restored_probability15

            - reference_probability15

        )

    )

)

print(

    "Reloaded model maximum probability difference:",

    restored_difference15

)

audit_pass(
    "Independent saved-model inference verification"
)

del (

    source_data15,

    test_frame15,

    X_final15,

    final_matrix15,

    final_model15

)

gc.collect()

provider_importance15 = float(

    shap_importance15.loc[

        shap_importance15["feature"] == "Provider",

        "relative_importance_percent"

    ].iloc[0]

)

audit_note(

    "Provider feature reliance",

    f"Provider accounts for {provider_importance15:.4f}% "
    "of normalized mean absolute SHAP importance. "
    "Interpretation is conditional on the claim-wise split."

)

# ------------------------------------------------------------
# 17. Verify final manuscript files and hardware report
# ------------------------------------------------------------

print("\nFINAL MANUSCRIPT ARTIFACTS")

result_index15 = read_json(

    REPORT_DIR / "cell14_final_results_index.json"

)

for category in [

    "final_tables",

    "available_figures"

]:

    for item in result_index15[category]:

        path = Path(item["path"])

        assert path.exists()

        assert sha256_file(path) == item["sha256"]

assert sha256_file(final_model_path15) == (

    result_index15["inference_ready_model"]["sha256"]

)

final_tables15 = sorted(

    FINAL_TABLE_DIR.glob("*.csv")

)

final_figures15 = sorted(

    FIGURE_DIR.glob("*.png")

)

assert len(final_tables15) == 9

assert len(final_figures15) >= 14

environment15 = read_json(

    REPORT_DIR / "cell14_experimental_environment.json"

)

assert environment15["hardware"]["gpu_description"]

assert environment15["hardware"]["python_version"]

assert environment15["software"]["xgboost"]

assert (

    REVISED_DIR / "requirements.txt"

).exists()

audit_pass(
    "Final manuscript tables and figures"
)

audit_pass(
    "Hardware and software documentation"
)

if not environment15.get("repository_url"):

    audit_note(

        "Public code repository",

        "Code is archived locally in Google Drive. "
        "A verified public repository URL is not "
        "available and must not be fabricated."

    )

print("Final manuscript tables:", len(final_tables15))

print("Available figures:", len(final_figures15))

# ------------------------------------------------------------
# 18. Final scientific reporting values
# ------------------------------------------------------------

print("\nFINAL MANUSCRIPT REPORTING VALUES")

full_runs15 = runs15.loc[

    (runs15["experiment"] == "Component ablation")

    &

    (runs15["configuration"] == "Complete framework")

]

assert len(full_runs15) == 5

final_reporting15 = {

    "model": "DP-enabled complete framework",

    "runs": 5,

    "seeds": expected_seeds15,

    "predictors": 42,

    "test_records": 111643,

    "metrics": {},

    "standalone_reference_epsilon":

        epsilon15(rho_reference15),

    "cell09_combined_epsilon":

        epsilon15(rho09),

    "final_cumulative_dp_only_epsilon":

        epsilon15(rho13),

    "delta": delta15,

    "whole_framework_dp_claim": False,

    "provider_overlap": provider_overlap15,

    "provider_shap_importance_percent":

        provider_importance15

}

for metric in METRICS15:

    values = full_runs15[metric].to_numpy()

    final_reporting15["metrics"][metric] = {

        "mean": float(np.mean(values)),

        "sd": float(np.std(values, ddof=1))

    }

    if metric == "roc_auc":

        print(

            f"{metric:12s}: "

            f"{np.mean(values):.6f} "

            f"+/- {np.std(values, ddof=1):.6f}"

        )

    else:

        print(

            f"{metric:12s}: "

            f"{100*np.mean(values):.2f}% "

            f"+/- {100*np.std(values, ddof=1):.2f}%"

        )

write_json(

    REPORT_DIR / "cell15_final_reporting_values.json",

    final_reporting15

)

print(

    "\nFinal cumulative DP-only epsilon:",

    epsilon15(rho13)

)

# ------------------------------------------------------------
# 19. Reviewer coding completion matrix
#
# Separate experimental completion from remaining
# manuscript-level requirements.
# ------------------------------------------------------------

reviewer_matrix15 = pd.DataFrame([

    {

        "comment": "Major 1",

        "requirement": "Federated learning experiments",

        "cells": "07, 08, 13",

        "coding_status": "Completed",

        "remaining": "Revise manuscript description"

    },

    {

        "comment": "Major 2",

        "requirement": "Blockchain scalability",

        "cells": "11, 13",

        "coding_status": "Completed",

        "remaining": "Report simulation scope"

    },

    {

        "comment": "Major 3",

        "requirement": "Abstract numerical consistency",

        "cells": "14, 15",

        "coding_status": "Verified results available",

        "remaining": "Correct manuscript values"

    },

    {

        "comment": "Major 4",

        "requirement": "Component ablation",

        "cells": "12, 13",

        "coding_status": "Completed",

        "remaining": "Add ablation results and limitations"

    },

    {

        "comment": "Major 5",

        "requirement": "Fair baseline comparison",

        "cells": "06, 08, 13",

        "coding_status": "Internal comparison completed",

        "remaining":

            "Remove unsupported external overfitting claim; "

            "discuss architectural differences"

    },

    {

        "comment": "Major 6",

        "requirement": "Repeated runs and statistics",

        "cells": "13, 15",

        "coding_status": "Completed for core experiments",

        "remaining": "Report fixed-split inference scope"

    },

    {

        "comment": "Major 7",

        "requirement": "Reproducibility",

        "cells": "14, 15",

        "coding_status": "Local documentation completed",

        "remaining": "Verified public repository URL"

    },

    {

        "comment": "Major 8",

        "requirement": "Federated aggregation",

        "cells": "07, 09",

        "coding_status": "Implemented",

        "remaining": "Correct mathematical formulation"

    },

    {

        "comment": "Major 9",

        "requirement": "Differential privacy",

        "cells": "09, 10, 12, 13, 15",

        "coding_status": "Conditional DP accounting completed",

        "remaining": "Correct privacy scope and budget claims"

    },

    {

        "comment": "Major 10",

        "requirement": "Preprocessing and data partition",

        "cells": "05, 15",

        "coding_status": "Required preprocessing implemented",

        "remaining": "Disclose provider overlap and encoding limits"

    },

    {

        "comment": "Minor 26",

        "requirement": "XGBoost hyperparameters",

        "cells": "05, 07, 09",

        "coding_status": "Documented",

        "remaining": "Update manuscript"

    },

    {

        "comment": "Minor 27",

        "requirement": "Feature selection clarification",

        "cells": "05, 12, 14",

        "coding_status": "Feature inventory documented",

        "remaining": "Clarify retained and removed features"

    },

    {

        "comment": "Minor 29",

        "requirement": "Consensus and Merkle root",

        "cells": "11, 15",

        "coding_status": "Completed",

        "remaining": "Update algorithm and methodology"

    }

])

reviewer_matrix15.to_csv(

    REPORT_DIR / "cell15_reviewer_completion_matrix.csv",

    index=False

)

print("\nREVIEWER CODING COMPLETION MATRIX")

print(

    reviewer_matrix15[

        ["comment", "requirement", "coding_status"]

    ].to_string(index=False)

)

# ------------------------------------------------------------
# 20. Verify existing results remain unchanged
# ------------------------------------------------------------

for path_string, original_hash in (

    previous_hashes15.items()

):

    assert sha256_file(

        Path(path_string)

    ) == original_hash

audit_pass(
    "Previous experimental tables unchanged"
)

# ------------------------------------------------------------
# 21. Save complete final audit report
# ------------------------------------------------------------

audit_df15 = pd.DataFrame(AUDIT_RECORDS)

audit_df15.to_csv(

    REPORT_DIR / "cell15_final_verification.csv",

    index=False

)

final_report15 = {

    "cell": 15,

    "status": "AUDIT_PASSED",

    "completed_utc":

        datetime.now(timezone.utc).isoformat(),

    "workspace": str(REVISED_DIR),

    "original_source_integrity": True,

    "previous_experimental_tables_unchanged": True,

    "original_predictors": 42,

    "dataset_verification": dataset_summary15,

    "predictive_run_records": len(runs15),

    "blockchain_run_records": len(block_runs15),

    "runs_per_configuration": 5,

    "statistical_tests": len(tests15),

    "final_manuscript_tables": len(final_tables15),

    "available_figures": len(final_figures15),

    "shap_additivity_verified": True,

    "saved_model_verified": True,

    "maximum_restored_probability_difference":

        restored_difference15,

    "final_reporting_values": final_reporting15,

    "audit_checks": AUDIT_RECORDS,

    "reporting_limitations": REPORTING_LIMITATIONS,

    "additional_model_training": False,

    "additional_private_training": False,

    "additional_privacy_rho": 0.0,

    "public_repository_url":

        environment15.get("repository_url"),

    "manuscript_revision_completed": False,

    "elapsed_seconds_before_backup":

        time.perf_counter() - START_TIME

}

write_json(

    REPORT_DIR / "cell15_final_research_audit.json",

    final_report15

)

# ------------------------------------------------------------
# 22. Prepare reproducibility README
# ------------------------------------------------------------

readme15 = """
# Health Insurance Fraud Detection
## Reviewer-only revised experimental project

### Authoritative workspace

03_Reviewer_Only_Revision

The original source project is preserved separately.
This revised branch contains reviewer-requested corrections.

### Execution order

Use the archived source scripts and original notebook
in the documented sequence.

Cells 05-14 contain completed experimental results.
Cell 15 provides final numerical and artifact verification.

Do not blindly rerun private training cells.

Additional private model releases require further
privacy accounting.

### Experimental results

Use the nine CSV files located in:

results/tables/final_manuscript/

Single-run results remain archived in the original
cell-specific tables.

### Model inference

Final reference model:

models/cell14_complete_framework_dp.ubj

Corresponding preprocessing:

models/differential_privacy/cell09_public_preprocessing.joblib

### Scientific limitations

The evaluation uses a claim-wise split with shared providers.

Federated learning uses simulated horizontal clients.

Blockchain measurements describe local PoA-style simulation.

The theoretical DP budget applies to the designated
protected claim records under fixed-data assumptions.

The reported DP-only budget excludes non-private releases.

No end-to-end framework privacy guarantee is established.

Five-run confidence intervals characterize stochastic
variability on the fixed experimental dataset.

### Data and code availability

Research scripts and results are archived in Google Drive.

A public repository URL has not yet been verified.

Do not publish raw claims, identifiers, per-record
predictions, non-private trained models, or restricted
explanation artifacts without authorization and review.

### Backup contents

The compact internal backup contains source scripts,
reports, checkpoints, statistical tables, figures,
dependency documentation, a reference inference model,
its preprocessing object, and a complete workspace
file manifest.

Large datasets and other model artifacts remain stored
in the authoritative Google Drive workspace.

This backup is not a public-release package.
"""

(
    REVISED_DIR / "README_REPRODUCIBILITY.md"
).write_text(

    readme15.strip() + "\n",

    encoding="utf-8"

)

# ------------------------------------------------------------
# 23. Save complete Cell 15 source
# ------------------------------------------------------------

source_saved15 = False

try:

    current_source15 = (

        get_ipython().user_ns["In"][-1]

    )

    if current_source15.strip():

        (

            CODE_DIR / "cell15_final_verification.py"

        ).write_text(

            current_source15,

            encoding="utf-8"

        )

        source_saved15 = True

except Exception as exc:

    print("Source archival warning:", exc)

print("Complete Cell 15 source saved:", source_saved15)

# ------------------------------------------------------------
# 24. Save preliminary checkpoint
# ------------------------------------------------------------

checkpoint15 = {

    "cell": 15,

    "status": "AUDIT_PASSED_BACKUP_PENDING",

    "reviewer_only_experimental_audit": True,

    "previous_results_modified": False,

    "original_source_modified": False,

    "additional_training": False,

    "additional_privacy_rho": 0.0,

    "final_cumulative_dp_only_epsilon":

        epsilon15(rho13),

    "delta": delta15,

    "source_saved": source_saved15,

    "repository_url_available":

        bool(environment15.get("repository_url")),

    "next_stage": "Manuscript and reviewer response revision"

}

write_json(

    CHECKPOINT_DIR / "cell15_completed.json",

    checkpoint15

)

# ------------------------------------------------------------
# 25. Generate complete workspace file manifest
#
# Every existing file is inventoried and hashed.
# No large dataset or model files are deleted.
# ------------------------------------------------------------

print("\nCOMPLETE WORKSPACE FILE INVENTORY")

manifest_path15 = (

    BACKUP_DIR / "final_workspace_sha256_manifest.csv"

)

manifest_records15 = []

for path in sorted(REVISED_DIR.rglob("*")):

    if not path.is_file():

        continue

    manifest_records15.append({

        "relative_path":

            str(path.relative_to(REVISED_DIR)),

        "size_bytes":

            path.stat().st_size,

        "sha256":

            sha256_file(path)

    })

manifest15 = pd.DataFrame(manifest_records15)

manifest15.to_csv(

    manifest_path15,

    index=False

)

print("Total archived workspace files:", len(manifest15))

print(

    "Workspace size (GiB):",

    round(

        manifest15["size_bytes"].sum() / 1024**3,

        4

    )

)

# ------------------------------------------------------------
# 26. Create compact internal research backup
#
# Include reproducibility essentials.
# Large datasets and repeated models remain in Drive.
# This archive is NOT intended for public upload.
# ------------------------------------------------------------

timestamp15 = datetime.now(

    timezone.utc

).strftime("%Y%m%dT%H%M%SZ")

backup_path15 = (

    BACKUP_DIR

    / f"Reviewer_Revision_Internal_Backup_{timestamp15}.zip"

)

backup_files15 = set()

for folder in [

    CODE_DIR,

    REPORT_DIR,

    CHECKPOINT_DIR,

    TABLE_DIR,

    FIGURE_DIR

]:

    backup_files15.update(

        path for path in folder.rglob("*")

        if path.is_file()

    )

for filename in [

    "requirements.txt",

    "requirements_full.txt",

    "README_REPRODUCIBILITY.md"

]:

    path = REVISED_DIR / filename

    if path.exists():

        backup_files15.add(path)

backup_files15.add(final_model_path15)

backup_files15.add(

    DP_DIR / "cell09_public_preprocessing.joblib"

)

assert all(

    path.exists()

    for path in backup_files15

)

# Finalize checkpoint before archiving it.

checkpoint15["status"] = "COMPLETED"

checkpoint15["backup_path"] = str(backup_path15)

checkpoint15["manifest_path"] = str(manifest_path15)

write_json(

    CHECKPOINT_DIR / "cell15_completed.json",

    checkpoint15

)

# Refresh checkpoint entry in the manifest.

checkpoint_path15 = (

    CHECKPOINT_DIR / "cell15_completed.json"

)

checkpoint_relative15 = str(

    checkpoint_path15.relative_to(REVISED_DIR)

)

mask15 = (

    manifest15["relative_path"]

    == checkpoint_relative15

)

assert mask15.sum() == 1

manifest15.loc[mask15, "size_bytes"] = (

    checkpoint_path15.stat().st_size

)

manifest15.loc[mask15, "sha256"] = (

    sha256_file(checkpoint_path15)

)

manifest15.to_csv(

    manifest_path15,

    index=False

)

temporary_backup15 = backup_path15.with_suffix(".tmp")

try:

    with zipfile.ZipFile(

        temporary_backup15,

        mode="w",

        compression=zipfile.ZIP_DEFLATED,

        compresslevel=6,

        allowZip64=True

    ) as archive:

        for path in sorted(backup_files15):

            archive.write(

                path,

                arcname=str(

                    path.relative_to(REVISED_DIR)

                )

            )

        archive.write(

            manifest_path15,

            arcname="final_workspace_sha256_manifest.csv"

        )

    with zipfile.ZipFile(

        temporary_backup15,

        mode="r"

    ) as archive:

        assert archive.testzip() is None

        assert len(archive.namelist()) == (

            len(backup_files15) + 1

        )

    os.replace(

        temporary_backup15,

        backup_path15

    )

except Exception:

    checkpoint15["status"] = (

        "AUDIT_PASSED_BACKUP_FAILED"

    )

    write_json(

        CHECKPOINT_DIR / "cell15_completed.json",

        checkpoint15

    )

    raise

backup_hash15 = sha256_file(

    backup_path15

)

backup_info15 = {

    "backup_path":

        str(backup_path15),

    "backup_sha256":

        backup_hash15,

    "backup_size_bytes":

        backup_path15.stat().st_size,

    "backup_files":

        len(backup_files15) + 1,

    "complete_workspace_inventory":

        str(manifest_path15),

    "workspace_file_count":

        len(manifest15),

    "zip_integrity_verified":

        True,

    "large_datasets_and_repeated_models":

        "Preserved separately in authoritative workspace",

    "public_release_package":

        False,

    "elapsed_seconds":

        time.perf_counter() - START_TIME

}

write_json(

    BACKUP_DIR / "final_backup_verification.json",

    backup_info15

)

print("\nCOMPACT INTERNAL RESEARCH BACKUP")

print("Backup path:", backup_path15)

print(

    "Backup size (MiB):",

    round(backup_path15.stat().st_size / 1024**2, 2)

)

print("ZIP integrity: PASS")

print("SHA-256:", backup_hash15)

print("Complete workspace manifest: SAVED")

print("Large datasets and all models: PRESERVED IN DRIVE")

# ------------------------------------------------------------
# 27. Final completion summary
# ------------------------------------------------------------

gc.collect()

print("\n" + "=" * 75)
print("CELL 15 COMPLETED SUCCESSFULLY")
print("=" * 75)

print("Reviewer-only experimental checkpoints: PASS")
print("Original source integrity: PASS")
print("Dataset consistency: PASS")
print("Centralized prediction metrics: PASS")
print("Federated prediction metrics: PASS")
print("Differential privacy accounting: PASS")
print("Component ablation verification: PASS")
print("Blockchain hashes and signatures: PASS")
print("Five-run statistical verification: PASS")
print("All 18 significance tests: PASS")
print("Holm correction: PASS")
print("SHAP numerical verification: PASS")
print("Reloaded final model inference: PASS")
print("Original experimental tables: UNCHANGED")
print("Original 42 predictors: RETAINED")
print("Additional model training: NONE")
print("Additional private training: NONE")
print("Hyperparameter tuning: NONE")
print("External dataset: NONE")
print("Final manuscript tables: 9")

print(

    "Available experimental figures:",

    len(final_figures15)

)

print("Hardware and software documentation: SAVED")
print("Reviewer completion matrix: SAVED")
print("Final numerical reporting values: SAVED")
print("Complete workspace SHA-256 manifest: SAVED")
print("Internal research backup: VERIFIED")
print("Public repository URL: PENDING")
print("Manuscript revision: PENDING")

print(

    "\nTotal execution seconds:",

    round(time.perf_counter() - START_TIME, 2)

)

print("\n" + "=" * 75)
print("REVIEWER-REQUESTED EXPERIMENTAL CODING STAGE COMPLETE")
print("READY FOR MANUSCRIPT REVISION AND RESPONSE SHEET")
print("=" * 75)
```

```python
import numpy as np
from pathlib import Path
from sklearn.metrics import confusion_matrix, roc_auc_score

# Load uploaded files
pred = np.load(
    "/content/cell12_complete_framework_predictions.npz",
    allow_pickle=False
)

test = np.load(
    "/content/cell05_xgboost_test_seed42.npz",
    allow_pickle=False
)

print("=" * 60)
print("PREDICTION FILE INSPECTION")
print("=" * 60)

for key in pred.files:
    print(key, "| Shape:", pred[key].shape,
          "| Type:", pred[key].dtype)

print("\n" + "=" * 60)
print("TEST FILE INSPECTION")
print("=" * 60)

for key in test.files:
    print(key, "| Shape:", test[key].shape,
          "| Type:", test[key].dtype)

# Identify actual labels
label_keys = [
    "y_true", "y_test", "actual", "target",
    "label", "labels", "PotentialFraud", "true_label"
]

label_key = next(
    (k for k in label_keys if k in test.files),
    None
)

if label_key is None:
    print("\nSTATUS: Actual label array not identified.")

elif "row_id" not in test.files:
    print("\nSTATUS: Test row IDs not available.")
    print("Row alignment must be verified.")

else:
    test_ids = test["row_id"].ravel()
    pred_ids = pred["row_id"].ravel()
    labels = test[label_key].ravel()
    y_prob = pred["probability"].ravel()

    assert len(test_ids) == len(labels)
    assert len(pred_ids) == len(y_prob) == 111643
    assert len(np.unique(test_ids)) == len(test_ids)
    assert len(np.unique(pred_ids)) == len(pred_ids)
    assert set(np.unique(labels)) == {0, 1}
    assert np.isfinite(y_prob).all()
    assert ((y_prob >= 0) & (y_prob <= 1)).all()

    label_map = dict(zip(test_ids, labels))

    assert set(pred_ids) == set(test_ids), \
        "Test row IDs do not match."

    y_true = np.array(
        [label_map[rid] for rid in pred_ids],
        dtype=int
    )

    y_pred = (y_prob >= 0.5).astype(int)

    cm = confusion_matrix(y_true, y_pred)
    auc = roc_auc_score(y_true, y_prob)

    print("\n" + "=" * 60)
    print("VERIFICATION RESULTS")
    print("=" * 60)

    print("Matched records:", len(y_true))
    print("Non-fraud records:", np.sum(y_true == 0))
    print("Fraud-labelled records:", np.sum(y_true == 1))
    print("Confusion matrix:\n", cm)
    print("ROC-AUC:", f"{auc:.6f}")

    expected_cm = np.array([
        [64982, 4101],
        [8576, 33984]
    ])

    cm_pass = np.array_equal(cm, expected_cm)
    auc_pass = abs(auc - 0.945909) < 0.0000005

    print("\nConfusion matrix match:", cm_pass)
    print("ROC-AUC match:", auc_pass)

    if cm_pass and auc_pass:
        np.savez_compressed(
            "/content/verified_figure_predictions.npz",
            y_true=y_true,
            y_prob=y_prob,
            y_pred=y_pred,
            row_id=pred_ids
        )

        print("\nFINAL STATUS: PASS")
        print("Verified predictions ready for figures.")

    else:
        print("\nFINAL STATUS: MISMATCH")
        print("Do not generate manuscript figures yet.")
```

```python
# ============================================================
# FINAL MANUSCRIPT FIGURES 3, 4 AND 7
# Verified predictions | 300 DPI | Direct ZIP download
# No Drive connection | No retraining
# ============================================================

import numpy as np
import matplotlib.pyplot as plt
import zipfile

from pathlib import Path
from sklearn.metrics import (
    roc_curve,
    roc_auc_score,
    precision_recall_curve,
    average_precision_score,
    confusion_matrix
)
from scipy.stats import gaussian_kde
from google.colab import files

# ------------------------------------------------------------
# 1. Load verified predictions
# ------------------------------------------------------------

data = np.load(
    "/content/verified_figure_predictions.npz",
    allow_pickle=False
)

y_true = data["y_true"]
y_prob = data["y_prob"]
y_pred = data["y_pred"]

auc = roc_auc_score(y_true, y_prob)

assert len(y_true) == 111643
assert abs(auc - 0.945909) < 0.0000005

assert np.array_equal(
    confusion_matrix(y_true, y_pred),
    np.array([[64982, 4101], [8576, 33984]])
)

print("Prediction verification: PASS")
print(f"ROC-AUC: {auc:.6f}")

# ------------------------------------------------------------
# 2. Figure settings
# ------------------------------------------------------------

OUT = Path("/content/Final_Manuscript_Figures")
OUT.mkdir(exist_ok=True)

plt.rcParams.update({
    "font.family": "DejaVu Serif",
    "font.size": 14,
    "axes.labelsize": 15,
    "xtick.labelsize": 13,
    "ytick.labelsize": 13,
    "legend.fontsize": 12,
    "axes.linewidth": 1.2,
    "figure.dpi": 120,
    "savefig.dpi": 300,
    "figure.facecolor": "white",
    "axes.facecolor": "white"
})

# ------------------------------------------------------------
# FIGURE 3: ROC Curve
# ------------------------------------------------------------

fpr, tpr, _ = roc_curve(y_true, y_prob)

fig, ax = plt.subplots(figsize=(7, 6))

ax.plot(
    fpr, tpr,
    color="#174A7E",
    linewidth=2.5,
    label=f"ROC-AUC = {auc:.6f}"
)

ax.plot(
    [0, 1], [0, 1],
    color="gray",
    linestyle="--",
    linewidth=1.5,
    label="Random classifier"
)

ax.set_xlabel("False Positive Rate")
ax.set_ylabel("True Positive Rate")

ax.set_xlim(0, 1)
ax.set_ylim(0, 1.02)

ax.legend(loc="lower right")
ax.grid(alpha=0.20)

fig.tight_layout()

fig3 = OUT / "Figure_03_ROC_Curve.png"

fig.savefig(fig3, dpi=300, bbox_inches="tight")
plt.show()
plt.close(fig)

# ------------------------------------------------------------
# FIGURE 4: Precision-Recall Curve
# ------------------------------------------------------------

precision, recall, _ = precision_recall_curve(
    y_true, y_prob
)

ap = average_precision_score(y_true, y_prob)
prevalence = y_true.mean()

fig, ax = plt.subplots(figsize=(7, 6))

ax.plot(
    recall, precision,
    color="#174A7E",
    linewidth=2.5,
    label=f"Average Precision = {ap:.6f}"
)

ax.axhline(
    prevalence,
    color="gray",
    linestyle="--",
    linewidth=1.5,
    label=f"Class prevalence = {prevalence:.4f}"
)

ax.set_xlabel("Recall")
ax.set_ylabel("Precision")

ax.set_xlim(0, 1)
ax.set_ylim(0, 1.02)

ax.legend(loc="lower left")
ax.grid(alpha=0.20)

fig.tight_layout()

fig4 = OUT / "Figure_04_Precision_Recall_Curve.png"

fig.savefig(fig4, dpi=300, bbox_inches="tight")
plt.show()
plt.close(fig)

# ------------------------------------------------------------
# FIGURE 7: Fraud Probability Density Distribution
# ------------------------------------------------------------

nonfraud = y_prob[y_true == 0]
fraud = y_prob[y_true == 1]

x = np.linspace(0, 1, 500)

# Boundary-corrected kernel density estimation
def bounded_kde(values, grid):

    kde = gaussian_kde(values)

    density = (
        kde(grid)
        + kde(-grid)
        + kde(2 - grid)
    )

    return density / np.trapezoid(density, grid)

density_nonfraud = bounded_kde(nonfraud, x)
density_fraud = bounded_kde(fraud, x)

fig, ax = plt.subplots(figsize=(7.5, 6))

ax.plot(
    x, density_nonfraud,
    color="#2166AC",
    linewidth=2.5,
    label="Non-fraud"
)

ax.fill_between(
    x, density_nonfraud,
    color="#2166AC",
    alpha=0.22
)

ax.plot(
    x, density_fraud,
    color="#D6604D",
    linewidth=2.5,
    label="Fraud-labelled"
)

ax.fill_between(
    x, density_fraud,
    color="#D6604D",
    alpha=0.22
)

ax.set_xlabel("Predicted Fraud Probability")
ax.set_ylabel("Probability Density")

ax.set_xlim(0, 1)
ax.set_ylim(bottom=0)

ax.legend(loc="best")
ax.grid(alpha=0.20)

fig.tight_layout()

fig7 = OUT / "Figure_07_Fraud_Probability_Density.png"

fig.savefig(fig7, dpi=300, bbox_inches="tight")
plt.show()
plt.close(fig)

# ------------------------------------------------------------
# 3. Create ZIP and download directly
# ------------------------------------------------------------

zip_path = Path("/content/Final_Figures_03_04_07.zip")

with zipfile.ZipFile(
    zip_path, "w", zipfile.ZIP_DEFLATED
) as z:

    for path in [fig3, fig4, fig7]:
        z.write(path, arcname=path.name)

with zipfile.ZipFile(zip_path, "r") as z:
    assert z.testzip() is None
    assert len(z.namelist()) == 3

print("\n" + "=" * 60)
print("FINAL FIGURE GENERATION: PASS")
print("=" * 60)

print("Test records:", len(y_true))
print(f"ROC-AUC: {auc:.6f}")
print(f"Average Precision: {ap:.6f}")
print("Figures generated: 3")
print("Resolution: 300 DPI")
print("Model retraining: NO")
print("Google Drive connection: NO")
print("Google Drive saving: NO")
print("ZIP integrity: PASS")

files.download(str(zip_path))
```

```python
from pathlib import Path
from google.colab import files
import zipfile

ROOT = Path("/content")

ZIP_PATH = ROOT / "Narayan_Patil_Current_Work_Backup.zip"

exclude_folders = {
    "drive", "sample_data", ".config",
    ".cache", "__pycache__", ".ipynb_checkpoints"
}

allowed_extensions = {
    ".csv", ".json", ".npz", ".npy",
    ".png", ".jpg", ".pdf", ".py",
    ".pkl", ".joblib", ".xlsx", ".txt",
    ".parquet", ".ipynb", ".zip"
}

saved_files = []

with zipfile.ZipFile(
    ZIP_PATH, "w", zipfile.ZIP_DEFLATED
) as z:

    for path in ROOT.iterdir():

        if path.name in exclude_folders:
            continue

        candidates = (
            path.rglob("*") if path.is_dir()
            else [path]
        )

        for f in candidates:

            if not f.is_file():
                continue

            if f == ZIP_PATH:
                continue

            if any(p in exclude_folders for p in f.parts):
                continue

            if f.suffix.lower() in allowed_extensions:

                z.write(f, arcname=str(f.relative_to(ROOT)))
                saved_files.append(f)

                print("Included:", f.relative_to(ROOT))

# Verify ZIP integrity
with zipfile.ZipFile(ZIP_PATH, "r") as z:
    assert z.testzip() is None

print("\n" + "=" * 60)
print("BACKUP COMPLETED SUCCESSFULLY")
print("=" * 60)
print("Total files:", len(saved_files))
print("ZIP size:", round(ZIP_PATH.stat().st_size / 1024**2, 2), "MB")
print("ZIP integrity: PASS")
print("Google Drive connection: NO")
print("Re-upload: NO")
print("Model retraining: NO")

files.download(str(ZIP_PATH))
```

```python
# ============================================================
# RESUME CELL 02 — MAJOR COMMENT 10: PREPROCESSING AUDIT
# ============================================================

import pandas as pd
import numpy as np
import json
from pathlib import Path

ROOT = Path("/content/Narayan_Patil_Resume")

MASTER = ROOT / "master_dataset (1).csv"
CLEANED = ROOT / "cleaned_dataset.csv"

OUT = ROOT / "Major_Comment_10_Audit"
OUT.mkdir(parents=True, exist_ok=True)

assert MASTER.exists(), "Master dataset missing."
assert CLEANED.exists(), "Cleaned dataset missing."

# ------------------------------------------------------------
# 1. Calculate exact missing-value statistics
# ------------------------------------------------------------

def audit_dataset(path):

    missing = None
    total_rows = 0

    for chunk in pd.read_csv(
        path,
        chunksize=50000,
        low_memory=False
    ):
        counts = chunk.isna().sum()

        missing = (
            counts if missing is None
            else missing.add(counts, fill_value=0)
        )

        total_rows += len(chunk)

    total_columns = len(missing)

    audit = pd.DataFrame({
        "Column": missing.index,
        "Missing_Count": missing.values.astype(int)
    })

    audit["Missing_Percentage"] = (
        audit["Missing_Count"] / total_rows * 100
    )

    audit["Exceeds_70_Percent"] = (
        audit["Missing_Percentage"] > 70
    )

    return total_rows, total_columns, audit


master_rows, master_cols, master_audit = audit_dataset(MASTER)

cleaned_rows, cleaned_cols, cleaned_audit = audit_dataset(CLEANED)

print("=" * 75)
print("DATASET DIMENSION VERIFICATION")
print("=" * 75)

print("Master dataset:", (master_rows, master_cols))
print("Cleaned dataset:", (cleaned_rows, cleaned_cols))

print("Expected master dimensions:",
      (master_rows, master_cols) == (558211, 55))

print("Expected cleaned dimensions:",
      (cleaned_rows, cleaned_cols) == (558211, 38))

print("Record count preserved:", master_rows == cleaned_rows)

# ------------------------------------------------------------
# 2. Identify removed, retained and added columns
# ------------------------------------------------------------

master_names = master_audit["Column"].tolist()
cleaned_names = cleaned_audit["Column"].tolist()

removed_columns = [
    c for c in master_names if c not in cleaned_names
]

retained_columns = [
    c for c in master_names if c in cleaned_names
]

added_columns = [
    c for c in cleaned_names if c not in master_names
]

master_audit["Column_Status"] = np.where(
    master_audit["Column"].isin(removed_columns),
    "Removed",
    "Retained"
)

removed_audit = master_audit[
    master_audit["Column_Status"] == "Removed"
].copy()

removed_audit["Threshold_Assessment"] = np.where(
    removed_audit["Exceeds_70_Percent"],
    "Consistent with >70% missingness rule",
    "Removal reason requires verification"
)

removed_audit = removed_audit.sort_values(
    "Missing_Percentage",
    ascending=False
)

retained_high_missing = master_audit[
    (master_audit["Column_Status"] == "Retained") &
    (master_audit["Exceeds_70_Percent"])
].copy()

print("\n" + "=" * 75)
print("COLUMN REMOVAL AUDIT")
print("=" * 75)

print("Removed columns:", len(removed_columns))
print("Retained columns:", len(retained_columns))
print("Added columns:", len(added_columns))
print("Added column names:", added_columns)

print("\nEXACT REMOVED COLUMN STATISTICS:\n")

print(
    removed_audit.to_string(
        index=False,
        float_format=lambda x: f"{x:.6f}"
    )
)

# ------------------------------------------------------------
# 3. Verify missingness threshold
# ------------------------------------------------------------

removed_above_70 = int(
    removed_audit["Exceeds_70_Percent"].sum()
)

removed_at_or_below_70 = (
    len(removed_columns) - removed_above_70
)

print("\n" + "=" * 75)
print("70% MISSINGNESS THRESHOLD VERIFICATION")
print("=" * 75)

print("Removed columns exceeding 70%:", removed_above_70)

print("Removed columns with 70% or less:",
      removed_at_or_below_70)

print("Retained columns exceeding 70%:",
      len(retained_high_missing))

if not retained_high_missing.empty:
    print(retained_high_missing.to_string(index=False))

# ------------------------------------------------------------
# 4. Overall missing-value statistics
# ------------------------------------------------------------

master_missing = int(
    master_audit["Missing_Count"].sum()
)

cleaned_missing = int(
    cleaned_audit["Missing_Count"].sum()
)

master_missing_pct = (
    master_missing / (master_rows * master_cols) * 100
)

cleaned_missing_pct = (
    cleaned_missing / (cleaned_rows * cleaned_cols) * 100
)

print("\n" + "=" * 75)
print("OVERALL MISSING-VALUE STATISTICS")
print("=" * 75)

print("Master missing values:", master_missing)
print(f"Master missing percentage: {master_missing_pct:.6f}%")

print("Cleaned missing values:", cleaned_missing)
print(f"Cleaned missing percentage: {cleaned_missing_pct:.6f}%")

# ------------------------------------------------------------
# 5. Save audit evidence
# ------------------------------------------------------------

master_audit.to_csv(
    OUT / "01_Master_Missing_Value_Audit.csv",
    index=False
)

cleaned_audit.to_csv(
    OUT / "02_Cleaned_Missing_Value_Audit.csv",
    index=False
)

removed_audit.to_csv(
    OUT / "03_Removed_Columns_Exact_Statistics.csv",
    index=False
)

retained_high_missing.to_csv(
    OUT / "04_Retained_High_Missing_Columns.csv",
    index=False
)

summary = {
    "master_rows": master_rows,
    "master_columns": master_cols,
    "cleaned_rows": cleaned_rows,
    "cleaned_columns": cleaned_cols,
    "master_missing_count": master_missing,
    "master_missing_percentage": master_missing_pct,
    "cleaned_missing_count": cleaned_missing,
    "cleaned_missing_percentage": cleaned_missing_pct,
    "removed_column_count": len(removed_columns),
    "removed_columns": removed_columns,
    "retained_column_count": len(retained_columns),
    "added_columns": added_columns,
    "removed_above_70_percent": removed_above_70,
    "removed_at_or_below_70_percent": removed_at_or_below_70,
    "retained_above_70_percent": (
        retained_high_missing["Column"].tolist()
    ),
    "missing_value_definition": "Pandas default NA parsing",
    "original_removal_rationale_verified": False,
    "model_retraining": False
}

with open(
    OUT / "05_Major_Comment_10_Audit_Report.json",
    "w"
) as f:
    json.dump(summary, f, indent=4)

print("\n" + "=" * 75)
print("SAVED AUDIT FILES")
print("=" * 75)

for f in sorted(OUT.iterdir()):
    print(f.name)

print("\nOutput folder:", OUT)
print("Original datasets modified: NO")
print("Model retraining: NO")
print("Re-upload required: NO")
print("AUDIT COMPUTATION STATUS: PASS")
```

```python
from pathlib import Path
import re

ROOT = Path("/content/Narayan_Patil_Resume")

print("=" * 70)
print("ORIGINAL PREPROCESSING SCRIPT VERIFICATION")
print("=" * 70)

scripts = list(ROOT.rglob("*.py")) + \
          list(ROOT.rglob("*.ipynb"))

print("Available code files:", len(scripts))

patterns = [
    r"dropna",
    r"isnull",
    r"isna",
    r"0\.7\b",
    r"0\.70\b",
    r"threshold",
    r"fillna",
    r"ClaimDuration",
    r"PatientAge",
    r"drop_cols",
    r"missing_percent"
]

matches = 0

for path in scripts:

    content = path.read_text(
        encoding="utf-8",
        errors="replace"
    )

    for i, line in enumerate(content.splitlines(), 1):

        if any(re.search(p, line, re.I) for p in patterns):

            print(f"\nFile: {path.name}")
            print(f"Line {i}: {line[:500]}")

            matches += 1

print("\nTotal matching code lines:", matches)

if matches == 0:
    print("Original preprocessing code not found.")
    print("Dataset-based threshold evidence remains valid.")

print("\nModel retraining: NO")
print("Original files modified: NO")
print("Re-upload: NO")
```
